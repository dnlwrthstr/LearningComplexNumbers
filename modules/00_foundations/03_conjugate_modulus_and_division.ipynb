# Conjugate, Modulus and Division

Notebook 02 could add, subtract and multiply complex numbers, but it postponed **division**. This notebook supplies the two tools that make division work: the **complex conjugate** $\bar z$, which flips the sign of the imaginary part, and the **modulus** $|z|$, the size (distance from $0$) of a complex number. They are tied together by the single identity $z \bar z = |z|^2$.

With them we can divide by any nonzero complex number, compute reciprocals, and prove that size is multiplicative: $|zw| = |z|\,|w|$.

This is notebook 03 of module 00 (**Foundations**) in the **Learning Complex Numbers** series.

## 1. Introduction

How do you compute $\dfrac{1}{3 + 4i}$? The denominator is not a real number, so "divide each part" makes no sense. The trick is the same one used to rationalise $\frac{1}{1 + \sqrt2}$: multiply top and bottom by a cleverly chosen partner that makes the denominator real. For $a + bi$ that partner is $a - bi$, the **conjugate**, because $(a + bi)(a - bi) = a^2 + b^2$.

The conjugate and the modulus appear constantly later on:

- In circuits, the magnitude $|Z|$ of an impedance gives the ratio of voltage to current amplitudes, and power is computed with a conjugate, $V \bar I$.
- In quantum mechanics, the probability of a state is $|\psi|^2 = \psi\bar\psi$.
- In linear algebra, the inner product of complex vectors needs conjugates (module 04).
- Polynomials with real coefficients have roots in conjugate pairs (module 02).

**Prerequisites**

- Cartesian form and complex multiplication — [Cartesian form and arithmetic](02_cartesian_form_and_arithmetic.ipynb) (this module).
- Real and imaginary parts, conjugate-pair roots of quadratics — [Why complex numbers exist](01_why_complex_numbers_exist.ipynb) (this module).

After completing this notebook you will be able to:

- Compute the conjugate and modulus of a complex number and interpret both geometrically.
- Use the rules $\overline{z + w} = \bar z + \bar w$, $\overline{zw} = \bar z\,\bar w$ and $z\bar z = |z|^2$.
- Extract real and imaginary parts with conjugates: $\mathrm{Re}\,z = \frac{z + \bar z}{2}$, $\mathrm{Im}\,z = \frac{z - \bar z}{2i}$.
- Divide complex numbers and find reciprocals by hand.
- Prove and use $|zw| = |z||w|$ and $|z/w| = |z|/|w|$.

## 2. Theory

### 2.1 The complex conjugate

For $z = a + bi$ the **conjugate** is

$$
\bar z = a - bi.
$$

(Some books write $z^*$, especially in physics.) Geometrically, $\bar z$ is the **mirror image of $z$ in the real axis**: same real part, opposite imaginary part.

Properties, all checked by writing $z = a + bi$, $w = c + di$ and expanding:

| Property | Formula |
|---|---|
| Involution | $\overline{\bar z} = z$ |
| Sum | $\overline{z + w} = \bar z + \bar w$ |
| Difference | $\overline{z - w} = \bar z - \bar w$ |
| Product | $\overline{zw} = \bar z\,\bar w$ |
| Quotient | $\overline{z / w} = \bar z / \bar w \quad (w \neq 0)$ |
| Powers | $\overline{z^n} = \bar z^{\,n}$ |
| Real numbers | $\bar z = z \iff z \in \mathbb{R}$ |
| Imaginary numbers | $\bar z = -z \iff z$ is purely imaginary |

**Proof of the product rule.** $zw = (ac - bd) + (ad + bc)i$, so $\overline{zw} = (ac - bd) - (ad + bc)i$. On the other side, $\bar z\,\bar w = (a - bi)(c - di) = (ac - bd) + (-ad - bc)i$. Same thing. $\blacksquare$

In words: **conjugation respects all arithmetic**. You may conjugate a whole expression by conjugating each piece. (This is exactly why non-real roots of real polynomials come in conjugate pairs — module 02.)

### 2.2 Real and imaginary parts via the conjugate

Adding and subtracting $z = a + bi$ and $\bar z = a - bi$:

$$
z + \bar z = 2a, \qquad z - \bar z = 2bi,
$$

so

$$
\mathrm{Re}\, z = \frac{z + \bar z}{2}, \qquad \mathrm{Im}\, z = \frac{z - \bar z}{2i}.
$$

These formulas let you work with real and imaginary parts without ever splitting $z$ into $a$ and $b$.

### 2.3 The modulus

The **modulus** (absolute value) of $z = a + bi$ is

$$
|z| = \sqrt{a^2 + b^2}.
$$

By Pythagoras this is the **distance from $0$ to the point $(a, b)$**. For a real number ($b = 0$) it reduces to the familiar $|a|$.

Basic facts:

- $|z| \ge 0$, and $|z| = 0 \iff z = 0$.
- $|\bar z| = |z|$ and $|-z| = |z|$ (reflections do not change distance from $0$).
- $|\mathrm{Re}\,z| \le |z|$ and $|\mathrm{Im}\,z| \le |z|$ (a leg is no longer than the hypotenuse).

### 2.4 The key identity: $z \bar z = |z|^2$

$$
z\bar z = (a + bi)(a - bi) = a^2 - abi + abi - b^2 i^2 = a^2 + b^2 = |z|^2.
$$

A complex number times its conjugate is always **real and non-negative**. This one line powers everything that follows.

### 2.5 Division

To divide by $w = c + di \neq 0$, multiply numerator and denominator by $\bar w$:

$$
\frac{z}{w} = \frac{z\,\bar w}{w\,\bar w} = \frac{z\,\bar w}{|w|^2}.
$$

The denominator is now the positive real number $c^2 + d^2$, so we just divide each part of $z\bar w$ by it. Explicitly,

$$
\frac{a + bi}{c + di} = \frac{(a + bi)(c - di)}{c^2 + d^2} = \frac{ac + bd}{c^2 + d^2} + \frac{bc - ad}{c^2 + d^2}\, i.
$$

**Reciprocal.** With $z = 1$:

$$
\frac{1}{w} = \frac{\bar w}{|w|^2}, \qquad \text{e.g.}\quad \frac{1}{i} = \frac{-i}{1} = -i.
$$

This is the multiplicative inverse promised by the field axioms in notebook 02. It exists for every $w \neq 0$ precisely because $|w|^2 > 0$ then.

Geometrically, $\frac{1}{w}$ points in the direction of $\bar w$ (mirror image of $w$) with length $\frac{|w|}{|w|^2} = \frac{1}{|w|}$.

### 2.6 The modulus is multiplicative

**Theorem.** $|zw| = |z|\,|w|$ for all $z, w \in \mathbb{C}$.

**Proof.** Using $z\bar z = |z|^2$ and the product rule for conjugates,

$$
|zw|^2 = (zw)\overline{(zw)} = z w \bar z \bar w = (z\bar z)(w\bar w) = |z|^2 |w|^2.
$$

Both sides are non-negative, so taking square roots gives $|zw| = |z||w|$. $\blacksquare$

Consequences:

- $\left|\dfrac{z}{w}\right| = \dfrac{|z|}{|w|}$ for $w \neq 0$ (apply the theorem to $\frac{z}{w} \cdot w = z$).
- $|z^n| = |z|^n$.
- $|iz| = |z|$: multiplying by $i$ does not change size — consistent with it being a rotation.

Written out in coordinates, $|zw|^2 = |z|^2|w|^2$ is the surprising real identity

$$
(ac - bd)^2 + (ad + bc)^2 = (a^2 + b^2)(c^2 + d^2),
$$

the **Brahmagupta–Fibonacci identity**: a product of two sums of two squares is again a sum of two squares.

Note that **$|z + w| \neq |z| + |w|$** in general. Size is multiplicative but not additive; for sums there is only an inequality (the triangle inequality, notebook 04).

## 3. Worked Examples

Let $z = 3 + 4i$ and $w = 1 - 2i$.

### Example 1 — conjugates and moduli

- $\bar z = 3 - 4i$, $\bar w = 1 + 2i$.
- $|z| = \sqrt{9 + 16} = 5$, $|w| = \sqrt{1 + 4} = \sqrt5$.
- $z\bar z = (3 + 4i)(3 - 4i) = 9 + 16 = 25 = |z|^2$. ✓

### Example 2 — conjugate of a product

$zw = (3 + 4i)(1 - 2i) = 3 - 6i + 4i - 8i^2 = 11 - 2i$, so $\overline{zw} = 11 + 2i$.

$\bar z\,\bar w = (3 - 4i)(1 + 2i) = 3 + 6i - 4i - 8i^2 = 11 + 2i$. ✓

### Example 3 — real and imaginary parts via conjugates

$\dfrac{z + \bar z}{2} = \dfrac{6}{2} = 3$ and $\dfrac{z - \bar z}{2i} = \dfrac{8i}{2i} = 4$. ✓

### Example 4 — a reciprocal

$$
\frac{1}{3 + 4i} = \frac{3 - 4i}{(3 + 4i)(3 - 4i)} = \frac{3 - 4i}{25} = \frac{3}{25} - \frac{4}{25}i = 0.12 - 0.16i.
$$

Check: $(3 + 4i)(0.12 - 0.16i) = 0.36 - 0.48i + 0.48i - 0.64i^2 = 0.36 + 0.64 = 1$. ✓

Its modulus is $\sqrt{0.0144 + 0.0256} = 0.2 = \frac{1}{5} = \frac{1}{|z|}$, as predicted.

### Example 5 — a quotient

$$
\frac{z}{w} = \frac{3 + 4i}{1 - 2i} = \frac{(3 + 4i)(1 + 2i)}{(1 - 2i)(1 + 2i)} = \frac{3 + 6i + 4i + 8i^2}{1 + 4} = \frac{-5 + 10i}{5} = -1 + 2i.
$$

Check by multiplying back: $(-1 + 2i)(1 - 2i) = -1 + 2i + 2i - 4i^2 = 3 + 4i$. ✓

Modulus check: $|-1 + 2i| = \sqrt5$ and $\frac{|z|}{|w|} = \frac{5}{\sqrt5} = \sqrt5$. ✓

### Example 6 — modulus of a product without multiplying out

$$
\left|(3 + 4i)^3 (1 - 2i)^2\right| = |3 + 4i|^3\, |1 - 2i|^2 = 5^3 \cdot 5 = 625.
$$

### Example 7 — solving an equation with a conjugate

Find all $z$ with $z + 2\bar z = 6 - i$. Write $z = x + yi$:

$$
(x + yi) + 2(x - yi) = 3x - yi = 6 - i.
$$

Comparing parts: $3x = 6$, $-y = -1$, so $z = 2 + i$.

## 4. Python Examples

### 4.1 Conjugate and modulus in Python

- built-in: `z.conjugate()` and `abs(z)`;
- numpy (works on arrays): `np.conj(z)` and `np.abs(z)`;
- sympy (exact): `sp.conjugate(z)` and `sp.Abs(z)`.

```python
import numpy as np
import sympy as sp

z = 3 + 4j
w = 1 - 2j

print("conj z       =", z.conjugate(), "  numpy:", np.conj(z))
print("|z|          =", abs(z), "  numpy:", np.abs(z))
print("z * conj z   =", z * z.conjugate())
print("Re via conj  =", (z + z.conjugate()) / 2)
print("Im via conj  =", (z - z.conjugate()) / 2j)

z_exact = 3 + 4 * sp.I
print("sympy: conj =", sp.conjugate(z_exact), " |z| =", sp.Abs(z_exact), " |1-2i| =", sp.Abs(1 - 2 * sp.I))
```

### 4.2 Division by the conjugate trick

Implement division using only multiplication, conjugation and division by a **real** number, then compare with Python's `/`.

```python
import numpy as np


def divide(z, w):
    if w == 0:
        raise ZeroDivisionError("cannot divide by 0")
    numerator = z * w.conjugate()          # complex
    denominator = (w * w.conjugate()).real  # |w|^2, a positive real number
    return complex(numerator.real / denominator, numerator.imag / denominator)


z = 3 + 4j
w = 1 - 2j
print("divide(z, w) =", divide(z, w), "  z / w =", z / w)
print("1 / (3+4j)   =", divide(1 + 0j, 3 + 4j))

rng = np.random.default_rng(seed=0)
samples = rng.normal(size=(10_000, 4)) * 5
worst = max(abs(divide(complex(a, b), complex(c, d)) - complex(a, b) / complex(c, d))
            for a, b, c, d in samples)
print(f"largest disagreement with '/' over 10,000 random pairs: {worst:.2e}")
```

### 4.3 Symbolic proofs of the rules

With real symbols, sympy confirms the conjugate rules, the division formula, and $|zw|^2 = |z|^2|w|^2$ (the Brahmagupta–Fibonacci identity) for all inputs.

```python
import sympy as sp

a, b, c, d = sp.symbols("a b c d", real=True)
z = a + b * sp.I
w = c + d * sp.I

print("conj(z+w) - (conj z + conj w) =", sp.simplify(sp.conjugate(z + w) - (sp.conjugate(z) + sp.conjugate(w))))
print("conj(zw)  - conj z * conj w   =", sp.simplify(sp.expand(sp.conjugate(z * w) - sp.conjugate(z) * sp.conjugate(w))))
print("z * conj z                    =", sp.expand(z * sp.conjugate(z)))

quotient = sp.expand(z * sp.conjugate(w)) / (c**2 + d**2)
print("Re(z/w) =", sp.simplify(sp.re(quotient)))
print("Im(z/w) =", sp.simplify(sp.im(quotient)))

product = sp.expand(z * w)
lhs = sp.re(product) ** 2 + sp.im(product) ** 2
rhs = (a**2 + b**2) * (c**2 + d**2)
print("|zw|^2 - |z|^2 |w|^2 =", sp.expand(lhs - rhs))
```

### 4.4 Size is multiplicative, not additive

On random samples, $|zw| = |z||w|$ holds to rounding error, while $|z + w|$ is almost never $|z| + |w|$ — it is only ever **at most** that.

```python
import numpy as np

rng = np.random.default_rng(seed=2)
z = rng.normal(size=1000) + 1j * rng.normal(size=1000)
w = rng.normal(size=1000) + 1j * rng.normal(size=1000)

print("max | |zw| - |z||w| |        :", np.max(np.abs(np.abs(z * w) - np.abs(z) * np.abs(w))))
print("max | |z/w| - |z|/|w| |      :", np.max(np.abs(np.abs(z / w) - np.abs(z) / np.abs(w))))
print("fraction with |z+w| = |z|+|w|:", np.mean(np.isclose(np.abs(z + w), np.abs(z) + np.abs(w))))
print("always |z+w| <= |z|+|w|      :", np.all(np.abs(z + w) <= np.abs(z) + np.abs(w) + 1e-12))
```

### 4.5 A classic pitfall: `abs(z)**2` versus `z**2`

$|z|^2 = z\bar z$ is real; $z^2$ is in general complex. They agree only when $z$ is real.

```python
z = 3 + 4j
print("abs(z)**2         =", abs(z) ** 2)
print("z * z.conjugate() =", z * z.conjugate())
print("z**2              =", z ** 2, "  <- not the same thing")
```

## 5. Visualizations

### 5.1 Conjugate as reflection, modulus as length

The conjugate $\bar z$ is the mirror image of $z$ in the real axis; both lie on the circle of radius $|z|$ around $0$.

```python
import matplotlib.pyplot as plt
import numpy as np

z = 3 + 4j
z_bar = np.conj(z)
radius = abs(z)

fig, ax = plt.subplots(figsize=(6, 6))
circle = radius * np.exp(1j * np.linspace(0, 2 * np.pi, 300))
ax.plot(circle.real, circle.imag, color="lightgray", label=f"$|z| = {radius:g}$")

for point, color, label in [(z, "tab:blue", "$z = 3 + 4i$"), (z_bar, "tab:orange", "$\\bar z = 3 - 4i$")]:
    ax.annotate("", xy=(point.real, point.imag), xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color=color, linewidth=2))
    ax.text(point.real + 0.2, point.imag, label, color=color, fontsize=12, va="center")

ax.plot([z.real, z_bar.real], [z.imag, z_bar.imag], "k--", linewidth=1, label="reflection in the real axis")
ax.plot([0, z.real], [0, 0], color="tab:green", linewidth=3, label="$\\mathrm{Re}\\, z = 3$")
ax.plot([z.real, z.real], [0, z.imag], color="tab:red", linewidth=3, label="$\\mathrm{Im}\\, z = 4$")

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-6, 7)
ax.set_ylim(-6, 6)
ax.set_aspect("equal")
ax.set_title("Conjugate = mirror image, modulus = distance from 0")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower left", fontsize=9)
plt.show()
```

### 5.2 The reciprocal: reflect, then invert the length

$\frac{1}{w} = \frac{\bar w}{|w|^2}$ points along $\bar w$ but has length $\frac{1}{|w|}$. Points outside the unit circle go inside and vice versa; points **on** the unit circle just reflect, because there $\frac{1}{w} = \bar w$.

```python
import matplotlib.pyplot as plt
import numpy as np

points = np.array([2 + 1j, 0.5 + 0.5j, np.exp(1j * 2.2)])
colors = ["tab:blue", "tab:green", "tab:purple"]

fig, ax = plt.subplots(figsize=(6.5, 6.5))
unit_circle = np.exp(1j * np.linspace(0, 2 * np.pi, 300))
ax.plot(unit_circle.real, unit_circle.imag, color="lightgray", label="unit circle $|w| = 1$")

for w, color in zip(points, colors):
    reciprocal = 1 / w
    ax.scatter(w.real, w.imag, s=70, color=color, zorder=3)
    ax.scatter(reciprocal.real, reciprocal.imag, s=70, color=color, marker="s", zorder=3)
    ax.annotate("", xy=(reciprocal.real, reciprocal.imag), xytext=(w.real, w.imag),
                arrowprops=dict(arrowstyle="->", color=color, linestyle="--"))
    ax.text(w.real + 0.08, w.imag + 0.08, f"$w$, $|w| = {abs(w):.2f}$", color=color, fontsize=9)
    ax.text(reciprocal.real + 0.08, reciprocal.imag - 0.18, f"$1/w$, length ${abs(reciprocal):.2f}$",
            color=color, fontsize=9)

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-2, 3.4)
ax.set_ylim(-2.2, 2)
ax.set_aspect("equal")
ax.set_title("Circles: $w$,  squares: $1/w = \\bar w / |w|^2$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower left")
plt.show()
```

### 5.3 Moduli multiply: circles map to circles

Take all points $z$ on the circle $|z| = 1.5$ and multiply each by $w = 1 + i$ (with $|w| = \sqrt2$). Every image lands on the circle of radius $1.5\sqrt2 \approx 2.12$ — the picture of $|zw| = |z||w|$. Adding $w$ instead just shifts the circle, and the moduli of the results vary: no rule like $|z + w| = |z| + |w|$.

```python
import matplotlib.pyplot as plt
import numpy as np

theta = np.linspace(0, 2 * np.pi, 400)
circle = 1.5 * np.exp(1j * theta)
w = 1 + 1j

fig, axes = plt.subplots(1, 2, figsize=(12, 6))

ax = axes[0]
ax.plot(circle.real, circle.imag, label="$|z| = 1.5$")
ax.plot((w * circle).real, (w * circle).imag, label="$wz$: radius $1.5\\sqrt{2}$")
ax.scatter([w.real], [w.imag], color="black", zorder=3)
ax.text(w.real + 0.1, w.imag + 0.1, "$w = 1 + i$")
ax.set_title("Multiplying by $w$: still centred at 0, radius times $|w|$")

ax = axes[1]
ax.plot(circle.real, circle.imag, label="$|z| = 1.5$")
ax.plot((w + circle).real, (w + circle).imag, label="$z + w$: shifted, not rescaled")
ax.scatter([w.real], [w.imag], color="black", zorder=3)
ax.text(w.real + 0.1, w.imag + 0.1, "$w = 1 + i$")
ax.set_title("Adding $w$: a translation")

for ax in axes:
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(-3, 3.2)
    ax.set_ylim(-3, 3.2)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="lower left")
plt.tight_layout()
plt.show()
```

## 7. Solutions

1. $\bar z = -2 - 5i$; $|z| = \sqrt{4 + 25} = \sqrt{29}$; $z\bar z = 29$.

4. $\dfrac{|1 + i|^6}{|3 - 4i|^2} = \dfrac{(\sqrt2)^6}{5^2} = \dfrac{8}{25}$.
5. With $z = x + yi$: $2x + 2yi - i(x - yi) = 2x + 2yi - xi + yi^2 = (2x - y) + (2y - x)i = 3$. So $2x - y = 3$ and $2y - x = 0$, giving $x = 2y$, $3y = 3$, $y = 1$, $x = 2$: $z = 2 + i$. Check: $2(2 + i) - i(2 - i) = 4 + 2i - 2i + i^2 = 3$ ✓.
6. On the unit circle $\frac{1}{w} = \frac{\bar w}{|w|^2} = \bar w$, so $\frac{1}{w}$ is simply the mirror image $0.6 - 0.8i$ — the same length, reflected in the real axis.
7. $\mathrm{Im}\,z = \frac{z - \bar z}{2i}$ is $0$ exactly when $z = \bar z$, i.e. $z$ is real. $\mathrm{Re}\,z = \frac{z + \bar z}{2}$ is $0$ exactly when $z = -\bar z$, i.e. $z$ is purely imaginary.
8. Let $q = z/w$, so $z = qw$. Conjugating, $\bar z = \overline{qw} = \bar q\,\bar w$. Since $w \neq 0$, $\bar w \neq 0$ (same modulus), so $\bar q = \bar z / \bar w$. $\blacksquare$
9. $(2 + i)(3 + 2i) = 6 + 4i + 3i + 2i^2 = 4 + 7i$, so $65 = 4^2 + 7^2$. $(2 + i)(3 - 2i) = 6 - 4i + 3i - 2i^2 = 8 - i$, so $65 = 8^2 + 1^2$.
10. $|z| = 1$ means $z\bar z = 1$, so $\bar z = 1/z$. Let $u = \frac{1 + z}{1 - z}$. Then $\bar u = \frac{1 + \bar z}{1 - \bar z} = \frac{1 + 1/z}{1 - 1/z} = \frac{z + 1}{z - 1} = -u$ (multiplying top and bottom by $z$). By Exercise 7, $u$ is purely imaginary.

## 8. Conclusion

- The **conjugate** $\bar z = a - bi$ is the reflection of $z$ in the real axis, and it respects all arithmetic: $\overline{z \pm w} = \bar z \pm \bar w$, $\overline{zw} = \bar z\bar w$, $\overline{z/w} = \bar z/\bar w$.
- $\mathrm{Re}\,z = \frac{z + \bar z}{2}$ and $\mathrm{Im}\,z = \frac{z - \bar z}{2i}$; $z$ is real iff $z = \bar z$.
- The **modulus** $|z| = \sqrt{a^2 + b^2}$ is the distance from $0$; $z\bar z = |z|^2$ is real and non-negative.
- **Division:** multiply top and bottom by the conjugate of the denominator, $\frac{z}{w} = \frac{z\bar w}{|w|^2}$; the reciprocal is $\frac{1}{w} = \frac{\bar w}{|w|^2}$ — reflect, then invert the length.
- The modulus is **multiplicative**, $|zw| = |z||w|$, but not additive.
- In Python: `z.conjugate()`, `abs(z)`, `np.conj`, `np.abs`, `sp.conjugate`, `sp.Abs`. Remember `abs(z)**2` is not `z**2`.
