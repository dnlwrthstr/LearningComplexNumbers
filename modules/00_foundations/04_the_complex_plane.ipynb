# The Complex Plane

So far complex numbers have been mostly algebra: $a + bi$, FOIL, conjugates. This notebook makes the geometry explicit. Every complex number $a + bi$ is a **point** $(a, b)$ in the plane and also an **arrow** (vector) from the origin to that point. In this picture addition is vector addition, negation and conjugation are reflections, $|z - w|$ is the distance between two points, and simple equations like $|z - c| = r$ describe circles.

The central theorem is the **triangle inequality** $|z + w| \le |z| + |w|$, which we prove algebraically and see geometrically.

This is notebook 04 of module 00 (**Foundations**) in the **Learning Complex Numbers** series.

## 1. Introduction

Wessel (1799), Argand (1806) and Gauss independently realised that complex numbers become far less mysterious once you **draw** them. The plane they used is called the **complex plane** or **Argand diagram**: the horizontal axis carries the real part, the vertical axis the imaginary part.

From here on, almost every statement in this series has a picture:

- addition = translation (this notebook),
- multiplication = rotation and scaling (module 01),
- complex functions = maps of the plane to itself (module 03),
- contour integrals = integrals along curves in the plane (module 06).

Reading equations and inequalities in $z$ as **regions** of the plane — discs, half-planes, annuli — is a skill you will use constantly, for example when describing where a power series converges or where a system is stable.

## 2. Theory

### 2.1 Points and vectors

The map $a + bi \;\longleftrightarrow\; (a, b)$ identifies $\mathbb{C}$ with the plane $\mathbb{R}^2$. A complex number can be viewed in two ways:

- as a **point** — a location in the plane;
- as a **vector** — an arrow with a length and a direction, which can be drawn starting anywhere. Its standard picture starts at the origin.

The real numbers form the horizontal **real axis**, the purely imaginary numbers the vertical **imaginary axis**.

### 2.2 Addition is vector addition

$$
(a + bi) + (c + di) = (a + c) + (b + d)i \quad\longleftrightarrow\quad (a, b) + (c, d) = (a + c, b + d).
$$

Two equivalent pictures:

- **Parallelogram rule**: draw $z$ and $w$ from $0$; $z + w$ is the fourth vertex of the parallelogram on $0, z, w$.
- **Translation (tip-to-tail)**: the map $T_w(z) = z + w$ shifts *every* point of the plane by the vector $w$.

Subtraction: $w - z$ is the vector **from $z$ to $w$**, since $z + (w - z) = w$.

Real scalar multiples $tz$ ($t \in \mathbb{R}$) stretch the arrow; $t < 0$ also reverses it. The points $z + t(w - z)$, $t \in \mathbb{R}$, form the **line** through $z$ and $w$; for $t \in [0, 1]$ the **segment**, with **midpoint** $\frac{z + w}{2}$ at $t = \frac12$.

### 2.3 Reflections

| Map | Coordinates | Geometry |
|---|---|---|
| $z \mapsto \bar z$ | $(a, b) \mapsto (a, -b)$ | reflection in the real axis |
| $z \mapsto -z$ | $(a, b) \mapsto (-a, -b)$ | point reflection through $0$ (rotation by 180°) |
| $z \mapsto -\bar z$ | $(a, b) \mapsto (-a, b)$ | reflection in the imaginary axis |

All three keep the distance to the origin: $|\bar z| = |-z| = |-\bar z| = |z|$.

### 2.4 Distance

The distance between the points $z = a + bi$ and $w = c + di$ is the length of the vector $z - w$:

$$
d(z, w) = |z - w| = \sqrt{(a - c)^2 + (b - d)^2}.
$$

This is exactly the Euclidean distance formula from coordinate geometry. It has the properties of a distance: $|z - w| \ge 0$ with equality only for $z = w$; $|z - w| = |w - z|$; and the triangle inequality below.

Translations preserve distances: $|(z + u) - (w + u)| = |z - w|$. So do the reflections of 2.3, e.g. $|\bar z - \bar w| = |\overline{z - w}| = |z - w|$.

### 2.5 The triangle inequality

**Theorem.** For all $z, w \in \mathbb{C}$,

$$
|z + w| \le |z| + |w|,
$$

with equality if and only if one of them is a non-negative real multiple of the other (they point in the same direction).

**Geometric meaning.** In the triangle with vertices $0$, $z$, $z + w$, the side lengths are $|z|$, $|w|$ and $|z + w|$. The direct route from $0$ to $z + w$ is never longer than the detour via $z$.

**Proof.** Both sides are non-negative, so it suffices to compare squares. Using $|u|^2 = u\bar u$:

$$
\begin{aligned}
|z + w|^2 &= (z + w)(\bar z + \bar w) \\
          &= z\bar z + z\bar w + w\bar z + w\bar w \\
          &= |z|^2 + \big(z\bar w + \overline{z\bar w}\big) + |w|^2 \\
          &= |z|^2 + 2\,\mathrm{Re}(z\bar w) + |w|^2.
\end{aligned}
$$

(We used $w\bar z = \overline{z\bar w}$ and $u + \bar u = 2\,\mathrm{Re}\,u$ from notebook 03.) Since a real part is never larger than the modulus,

$$
\mathrm{Re}(z\bar w) \le |z\bar w| = |z|\,|\bar w| = |z|\,|w|.
$$

Therefore

$$
|z + w|^2 \le |z|^2 + 2|z||w| + |w|^2 = \big(|z| + |w|\big)^2,
$$

and taking square roots gives $|z + w| \le |z| + |w|$. $\blacksquare$

**Equality** holds exactly when $\mathrm{Re}(z\bar w) = |z\bar w|$, i.e. when $z\bar w$ is a non-negative real number. If $w \neq 0$ that means $z = \frac{z\bar w}{|w|^2}\, w$ is a non-negative real multiple of $w$.

**Reverse triangle inequality.** Applying the theorem to $z = (z - w) + w$ gives $|z| \le |z - w| + |w|$, i.e. $|z| - |w| \le |z - w|$; swapping roles gives the same with signs flipped. Together:

$$
\big|\, |z| - |w| \,\big| \le |z - w|.
$$

Combined, for any $z, w$:

$$
\big|\, |z| - |w| \,\big| \;\le\; |z \pm w| \;\le\; |z| + |w|.
$$

The same argument with more terms gives $|z_1 + z_2 + \dots + z_n| \le |z_1| + |z_2| + \dots + |z_n|$ — the straight path is never longer than a polygonal path.

### 2.6 Circles, discs and other regions

Because $|z - c|$ is the distance from $z$ to $c$:

| Set | Description |
|---|---|
| $\lvert z - c\rvert = r$ | circle with centre $c$, radius $r$ |
| $\lvert z - c\rvert < r$ | open disc (interior of the circle) |
| $\lvert z - c\rvert \le r$ | closed disc (interior plus circle) |
| $r_1 < \lvert z - c\rvert < r_2$ | annulus (ring) |
| $\lvert z - a\rvert = \lvert z - b\rvert$ | perpendicular bisector of the segment from $a$ to $b$ |
| $\mathrm{Re}\,z > k$ | half-plane right of the vertical line $x = k$ |
| $\mathrm{Im}\,z \le k$ | half-plane on or below the horizontal line $y = k$ |

To confirm $|z - c| = r$ is the familiar circle, write $z = x + yi$ and $c = c_1 + c_2 i$:

$$
|z - c| = r \iff (x - c_1)^2 + (y - c_2)^2 = r^2.
$$

A useful special case is the **unit circle** $|z| = 1$, which will appear in nearly every notebook from module 01 on.

## 3. Worked Examples

### Example 1 — the parallelogram

$z = 1 + 3i$, $w = 4 + i$. Then $z + w = 5 + 4i$; the parallelogram has vertices $0$, $1 + 3i$, $5 + 4i$, $4 + i$. The vector from $z$ to $w$ is $w - z = 3 - 2i$, which is the other diagonal.

### Example 2 — distance and midpoint

Distance between $z = 1 + 3i$ and $w = 4 - i$:

$$
|z - w| = |-3 + 4i| = \sqrt{9 + 16} = 5.
$$

Midpoint: $\frac{z + w}{2} = \frac{5 + 2i}{2} = 2.5 + i$.

### Example 3 — checking the triangle inequality

$z = 3 + 4i$, $w = -1 + 2i$: $|z| = 5$, $|w| = \sqrt5 \approx 2.236$, $z + w = 2 + 6i$, $|z + w| = \sqrt{40} \approx 6.325$.

$$
\big|5 - 2.236\big| = 2.764 \;\le\; 6.325 \;\le\; 7.236 = 5 + 2.236. \checkmark
$$

Equality case: $z = 3 + 4i$, $w = 6 + 8i = 2z$. $|z + w| = |9 + 12i| = 15 = 5 + 10$.

### Example 4 — identifying a circle

Describe $\{z : |z - 2 + i| = 3\}$. Rewrite $z - 2 + i = z - (2 - i)$: the circle with centre $2 - i$ and radius $3$. In coordinates: $(x - 2)^2 + (y + 1)^2 = 9$.

### Example 5 — completing the square

Describe $\{z : |z|^2 - 4\,\mathrm{Re}\,z + 2\,\mathrm{Im}\,z = 4\}$. With $z = x + yi$:

$$
x^2 + y^2 - 4x + 2y = 4 \iff (x - 2)^2 + (y + 1)^2 = 9,
$$

the same circle as in Example 4.

### Example 6 — a perpendicular bisector

Describe $\{z : |z - 1| = |z - i|\}$: all points equidistant from $1$ and $i$. Squaring,

$$
(x - 1)^2 + y^2 = x^2 + (y - 1)^2 \iff -2x + 1 = -2y + 1 \iff y = x.
$$

The line $\mathrm{Im}\,z = \mathrm{Re}\,z$, the perpendicular bisector of the segment from $1$ to $i$.

### Example 7 — bounding with the triangle inequality

If $|z| \le 2$, how large can $|z^2 + 3z - 1|$ be? By the triangle inequality and $|z^2| = |z|^2$,

$$
|z^2 + 3z - 1| \le |z|^2 + 3|z| + 1 \le 4 + 6 + 1 = 11.
$$

Is $11$ actually reached? Equality would need $z^2$, $3z$ and $-1$ to point in the same direction with $|z| = 2$; but $3z$ and $-1$ point the same way only for $z$ a negative real, $z = -2$, and there $z^2 = 4$ points the other way. So $11$ is a safe **upper bound**, not the maximum. Try $z = 2$: $4 + 6 - 1 = 9$. Section 4 confirms numerically that $9$ is the true maximum.

## 4. Python Examples

### 4.1 Distances and midpoints

```python
import numpy as np

z = 1 + 3j
w = 4 - 1j

print("|z - w|        =", abs(z - w))
print("np.hypot check =", np.hypot(z.real - w.real, z.imag - w.imag))
print("midpoint       =", (z + w) / 2)

# Translations and conjugation preserve distance
shift = -7 + 2.5j
print("after a shift  =", abs((z + shift) - (w + shift)))
print("after conj     =", abs(np.conj(z) - np.conj(w)))
```

### 4.2 Testing the triangle inequality on random numbers

Check $\big||z| - |w|\big| \le |z + w| \le |z| + |w|$ on a million random pairs, and see how close to equality the ratio $\frac{|z + w|}{|z| + |w|}$ gets.

```python
import numpy as np

rng = np.random.default_rng(seed=0)
size = 1_000_000
z = rng.normal(size=size) + 1j * rng.normal(size=size)
w = rng.normal(size=size) + 1j * rng.normal(size=size)

upper_ok = np.all(np.abs(z + w) <= np.abs(z) + np.abs(w) + 1e-12)
lower_ok = np.all(np.abs(np.abs(z) - np.abs(w)) <= np.abs(z - w) + 1e-12)
ratio = np.abs(z + w) / (np.abs(z) + np.abs(w))

print("triangle inequality holds for all pairs:        ", upper_ok)
print("reverse triangle inequality holds for all pairs:", lower_ok)
print(f"ratio |z+w| / (|z|+|w|): min {ratio.min():.4f}, max {ratio.max():.6f}")

# Equality when w is a positive multiple of z
z0 = 3 + 4j
print("equality case:", abs(z0 + 2 * z0), "=", abs(z0) + abs(2 * z0))
```

### 4.3 The key step of the proof, symbolically

The proof rests on $|z + w|^2 = |z|^2 + 2\,\mathrm{Re}(z\bar w) + |w|^2$. Sympy confirms it for all real coordinates.

```python
import sympy as sp

a, b, c, d = sp.symbols("a b c d", real=True)
z = a + b * sp.I
w = c + d * sp.I

lhs = sp.expand((z + w) * sp.conjugate(z + w))
rhs = sp.expand(z * sp.conjugate(z) + 2 * sp.re(z * sp.conjugate(w)) + w * sp.conjugate(w))
print("|z+w|^2                       =", lhs)
print("|z|^2 + 2 Re(z conj w) + |w|^2 =", rhs)
print("difference                    =", sp.simplify(lhs - rhs))
```

### 4.4 Region membership as boolean masks

A region given by an inequality in $z$ is just a boolean test. With numpy we can test a whole grid of points at once — the basis of the region plots in Section 5.

```python
import numpy as np


def in_disc(z, centre, radius):
    return np.abs(z - centre) < radius


points = np.array([0, 2 - 1j, 4 + 1j, 5 - 1j, 2 + 2j])
print("points:          ", points)
print("|z - (2-i)| < 3: ", in_disc(points, 2 - 1j, 3))
print("on the circle:   ", np.isclose(np.abs(points - (2 - 1j)), 3))
print("Re z > 1:        ", points.real > 1)
```

### 4.5 How sharp was the bound in Example 7?

Sample the closed disc $|z| \le 2$ densely and compute the largest value of $|z^2 + 3z - 1|$.

```python
import numpy as np

x = np.linspace(-2, 2, 801)
X, Y = np.meshgrid(x, x)
Z = X + 1j * Y
inside = np.abs(Z) <= 2

values = np.abs(Z**2 + 3 * Z - 1)
values[~inside] = 0
index = np.unravel_index(np.argmax(values), values.shape)

print(f"triangle-inequality bound: 11")
print(f"largest value found:       {values[index]:.4f} at z = {Z[index]:.3f}")
```

The true maximum, $9$ at $z = 2$ on the boundary circle, is smaller than the bound $11$: the triangle inequality gives quick, safe estimates rather than exact answers.

## 5. Visualizations

### 5.1 Addition as parallelogram and as translation

Two small helpers draw the standard Argand-diagram frame and arrows.

```python
import matplotlib.pyplot as plt
import numpy as np


def argand_axes(ax, limits, title):
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(*limits[:2])
    ax.set_ylim(*limits[2:])
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)


def arrow(ax, start, end, color, label=None, offset=(0.15, 0.15)):
    ax.annotate("", xy=(end.real, end.imag), xytext=(start.real, start.imag),
                arrowprops=dict(arrowstyle="->", color=color, linewidth=2))
    if label:
        ax.text(end.real + offset[0], end.imag + offset[1], label, color=color, fontsize=11)


z = 1 + 3j
w = 4 + 1j

fig, axes = plt.subplots(1, 2, figsize=(13, 6))

ax = axes[0]
arrow(ax, 0, z, "tab:blue", "$z$")
arrow(ax, 0, w, "tab:orange", "$w$")
arrow(ax, 0, z + w, "tab:green", "$z + w$")
arrow(ax, z, w, "tab:red", "$w - z$", offset=(-0.6, 0.6))
ax.plot([z.real, (z + w).real, w.real], [z.imag, (z + w).imag, w.imag], "k--", linewidth=1)
argand_axes(ax, (-1, 6, -1, 5), "Parallelogram rule; $w - z$ points from $z$ to $w$")

ax = axes[1]
square = np.array([0, 1, 1 + 1j, 1j, 0])
ax.fill(square.real, square.imag, alpha=0.3, color="tab:blue", label="square $S$")
ax.fill((square + w).real, (square + w).imag, alpha=0.3, color="tab:green", label="$S + w$")
for corner in square[:-1]:
    arrow(ax, corner, corner + w, "tab:orange")
argand_axes(ax, (-1, 6, -1, 5), "Adding $w = 4 + i$ translates every point")
ax.legend(loc="upper left")

plt.tight_layout()
plt.show()
```

### 5.2 Negation and conjugation are reflections

```python
import matplotlib.pyplot as plt
import numpy as np

# A non-symmetric "flag" shape makes the reflections easy to tell apart
flag = np.array([1 + 0.5j, 1 + 2.5j, 2.5 + 2j, 1 + 1.5j])
maps = [
    ("$z$", flag, "tab:blue"),
    ("$\\bar z$ (reflect in real axis)", np.conj(flag), "tab:orange"),
    ("$-z$ (point reflection through 0)", -flag, "tab:green"),
    ("$-\\bar z$ (reflect in imaginary axis)", -np.conj(flag), "tab:red"),
]

fig, ax = plt.subplots(figsize=(7, 7))
for label, shape, color in maps:
    closed = np.append(shape, shape[0])
    ax.fill(closed.real, closed.imag, alpha=0.35, color=color, label=label)
    ax.plot(closed.real, closed.imag, color=color)
    ax.scatter(shape[0].real, shape[0].imag, color=color, s=40, zorder=3)

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-3.5, 3.5)
ax.set_ylim(-3.5, 3.5)
ax.set_aspect("equal")
ax.set_title("Three reflections of the same flag")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left", bbox_to_anchor=(1.02, 1))
plt.show()
```

### 5.3 The triangle inequality

The direct side $|z + w|$ versus the detour $|z| + |w|$. On the right, $w$ rotates around the tip of $z$: the length $|z + w|$ swings between $\big||z| - |w|\big|$ and $|z| + |w|$, hitting the extremes exactly when $w$ points along or against $z$.

```python
import matplotlib.pyplot as plt
import numpy as np

z = 3 + 1j
w = 0.5 + 2j

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))

ax = axes[0]
for start, end, color, label in [(0, z, "tab:blue", f"$|z| = {abs(z):.2f}$"),
                                 (z, z + w, "tab:orange", f"$|w| = {abs(w):.2f}$"),
                                 (0, z + w, "tab:green", f"$|z + w| = {abs(z + w):.2f}$")]:
    ax.annotate("", xy=(end.real, end.imag), xytext=(start.real, start.imag),
                arrowprops=dict(arrowstyle="->", color=color, linewidth=2))
    middle = (start + end) / 2
    ax.text(middle.real + 0.1, middle.imag - 0.35, label, color=color, fontsize=11)
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-0.5, 4.5)
ax.set_ylim(-1, 3.5)
ax.set_aspect("equal")
ax.set_title(f"$|z + w| = {abs(z + w):.2f} \\leq |z| + |w| = {abs(z) + abs(w):.2f}$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)

ax = axes[1]
# Angle between w and z runs over (-pi, pi]; w keeps its length |w|
angle_between = np.linspace(-np.pi, np.pi, 400)
w_rotated = abs(w) * np.exp(1j * (np.angle(z) + angle_between))
ax.plot(angle_between, np.abs(z + w_rotated), label="$|z + w|$")
ax.axhline(abs(z) + abs(w), color="tab:red", linestyle="--", label="$|z| + |w|$")
ax.axhline(abs(abs(z) - abs(w)), color="tab:purple", linestyle="--", label="$||z| - |w||$")
ax.set_xlabel("angle between $w$ and $z$ (radians)")
ax.set_ylabel("length")
ax.set_title("Rotating $w$: the triangle inequality is sharp")
ax.grid(True, alpha=0.3)
ax.legend(loc="center right")

plt.tight_layout()
plt.show()
```

### 5.4 Circles, discs, half-planes and bisectors

The regions from Section 2.6, shaded by testing each grid point. Solid curves are boundaries that belong to the set, dashed ones do not.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-4, 6, 500)
X, Y = np.meshgrid(x, x - 1)
Z = X + 1j * Y
theta = np.linspace(0, 2 * np.pi, 300)

panels = [
    ("$|z - (2 - i)| \\leq 3$", np.abs(Z - (2 - 1j)) <= 3, [(2 - 1j, 3, "-")]),
    ("$1 < |z - 1| < 3$", (np.abs(Z - 1) > 1) & (np.abs(Z - 1) < 3), [(1, 1, "--"), (1, 3, "--")]),
    ("$\\mathrm{Re}\\, z > 1$ and $\\mathrm{Im}\\, z \\leq 2$", (Z.real > 1) & (Z.imag <= 2), []),
    ("$|z - 1| < |z - i|$", np.abs(Z - 1) < np.abs(Z - 1j), []),
]

fig, axes = plt.subplots(2, 2, figsize=(11, 11))
for ax, (title, mask, circles) in zip(axes.flat, panels):
    ax.contourf(X, Y, mask.astype(float), levels=[0.5, 1.5], colors=["tab:blue"], alpha=0.3)
    for centre, radius, style in circles:
        boundary = centre + radius * np.exp(1j * theta)
        ax.plot(boundary.real, boundary.imag, style, color="tab:blue")
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(-4, 6)
    ax.set_ylim(-5, 5)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)

# Boundaries for the half-plane and bisector panels
axes[1, 0].plot([1, 1], [-5, 2], "--", color="tab:blue")   # Re z = 1 is excluded
axes[1, 0].plot([1, 6], [2, 2], "-", color="tab:blue")    # Im z = 2 is included
axes[1, 1].plot([-5, 6], [-5, 6], "--", color="tab:blue")
axes[1, 1].scatter([1, 0], [0, 1], color="black", zorder=3)
axes[1, 1].text(1.15, -0.4, "$1$")
axes[1, 1].text(0.15, 1.15, "$i$")

plt.tight_layout()
plt.show()
```

## 7. Solutions

2. $\frac{(-1 - 4i) + (5 + 2i)}{2} = \frac{4 - 2i}{2} = 2 - i$.
3. (a) $|z - (-3i)| = 2$: the circle with centre $-3i$ and radius $2$. (b) The closed disc with centre $1 + i$ and radius $1$. (c) The open half-plane strictly below the horizontal line $\mathrm{Im}\,z = -1$.
4. $|z + 2i| = |z - (-2i)|$, so the set is all points equidistant from $2$ and $-2i$: the perpendicular bisector of that segment, a line through the midpoint $1 - i$ perpendicular to the direction $-2 - 2i$, i.e. of slope $-1$. Algebra: $(x - 2)^2 + y^2 = x^2 + (y + 2)^2 \iff -4x = 4y \iff y = -x$. It passes through $0$ and $1 - i$ ✓.
5. $|z| = 13$, $|w| = 5$, $z + w = 2 - 8i$, $|z + w| = \sqrt{68} \approx 8.25$, $z - w = 8 - 16i$, $|z - w| = \sqrt{320} \approx 17.89$. Then $8 \le 8.25 \le 18$ and $8 \le 17.89 \le 18$ ✓. Equality needs $w'$ a positive multiple of $z$ with length 5: $w' = \frac{5}{13}(5 - 12i) = \frac{25}{13} - \frac{60}{13}i$; then $|z + w'| = \frac{18}{13} \cdot 13 = 18$.
6. $|z - c|^2 = (z - c)(\bar z - \bar c) = z\bar z - z\bar c - c\bar z + c\bar c = |z|^2 - 2\,\mathrm{Re}(\bar c z) + |c|^2$, because $\bar c z + \overline{\bar c z} = 2\,\mathrm{Re}(\bar c z)$. So the equation says $|z - c|^2 = r^2$, i.e. $|z - c| = r$.
7. $|z^3 - 2z + 5| \le |z|^3 + 2|z| + 5 \le 1 + 2 + 5 = 8$. For the second part, $|z^3 - 2z| \le |z|^3 + 2|z| \le 3$, so by the reverse inequality $|z^3 - 2z + 5| \ge 5 - |z^3 - 2z| \ge 5 - 3 = 2 > 0$. No root lies in the closed unit disc.
8. Expanding as in the proof of 2.5: $|z + w|^2 = |z|^2 + 2\,\mathrm{Re}(z\bar w) + |w|^2$ and $|z - w|^2 = |z|^2 - 2\,\mathrm{Re}(z\bar w) + |w|^2$. Adding cancels the middle terms. Geometrically: in a parallelogram, the sum of the squares of the two diagonals ($|z + w|$, $|z - w|$) equals the sum of the squares of the four sides.
9. $|z_1 + z_2 + z_3| = |(z_1 + z_2) + z_3| \le |z_1 + z_2| + |z_3| \le |z_1| + |z_2| + |z_3|$. Picture: placing the three arrows tip-to-tail gives a polygonal path from $0$ to $z_1 + z_2 + z_3$; the straight segment is never longer.
10. Foci $\pm 1$, so $c = 1$; the sum of distances is $2a = 4$, so the semi-major axis is $a = 2$ (along the real axis) and the semi-minor axis is $b = \sqrt{a^2 - c^2} = \sqrt3$. Equation: $\frac{x^2}{4} + \frac{y^2}{3} = 1$. Check: $z = 2$ gives $1 + 3 = 4$ ✓; $z = i\sqrt3$ gives $2\sqrt{1 + 3} = 4$ ✓.

## 8. Conclusion

- $a + bi \leftrightarrow (a, b)$: complex numbers are points of the plane and vectors from the origin.
- **Addition** is vector addition (parallelogram rule) and adding a fixed $w$ **translates** the plane; $w - z$ is the vector from $z$ to $w$.
- $\bar z$, $-z$, $-\bar z$ are **reflections** in the real axis, through $0$, and in the imaginary axis.
- $|z - w|$ is the **distance** between $z$ and $w$; $\frac{z + w}{2}$ is the midpoint.
- **Triangle inequality** $\big||z| - |w|\big| \le |z \pm w| \le |z| + |w|$, proved from $|z + w|^2 = |z|^2 + 2\,\mathrm{Re}(z\bar w) + |w|^2$ and $\mathrm{Re}\,u \le |u|$; equality iff $z, w$ point the same way.
- $|z - c| = r$ is a circle, $|z - c| < r$ an open disc, $|z - a| = |z - b|$ a perpendicular bisector, $\mathrm{Re}\,z > k$ a half-plane. Sketch first, then compute.
