This notebook makes the arithmetic systematic: how to **add**, **subtract** and **multiply** complex numbers in Cartesian form, why the result is always again of the form $a + bi$, which familiar laws of algebra survive the extension (all the field axioms) and which one does not (there is no ordering $<$ on $\mathbb{C}$).

## 1. Introduction

A number system is only useful if you can calculate in it. For complex numbers the rules turn out to be pleasantly short: treat $i$ like an algebraic symbol, multiply out brackets as usual, and replace every $i^2$ by $-1$. Everything else follows.

These rules are what every piece of software that handles complex numbers implements — from Python's `complex` type to the arithmetic units in signal-processing chips. An AC circuit simulator adds impedances with exactly the addition rule below; a Fourier transform is mostly complex multiplications.

## 2. Theory

### 2.1 Cartesian form

Every complex number has a unique **Cartesian** (or rectangular) form

$$
z = a + bi, \qquad a = \mathrm{Re}\, z \in \mathbb{R},\quad b = \mathrm{Im}\, z \in \mathbb{R}.
$$

Uniqueness is what makes "comparing parts" legitimate: $a + bi = c + di$ exactly when $a = c$ and $b = d$. So a single complex equation is really **two real equations**.

### 2.2 Addition and subtraction: componentwise

Collect real parts and imaginary parts separately:

$$
(a + bi) + (c + di) = (a + c) + (b + d)\,i,
$$

$$
(a + bi) - (c + di) = (a - c) + (b - d)\,i.
$$

This is exactly vector addition of the pairs $(a, b)$ and $(c, d)$ — the geometric meaning is the subject of notebook 04.

### 2.3 Multiplication: FOIL, then $i^2 = -1$

Expand with the distributive law (First, Outer, Inner, Last) and use $i^2 = -1$:

$$
(a + bi)(c + di) = ac + adi + bci + bd\,i^2 = (ac - bd) + (ad + bc)\,i.
$$

Two special cases are worth memorising:

- **Real scalar:** $k(a + bi) = ka + kbi$ — scale both parts.
- **Times $i$:** $i(a + bi) = ai + bi^2 = -b + ai$. The pair $(a, b)$ becomes $(-b, a)$: a quarter turn counter-clockwise.

And one product that removes the imaginary part entirely:

$$
(a + bi)(a - bi) = a^2 - (bi)^2 = a^2 + b^2 \in \mathbb{R}.
$$

This is the key to division (notebook 03).

### 2.4 The field axioms

A **field** is a set with $+$ and $\times$ satisfying, for all $z, w, u$:

| Law | Addition | Multiplication |
|---|---|---|
| Closure | $z + w \in \mathbb{C}$ | $zw \in \mathbb{C}$ |
| Associative | $(z + w) + u = z + (w + u)$ | $(zw)u = z(wu)$ |
| Commutative | $z + w = w + z$ | $zw = wz$ |
| Identity | $z + 0 = z$ | $1 \cdot z = z$ |
| Inverse | $z + (-z) = 0$ | $z \cdot z^{-1} = 1$ for $z \neq 0$ |
| Distributive | $z(w + u) = zw + zu$ | |

$\mathbb{C}$ satisfies all of them:

- Closure is visible in the formulas: sums and products come out as (real) $+$ (real)$\,i$.
- Commutativity of $\times$: swap $(a, b) \leftrightarrow (c, d)$ in $(ac - bd) + (ad + bc)i$ and nothing changes.
- Associativity and distributivity reduce to the corresponding laws for real numbers after expanding (Section 4 checks them symbolically with sympy).
- The additive inverse of $a + bi$ is $-a - bi$; the multiplicative inverse of $a + bi \neq 0$ is $\dfrac{a - bi}{a^2 + b^2}$ (derived in notebook 03).

Consequence: **every algebraic identity you know from real numbers that uses only $+, -, \times, \div$ still holds** — the binomial theorem, $(z - w)(z + w) = z^2 - w^2$, the formula for a geometric series, and so on.

### 2.5 What is lost: no ordering

The reals are an **ordered field**: there is a relation $<$ such that

1. for any $x$, exactly one of $x < 0$, $x = 0$, $x > 0$ holds (trichotomy);
2. if $x > 0$ and $y > 0$ then $x + y > 0$ and $xy > 0$.

From (2) it follows that every nonzero square is positive: if $x > 0$ then $x^2 = x \cdot x > 0$; if $x < 0$ then $-x > 0$ and $x^2 = (-x)(-x) > 0$.

**Claim.** No relation $<$ on $\mathbb{C}$ satisfies (1) and (2).

**Proof.** $i \neq 0$, so its square must be positive: $i^2 = -1 > 0$. But $1 = 1^2$ is also a nonzero square, so $1 > 0$. By (2), $-1 + 1 = 0 > 0$, contradicting trichotomy. $\blacksquare$

So "$z < w$" is meaningless for complex numbers. You *can* compare real-valued quantities derived from them — moduli $|z| < |w|$ (notebook 03), real parts, imaginary parts — but not the numbers themselves. Python enforces this: `1j < 2j` raises a `TypeError`.

### 2.6 Powers of $i$, revisited

Since $i^4 = 1$, for any integer $n$ write $n = 4q + r$ with $r \in \{0, 1, 2, 3\}$ (this works for negative $n$ too, with $r = n \bmod 4$ taken non-negative):

$$
i^n = (i^4)^q\, i^r = i^r = \begin{cases} 1 & r = 0 \\ i & r = 1 \\ -1 & r = 2 \\ -i & r = 3. \end{cases}
$$

Negative powers: $i^{-1} = \frac{1}{i} = -i$ because $i(-i) = -i^2 = 1$; consistent with $-1 \bmod 4 = 3$ and $i^3 = -i$.

A useful consequence: four consecutive powers always sum to zero,

$$
i^n + i^{n+1} + i^{n+2} + i^{n+3} = i^n(1 + i - 1 - i) = 0.
$$

## 3. Worked Examples

Let $z = 3 + 2i$ and $w = 1 - 4i$.

### Example 1 — sum and difference

$$
z + w = (3 + 1) + (2 - 4)i = 4 - 2i,
$$

$$
z - w = (3 - 1) + (2 - (-4))i = 2 + 6i.
$$

Watch the sign in the imaginary part of $z - w$: subtracting $-4i$ adds $4i$.

### Example 2 — product by FOIL

$$
\begin{aligned}
zw &= (3 + 2i)(1 - 4i) \\
   &= 3 \cdot 1 + 3 \cdot (-4i) + 2i \cdot 1 + 2i \cdot (-4i) \\
   &= 3 - 12i + 2i - 8i^2 \\
   &= 3 - 10i + 8 \\
   &= 11 - 10i.
\end{aligned}
$$

Check with the formula $(ac - bd) + (ad + bc)i$, where $a = 3, b = 2, c = 1, d = -4$: $ac - bd = 3 + 8 = 11$, $ad + bc = -12 + 2 = -10$. ✓

### Example 3 — a square

$$
(2 - 3i)^2 = 4 - 12i + 9i^2 = 4 - 12i - 9 = -5 - 12i.
$$

### Example 4 — a product that is real

$$
(3 + 2i)(3 - 2i) = 9 - 6i + 6i - 4i^2 = 9 + 4 = 13.
$$

### Example 5 — solving a linear equation by comparing parts

Find real $x, y$ with $(x + yi)(2 + i) = 1 + 8i$.

Left side: $(2x - y) + (x + 2y)i$. Comparing parts gives two real equations:

$$
2x - y = 1, \qquad x + 2y = 8.
$$

From the first, $y = 2x - 1$; substituting, $x + 4x - 2 = 8$, so $x = 2$ and $y = 3$. Check: $(2 + 3i)(2 + i) = 4 + 2i + 6i + 3i^2 = 1 + 8i$. ✓

### Example 6 — large and negative powers of $i$

- $i^{99}$: $99 = 4 \cdot 24 + 3$, so $i^{99} = i^3 = -i$.
- $i^{-6}$: $-6 = 4 \cdot (-2) + 2$, so $i^{-6} = i^2 = -1$. Check: $i^{-6} = 1 / i^6 = 1/(-1) = -1$. ✓
- $\sum_{n=0}^{101} i^n$: the first $100$ terms form $25$ full cycles summing to $0$, leaving $i^{100} + i^{101} = 1 + i$.

### Example 7 — the binomial theorem still works

$$
(1 + i)^4 = \big((1 + i)^2\big)^2 = (1 + 2i + i^2)^2 = (2i)^2 = -4.
$$

Via the binomial theorem: $1 + 4i + 6i^2 + 4i^3 + i^4 = 1 + 4i - 6 - 4i + 1 = -4$. ✓

## 4. Python Examples

### 4.1 Built-in arithmetic

Python's `complex` type implements exactly the rules of Section 2.

```python
z = 3 + 2j
w = 1 - 4j

print("z + w  =", z + w)
print("z - w  =", z - w)
print("z * w  =", z * w)
print("(2-3j)**2 =", (2 - 3j) ** 2)
print("(3+2j)*(3-2j) =", (3 + 2j) * (3 - 2j))
print("1j * z =", 1j * z, "  (the pair (3, 2) became (-2, 3))")
```

### 4.2 The multiplication rule, derived symbolically

Let sympy do the FOIL expansion with real symbols and collect real and imaginary parts.

```python
import sympy as sp

a, b, c, d = sp.symbols("a b c d", real=True)
product = sp.expand((a + b * sp.I) * (c + d * sp.I))

print("product       =", product)
print("real part     =", sp.re(product))
print("imaginary part=", sp.im(product))
```

### 4.3 Checking the field axioms symbolically

With six real symbols we can verify associativity, commutativity and distributivity for **all** complex numbers at once: each difference simplifies to exactly zero.

```python
import sympy as sp

a, b, c, d, e, f = sp.symbols("a b c d e f", real=True)
z = a + b * sp.I
w = c + d * sp.I
u = e + f * sp.I

checks = {
    "z + w = w + z": (z + w) - (w + z),
    "zw = wz": z * w - w * z,
    "(z + w) + u = z + (w + u)": ((z + w) + u) - (z + (w + u)),
    "(zw)u = z(wu)": (z * w) * u - z * (w * u),
    "z(w + u) = zw + zu": z * (w + u) - (z * w + z * u),
    "z * (a - bi)/(a^2 + b^2) = 1": z * (a - b * sp.I) / (a**2 + b**2) - 1,
}
for law, difference in checks.items():
    print(f"{law:32s} difference simplifies to {sp.simplify(sp.expand(difference))}")
```

### 4.4 No ordering

Python refuses to compare complex numbers with `<`. It still lets you compare derived real quantities.

```python
z = 1j
w = 2j

try:
    print(z < w)
except TypeError as error:
    print("1j < 2j  -> TypeError:", error)

print("abs(z) < abs(w):", abs(z) < abs(w))
print("z.imag < w.imag:", z.imag < w.imag)
```

### 4.5 Powers of $i$, including negative exponents

`sympy` computes $i^n$ exactly; the floating-point `1j ** n` agrees up to rounding.

```python
import numpy as np
import sympy as sp

lookup = {0: 1, 1: 1j, 2: -1, 3: complex(0, -1)}
for n in [99, 100, 2026, -1, -6, -7]:
    by_rule = lookup[n % 4]            # Python's % is non-negative for a positive modulus
    exact = sp.I ** n
    numeric = 1j ** n
    print(f"i^{n:5d}: n mod 4 = {n % 4},  rule -> {by_rule!s:5},  sympy -> {exact!s:3},"
          f"  1j**n close: {np.isclose(numeric, by_rule)}")

print("sum_{n=0}^{101} i^n =", sum(sp.I ** n for n in range(102)))
```

### 4.6 A complex number class from scratch

A complex number is nothing more than a pair of reals with the arithmetic rules of Section 2. Here is a minimal class that knows only that — no `1j` anywhere inside it. Python's operator methods (`__add__`, `__mul__`, …) let us write `z * w` with our own type.

```python
class ComplexNumber:
    # A complex number stored as a pair of real numbers (re, im).

    def __init__(self, re, im=0.0):
        self.re = float(re)
        self.im = float(im)

    def __add__(self, other):
        other = _as_complex_number(other)
        return ComplexNumber(self.re + other.re, self.im + other.im)

    def __sub__(self, other):
        other = _as_complex_number(other)
        return ComplexNumber(self.re - other.re, self.im - other.im)

    def __mul__(self, other):
        other = _as_complex_number(other)
        a, b, c, d = self.re, self.im, other.re, other.im
        return ComplexNumber(a * c - b * d, a * d + b * c)

    def __neg__(self):
        return ComplexNumber(-self.re, -self.im)

    def __pow__(self, n):
        # Non-negative integer powers by repeated multiplication
        result = ComplexNumber(1, 0)
        for _ in range(n):
            result = result * self
        return result

    # Allow 2 + z, 2 * z with a plain number on the left
    __radd__ = __add__
    __rmul__ = __mul__

    def __rsub__(self, other):
        return _as_complex_number(other) - self

    def __repr__(self):
        sign = "+" if self.im >= 0 else "-"
        return f"({self.re + 0.0:g} {sign} {abs(self.im):g}i)"   # + 0.0 turns -0.0 into 0.0


def _as_complex_number(value):
    if isinstance(value, ComplexNumber):
        return value
    return ComplexNumber(value, 0)


z = ComplexNumber(3, 2)
w = ComplexNumber(1, -4)
i = ComplexNumber(0, 1)

print("z + w =", z + w)
print("z - w =", z - w)
print("z * w =", z * w)
print("i * i =", i * i)
print("i**99 =", i ** 99)
print("2 - z =", 2 - z)
```

Now test it against the built-in type on many random inputs. If our rules are right, the results agree up to floating-point rounding.

```python
import numpy as np

# Uses the ComplexNumber class defined in the previous cell.
rng = np.random.default_rng(seed=0)
worst_error = 0.0

for _ in range(10_000):
    a, b, c, d = rng.uniform(-10, 10, size=4)
    mine_sum = ComplexNumber(a, b) + ComplexNumber(c, d)
    mine_product = ComplexNumber(a, b) * ComplexNumber(c, d)
    builtin_sum = complex(a, b) + complex(c, d)
    builtin_product = complex(a, b) * complex(c, d)
    worst_error = max(
        worst_error,
        abs(complex(mine_sum.re, mine_sum.im) - builtin_sum),
        abs(complex(mine_product.re, mine_product.im) - builtin_product),
    )

print(f"largest disagreement over 10,000 random pairs: {worst_error:.2e}")
print("agrees with built-in complex:", worst_error < 1e-12)
```

## 5. Visualizations

Previewing notebook 04: plot $a + bi$ as the point (or arrow) $(a, b)$, real part across, imaginary part up.

### 5.1 Addition is adding arrows

$z + w$ is reached by walking along $z$, then along $w$ — the diagonal of the parallelogram they span.

```python
import matplotlib.pyplot as plt

z = 3 + 2j
w = 1 - 4j
total = z + w


def arrow(ax, start, end, color, label=None):
    ax.annotate("", xy=(end.real, end.imag), xytext=(start.real, start.imag),
                arrowprops=dict(arrowstyle="->", color=color, linewidth=2))
    if label:
        ax.text(end.real + 0.15, end.imag + 0.15, label, color=color, fontsize=12)


fig, ax = plt.subplots(figsize=(6, 6))
arrow(ax, 0, z, "tab:blue", "$z = 3 + 2i$")
arrow(ax, 0, w, "tab:orange", "$w = 1 - 4i$")
arrow(ax, z, total, "tab:orange")
arrow(ax, w, total, "tab:blue")
arrow(ax, 0, total, "tab:green", "$z + w = 4 - 2i$")

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-1, 6)
ax.set_ylim(-5, 3.5)
ax.set_aspect("equal")
ax.set_title("Addition: tip-to-tail")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
plt.show()
```

### 5.2 Multiplying by $i$ turns the whole plane a quarter turn

Take a small "house" shape made of complex numbers and multiply every point by $i$, then by $i$ again. Algebraically $(a, b) \mapsto (-b, a)$; geometrically a 90° rotation about the origin.

```python
import matplotlib.pyplot as plt
import numpy as np

house = np.array([1 + 0.5j, 3 + 0.5j, 3 + 2j, 2 + 3j, 1 + 2j, 1 + 0.5j])

fig, ax = plt.subplots(figsize=(6, 6))
for power, color in zip(range(4), ["tab:blue", "tab:orange", "tab:green", "tab:red"]):
    shape = (1j ** power) * house
    ax.fill(shape.real, shape.imag, alpha=0.3, color=color)
    ax.plot(shape.real, shape.imag, color=color, label=f"$i^{power} \\cdot$ house")

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)
ax.set_aspect("equal")
ax.set_title("Repeated multiplication by $i$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left", bbox_to_anchor=(1.02, 1))
plt.show()
```

### 5.3 Multiplying by a general number: rotate and scale

Multiply the same house by $w = 1 + i$. The shape is rotated (by 45°) and enlarged (by $\sqrt{2} \approx 1.41$) — but not distorted. Module 01 explains exactly why; for now note that the FOIL formula hides a very rigid geometric operation.

```python
import matplotlib.pyplot as plt
import numpy as np

house = np.array([1 + 0.5j, 3 + 0.5j, 3 + 2j, 2 + 3j, 1 + 2j, 1 + 0.5j])
multiplier = 1 + 1j
image = multiplier * house

fig, ax = plt.subplots(figsize=(6, 6))
ax.fill(house.real, house.imag, alpha=0.3, color="tab:blue")
ax.plot(house.real, house.imag, color="tab:blue", label="house")
ax.fill(image.real, image.imag, alpha=0.3, color="tab:purple")
ax.plot(image.real, image.imag, color="tab:purple", label="$(1 + i) \\cdot$ house")

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-3, 5)
ax.set_ylim(-1, 7)
ax.set_aspect("equal")
ax.set_title("Multiplying by $1 + i$: rotate 45°, scale by $\\sqrt{2}$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper right")
plt.show()
```

### 5.4 The cycle of $i^n$

Plotting $i^n$ for $n = -4, \dots, 7$ shows only four distinct points, visited over and over — the picture of "$i^n$ depends only on $n \bmod 4$".

```python
import matplotlib.pyplot as plt
import numpy as np

exponents = np.arange(-4, 8)
powers = np.array([complex(1j ** int(n)) for n in exponents])

fig, ax = plt.subplots(figsize=(6, 6))
circle = np.exp(1j * np.linspace(0, 2 * np.pi, 200))
ax.plot(circle.real, circle.imag, color="lightgray")
ax.scatter(powers.real, powers.imag, s=120, color="tab:blue", zorder=3)

# Gather the exponents that land on each of the four points
for point, offset in [(1, (0.15, 0.1)), (1j, (0.1, 0.15)), (-1, (-0.75, 0.1)), (-1j, (0.1, -0.3))]:
    hits = [int(n) for n, p in zip(exponents, powers) if abs(p - point) < 1e-12]
    ax.text(point.real + offset[0], point.imag + offset[1], "n = " + ", ".join(map(str, hits)), fontsize=9)

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-1.8, 1.8)
ax.set_ylim(-1.8, 1.8)
ax.set_aspect("equal")
ax.set_title("$i^n$ for $n = -4, \\dots, 7$: four points only")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
plt.show()
```

Sums: if $x, y > 0$ each has positive real part, or zero real part and positive imaginary part; in every combination $x + y$ is positive too. ## 8. Conclusion

- Cartesian form $a + bi$ is unique, so one complex equation is two real equations ("compare parts").
- Add and subtract componentwise; multiply by FOIL and $i^2 = -1$: $(a + bi)(c + di) = (ac - bd) + (ad + bc)i$.
- $\mathbb{C}$ is a **field**: all the laws of $+, -, \times, \div$ survive, so every real algebraic identity carries over.
- $\mathbb{C}$ cannot be **ordered** compatibly with its arithmetic, because $i^2 = -1$ would have to be positive. Compare moduli or parts instead.
- $i^n = i^{\,n \bmod 4}$, for negative $n$ too.
- A complex number is just a pair of reals plus these rules; our 40-line `ComplexNumber` class matches Python's `complex` to rounding error.
- Geometrically, addition is tip-to-tail and multiplication by $i$ is a quarter turn.
