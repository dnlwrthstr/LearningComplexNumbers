# Hermitian and Unitary Matrices

Two families of complex matrices behave especially well with the Hermitian inner product. **Hermitian** matrices ($A^H = A$) are the complex version of symmetric matrices: their eigenvalues are real and their eigenvectors orthogonal. **Unitary** matrices ($U^HU = I$) are the complex version of orthogonal matrices: they preserve lengths and angles, and their eigenvalues lie on the unit circle. Together they give the **spectral theorem** $A = U\Lambda U^H$, the backbone of quantum mechanics, signal processing and numerical linear algebra.

This is notebook 04 of module 04 (**Complex linear algebra**) in the **Learning Complex Numbers** series.

## 1. Introduction

A useful way to remember this notebook is a dictionary between numbers and matrices:

| Complex numbers | Matrices |
|---|---|
| conjugate $\bar z$ | conjugate transpose $A^H$ |
| real numbers, $\bar z = z$ | **Hermitian** matrices, $A^H = A$ |
| unit circle, $\bar z z = 1$ | **unitary** matrices, $U^H U = I$ |
| $e^{i\theta}$ with $\theta$ real | $e^{iH}$ with $H$ Hermitian |

The analogy is not loose: the eigenvalues of a Hermitian matrix are real numbers, the eigenvalues of a unitary matrix are on the unit circle, and $e^{iH}$ is always unitary. In quantum mechanics observables are Hermitian (measurements must be real) and time evolution is unitary (probabilities must add up to 1). In signal processing the discrete Fourier transform is unitary (notebook 05).

## 2. Theory

### 2.1 The conjugate transpose

For a complex $m\times n$ matrix $A$, the **conjugate transpose** (also *Hermitian adjoint*, written $A^H$, $A^*$ or $A^\dagger$) is

$$
A^H = \bar A^T, \qquad (A^H)_{jk} = \overline{a_{kj}}.
$$

Rules: $(A^H)^H = A$, $(A + B)^H = A^H + B^H$, $(cA)^H = \bar c A^H$, $(AB)^H = B^H A^H$. For a column vector, $u^H$ is the conjugated row, so $\langle u, v\rangle = u^Hv$ (notebook 03). The defining property of the adjoint is

$$
\langle u, Av\rangle = u^H A v = (A^H u)^H v = \langle A^H u, v\rangle.
$$

"Move a matrix to the other slot of the inner product and it becomes its adjoint." For a real matrix, $A^H = A^T$.

### 2.2 Hermitian matrices

$A$ is **Hermitian** if $A^H = A$: $a_{kj} = \overline{a_{jk}}$. So the diagonal is **real** ($a_{jj} = \overline{a_{jj}}$), and entries mirrored across the diagonal are conjugates:

$$
\begin{pmatrix} 2 & 1 - i \\ 1 + i & 3 \end{pmatrix}, \qquad
\sigma_x = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \quad
\sigma_y = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}, \quad
\sigma_z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}.
$$

(The $\sigma$'s are the Pauli matrices of quantum mechanics.) Real symmetric matrices are exactly the real Hermitian matrices. For Hermitian $A$, $\langle u, Av\rangle = \langle Au, v\rangle$, and in particular $\langle v, Av\rangle = \overline{\langle v, Av\rangle}$ is **real** for every $v$.

**Theorem (real eigenvalues).** If $A^H = A$ and $Av = \lambda v$ with $v \ne 0$, then $\lambda \in \mathbb{R}$.

*Proof.* $\lambda\langle v, v\rangle = \langle v, Av\rangle = \langle Av, v\rangle = \langle\lambda v, v\rangle = \bar\lambda\langle v, v\rangle$. Since $\langle v, v\rangle > 0$, $\lambda = \bar\lambda$. $\square$

**Theorem (orthogonal eigenvectors).** If $Av = \lambda v$ and $Aw = \mu w$ with $\lambda \ne \mu$, then $\langle v, w\rangle = 0$.

*Proof.* $\mu\langle v, w\rangle = \langle v, Aw\rangle = \langle Av, w\rangle = \bar\lambda\langle v, w\rangle = \lambda\langle v, w\rangle$ (as $\lambda$ is real). So $(\mu - \lambda)\langle v, w\rangle = 0$ and $\mu \ne \lambda$. $\square$

Both proofs rely on the conjugation in the inner product: the first slot pulls out $\bar\lambda$, the second $\lambda$.

### 2.3 Unitary matrices

$U$ ($n\times n$) is **unitary** if

$$
U^H U = I \iff U^{-1} = U^H \iff \text{the columns of } U \text{ are orthonormal in } \mathbb{C}^n.
$$

(The $(j,k)$ entry of $U^HU$ is $\langle u_j, u_k\rangle$.) Real unitary matrices are exactly the orthogonal matrices, e.g. rotations $R(\theta)$. Other examples:

$$
\begin{pmatrix} e^{i\alpha} & 0 \\ 0 & e^{i\beta} \end{pmatrix}, \qquad
\frac{1}{\sqrt2}\begin{pmatrix} 1 & i \\ i & 1 \end{pmatrix}, \qquad
\frac{1}{\sqrt2}\begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}, \qquad \sigma_x, \sigma_y, \sigma_z.
$$

**Theorem (isometry).** $\langle Uu, Uv\rangle = (Uu)^H Uv = u^H U^H U v = \langle u, v\rangle$. In particular $\lVert Uv\rVert = \lVert v\rVert$: unitary maps preserve lengths and (complex) angles.

**Theorem (eigenvalues on the unit circle).** If $Uv = \lambda v$, $v \ne 0$, then $\lVert v\rVert = \lVert Uv\rVert = |\lambda|\,\lVert v\rVert$, so $|\lambda| = 1$, i.e. $\lambda = e^{i\theta}$. Also $|\det U|^2 = \det U^H \det U = \det(U^HU) = 1$.

**Theorem (orthogonal eigenvectors).** If $Uv = \lambda v$, $Uw = \mu w$, $\lambda \ne \mu$: $\langle v, w\rangle = \langle Uv, Uw\rangle = \bar\lambda\mu\langle v, w\rangle$. Since $|\lambda| = 1$, $\bar\lambda = 1/\lambda$, so $\bar\lambda\mu = \mu/\lambda \ne 1$ and $\langle v, w\rangle = 0$. $\square$

This explains notebook 02: the rotation $R(\theta)$ is unitary, so its eigenvalues $e^{\pm i\theta}$ lie on the unit circle and its eigenvectors $(1, \mp i)$ are orthogonal (notebook 03, Example 3).

### 2.4 The spectral theorem

**Theorem.** If $A$ is Hermitian, there is a unitary $U$ and a real diagonal $\Lambda$ with

$$
A = U\Lambda U^H = \sum_{k=1}^n \lambda_k\, u_k u_k^H, \qquad \lambda_k \in \mathbb{R},\quad \langle u_j, u_k\rangle = \delta_{jk}.
$$

*Sketch of proof.* Every square complex matrix has a **Schur form** $A = QTQ^H$ with $Q$ unitary and $T$ upper triangular (proved by induction: take one eigenvector, extend to an orthonormal basis, recurse on the rest — possible because over $\mathbb{C}$ an eigenvalue always exists). If $A$ is Hermitian then $T = Q^HAQ$ is Hermitian too; an upper-triangular Hermitian matrix is diagonal with real diagonal. $\square$

Each $u_ku_k^H$ is the orthogonal projection onto the line through $u_k$, so a Hermitian matrix is "stretch by the real number $\lambda_k$ along each of $n$ orthogonal complex lines". More generally a matrix is unitarily diagonalisable iff it is **normal** ($AA^H = A^HA$); unitary matrices are normal, so $U = V\operatorname{diag}(e^{i\theta_1}, \dots, e^{i\theta_n})V^H$ with $V$ unitary.

### 2.5 From Hermitian to unitary: $e^{iH}$

Functions of a Hermitian matrix act on its eigenvalues: $f(A) = U\operatorname{diag}(f(\lambda_k))U^H$. With $f(x) = e^{ix}$:

$$
e^{iH} = U\operatorname{diag}(e^{i\lambda_1}, \dots, e^{i\lambda_n})U^H,
$$

a unitary matrix (product of unitaries, with a diagonal of unit-modulus entries). This is the matrix version of "real $\theta$ gives $e^{i\theta}$ on the unit circle", and $t \mapsto e^{-itH}$ is exactly how quantum states evolve in time.

### 2.6 `eigh` versus `eig`

`np.linalg.eigh(A)` is specialised for Hermitian input. It returns **real** eigenvalues in ascending order and an **orthonormal** (unitary) eigenvector matrix, and is faster and more accurate than `eig`. It reads only one triangle of `A` and silently assumes the rest — so feeding it a non-Hermitian matrix gives a wrong answer without any error. `np.linalg.eig` works for any matrix but returns complex dtype and eigenvectors that are normalised but not necessarily orthogonal to each other when eigenvalues repeat.

## 3. Worked Examples

### Example 1 — Spectral decomposition of a $2\times 2$ Hermitian matrix

$$
A = \begin{pmatrix} 2 & 1 - i \\ 1 + i & 3 \end{pmatrix}.
$$

Characteristic polynomial: $(2 - \lambda)(3 - \lambda) - (1 - i)(1 + i) = \lambda^2 - 5\lambda + 6 - 2 = (\lambda - 1)(\lambda - 4)$. Real eigenvalues $1$ and $4$, as promised.

- $\lambda = 1$: $(A - I)v = 0$, first row $v_1 + (1 - i)v_2 = 0$, so $v = (-1 + i,\ 1)$, $\lVert v\rVert^2 = 2 + 1 = 3$.
- $\lambda = 4$: first row $-2v_1 + (1 - i)v_2 = 0$, so $w = (1 - i,\ 2)$, $\lVert w\rVert^2 = 2 + 4 = 6$.

Orthogonality: $\langle v, w\rangle = \overline{(-1 + i)}(1 - i) + 1\cdot 2 = (-1 - i)(1 - i) + 2 = (-1 + i - i + i^2) + 2 = 0$. ✓ So

$$
U = \begin{pmatrix} \frac{-1 + i}{\sqrt3} & \frac{1 - i}{\sqrt6} \\ \frac{1}{\sqrt3} & \frac{2}{\sqrt6} \end{pmatrix}, \qquad
A = U\begin{pmatrix} 1 & 0 \\ 0 & 4 \end{pmatrix}U^H.
$$

### Example 2 — The quadratic form is real

With the same $A$ and $x = (1, i)$: $Ax = (2 + (1 - i)i,\ (1 + i) + 3i) = (3 + i,\ 1 + 4i)$ and

$$
\langle x, Ax\rangle = 1\cdot(3 + i) + \bar i\,(1 + 4i) = 3 + i - i - 4i^2 = 7.
$$

Real, as it must be. The **Rayleigh quotient** $\langle x, Ax\rangle/\langle x, x\rangle = 7/2 = 3.5$ lies between the smallest and largest eigenvalues, $1$ and $4$.

### Example 3 — $\sigma_y$ is both Hermitian and unitary

$\sigma_y^H = \overline{\begin{pmatrix} 0 & i \\ -i & 0 \end{pmatrix}} = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix} = \sigma_y$ and $\sigma_y^2 = \begin{pmatrix} -i\cdot i & 0 \\ 0 & i\cdot(-i) \end{pmatrix} = I$, so $\sigma_y^H\sigma_y = I$. Its eigenvalues must be real **and** of modulus 1: $\pm 1$. Indeed $\sigma_y(1, i) = (-i\cdot i,\ i) = (1, i)$ and $\sigma_y(1, -i) = (-1,\ i) = -(1, -i)$.

### Example 4 — A unitary matrix and its eigenvalues

$$
U = \frac{1}{\sqrt2}\begin{pmatrix} 1 & i \\ i & 1 \end{pmatrix}, \qquad
U^HU = \frac12\begin{pmatrix} 1 & -i \\ -i & 1 \end{pmatrix}\begin{pmatrix} 1 & i \\ i & 1 \end{pmatrix} = \frac12\begin{pmatrix} 1 + 1 & i - i \\ -i + i & 1 + 1 \end{pmatrix} = I.
$$

$U(1, 1) = \frac{1 + i}{\sqrt2}(1, 1)$ and $U(1, -1) = \frac{1 - i}{\sqrt2}(1, -1)$: eigenvalues $e^{\pm i\pi/4}$ on the unit circle, eigenvectors $(1, 1) \perp (1, -1)$. Note $U = e^{i(\pi/4)\sigma_x}$: the Hermitian $\sigma_x$ has eigenvalues $\pm 1$ on the same eigenvectors, and $e^{\pm i\pi/4}$ are their images under $x \mapsto e^{i\pi x/4}$.

## 4. Python Examples

### 4.1 Conjugate transpose and the adjoint identity

```python
import numpy as np

A = np.array([[1 + 2j, 3], [-1j, 4 - 1j]])
print("A^H = A.conj().T =\n", A.conj().T + (0 + 0j))   # + 0 tidies -0 signs in the display

rng = np.random.default_rng(0)
B = rng.normal(size=(3, 3)) + 1j * rng.normal(size=(3, 3))
u = rng.normal(size=3) + 1j * rng.normal(size=3)
v = rng.normal(size=3) + 1j * rng.normal(size=3)
print("<u, Bv> = <B^H u, v>:", np.isclose(np.vdot(u, B @ v), np.vdot(B.conj().T @ u, v)))
C = rng.normal(size=(3, 3)) + 1j * rng.normal(size=(3, 3))
print("(BC)^H = C^H B^H     :", np.allclose((B @ C).conj().T, C.conj().T @ B.conj().T))
print("(BC)^H = B^H C^H ?   :", np.allclose((B @ C).conj().T, B.conj().T @ C.conj().T), " <- order matters")
```

### 4.2 Example 1 exactly, with sympy

```python
import sympy as sp

A = sp.Matrix([[2, 1 - sp.I], [1 + sp.I, 3]])
print("Hermitian:", A == A.H)
print("char. polynomial:", sp.factor(A.charpoly().as_expr()))

U_columns = []
for eigenvalue, multiplicity, vectors in A.eigenvects():
    vector = vectors[0]
    unit = sp.simplify(vector / sp.sqrt((vector.H * vector)[0, 0]))
    U_columns.append(unit)
    print(f"lambda = {eigenvalue}: unit eigenvector {list(unit)}")

U = sp.Matrix.hstack(*U_columns)
print("<u1, u2> =", sp.simplify((U_columns[0].H * U_columns[1])[0, 0]))
print("U^H U =", sp.simplify(U.H * U))
print("U diag U^H - A =", sp.simplify(U * sp.diag(1, 4) * U.H - A))
```

### 4.3 `eigh` on a random Hermitian matrix

$(B + B^H)/2$ is Hermitian for any square $B$. `eigh` returns real ascending eigenvalues and a unitary eigenvector matrix; `eig` finds the same eigenvalues but with tiny imaginary rounding noise and in no particular order.

```python
import numpy as np

rng = np.random.default_rng(1)
B = rng.normal(size=(4, 4)) + 1j * rng.normal(size=(4, 4))
H = (B + B.conj().T) / 2
print("Hermitian:", np.allclose(H, H.conj().T), "   diagonal:", np.diag(H).round(4))

eigenvalues, U = np.linalg.eigh(H)
print("eigh eigenvalues (real dtype):", eigenvalues.round(6), eigenvalues.dtype)
print("U unitary:", np.allclose(U.conj().T @ U, np.eye(4)))
print("H = U diag U^H:", np.allclose(H, U @ np.diag(eigenvalues) @ U.conj().T))

eig_values = np.linalg.eigvals(H)
print("eig eigenvalues:", eig_values.round(6))
print("largest |imag part| from eig:", np.abs(eig_values.imag).max())

x = rng.normal(size=4) + 1j * rng.normal(size=4)
rayleigh = np.vdot(x, H @ x) / np.vdot(x, x)
print(f"Rayleigh quotient {rayleigh:.4f}: real and between {eigenvalues[0]:.3f} and {eigenvalues[-1]:.3f}")
```

### 4.4 Unitary matrices preserve everything

A random unitary matrix comes from the QR factorisation of a random complex matrix (its $Q$ has orthonormal columns).

```python
import numpy as np

rng = np.random.default_rng(2)
Q, _ = np.linalg.qr(rng.normal(size=(4, 4)) + 1j * rng.normal(size=(4, 4)))
print("Q^H Q = I:", np.allclose(Q.conj().T @ Q, np.eye(4)))

u = rng.normal(size=4) + 1j * rng.normal(size=4)
v = rng.normal(size=4) + 1j * rng.normal(size=4)
print("<Qu, Qv> =", np.round(np.vdot(Q @ u, Q @ v), 10), "   <u, v> =", np.round(np.vdot(u, v), 10))
print("||Qu|| = ||u||:", np.isclose(np.linalg.norm(Q @ u), np.linalg.norm(u)))

eigenvalues, V = np.linalg.eig(Q)
print("|eigenvalues|:", np.abs(eigenvalues).round(12))
print("|det Q| =", abs(np.linalg.det(Q)).round(12))
print("eigenvectors orthonormal (Q is normal):", np.allclose(V.conj().T @ V, np.eye(4)))
```

### 4.5 $e^{iH}$ is unitary

Exponentiate a Hermitian matrix through its spectral decomposition and check the result. For $H = \frac{\pi}{4}\sigma_x$ we recover Example 4's matrix.

```python
import numpy as np


def exp_i_hermitian(H):
    eigenvalues, U = np.linalg.eigh(H)
    return U @ np.diag(np.exp(1j * eigenvalues)) @ U.conj().T


sigma_x = np.array([[0, 1], [1, 0]], dtype=complex)
U = exp_i_hermitian(np.pi / 4 * sigma_x)
print("e^{i pi/4 sigma_x} * sqrt(2) =\n", np.round(U * np.sqrt(2), 12) + (0 + 0j))
print("unitary:", np.allclose(U.conj().T @ U, np.eye(2)))
print("eigenvalues:", np.round(np.linalg.eigvals(U), 6), " e^{i pi/4} =", np.round(np.exp(1j * np.pi / 4), 6))

rng = np.random.default_rng(3)
B = rng.normal(size=(5, 5)) + 1j * rng.normal(size=(5, 5))
H = (B + B.conj().T) / 2
W = exp_i_hermitian(H)
print("random e^{iH} unitary:", np.allclose(W.conj().T @ W, np.eye(5)))
print("angles of its eigenvalues = eigenvalues of H (mod 2 pi):",
      np.allclose(np.sort(np.mod(np.angle(np.linalg.eigvals(W)), 2 * np.pi)),
                  np.sort(np.mod(np.linalg.eigvalsh(H), 2 * np.pi))))
```

### 4.6 The `eigh` trap

`eigh` trusts you. Give it a non-Hermitian matrix and it quietly answers a different question: it reads only the lower triangle and mirrors it, so here it diagonalises $\begin{pmatrix} 1 & -5 \\ -5 & 2 \end{pmatrix}$ instead of the matrix we passed.

```python
import numpy as np

not_hermitian = np.array([[1.0, 5.0], [-5.0, 2.0]])
mirrored_lower = np.array([[1.0, -5.0], [-5.0, 2.0]])
print("true eigenvalues (eig) :", np.linalg.eigvals(not_hermitian).round(4))
print("eigh's answer          :", np.linalg.eigh(not_hermitian)[0].round(4))
print("eigenvalues of mirrored:", np.linalg.eigvalsh(mirrored_lower).round(4), "  <- what eigh actually computed")
print("check before trusting eigh:", np.allclose(not_hermitian, not_hermitian.conj().T))
```

## 5. Visualizations

### 5.1 Where the eigenvalues live

Eigenvalues of 40 random $6\times 6$ matrices of each kind. Hermitian ones fall on the **real axis**, unitary ones on the **unit circle**, generic complex matrices anywhere — the matrix version of "real numbers" and "unit complex numbers".

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(4)
hermitian_values, unitary_values, general_values = [], [], []
for _ in range(40):
    B = (rng.normal(size=(6, 6)) + 1j * rng.normal(size=(6, 6))) / np.sqrt(6)
    hermitian_values.extend(np.linalg.eigvalsh((B + B.conj().T) / 2))
    Q, _ = np.linalg.qr(B)
    unitary_values.extend(np.linalg.eigvals(Q))
    general_values.extend(np.linalg.eigvals(B))
hermitian_values = np.array(hermitian_values)
unitary_values = np.array(unitary_values)
general_values = np.array(general_values)

circle = np.exp(1j * np.linspace(0, 2 * np.pi, 300))
fig, ax = plt.subplots(figsize=(7.5, 7.5))
ax.plot(circle.real, circle.imag, color="lightgray")
ax.scatter(general_values.real, general_values.imag, s=10, color="gray", alpha=0.5, label="general complex $B$")
ax.scatter(unitary_values.real, unitary_values.imag, s=14, color="tab:orange", label="unitary $Q$: $|\\lambda| = 1$")
ax.scatter(hermitian_values.real, np.zeros_like(hermitian_values), s=14, color="tab:blue", marker="|",
           label="Hermitian $(B + B^H)/2$: $\\lambda$ real")
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-2.2, 2.2)
ax.set_ylim(-2.2, 2.2)
ax.set_aspect("equal")
ax.set_title("Eigenvalues: Hermitian on $\\mathbb{R}$, unitary on the unit circle")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left", fontsize=9)
plt.show()
print("max |Im| Hermitian: 0 by construction (eigvalsh);  max ||lambda| - 1| unitary:",
      np.abs(np.abs(unitary_values) - 1).max())
```

### 5.2 The quadratic form $\langle x, Ax\rangle / \lVert x\rVert^2$

For many random unit vectors $x \in \mathbb{C}^2$ plot the complex number $\langle x, Ax\rangle$. For the Hermitian matrix of Example 1 every value is **real** and lies in $[\lambda_{\min}, \lambda_{\max}] = [1, 4]$. For a non-Hermitian matrix with the same eigenvalues the values fill a 2-D region (an ellipse, the *numerical range*) and leave the real axis.

```python
import matplotlib.pyplot as plt
import numpy as np

A_hermitian = np.array([[2, 1 - 1j], [1 + 1j, 3]])
A_triangular = np.array([[1, 2 + 1j], [0, 4]])          # eigenvalues 1 and 4, not Hermitian

rng = np.random.default_rng(6)
x = rng.normal(size=(2, 3000)) + 1j * rng.normal(size=(2, 3000))
x = x / np.linalg.norm(x, axis=0)
form_hermitian = np.sum(np.conj(x) * (A_hermitian @ x), axis=0)
form_triangular = np.sum(np.conj(x) * (A_triangular @ x), axis=0)

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(form_triangular.real, form_triangular.imag, s=3, alpha=0.3, color="gray", label="non-Hermitian $[[1, 2+i], [0, 4]]$")
ax.scatter(form_hermitian.real, form_hermitian.imag, s=3, color="tab:blue", label="Hermitian $[[2, 1-i], [1+i, 3]]$")
ax.plot([1, 4], [0, 0], "o", color="crimson", markersize=8, label="eigenvalues 1 and 4")
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_aspect("equal")
ax.set_title("$\\langle x, Ax\\rangle$ for unit vectors $x$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left", fontsize=9)
plt.show()
print("Hermitian form: max |Im| =", np.abs(form_hermitian.imag).max(), " range:",
      form_hermitian.real.min().round(3), "to", form_hermitian.real.max().round(3))
```

### 5.3 Unitary time evolution $x(t) = e^{-itH}x_0$

Take a $3\times 3$ Hermitian $H$ and evolve a unit vector. Left: the eigenvalues $e^{-it\lambda_k}$ of $e^{-itH}$ run around the unit circle at speeds $\lambda_k$. Right: the squared moduli $|x_j(t)|^2$ of the three components slosh back and forth, but their **sum stays exactly 1** — the norm is preserved. (In quantum mechanics these are the probabilities of the three outcomes; module 07 returns to this.)

```python
import matplotlib.pyplot as plt
import numpy as np

H = np.array([[1.0, 0.5 - 0.5j, 0.0], [0.5 + 0.5j, 0.0, 0.8j], [0.0, -0.8j, -1.0]])
eigenvalues, U = np.linalg.eigh(H)
x0 = np.array([1, 0, 0], dtype=complex)
times = np.linspace(0, 8, 400)

states = np.array([U @ (np.exp(-1j * t * eigenvalues) * (U.conj().T @ x0)) for t in times])
probabilities = np.abs(states) ** 2

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
ax = axes[0]
circle = np.exp(1j * np.linspace(0, 2 * np.pi, 300))
ax.plot(circle.real, circle.imag, color="lightgray")
short = times[times <= 1.5]
for value, color in zip(eigenvalues, ["tab:blue", "tab:orange", "tab:green"]):
    path = np.exp(-1j * short * value)
    ax.plot(path.real, path.imag, color=color, linewidth=2, label=f"$e^{{-it\\lambda}}$, $\\lambda = {value:.2f}$")
    ax.plot(path[-1].real, path[-1].imag, "o", color=color)
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-1.3, 1.3)
ax.set_ylim(-1.3, 1.3)
ax.set_aspect("equal")
ax.set_title("Eigenvalues of $e^{-itH}$ for $0 \\leq t \\leq 1.5$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower left", fontsize=9)

ax = axes[1]
for j in range(3):
    ax.plot(times, probabilities[:, j], label=f"$|x_{j + 1}(t)|^2$")
ax.plot(times, probabilities.sum(axis=1), "k--", label="sum $= \\|x(t)\\|^2$")
ax.set_title("Components change, the norm does not")
ax.set_xlabel("t")
ax.set_ylabel("squared modulus")
ax.set_ylim(-0.05, 1.1)
ax.grid(True, alpha=0.3)
ax.legend(loc="center right", fontsize=9)
plt.tight_layout()
plt.show()
print("max deviation of the norm from 1:", np.abs(probabilities.sum(axis=1) - 1).max())
```

## 7. Solutions

1. $\begin{pmatrix} 1 & i \\ -i & 1 \end{pmatrix}$: Hermitian ($\bar a_{12} = -i = a_{21}$), not unitary (its square is $\begin{pmatrix} 2 & 2i \\ -2i & 2 \end{pmatrix} \ne I$). $\begin{pmatrix} 0 & i \\ i & 0 \end{pmatrix}$: its adjoint is $\begin{pmatrix} 0 & -i \\ -i & 0 \end{pmatrix} = -A$, so not Hermitian, but $A^HA = I$: unitary. $\sigma_x$: both. $\begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}$: neither (not symmetric; first column and second column are not orthogonal).
2. $A^H = \begin{pmatrix} 1 - 2i & i \\ 3 & 4 + i \end{pmatrix}$.
3. $\square = \overline{2 - 5i} = 2 + 5i$. Diagonal entries satisfy $a_{jj} = \overline{a_{jj}}$, so they must be real.
4. $(1 - \lambda)^2 - (i)(-i) = (1 - \lambda)^2 - 1 = 0$, so $\lambda = 0, 2$. $\lambda = 0$: $v_1 + iv_2 = 0$, $u_1 = \frac{1}{\sqrt2}(1, i)$. $\lambda = 2$: $-v_1 + iv_2 = 0$, $u_2 = \frac{1}{\sqrt2}(1, -i)$. They are orthogonal (notebook 03, Example 3). $H = U\operatorname{diag}(0, 2)U^H$ with $U = \frac{1}{\sqrt2}\begin{pmatrix} 1 & 1 \\ i & -i \end{pmatrix}$.
5. Both components of $v$ point along the positive real axis with length $1/\sqrt2$. $U$ turns the first by $+60°$ and the second by $-90°$; lengths are unchanged, so $\lVert Uv\rVert = 1$. $\langle v, Uv\rangle = \frac12(e^{i\pi/3} + e^{-i\pi/2}) = \frac12\left(\frac12 + i\left(\frac{\sqrt3}{2} - 1\right)\right) \approx 0.25 - 0.067i$.
6. $(UV)^H(UV) = V^HU^HUV = V^HV = I$; $(U^{-1})^H U^{-1} = (U^H)^H U^H = UU^H = I$ (a left inverse of a square matrix is a right inverse). $e^{i\pi H/2} = U\operatorname{diag}(e^{0}, e^{i\pi})U^H = u_1u_1^H - u_2u_2^H$. With $u_1u_1^H = \frac12\begin{pmatrix} 1 & -i \\ i & 1 \end{pmatrix}$ and $u_2u_2^H = \frac12\begin{pmatrix} 1 & i \\ -i & 1 \end{pmatrix}$ this is $\begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix} = \sigma_y$, which is unitary (Example 3).
7. Let $B = A - A^H$. Since $\langle v, A^Hv\rangle = \langle Av, v\rangle = \overline{\langle v, Av\rangle}$, we get $\langle v, Bv\rangle = 2i\operatorname{Im}\langle v, Av\rangle = 0$ for every $v$. With $b_{jk} = \langle e_j, Be_k\rangle$: $v = e_j$ gives $b_{jj} = 0$; $v = e_j + e_k$ gives $b_{jk} + b_{kj} = 0$; $v = e_j + ie_k$ gives $ib_{jk} - ib_{kj} = 0$. Together $b_{jk} = 0$, so $B = 0$ and $A = A^H$. (Over $\mathbb{R}$ this fails: a rotation by $90°$ has $x \cdot Jx = 0$ for all real $x$.)
8. $U^2 = U^HU = I$. If $Uv = \lambda v$ then $\lambda$ is real (Hermitian) and $|\lambda| = 1$ (unitary), so $\lambda = \pm 1$. For $P = I - 2ww^H$: $P^H = I - 2ww^H = P$, and $P^2 = I - 4ww^H + 4w(w^Hw)w^H = I$. It reflects $w \mapsto -w$ and fixes everything orthogonal to $w$.
9. $R(\theta)$ is real orthogonal, hence unitary, and for $\sin\theta \ne 0$ its eigenvalues $e^{i\theta} \ne e^{-i\theta}$ are distinct, so by §2.3 the eigenvectors are orthogonal. $\sigma_y$ has eigenvalues $\pm 1$ with eigenvectors $\frac{1}{\sqrt2}(1, i)$ and $\frac{1}{\sqrt2}(1, -i)$ (Example 3), so $e^{-i\theta\sigma_y}$ has the same eigenvectors with eigenvalues $e^{-i\theta}$ on $(1, i)$ and $e^{i\theta}$ on $(1, -i)$ — exactly the eigen-data of $R(\theta)$ from notebook 02. Two diagonalisable matrices with the same eigenvectors and eigenvalues are equal. Equivalently, $-i\sigma_y = \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix} = J$ and $e^{\theta J} = R(\theta)$ (notebook 01).

Numerical checks:

```python
import numpy as np


def is_hermitian(M):
    return np.allclose(M, M.conj().T)


def is_unitary(M):
    return np.allclose(M.conj().T @ M, np.eye(len(M)))


matrices = {"[[1,i],[-i,1]]": np.array([[1, 1j], [-1j, 1]]), "[[0,i],[i,0]]": np.array([[0, 1j], [1j, 0]]),
            "sigma_x": np.array([[0, 1], [1, 0]]), "[[1,1],[0,1]]": np.array([[1, 1], [0, 1]])}
for name, M in matrices.items():
    print(f"Ex 1: {name:15s} Hermitian={is_hermitian(M)!s:5}  unitary={is_unitary(M)}")

H = np.array([[1, 1j], [-1j, 1]])
values, U = np.linalg.eigh(H)
print("Ex 4: eigenvalues", values.round(12) + 0.0, "  H = U diag U^H:", np.allclose(H, U @ np.diag(values) @ U.conj().T))

D = np.diag([np.exp(1j * np.pi / 3), np.exp(-1j * np.pi / 2)])
v = np.array([1, 1]) / np.sqrt(2)
print("Ex 5: ||Uv|| =", round(np.linalg.norm(D @ v), 12), "  <v, Uv> =", np.round(np.vdot(v, D @ v), 4))

E = U @ np.diag(np.exp(1j * np.pi / 2 * values)) @ U.conj().T
print("Ex 6: e^{i pi H/2} =", (np.round(E, 12) + (0 + 0j)).tolist(), " unitary:", is_unitary(E))

w = np.array([1, 2j, -1]) / np.sqrt(6)
P = np.eye(3) - 2 * np.outer(w, w.conj())
print("Ex 8: Householder Hermitian:", is_hermitian(P), " unitary:", is_unitary(P), " eigenvalues:", np.linalg.eigvalsh(P).round(12))

theta = 0.9
sigma_y = np.array([[0, -1j], [1j, 0]])
values_y, V = np.linalg.eigh(-theta * sigma_y)
R = V @ np.diag(np.exp(1j * values_y)) @ V.conj().T
print("Ex 9: e^{-i theta sigma_y} = R(theta):",
      np.allclose(R, [[np.cos(theta), -np.sin(theta)], [np.sin(theta), np.cos(theta)]]))
```

## 8. Conclusion

- The conjugate transpose $A^H = \bar A^T$ is the adjoint: $\langle u, Av\rangle = \langle A^Hu, v\rangle$.
- **Hermitian** ($A^H = A$): real diagonal, $\langle v, Av\rangle$ real, **real eigenvalues**, eigenvectors for distinct eigenvalues **orthogonal** — both proved in one line each from the inner product.
- **Unitary** ($U^HU = I$): orthonormal columns, $U^{-1} = U^H$, preserves inner products and norms, eigenvalues on the **unit circle**, $|\det U| = 1$, orthogonal eigenvectors.
- **Spectral theorem:** $A = U\Lambda U^H = \sum\lambda_k u_ku_k^H$ for Hermitian $A$; normal matrices (including unitary ones) are unitarily diagonalisable too. $e^{iH}$ is unitary.
- In numpy use `np.linalg.eigh` for Hermitian matrices (real sorted eigenvalues, unitary eigenvectors) — and only for Hermitian matrices.
- Dictionary: $A^H \leftrightarrow \bar z$, Hermitian $\leftrightarrow$ real, unitary $\leftrightarrow$ $|z| = 1$.
