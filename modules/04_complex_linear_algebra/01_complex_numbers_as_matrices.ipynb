# Complex Numbers as $2\times 2$ Real Matrices

Multiplying by a fixed complex number $w$ is a **linear map** of the plane, so it has a $2\times 2$ matrix. This notebook shows that the correspondence

$$
a + bi \;\longleftrightarrow\; \begin{pmatrix} a & -b \\ b & a \end{pmatrix}
$$

turns complex arithmetic into matrix arithmetic *exactly*: sums go to sums, products to products, the modulus squared becomes the determinant, and $i$ becomes the 90° rotation matrix whose square is $-I$. The complex numbers are, quite literally, a family of rotation-scaling matrices.

This is notebook 01 of module 04 (**Complex linear algebra**) in the **Learning Complex Numbers** series.

## 1. Introduction

Module 01 showed that multiplying by $w = re^{i\varphi}$ rotates by $\varphi$ and scales by $r$. Linear algebra has its own language for "rotate and scale": a matrix. Putting the two side by side answers questions that are otherwise a little mysterious:

- *Is $i$ really a number?* It is at least as real as the matrix $\begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix}$, which visibly squares to $-I$.
- *Why is complex multiplication commutative when matrix multiplication is not?* Because these particular matrices commute with each other.
- *Why does $|zw| = |z||w|$?* Because $\det(AB) = \det A \det B$.

The matrix picture is also the doorway to the rest of this module: complex eigenvalues of real matrices (notebook 02) are exactly these rotation-scaling blocks hiding inside bigger matrices.

## 2. Theory

### 2.1 Multiplication by $w$ is a linear map of $\mathbb{R}^2$

Identify $z = x + iy$ with the column vector $\begin{pmatrix} x \\ y \end{pmatrix}$. Fix $w = a + bi$ and look at the map $T_w(z) = wz$. It is linear over $\mathbb{R}$:

$$
w(z_1 + z_2) = wz_1 + wz_2, \qquad w(cz) = c(wz) \quad (c \in \mathbb{R}).
$$

A linear map is determined by what it does to the basis vectors. The basis of the plane is $1 \leftrightarrow \begin{pmatrix} 1 \\ 0 \end{pmatrix}$ and $i \leftrightarrow \begin{pmatrix} 0 \\ 1 \end{pmatrix}$:

$$
w \cdot 1 = a + bi \;\leftrightarrow\; \begin{pmatrix} a \\ b \end{pmatrix}, \qquad
w \cdot i = -b + ai \;\leftrightarrow\; \begin{pmatrix} -b \\ a \end{pmatrix}.
$$

These are the columns of the matrix:

$$
M(w) = \begin{pmatrix} a & -b \\ b & a \end{pmatrix}, \qquad
M(w)\begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} ax - by \\ bx + ay \end{pmatrix},
$$

which is exactly the Cartesian product rule $(a + bi)(x + iy) = (ax - by) + i(bx + ay)$.

### 2.2 The dictionary: $1 \mapsto I$, $i \mapsto J$

Split the matrix into its two parts:

$$
M(a + bi) = a\begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix} + b\begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix} = aI + bJ.
$$

So $1$ corresponds to the identity $I$ and $i$ corresponds to

$$
J = M(i) = \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix}, \qquad J^2 = \begin{pmatrix} -1 & 0 \\ 0 & -1 \end{pmatrix} = -I.
$$

$J$ is the rotation by $90°$ counter-clockwise; doing it twice is a half turn, i.e. $-I$. The defining equation $i^2 = -1$ becomes a fact about rotations that you can check with a pencil.

### 2.3 The map respects $+$ and $\times$

For $z = a + bi$ and $w = c + di$:

**Sums.** $M(z) + M(w) = (a + c)I + (b + d)J = M(z + w)$, because $M$ is linear in $(a, b)$.

**Products.** Using only $IJ = JI = J$ and $J^2 = -I$:

$$
M(z)M(w) = (aI + bJ)(cI + dJ) = acI + (ad + bc)J + bdJ^2 = (ac - bd)I + (ad + bc)J = M(zw).
$$

Since $zw = wz$, it follows that $M(z)M(w) = M(w)M(z)$: these matrices **commute**, unlike matrices in general. Other entries of the dictionary:

| Complex | Matrix |
|---|---|
| $z + w$, $zw$ | $M(z) + M(w)$, $M(z)M(w)$ |
| $0$, $1$, $i$ | $0$, $I$, $J$ |
| $\bar z = a - bi$ | $M(\bar z) = M(z)^T$ |
| $\lvert z\rvert^2 = a^2 + b^2$ | $\det M(z) = a^2 + b^2$ |
| $z^{-1} = \bar z / \lvert z\rvert^2$ | $M(z)^{-1} = M(z)^T / \det M(z)$ |
| $z\bar z = \lvert z\rvert^2$ | $M(z)M(z)^T = \lvert z\rvert^2 I$ |

In algebra language: $M$ is an injective ring homomorphism $\mathbb{C} \to \mathbb{R}^{2\times 2}$, and its image is a copy of $\mathbb{C}$ living inside the $2\times 2$ real matrices.

### 2.4 Determinant $= |z|^2$, so $|zw| = |z||w|$ for free

$$
\det M(a + bi) = a \cdot a - (-b) \cdot b = a^2 + b^2 = |z|^2.
$$

The determinant is the area scale factor of a linear map. Multiplying by $z$ scales lengths by $|z|$, hence areas by $|z|^2$ — consistent. And because $\det$ is multiplicative,

$$
|zw|^2 = \det M(zw) = \det\big(M(z)M(w)\big) = \det M(z)\det M(w) = |z|^2|w|^2.
$$

### 2.5 Polar form: rotation-scaling matrices

Write $z = re^{i\varphi} = r\cos\varphi + i\,r\sin\varphi$. Then

$$
M(z) = r\begin{pmatrix} \cos\varphi & -\sin\varphi \\ \sin\varphi & \cos\varphi \end{pmatrix} = r\,R(\varphi),
$$

a rotation by $\varphi$ followed by a scaling by $r$ (in either order, they commute). Every nonzero $M(z)$ is such a **rotation-scaling matrix**, and every rotation-scaling matrix is some $M(z)$. Multiplying $M(z_1)M(z_2) = r_1 r_2 R(\varphi_1 + \varphi_2)$ reproduces "multiply the moduli, add the arguments". Even Euler's formula has a matrix twin:

$$
e^{\varphi J} = \sum_{k \ge 0} \frac{(\varphi J)^k}{k!} = \cos\varphi\, I + \sin\varphi\, J = R(\varphi),
$$

by the same even/odd split of the series used for $e^{i\varphi}$, with $J^2 = -I$ playing the role of $i^2 = -1$.

### 2.6 Which real matrices are complex numbers?

A general real $2\times 2$ matrix has four free entries; $M(z)$ has only two. Exactly:

$$
A = \begin{pmatrix} p & q \\ r & s \end{pmatrix} \text{ is some } M(z) \iff p = s \text{ and } r = -q \iff AJ = JA.
$$

So reflections $\begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$, shears $\begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}$ and non-uniform stretches $\begin{pmatrix} 2 & 0 \\ 0 & 1 \end{pmatrix}$ are **not** multiplication by any complex number. Geometrically: multiplication by $z$ preserves angles and orientation (it is a *similarity*), and these maps do not. (Conjugation $z \mapsto \bar z$ is the reflection — real-linear but not "multiply by a constant".)

## 3. Worked Examples

### Example 1 — The matrix of $3 + 4i$

$$
M(3 + 4i) = \begin{pmatrix} 3 & -4 \\ 4 & 3 \end{pmatrix}, \qquad \det = 9 + 16 = 25 = |3 + 4i|^2.
$$

Apply it to $z = 1 + 2i$: $\begin{pmatrix} 3 & -4 \\ 4 & 3 \end{pmatrix}\begin{pmatrix} 1 \\ 2 \end{pmatrix} = \begin{pmatrix} -5 \\ 10 \end{pmatrix}$, and indeed $(3 + 4i)(1 + 2i) = 3 + 6i + 4i + 8i^2 = -5 + 10i$. ✓

### Example 2 — A product, both ways

$z = 1 + 2i$, $w = 3 - i$. Complex: $zw = 3 - i + 6i - 2i^2 = 5 + 5i$. Matrices:

$$
\begin{pmatrix} 1 & -2 \\ 2 & 1 \end{pmatrix}\begin{pmatrix} 3 & 1 \\ -1 & 3 \end{pmatrix}
= \begin{pmatrix} 3 + 2 & 1 - 6 \\ 6 - 1 & 2 + 3 \end{pmatrix}
= \begin{pmatrix} 5 & -5 \\ 5 & 5 \end{pmatrix} = M(5 + 5i). \checkmark
$$

### Example 3 — Inverse

For $z = 3 + 4i$: $M(z)^{-1} = \frac{1}{25}\begin{pmatrix} 3 & 4 \\ -4 & 3 \end{pmatrix} = M\!\left(\frac{3 - 4i}{25}\right)$, and $\frac{1}{3 + 4i} = \frac{3 - 4i}{25}$. ✓ The matrix inverse formula "swap the diagonal, negate the off-diagonal, divide by the determinant" *is* the rule "conjugate and divide by $|z|^2$".

### Example 4 — Polar form of $1 + i$

$1 + i = \sqrt2\, e^{i\pi/4}$, so

$$
M(1 + i) = \begin{pmatrix} 1 & -1 \\ 1 & 1 \end{pmatrix} = \sqrt2 \begin{pmatrix} \tfrac{1}{\sqrt2} & -\tfrac{1}{\sqrt2} \\ \tfrac{1}{\sqrt2} & \tfrac{1}{\sqrt2} \end{pmatrix} = \sqrt2\, R(\pi/4).
$$

Then $M(1 + i)^8 = (\sqrt2)^8 R(2\pi) = 16I$, matching $(1 + i)^8 = 16$.

### Example 5 — Powers of $J$

$J^2 = -I$, $J^3 = -J$, $J^4 = I$: the same period-4 cycle as $1, i, -1, -i$. Applied to $\begin{pmatrix} 1 \\ 0 \end{pmatrix}$ they give $\begin{pmatrix} 1 \\ 0 \end{pmatrix}, \begin{pmatrix} 0 \\ 1 \end{pmatrix}, \begin{pmatrix} -1 \\ 0 \end{pmatrix}, \begin{pmatrix} 0 \\ -1 \end{pmatrix}$ — quarter turns.

### Example 6 — Is $\begin{pmatrix} 2 & 3 \\ -3 & 2 \end{pmatrix}$ a complex number? Is $\begin{pmatrix} 2 & 3 \\ 3 & 2 \end{pmatrix}$?

The first has equal diagonal entries and $r = -q$ ($-3 = -3$): it is $M(2 - 3i)$. The second has $r = q$, not $-q$: it is a symmetric matrix that stretches along the diagonals (eigenvalues $5$ and $-1$) and is not multiplication by any complex number.

## 4. Python Examples

### 4.1 Converting both ways

```python
import numpy as np


def to_matrix(z):
    # The 2x2 real matrix of 'multiply by z'.
    z = complex(z)
    return np.array([[z.real, -z.imag],
                     [z.imag, z.real]]) + 0.0   # + 0.0 turns -0.0 into 0.0 for display


def from_matrix(matrix, tolerance=1e-12):
    # Inverse map; refuses matrices that are not of the form [[a, -b], [b, a]].
    a, minus_b = matrix[0]
    b, a_again = matrix[1]
    if abs(a - a_again) > tolerance or abs(minus_b + b) > tolerance:
        raise ValueError("not a rotation-scaling matrix")
    return complex(a, b)


print("M(3+4i) =\n", to_matrix(3 + 4j))
print("back    =", from_matrix(to_matrix(3 + 4j)))
print("M(i)    =\n", to_matrix(1j))
print("M(i)^2  =\n", to_matrix(1j) @ to_matrix(1j))
```

### 4.2 Checking the dictionary on random numbers

Every row of the table in §2.3 becomes an `np.allclose` test.

```python
import numpy as np


def to_matrix(z):
    z = complex(z)
    return np.array([[z.real, -z.imag], [z.imag, z.real]]) + 0.0


rng = np.random.default_rng(0)
z = complex(*rng.normal(size=2))
w = complex(*rng.normal(size=2))
print("z =", z, "  w =", w)

checks = {
    "M(z+w) = M(z)+M(w)": np.allclose(to_matrix(z + w), to_matrix(z) + to_matrix(w)),
    "M(zw) = M(z)M(w)": np.allclose(to_matrix(z * w), to_matrix(z) @ to_matrix(w)),
    "M(z)M(w) = M(w)M(z)": np.allclose(to_matrix(z) @ to_matrix(w), to_matrix(w) @ to_matrix(z)),
    "M(conj z) = M(z)^T": np.allclose(to_matrix(np.conj(z)), to_matrix(z).T),
    "det M(z) = |z|^2": np.isclose(np.linalg.det(to_matrix(z)), abs(z) ** 2),
    "M(1/z) = M(z)^-1": np.allclose(to_matrix(1 / z), np.linalg.inv(to_matrix(z))),
    "M(z)M(z)^T = |z|^2 I": np.allclose(to_matrix(z) @ to_matrix(z).T, abs(z) ** 2 * np.eye(2)),
}
for name, passed in checks.items():
    print(f"{name:24s} {passed}")
```

### 4.3 A symbolic proof with sympy

Random tests are reassuring; sympy can check the identities for *all* $a, b, c, d$ at once.

```python
import sympy as sp

a, b, c, d = sp.symbols("a b c d", real=True)


def M(re, im):
    return sp.Matrix([[re, -im], [im, re]])


z = a + sp.I * b
w = c + sp.I * d
product = sp.expand(z * w)

print("M(z)M(w) - M(zw) =", sp.simplify(M(a, b) * M(c, d) - M(sp.re(product), sp.im(product))))
print("M(z)M(w) - M(w)M(z) =", sp.simplify(M(a, b) * M(c, d) - M(c, d) * M(a, b)))
print("det M(z) =", M(a, b).det())
print("inverse  =", M(a, b).inv().applyfunc(sp.simplify))
```

### 4.4 Euler's formula for matrices: $e^{\varphi J} = R(\varphi)$

sympy can exponentiate a matrix exactly. The result is the rotation matrix, the matrix twin of $e^{i\varphi} = \cos\varphi + i\sin\varphi$.

```python
import sympy as sp

phi = sp.symbols("varphi", real=True)
J = sp.Matrix([[0, -1], [1, 0]])

print("J^2 =", J**2)
print("exp(phi J) =", sp.simplify((phi * J).exp()))
```

### 4.5 Rotation-scaling decomposition

From a matrix $\begin{pmatrix} a & -b \\ b & a \end{pmatrix}$ read off $r = \sqrt{\det}$ and $\varphi = \operatorname{atan2}(b, a)$ — the same numbers `abs` and `cmath.phase` give for $a + bi$.

```python
import cmath

import numpy as np


def rotation(angle):
    return np.array([[np.cos(angle), -np.sin(angle)], [np.sin(angle), np.cos(angle)]])


matrix = np.array([[1.0, -1.0], [1.0, 1.0]])           # M(1 + i)
scale = np.sqrt(np.linalg.det(matrix))
angle = np.arctan2(matrix[1, 0], matrix[0, 0])
print(f"from the matrix:  r = {scale:.6f}, phi = {angle:.6f}")
print(f"from 1 + i:       r = {abs(1 + 1j):.6f}, phi = {cmath.phase(1 + 1j):.6f}")
print("r R(phi) reproduces the matrix:", np.allclose(scale * rotation(angle), matrix))
print("M(1+i)^8 =\n", np.linalg.matrix_power(matrix, 8))
```

### 4.6 Which matrices commute with $J$?

The test $AJ = JA$ separates complex numbers from other linear maps.

```python
import numpy as np

J = np.array([[0, -1], [1, 0]])
candidates = {
    "M(2-3i)  [[2,3],[-3,2]]": np.array([[2, 3], [-3, 2]]),
    "symmetric [[2,3],[3,2]]": np.array([[2, 3], [3, 2]]),
    "reflection diag(1,-1)": np.array([[1, 0], [0, -1]]),
    "shear [[1,1],[0,1]]": np.array([[1, 1], [0, 1]]),
    "stretch diag(2,1)": np.array([[2, 0], [0, 1]]),
}
for name, matrix in candidates.items():
    print(f"{name:26s} commutes with J: {np.array_equal(matrix @ J, J @ matrix)}")
```

## 5. Visualizations

### 5.1 Same picture, two computations

Take a small "house" shape in the plane. Transform it once by **complex multiplication** $z \mapsto wz$ and once by the **matrix** $M(w)$ acting on coordinate vectors. The two images coincide point for point: a rotation by $\arg w$ and a scaling by $|w|$.

```python
import matplotlib.pyplot as plt
import numpy as np

house = np.array([0, 1, 1 + 1j, 0.5 + 1.5j, 1j, 0])   # floor, right wall, roof, left wall
w = 1 + 1j
M_w = np.array([[w.real, -w.imag], [w.imag, w.real]])

image_complex = w * house
image_matrix = M_w @ np.vstack([house.real, house.imag])

fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(house.real, house.imag, "o-", color="tab:blue", label="original")
ax.plot(image_complex.real, image_complex.imag, "-", color="tab:orange", linewidth=4, alpha=0.5,
        label="$w z$ (complex multiply)")
ax.plot(image_matrix[0], image_matrix[1], "k--", label="$M(w)$ applied to vectors")
ax.annotate("", xy=(w.real, w.imag), xytext=(0, 0), arrowprops=dict(arrowstyle="->", color="crimson", linewidth=2))
ax.text(w.real + 0.05, w.imag - 0.15, "$w = 1 + i$", color="crimson", fontsize=12)

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-2.5, 2)
ax.set_ylim(-0.5, 3)
ax.set_aspect("equal")
ax.set_title("Multiplying by $1+i$: rotate by $45^\\circ$, scale by $\\sqrt{2}$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left")
plt.show()
```

### 5.2 $J$ is a quarter turn, $J^2 = -I$ a half turn

The unit square's images under $I, J, J^2, J^3$. The corner marked with a dot shows the orientation is preserved — these are rotations, not reflections.

```python
import matplotlib.pyplot as plt
import numpy as np

J = np.array([[0, -1], [1, 0]])
square = np.array([[0, 1, 1, 0, 0], [0, 0, 1, 1, 0]], dtype=float)
marker = np.array([[1.0], [0.0]])
labels = ["$I$", "$J$", "$J^2 = -I$", "$J^3 = -J$"]
colors = ["tab:blue", "tab:orange", "tab:green", "tab:red"]

fig, ax = plt.subplots(figsize=(6, 6))
power = np.eye(2)
for label, color in zip(labels, colors):
    image = power @ square
    corner = power @ marker
    ax.fill(image[0], image[1], color=color, alpha=0.3, label=label)
    ax.plot(corner[0], corner[1], "o", color=color, markersize=9)
    power = J @ power

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect("equal")
ax.set_title("Powers of $J = M(i)$ acting on the unit square")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left")
plt.show()
```

### 5.3 Complex numbers versus other linear maps

A grid of small squares under four matrices. Only $M(w)$ (left) keeps squares square — angles preserved, area scaled by $\det = |w|^2$. The symmetric matrix, the shear and the reflection are perfectly good linear maps, but not multiplication by a complex number.

```python
import matplotlib.pyplot as plt
import numpy as np

matrices = {
    "$M(1.2+0.5i)$": np.array([[1.2, -0.5], [0.5, 1.2]]),
    "symmetric $[[2,3],[3,2]]/3$": np.array([[2, 3], [3, 2]]) / 3,
    "shear $[[1,1],[0,1]]$": np.array([[1, 1], [0, 1]]),
    "reflection diag$(1,-1)$": np.array([[1, 0], [0, -1]]),
}
grid_values = np.linspace(0, 1, 5)

fig, axes = plt.subplots(1, 4, figsize=(16, 4.5))
for ax, (title, matrix) in zip(axes, matrices.items()):
    for value in grid_values:
        for line in (np.array([[value, value], [0, 1]]), np.array([[0, 1], [value, value]])):
            image = matrix @ line
            ax.plot(*line, color="lightgray", linewidth=1)
            ax.plot(*image, color="tab:blue", linewidth=1.5)
    corner = matrix @ np.array([1, 0])
    ax.plot(corner[0], corner[1], "o", color="crimson")
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(-1, 2.2)
    ax.set_ylim(-1.2, 2.2)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.4 Powers of a rotation-scaling matrix

$M(w)^k \begin{pmatrix} 1 \\ 0 \end{pmatrix}$ is the vector of $w^k$. With $|w| = 0.95 < 1$ the points spiral into the origin; with $|w| = 1$ they stay on the unit circle. This is a first glimpse of notebook 02's discrete dynamical systems.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(7, 7))
start = np.array([1.0, 0.0])
for w, color in [(0.95 * np.exp(0.4j), "tab:blue"), (np.exp(0.4j), "tab:orange")]:
    M_w = np.array([[w.real, -w.imag], [w.imag, w.real]])
    points = [start]
    for _ in range(40):
        points.append(M_w @ points[-1])
    points = np.array(points)
    ax.plot(points[:, 0], points[:, 1], "o-", markersize=4, color=color,
            label=f"$|w| = {abs(w):.2f}$, $\\arg w = 0.4$")

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_aspect("equal")
ax.set_title("Iterating $M(w)$: the vectors of $w^k$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left")
plt.show()
```

4. $M(-1 + i) = \begin{pmatrix} -1 & -1 \\ 1 & -1 \end{pmatrix}$. Corners: $1 \mapsto (-1, 1)$, $i \mapsto (-1, -1)$, $1 + i \mapsto (-2, 0)$. The square is rotated by $135°$ and blown up by $\sqrt2$; its diagonal (length $\sqrt2$) now lies along the negative real axis with length $2$.
5. $|\sqrt3 + i| = 2$, $\arg = \pi/6$, so $M = 2R(\pi/6)$ and $M^6 = 64 R(\pi) = -64 I$, i.e. $(\sqrt3 + i)^6 = -64$.
6. $M(z)^T M(z) = M(\bar z)M(z) = M(\bar z z) = M(|z|^2) = |z|^2 I$. The columns are orthogonal and each has length $|z|$. When $|z| = 1$, $M(z)^T M(z) = I$: $M(z)$ is an **orthogonal** matrix with determinant $+1$, a pure rotation.
7. Let $A = \begin{pmatrix} p & q \\ r & s \end{pmatrix}$. Then $AJ = \begin{pmatrix} q & -p \\ s & -r \end{pmatrix}$ and $JA = \begin{pmatrix} -r & -s \\ p & q \end{pmatrix}$. Equality means $q = -r$ and $p = s$ (the other two equations repeat these). So $A = \begin{pmatrix} p & -r \\ r & p \end{pmatrix} = M(p + ri)$. Conversely every $M(z) = aI + bJ$ commutes with $J$.
8. By Cayley–Hamilton $X^2 = (\operatorname{tr}X)X - (\det X)I$. Setting this equal to $-I$ gives $(\operatorname{tr}X)X = (\det X - 1)I$. If $\operatorname{tr}X \ne 0$, $X$ would be a multiple of $I$, say $cI$, and $c^2 = -1$ has no real solution. So $\operatorname{tr}X = 0$ and $\det X = 1$: $X = \begin{pmatrix} p & q \\ r & -p \end{pmatrix}$ with $-p^2 - qr = 1$. Infinitely many, e.g. $\begin{pmatrix} 1 & -2 \\ 1 & -1 \end{pmatrix}$. Among them, $M(z)$ form requires $p = -p$ and $r = -q$, so $p = 0$, $q^2 = 1$: only $X = \pm J$. Each other solution is $J$ seen in a skewed coordinate system ($X = PJP^{-1}$).
9. $\det(M - \lambda I) = (a - \lambda)^2 + b^2 = 0 \iff \lambda = a \pm bi$. Check: $\begin{pmatrix} a & -b \\ b & a \end{pmatrix}\begin{pmatrix} 1 \\ -i \end{pmatrix} = \begin{pmatrix} a + bi \\ b - ai \end{pmatrix} = (a + bi)\begin{pmatrix} 1 \\ -i \end{pmatrix}$, and conjugating gives the other pair. A real eigenvector would be a direction mapped to itself, but $M(z)$ rotates every direction by $\arg z \notin \{0, \pi\}$ when $b \ne 0$; no line through the origin is fixed.

```python
import numpy as np


def to_matrix(z):
    z = complex(z)
    return np.array([[z.real, -z.imag], [z.imag, z.real]]) + 0.0


print("Ex 1:", to_matrix(2 - 5j).tolist(), to_matrix(-1j).tolist(), to_matrix(7).tolist())
print("Ex 2: det =", np.linalg.det(np.array([[-1, 4], [-4, -1]])), " |z|^2 =", abs(-1 - 4j) ** 2)
print("Ex 3:", (to_matrix(2 + 1j) @ to_matrix(1 - 3j)).tolist(), " direct:", (2 + 1j) * (1 - 3j))
corners = np.array([[1, 1, 0], [0, 1, 1]])
print("Ex 4: images of 1, 1+i, i =", (to_matrix(-1 + 1j) @ corners).T.tolist())
print("Ex 5:", np.linalg.matrix_power(to_matrix(np.sqrt(3) + 1j), 6).round(10).tolist(), (np.sqrt(3) + 1j) ** 6)
z = 0.3 - 2.2j
print("Ex 6:", np.allclose(to_matrix(z).T @ to_matrix(z), abs(z) ** 2 * np.eye(2)))
X = np.array([[1, -2], [1, -1]])
print("Ex 8: X^2 =", (X @ X).tolist())
eigenvalues, eigenvectors = np.linalg.eig(to_matrix(2 + 3j))
print("Ex 9: eigenvalues of M(2+3i):", eigenvalues)
print("      eigenvectors scaled so first entry is 1:", (eigenvectors / eigenvectors[0]).T.round(12).tolist())
```

## 8. Conclusion

- Multiplication by $w = a + bi$ is a linear map of $\mathbb{R}^2$ with matrix $M(w) = \begin{pmatrix} a & -b \\ b & a \end{pmatrix} = aI + bJ$, read off from $w \cdot 1$ and $w \cdot i$.
- $M$ respects sums and products, so complex arithmetic *is* arithmetic with these matrices; they commute because complex numbers do.
- $i \leftrightarrow J$, the 90° rotation, and $J^2 = -I$ is the matrix version of $i^2 = -1$; $e^{\varphi J} = R(\varphi)$ is Euler's formula.
- $\det M(z) = |z|^2$ (area scaling), $M(\bar z) = M(z)^T$, $M(z^{-1}) = M(z)^{-1}$, and $|zw| = |z||w|$ follows from $\det(AB) = \det A\det B$.
- In polar form $M(re^{i\varphi}) = rR(\varphi)$: complex numbers are exactly the rotation-scaling matrices, the real $2\times 2$ matrices commuting with $J$. Reflections, shears and uneven stretches are not complex numbers.
