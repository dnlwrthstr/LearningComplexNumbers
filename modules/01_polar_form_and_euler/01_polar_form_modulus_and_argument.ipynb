# Polar Form, Modulus and Argument

A complex number $a + bi$ is a point in the plane, and a point in the plane can be located in two ways: by walking $a$ across and $b$ up (**Cartesian form**), or by turning to face a direction $\theta$ and walking a distance $r$ (**polar form**). This notebook introduces the polar description $z = r(\cos\theta + i\sin\theta)$, the **modulus** $r = |z|$ and the **argument** $\theta = \arg z$, and the traps hiding in the innocent-looking formula $\theta = \arctan(b/a)$.

Polar form is the key that unlocks the rest of this module: multiplication becomes rotation (notebook 02), Euler's formula compresses everything into $re^{i\theta}$ (notebook 03), and powers and roots become easy (notebooks 04 and 05).

This is notebook 01 of module 01 (**Polar Form and Euler's Formula**) in the **Learning Complex Numbers** series.

## 1. Introduction

In Cartesian form, adding is easy — $(1 + 2i) + (3 - i) = 4 + i$ — but multiplying, raising to powers and taking roots are messy. Try computing $(1 + i)^{10}$ by expanding brackets, or finding all three cube roots of $2 + 11i$ (Bombelli's number from notebook 00/01).

Polar form makes those operations easy because it separates a complex number into **how big** it is ($r$) and **which way it points** ($\theta$). Engineers use exactly this split: an AC voltage is described by its amplitude and phase, a signal's Fourier coefficient by its magnitude and phase, a quantum amplitude by its size and its phase.

## 2. Theory

### 2.1 From a point to a distance and a direction

Plot $z = a + bi$ as the point $(a, b)$. Its distance from the origin is the **modulus**

$$
r = |z| = \sqrt{a^2 + b^2} \;\ge\; 0,
$$

and the angle from the positive real axis to the arrow $0 \to z$, measured counter-clockwise in radians, is an **argument** $\theta$ of $z$. Right-triangle trigonometry gives back the coordinates:

$$
a = r\cos\theta, \qquad b = r\sin\theta,
$$

so

$$
\boxed{\,z = r(\cos\theta + i\sin\theta)\,}
$$

This is the **polar form** (also called trigonometric form; often abbreviated $r\,\mathrm{cis}\,\theta$). The bracket $\cos\theta + i\sin\theta$ is a number of modulus $1$ — a point on the **unit circle** — so polar form says: *a unit direction, stretched by $r$*.

### 2.2 The argument is not unique

Turning by a full turn $2\pi$ brings you back to the same direction, so if $\theta$ is an argument of $z$, so is $\theta + 2\pi k$ for every integer $k$. Strictly, $\arg z$ is a **set**:

$$
\arg z = \{\, \theta + 2\pi k : k \in \mathbb{Z} \,\}.
$$

For example $\arg i = \{\dots, -\tfrac{3\pi}{2}, \tfrac{\pi}{2}, \tfrac{5\pi}{2}, \dots\}$.

To get a single number we pick a representative. This series always uses the **principal argument**

$$
\operatorname{Arg} z \in (-\pi, \pi],
$$

the value returned by `cmath.phase` and `np.angle`. The interval is half-open: the negative real axis gets $\operatorname{Arg}(-1) = \pi$, not $-\pi$. (Some books use $[0, 2\pi)$ instead; always check which convention is in force.)

$z = 0$ has modulus $0$ and **no** argument — every direction works, so none is singled out. (Python returns `0.0` by convention; do not read anything into it.)

### 2.3 Why $\arctan(b/a)$ is not enough

The tempting formula $\theta = \arctan(b/a)$ has two defects:

1. $\arctan$ only returns values in $(-\tfrac{\pi}{2}, \tfrac{\pi}{2})$ — the right half-plane. It cannot tell $1 + i$ from $-1 - i$, since both have $b/a = 1$.
2. It divides by $a$, which fails on the imaginary axis.

The fix is the two-argument arctangent $\operatorname{atan2}(b, a)$, which looks at the **signs of $a$ and $b$ separately** and so knows the quadrant:

$$
\operatorname{Arg}(a + bi) = \operatorname{atan2}(b, a) =
\begin{cases}
\arctan(b/a) & a > 0 \\
\arctan(b/a) + \pi & a < 0,\ b \ge 0 \\
\arctan(b/a) - \pi & a < 0,\ b < 0 \\
+\pi/2 & a = 0,\ b > 0 \\
-\pi/2 & a = 0,\ b < 0
\end{cases}
$$

Note the argument order: `atan2(y, x)` — imaginary part first.

### 2.4 Conversions at a glance

| Direction | Formulas | Python |
|---|---|---|
| Cartesian $\to$ polar | $r = \sqrt{a^2 + b^2}$, $\ \theta = \operatorname{atan2}(b, a)$ | `cmath.polar(z)`, `abs(z)`, `cmath.phase(z)`, `np.abs`, `np.angle` |
| Polar $\to$ Cartesian | $a = r\cos\theta$, $\ b = r\sin\theta$ | `cmath.rect(r, theta)`, `r * np.exp(1j * theta)` |

### 2.5 Equality in polar form

Two nonzero complex numbers are equal exactly when

$$
r_1(\cos\theta_1 + i\sin\theta_1) = r_2(\cos\theta_2 + i\sin\theta_2)
\iff r_1 = r_2 \ \text{ and } \ \theta_1 - \theta_2 = 2\pi k \text{ for some } k \in \mathbb{Z}.
$$

Equal moduli, arguments equal **up to whole turns** — this "$+2\pi k$" is exactly what produces $n$ different $n$-th roots in notebook 05.

## 3. Worked Examples

### Example 1 — $z = 1 + i$ (first quadrant)

$r = \sqrt{1 + 1} = \sqrt2$. Since $a > 0$, $\theta = \arctan(1) = \tfrac{\pi}{4}$. So

$$
1 + i = \sqrt2\left(\cos\tfrac{\pi}{4} + i\sin\tfrac{\pi}{4}\right).
$$

### Example 2 — $z = -1 + \sqrt3\, i$ (second quadrant, the classic trap)

$r = \sqrt{1 + 3} = 2$. The naive formula gives $\arctan(\sqrt3 / -1) = \arctan(-\sqrt3) = -\tfrac{\pi}{3}$, which points into the **fourth** quadrant — wrong. The point is up and to the left, so add $\pi$:

$$
\operatorname{Arg} z = -\tfrac{\pi}{3} + \pi = \tfrac{2\pi}{3}, \qquad -1 + \sqrt3\, i = 2\left(\cos\tfrac{2\pi}{3} + i\sin\tfrac{2\pi}{3}\right).
$$

Check: $2\cos\tfrac{2\pi}{3} = 2 \cdot (-\tfrac12) = -1$ and $2\sin\tfrac{2\pi}{3} = 2 \cdot \tfrac{\sqrt3}{2} = \sqrt3$. ✓

### Example 3 — $z = -2 - 2i$ (third quadrant)

$r = \sqrt{8} = 2\sqrt2$. Here $b/a = 1$, so $\arctan$ says $\tfrac{\pi}{4}$ — the same as for $1 + i$, though the point is in the opposite direction. With $a < 0,\ b < 0$ we subtract $\pi$: $\operatorname{Arg} z = \tfrac{\pi}{4} - \pi = -\tfrac{3\pi}{4}$.

The angle $\tfrac{5\pi}{4}$ also describes this direction and is a perfectly good element of $\arg z$; it is just not the *principal* one because it lies outside $(-\pi, \pi]$.

### Example 4 — points on the axes

| $z$ | $r$ | $\operatorname{Arg} z$ |
|---|---|---|
| $5$ | $5$ | $0$ |
| $3i$ | $3$ | $\tfrac{\pi}{2}$ |
| $-4$ | $4$ | $\pi$ (not $-\pi$) |
| $-i$ | $1$ | $-\tfrac{\pi}{2}$ |

### Example 5 — polar to Cartesian

$z = 4\left(\cos\left(-\tfrac{5\pi}{6}\right) + i\sin\left(-\tfrac{5\pi}{6}\right)\right)$. With $\cos\tfrac{5\pi}{6} = -\tfrac{\sqrt3}{2}$ and $\sin\tfrac{5\pi}{6} = \tfrac12$ (cosine is even, sine is odd):

$$
z = 4\left(-\tfrac{\sqrt3}{2} - \tfrac12 i\right) = -2\sqrt3 - 2i.
$$

### Example 6 — the same number, many arguments

$\theta = \tfrac{7\pi}{3}$ is an argument of $z = 2(\cos\tfrac{7\pi}{3} + i\sin\tfrac{7\pi}{3})$, but it is not principal. Subtract one full turn: $\tfrac{7\pi}{3} - 2\pi = \tfrac{\pi}{3} \in (-\pi, \pi]$, so $\operatorname{Arg} z = \tfrac{\pi}{3}$ and $z = 1 + \sqrt3\, i$.

### Example 7 — Bombelli's number

$w = 2 + 11i$ has $r = \sqrt{4 + 121} = \sqrt{125} = 5\sqrt5$ and $\operatorname{Arg} w = \arctan\tfrac{11}{2} \approx 1.3909$ (about $79.7°$). Notebook 05 uses exactly these two numbers to find all three cube roots of $w$.

## 4. Python Examples

### 4.1 Modulus and argument of a scalar

`abs` gives the modulus, `cmath.phase` the principal argument, and `cmath.polar` returns both as a tuple $(r, \theta)$.

```python
import cmath
import math

examples = [1 + 1j, -1 + math.sqrt(3) * 1j, -2 - 2j, 3j, -4, -1j, 2 + 11j]

print(f"{'z':>26}  {'r = |z|':>10}  {'Arg z':>10}  {'Arg z / pi':>11}")
for z in examples:
    z = complex(z)
    r, theta = cmath.polar(z)
    print(f"{str(z):>26}  {r:10.6f}  {theta:10.6f}  {theta / math.pi:11.6f}")
```

### 4.2 `atan` versus `atan2`: the quadrant pitfall

One point in each quadrant. The naive `math.atan(b / a)` is right only in the right half-plane.

```python
import cmath
import math

points = [1 + 1j, -1 + 1j, -1 - 1j, 1 - 1j]

print(f"{'z':>10}  {'atan(b/a)':>10}  {'atan2(b,a)':>11}  {'cmath.phase':>11}  naive correct?")
for z in points:
    naive = math.atan(z.imag / z.real)
    correct = math.atan2(z.imag, z.real)
    same = math.isclose(naive, correct)
    print(f"{str(z):>10}  {naive:10.4f}  {correct:11.4f}  {cmath.phase(z):11.4f}  {same}")

try:
    math.atan(3 / 0)
except ZeroDivisionError as error:
    print("\natan(b/a) on the imaginary axis -> ZeroDivisionError:", error)
print("atan2(3, 0) =", math.atan2(3, 0), "= pi/2")
```

### 4.3 The negative real axis and signed zero

$\operatorname{Arg}(-1) = \pi$. Floating point has a subtlety: IEEE numbers have a **signed zero**, and `atan2` respects it. $-1 - 0i$ (with a negative zero imaginary part) is reported as $-\pi$. Mathematically it is the same number; numerically it sits "just below" the cut at $\pm\pi$. This is a first glimpse of the **branch cut** studied in notebook 03/03.

```python
import cmath

print("phase(-1 + 0j)          =", cmath.phase(complex(-1, 0.0)))
print("phase(-1 - 0j)          =", cmath.phase(complex(-1, -0.0)))
print("phase(-1 + 1e-12j)      =", cmath.phase(complex(-1, 1e-12)))
print("phase(-1 - 1e-12j)      =", cmath.phase(complex(-1, -1e-12)))
print("phase(0)  (no meaning)  =", cmath.phase(0))
```

Two points a hair apart, on either side of the negative real axis, get arguments that differ by almost $2\pi$. $\operatorname{Arg}$ is **discontinuous** across the negative real axis — the price of forcing a single value.

### 4.4 Converting both ways and round-tripping

`cmath.rect(r, theta)` builds $r(\cos\theta + i\sin\theta)$. Converting to polar and back recovers $z$ up to rounding.

```python
import cmath
import math

z = -1 + math.sqrt(3) * 1j
r, theta = cmath.polar(z)
back = cmath.rect(r, theta)

print("z                 =", z)
print("polar(z)          =", (r, theta), " theta/pi =", theta / math.pi)
print("rect(r, theta)    =", back)
print("round trip error  =", abs(back - z))

# Example 5: 4(cos(-5pi/6) + i sin(-5pi/6))
w = cmath.rect(4, -5 * math.pi / 6)
print("\nrect(4, -5pi/6)   =", w, "  expected", complex(-2 * math.sqrt(3), -2))
print("close?            =", abs(w - complex(-2 * math.sqrt(3), -2)) < 1e-12)
```

### 4.5 Arrays: `np.abs` and `np.angle`

numpy applies the same operations element-wise. `np.angle(z, deg=True)` reports degrees.

```python
import numpy as np

z = np.array([1 + 1j, -1 + np.sqrt(3) * 1j, -2 - 2j, 3j, -4 + 0j, -1j])
r = np.abs(z)
theta = np.angle(z)

print("r          =", np.round(r, 6))
print("Arg z      =", np.round(theta, 6))
print("Arg in deg =", np.round(np.angle(z, deg=True), 3))
print("rebuilt    =", np.round(r * (np.cos(theta) + 1j * np.sin(theta)), 12))
print("all close  =", np.allclose(r * (np.cos(theta) + 1j * np.sin(theta)), z))
```

### 4.6 Reducing any angle to the principal argument

Given an arbitrary angle $\theta$, the principal representative is the unique $\theta - 2\pi k$ in $(-\pi, \pi]$. A robust way is to let `cmath.phase` do the work on the unit vector, or to use modular arithmetic carefully so that $\pi$ maps to $\pi$ rather than $-\pi$.

```python
import cmath
import math


def principal_argument(theta):
    # Map any real angle to (-pi, pi]
    reduced = math.fmod(theta + math.pi, 2 * math.pi)
    if reduced <= 0:
        reduced += 2 * math.pi
    return reduced - math.pi


for theta in [7 * math.pi / 3, 5 * math.pi / 4, -3 * math.pi, math.pi, 10.0]:
    via_phase = cmath.phase(cmath.rect(1, theta))
    print(f"theta = {theta:9.5f}  ->  Arg = {principal_argument(theta):8.5f}   (phase of rect: {via_phase:8.5f})")
```

For $\theta = -3\pi$ (an odd multiple of $\pi$) our function returns $\pi$ as the convention demands, while `cmath.phase` of the rounded point may return $-\pi$: $\sin(-3\pi)$ comes out as a tiny negative number rather than exactly $0$. Exactly on the cut, rounding decides.

### 4.7 Exact polar form with sympy

`sympy.Abs` and `sympy.arg` give exact moduli and principal arguments.

```python
import sympy as sp

for z in [1 + sp.I, -1 + sp.sqrt(3) * sp.I, -2 - 2 * sp.I, -4, 2 + 11 * sp.I]:
    r = sp.Abs(z)
    theta = sp.arg(z)
    rebuilt = sp.simplify(r * (sp.cos(theta) + sp.I * sp.sin(theta)))
    print(f"z = {str(z):>16}   r = {str(r):>8}   Arg z = {str(theta):>12}   rebuilt = {rebuilt}")
```

## 5. Visualizations

### 5.1 Modulus and argument on the Argand diagram

Each number is drawn as an arrow from the origin; the arc shows its principal argument. The arcs go counter-clockwise for $\operatorname{Arg} > 0$ and clockwise for $\operatorname{Arg} < 0$.

```python
import matplotlib.pyplot as plt
import numpy as np

numbers = {
    "$1 + i$": 1 + 1j,
    "$-1 + \\sqrt{3}\\,i$": -1 + np.sqrt(3) * 1j,
    "$-2 - 2i$": -2 - 2j,
    "$2 - i$": 2 - 1j,
}
colors = ["tab:blue", "tab:orange", "tab:green", "tab:red"]

fig, ax = plt.subplots(figsize=(7, 7))
for (label, z), color in zip(numbers.items(), colors):
    r, theta = np.abs(z), np.angle(z)
    ax.annotate("", xy=(z.real, z.imag), xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color=color, linewidth=2))
    arc_radius = 0.35 + 0.18 * colors.index(color)
    arc = arc_radius * np.exp(1j * np.linspace(0, theta, 60))
    ax.plot(arc.real, arc.imag, color=color, linewidth=1.5)
    ax.text(z.real * 1.12, z.imag * 1.12,
            f"{label}\n$r={r:.3f},\\ \\theta={theta / np.pi:.3f}\\pi$",
            color=color, ha="center", va="center", fontsize=10)

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-3.5, 3.5)
ax.set_ylim(-3.5, 3.5)
ax.set_aspect("equal")
ax.set_title("Polar coordinates: distance $r$ and principal argument $\\theta$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
plt.show()
```

### 5.2 Why $\arctan(b/a)$ fails: a map of the plane

Colour every point $a + bi$ by its argument. Left: the true principal argument $\operatorname{atan2}(b, a)$ sweeps once around the origin, jumping only across the negative real axis (the cut). Right: $\arctan(b/a)$ repeats itself in the left half-plane — opposite directions get the same colour.

```python
import matplotlib.pyplot as plt
import numpy as np

grid = np.linspace(-2, 2, 401)
a, b = np.meshgrid(grid, grid)
with np.errstate(divide="ignore", invalid="ignore"):
    naive = np.arctan(b / a)
true_arg = np.arctan2(b, a)

fig, axes = plt.subplots(1, 2, figsize=(12, 5.5))
for ax, values, title in [(axes[0], true_arg, "$\\mathrm{Arg}\\,z = \\mathrm{atan2}(b, a)$"),
                          (axes[1], naive, "$\\arctan(b/a)$: only half the circle")]:
    image = ax.imshow(values, extent=[-2, 2, -2, 2], origin="lower", cmap="hsv", vmin=-np.pi, vmax=np.pi)
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
colorbar = fig.colorbar(image, ax=axes, shrink=0.85, ticks=[-np.pi, -np.pi / 2, 0, np.pi / 2, np.pi])
colorbar.ax.set_yticklabels(["$-\\pi$", "$-\\pi/2$", "$0$", "$\\pi/2$", "$\\pi$"])
plt.show()
```

### 5.3 Polar grid: circles of constant $r$, rays of constant $\theta$

Cartesian coordinates tile the plane with a square grid; polar coordinates tile it with circles $|z| = r$ and rays $\operatorname{Arg} z = \theta$. The highlighted point is $2\left(\cos\tfrac{2\pi}{3} + i\sin\tfrac{2\pi}{3}\right) = -1 + \sqrt3\,i$, at the crossing of the circle $r = 2$ and the ray $\theta = \tfrac{2\pi}{3}$.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(7, 7))
angles = np.linspace(0, 2 * np.pi, 300)
for radius in [0.5, 1, 1.5, 2, 2.5]:
    ax.plot(radius * np.cos(angles), radius * np.sin(angles), color="lightgray", linewidth=1)
for ray_angle in np.arange(0, 2 * np.pi, np.pi / 6):
    ax.plot([0, 2.7 * np.cos(ray_angle)], [0, 2.7 * np.sin(ray_angle)], color="lightgray", linewidth=1)

circle_r2 = 2 * np.exp(1j * angles)
ax.plot(circle_r2.real, circle_r2.imag, color="tab:blue", linewidth=2, label="circle $r = 2$")
ray = np.array([0, 2.7]) * np.exp(1j * 2 * np.pi / 3)
ax.plot(ray.real, ray.imag, color="tab:orange", linewidth=2, label="ray $\\theta = 2\\pi/3$")
z = 2 * np.exp(1j * 2 * np.pi / 3)
ax.scatter([z.real], [z.imag], color="crimson", s=80, zorder=3, label="$-1 + \\sqrt{3}\\,i$")

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.set_aspect("equal")
ax.set_title("Polar coordinates: circles of constant $r$, rays of constant $\\theta$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.legend(loc="lower right")
plt.show()
```

### 5.4 $\operatorname{Arg}$ jumps, $\arg$ does not

Walk once around the unit circle, $z(t) = \cos t + i\sin t$ for $t \in [0, 2\pi]$, and record $\operatorname{Arg} z(t)$. It climbs to $\pi$, then jumps down to $-\pi$ as the path crosses the negative real axis. `np.unwrap` removes the jumps by adding multiples of $2\pi$ — it picks, at each step, the element of the *set* $\arg z$ closest to the previous value, recovering the continuous angle $t$.

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 4 * np.pi, 800)
path = np.cos(t) + 1j * np.sin(t)
principal = np.angle(path)
continuous = np.unwrap(principal)

plt.figure(figsize=(9, 4.5))
plt.plot(t, principal, label="principal $\\mathrm{Arg}\\,z(t) \\in (-\\pi, \\pi]$")
plt.plot(t, continuous, "--", label="unwrapped: a continuous choice from $\\arg z(t)$")
plt.yticks([-np.pi, 0, np.pi, 2 * np.pi, 3 * np.pi, 4 * np.pi],
           ["$-\\pi$", "$0$", "$\\pi$", "$2\\pi$", "$3\\pi$", "$4\\pi$"])
plt.xlabel("t  (two laps around the unit circle)")
plt.ylabel("angle")
plt.title("The principal argument jumps by $2\\pi$ at the negative real axis")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

Since $a < 0$ and $b < 0$, subtract $\pi$: $\operatorname{Arg} = \tfrac{\pi}{6} - \pi = -\tfrac{5\pi}{6}$.
6. $\bar z = a - bi$ is the reflection of $z$ in the real axis, so $|\bar z| = \sqrt{a^2 + (-b)^2} = |z|$. With $z = r(\cos\theta + i\sin\theta)$, $\bar z = r(\cos\theta - i\sin\theta) = r(\cos(-\theta) + i\sin(-\theta))$, so $-\theta$ is an argument of $\bar z$. If $\theta \in (-\pi, \pi)$ then $-\theta$ is also in that open interval, hence principal. On the negative real axis $\operatorname{Arg} z = \pi$ and $\bar z = z$, so $\operatorname{Arg}\bar z = \pi \neq -\pi$: the formula fails because $-\pi$ is excluded from the principal range.
7. $\operatorname{Arg}(1 - i) = -\tfrac{\pi}{4}$, so $z = 2(\cos(-\tfrac{\pi}{4}) + i\sin(-\tfrac{\pi}{4})) = \sqrt2 - \sqrt2\, i$ — exactly one number. $\{\operatorname{Arg} z = \tfrac{\pi}{3}\}$ is the open **ray** from the origin at angle $60°$ (origin excluded). $\{|z - 1| = 1\}$ is the **circle** of radius $1$ centred at $1$ (it passes through $0$ and $2$).
8. $-z = z \cdot (-1)$ points in the opposite direction, so its arguments are $\operatorname{Arg} z \pm \pi$; we must pick the one inside $(-\pi, \pi]$. If $\operatorname{Arg} z \in (0, \pi]$, then $\operatorname{Arg} z - \pi \in (-\pi, 0]$ ✓ while $\operatorname{Arg} z + \pi > \pi$ ✗. If $\operatorname{Arg} z \in (-\pi, 0]$, then $\operatorname{Arg} z + \pi \in (0, \pi]$ ✓. Check: $\operatorname{Arg}(1 + i) = \tfrac{\pi}{4}$ and $\operatorname{Arg}(-1 - i) = -\tfrac{3\pi}{4} = \tfrac{\pi}{4} - \pi$ ✓; $\operatorname{Arg}(1 - i) = -\tfrac{\pi}{4}$ and $\operatorname{Arg}(-1 + i) = \tfrac{3\pi}{4} = -\tfrac{\pi}{4} + \pi$ ✓.
9. For $-a + bi$ with $a > 0$: $\arctan\tfrac{b}{-a} = -\arctan\tfrac{b}{a}$. If $b > 0$ (second quadrant), $\operatorname{Arg} = \pi - \arctan\tfrac{b}{a}$; if $b < 0$ (third quadrant), $\operatorname{Arg} = -\pi - \arctan\tfrac{b}{a}$. (For $b = 0$ the answer is $\pi$.) The function is in the check cell below.

Numerical checks:

```python
import cmath
import math
import random

print("Ex 1:", [cmath.polar(z) for z in (math.sqrt(3) + 1j, -5 + 0j, -3j)], " pi/6 =", math.pi / 6)
print("Ex 2:", cmath.rect(6, 3 * math.pi / 4), " expected", complex(-3 * math.sqrt(2), 3 * math.sqrt(2)))

target = -1 - 1j
for theta in (-3 * math.pi / 4, 5 * math.pi / 4, math.pi / 4, -11 * math.pi / 4):
    is_argument = abs(cmath.rect(abs(target), theta) - target) < 1e-12
    print(f"Ex 3: theta/pi = {theta / math.pi:6.3f} is an argument of -1-i: {is_argument}")

r, theta = cmath.polar(-3 + 4j)
print("Ex 4:", r, theta, math.degrees(theta))
print("Ex 5:", cmath.phase(-3 - math.sqrt(3) * 1j), " -5pi/6 =", -5 * math.pi / 6)
print("Ex 6:", cmath.phase((2 + 1j).conjugate()), -cmath.phase(2 + 1j), "| on the cut:", cmath.phase(complex(-2, 0.0)))
print("Ex 7:", cmath.rect(2, cmath.phase(1 - 1j)))
print("Ex 8:", cmath.phase(-(1 + 1j)), cmath.phase(1 + 1j) - math.pi, "|", cmath.phase(-(1 - 1j)), cmath.phase(1 - 1j) + math.pi)


def my_arg(z):
    a, b = z.real, z.imag
    if a > 0:
        return math.atan(b / a)
    if a < 0:
        return math.atan(b / a) + math.pi if b >= 0 else math.atan(b / a) - math.pi
    return math.pi / 2 if b > 0 else -math.pi / 2


random.seed(0)
samples = [complex(random.uniform(-5, 5), random.uniform(-5, 5)) for _ in range(1000)]
worst = max(abs(my_arg(z) - cmath.phase(z)) for z in samples)
print("Ex 9: largest difference my_arg vs cmath.phase over 1000 points:", worst)
```

## 8. Conclusion

- Every complex number $z = a + bi$ can be written as $z = r(\cos\theta + i\sin\theta)$: a distance $r = |z| = \sqrt{a^2 + b^2}$ and a direction $\theta$.
- The argument is only defined up to whole turns: $\arg z = \{\theta + 2\pi k\}$ is a set. The **principal argument** $\operatorname{Arg} z \in (-\pi, \pi]$ picks one, with $\operatorname{Arg}(-1) = \pi$; $z = 0$ has no argument.
- $\arctan(b/a)$ is only correct in the right half-plane. Use $\operatorname{atan2}(b, a)$ (imaginary part first), which handles all quadrants and the imaginary axis.
- $\operatorname{Arg}$ jumps by $2\pi$ across the negative real axis — the first appearance of a branch cut.
- In Python: `abs`, `cmath.phase`, `cmath.polar`, `cmath.rect` for scalars; `np.abs`, `np.angle`, `np.unwrap` for arrays; `sp.Abs`, `sp.arg` for exact values.
