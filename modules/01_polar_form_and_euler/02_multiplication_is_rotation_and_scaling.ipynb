# Multiplication Is Rotation and Scaling

In Cartesian form the product $(a + bi)(c + di) = (ac - bd) + (ad + bc)i$ looks like an arbitrary rule. In polar form it reveals what it really is: **moduli multiply and arguments add**. Multiplying by a complex number $w$ rotates the plane by $\operatorname{Arg} w$ and scales it by $|w|$. This notebook proves that with the angle-addition identities, extends it to division, applies it to whole shapes, and explains why multiplying by $i$ is a quarter turn.

This is notebook 02 of module 01 (**Polar Form and Euler's Formula**) in the **Learning Complex Numbers** series.

## 1. Introduction

Addition of complex numbers has an obvious geometric meaning — it is vector addition, a **translation** of the plane. Multiplication has an equally clean meaning, but it is hidden by Cartesian coordinates. Once uncovered, it explains:

- why $i^2 = -1$: two quarter turns make a half turn, and a half turn sends $1$ to $-1$;
- why $|zw| = |z||w|$;
- why powers of a complex number trace spirals (notebook 04) and its roots form regular polygons (notebook 05);
- why engineers multiply phasors to apply a gain and a phase shift in one step, and why rotations in 2-D graphics can be coded as a single complex multiplication.

## 2. Theory

### 2.1 The product in polar form

Write
$$
z = r(\cos\alpha + i\sin\alpha), \qquad w = s(\cos\beta + i\sin\beta).
$$

Multiply out, using $i^2 = -1$:

$$
\begin{aligned}
zw &= rs\,(\cos\alpha + i\sin\alpha)(\cos\beta + i\sin\beta) \\
   &= rs\,\big[(\cos\alpha\cos\beta - \sin\alpha\sin\beta) + i(\sin\alpha\cos\beta + \cos\alpha\sin\beta)\big].
\end{aligned}
$$

The two brackets are exactly the **angle-addition identities**

$$
\cos(\alpha + \beta) = \cos\alpha\cos\beta - \sin\alpha\sin\beta, \qquad
\sin(\alpha + \beta) = \sin\alpha\cos\beta + \cos\alpha\sin\beta,
$$

so

$$
\boxed{\, zw = rs\,\big(\cos(\alpha + \beta) + i\sin(\alpha + \beta)\big) \,}
$$

In words:

$$
|zw| = |z|\,|w|, \qquad \arg(zw) = \arg z + \arg w.
$$

The multiplication rule for complex numbers *is* the angle-addition formula in disguise. (Read backwards, it gives a way to remember those identities: expand $(\cos\alpha + i\sin\alpha)(\cos\beta + i\sin\beta)$.)

### 2.2 Division

If $w \ne 0$, the quotient $q = z/w$ is the number with $qw = z$. By 2.1, $|q|\,s = r$ and $\arg q + \beta = \alpha$, so

$$
\frac{z}{w} = \frac{r}{s}\big(\cos(\alpha - \beta) + i\sin(\alpha - \beta)\big).
$$

Special case, the reciprocal: $\dfrac{1}{w} = \dfrac{1}{s}\big(\cos(-\beta) + i\sin(-\beta)\big)$ — invert the length, reverse the angle. Compare with $\bar w = s(\cos(-\beta) + i\sin(-\beta))$: so $1/w = \bar w / |w|^2$, the formula from notebook 00/03.

### 2.3 Multiplication by $w$ as a map of the plane

Fix $w = s(\cos\beta + i\sin\beta)$ and consider the function $M_w(z) = wz$ on the whole plane. For every $z$, $M_w(z)$ has modulus $s|z|$ and argument $\arg z + \beta$. So $M_w$:

1. **rotates** every point about the origin by the angle $\beta = \operatorname{Arg} w$, and
2. **scales** every distance from the origin by $s = |w|$.

This is a **spiral similarity** centred at $0$. It preserves angles and shapes (a square goes to a square), changes sizes by the factor $|w|$, and fixes the origin. The order of the two steps does not matter.

Some special multipliers:

| $w$ | $\lvert w\rvert$ | $\operatorname{Arg} w$ | Effect of $z \mapsto wz$ |
|---|---|---|---|
| $2$ | $2$ | $0$ | stretch by 2 |
| $-1$ | $1$ | $\pi$ | half turn (point reflection through $0$) |
| $i$ | $1$ | $\pi/2$ | quarter turn counter-clockwise |
| $-i$ | $1$ | $-\pi/2$ | quarter turn clockwise |
| $\cos\beta + i\sin\beta$ | $1$ | $\beta$ | pure rotation by $\beta$ |
| $1 + i$ | $\sqrt2$ | $\pi/4$ | rotate $45°$, stretch by $\sqrt2$ |

### 2.4 Multiplication by $i$

$i(a + bi) = -b + ai$: the point $(a, b)$ goes to $(-b, a)$, which is precisely the rotation of $(a, b)$ by $90°$ counter-clockwise. Two quarter turns give a half turn: $i^2 z = -z$. **That is the geometric meaning of $i^2 = -1$**: "the number which, applied twice, turns $1$ around to $-1$."

### 2.5 The matrix view

Write $z = x + iy$ as the vector $(x, y)$ and $w = c + di$. Then $wz = (cx - dy) + i(dx + cy)$, i.e.

$$
\begin{pmatrix} c & -d \\ d & c \end{pmatrix}\begin{pmatrix} x \\ y \end{pmatrix}
= s\begin{pmatrix} \cos\beta & -\sin\beta \\ \sin\beta & \cos\beta \end{pmatrix}\begin{pmatrix} x \\ y \end{pmatrix},
$$

a scalar $s$ times a **rotation matrix**. Module 04 develops this correspondence in full.

### 2.6 A warning about principal values

$\arg(zw) = \arg z + \arg w$ is an identity between **sets** (every sum is an argument of the product). For principal values the equation can be off by $2\pi$, because the sum may leave $(-\pi, \pi]$:

$$
\operatorname{Arg}(zw) = \operatorname{Arg} z + \operatorname{Arg} w + 2\pi k, \qquad k \in \{-1, 0, 1\}.
$$

Example: $z = w = -1 + i$ with $\operatorname{Arg} = \tfrac{3\pi}{4}$; the sum is $\tfrac{3\pi}{2}$, but $zw = -2i$ has $\operatorname{Arg} = -\tfrac{\pi}{2} = \tfrac{3\pi}{2} - 2\pi$.

## 3. Worked Examples

### Example 1 — $(1 + i)(\sqrt3 + i)$ both ways

**Cartesian:** $(1 + i)(\sqrt3 + i) = \sqrt3 + i + \sqrt3 i + i^2 = (\sqrt3 - 1) + (\sqrt3 + 1)i$.

**Polar:** $1 + i = \sqrt2\,(\cos\tfrac{\pi}{4} + i\sin\tfrac{\pi}{4})$ and $\sqrt3 + i = 2\,(\cos\tfrac{\pi}{6} + i\sin\tfrac{\pi}{6})$, so the product is

$$
2\sqrt2\left(\cos\tfrac{5\pi}{12} + i\sin\tfrac{5\pi}{12}\right).
$$

**Bonus:** comparing real parts, $2\sqrt2\cos\tfrac{5\pi}{12} = \sqrt3 - 1$, so

$$
\cos 75° = \frac{\sqrt3 - 1}{2\sqrt2} = \frac{\sqrt6 - \sqrt2}{4}.
$$

The multiplication produced an exact value of $\cos 75°$ for free.

### Example 2 — a quotient

$$
\frac{4\left(\cos\frac{5\pi}{6} + i\sin\frac{5\pi}{6}\right)}{2\left(\cos\frac{\pi}{3} + i\sin\frac{\pi}{3}\right)}
= 2\left(\cos\tfrac{\pi}{2} + i\sin\tfrac{\pi}{2}\right) = 2i.
$$

Check in Cartesian form: numerator $= -2\sqrt3 + 2i$, denominator $= 1 + \sqrt3 i$, and $2i(1 + \sqrt3 i) = 2i - 2\sqrt3 = -2\sqrt3 + 2i$ ✓.

### Example 3 — rotate a point by $60°$

To rotate $z = 3 + i$ by $60°$ about the origin, multiply by $u = \cos\tfrac{\pi}{3} + i\sin\tfrac{\pi}{3} = \tfrac12 + \tfrac{\sqrt3}{2}i$:

$$
uz = \left(\tfrac12 + \tfrac{\sqrt3}{2} i\right)(3 + i) = \tfrac32 + \tfrac12 i + \tfrac{3\sqrt3}{2} i - \tfrac{\sqrt3}{2}
= \frac{3 - \sqrt3}{2} + \frac{1 + 3\sqrt3}{2}\, i.
$$

$|uz| = |z| = \sqrt{10}$, since $|u| = 1$.

### Example 4 — rotation about a point other than $0$

Multiplication rotates about the origin. To rotate $z$ by angle $\beta$ about a centre $c$: translate $c$ to $0$, rotate, translate back:

$$
z \mapsto c + u(z - c), \qquad u = \cos\beta + i\sin\beta.
$$

Rotating $z = 2 + i$ by $90°$ about $c = 1$: $1 + i\,(1 + i) = 1 + i - 1 = i$.

### Example 5 — repeated multiplication by $i$

$z = 2 + i \to i z = -1 + 2i \to i^2 z = -2 - i \to i^3 z = 1 - 2i \to i^4 z = 2 + i$. Four quarter turns bring the point back: the four points are the corners of a square centred at the origin.

### Example 6 — the principal-value pitfall

$z = -1 + i$ (Arg $\tfrac{3\pi}{4}$), $w = -1$ (Arg $\pi$). Then $zw = 1 - i$ with $\operatorname{Arg} = -\tfrac{\pi}{4}$, while $\operatorname{Arg} z + \operatorname{Arg} w = \tfrac{7\pi}{4}$. They differ by $2\pi$: same direction, different representative.

## 4. Python Examples

### 4.1 Moduli multiply, arguments add

Check the rule on random complex numbers. We compare **directions** with a tolerance by checking that the unit vectors agree, which sidesteps the $2\pi$ ambiguity of principal values.

```python
import cmath

import numpy as np

rng = np.random.default_rng(seed=1)
z = rng.normal(size=6) + 1j * rng.normal(size=6)
w = rng.normal(size=6) + 1j * rng.normal(size=6)
product = z * w

moduli_multiply = np.allclose(np.abs(product), np.abs(z) * np.abs(w))
# exp(i * angle) is the unit vector in that direction; equal directions <=> angles agree mod 2*pi
arguments_add = np.allclose(np.exp(1j * np.angle(product)), np.exp(1j * (np.angle(z) + np.angle(w))))
print("|zw| = |z||w|           :", moduli_multiply)
print("arg(zw) = arg z + arg w :", arguments_add, "(as directions, i.e. mod 2pi)")

sum_of_args = np.angle(z) + np.angle(w)
offsets = (np.angle(product) - sum_of_args) / (2 * np.pi)
print("Arg(zw) - (Arg z + Arg w), in units of 2pi:", np.round(offsets, 12))
```

The last line shows the warning of §2.6 in action: the difference between $\operatorname{Arg}(zw)$ and $\operatorname{Arg} z + \operatorname{Arg} w$ is always a whole number of turns, usually $0$ but sometimes $\pm 1$.

### 4.2 Symbolic proof with sympy

Expand the product of two unit-circle numbers and let sympy recognise the angle-addition formulas.

```python
import sympy as sp

alpha, beta = sp.symbols("alpha beta", real=True)
r, s = sp.symbols("r s", positive=True)

z = r * (sp.cos(alpha) + sp.I * sp.sin(alpha))
w = s * (sp.cos(beta) + sp.I * sp.sin(beta))
product = sp.expand(z * w)

real_part = sp.simplify(sp.re(product))
imag_part = sp.simplify(sp.im(product))
print("Re(zw) =", sp.expand(sp.re(product)), " ->", real_part)
print("Im(zw) =", sp.expand(sp.im(product)), " ->", imag_part)

expected = r * s * (sp.cos(alpha + beta) + sp.I * sp.sin(alpha + beta))
print("zw - rs(cos(a+b) + i sin(a+b)) simplifies to:", sp.simplify(sp.expand_trig(expected) - product))
```

### 4.3 Worked Example 1 and the exact value of $\cos 75°$

```python
import sympy as sp

product = sp.expand((1 + sp.I) * (sp.sqrt(3) + sp.I))
print("(1 + i)(sqrt3 + i) =", product)
print("modulus            =", sp.simplify(sp.Abs(product)), " (= sqrt2 * 2)")
print("argument           =", sp.simplify(sp.arg(product)), " (= pi/4 + pi/6)")

cos_75 = sp.re(product) / sp.Abs(product)
target = (sp.sqrt(6) - sp.sqrt(2)) / 4
print("cos(5pi/12) from the product =", sp.radsimp(cos_75), "=", sp.N(cos_75))
print("equals (sqrt6 - sqrt2)/4 ?    ", sp.simplify(sp.radsimp(cos_75) - target) == 0)
print("cos(5pi/12) directly         =", sp.N(sp.cos(5 * sp.pi / 12)))
```

### 4.4 Division and reciprocals

```python
import cmath
import math

z = cmath.rect(4, 5 * math.pi / 6)
w = cmath.rect(2, math.pi / 3)
quotient = z / w
print("z / w            =", quotient, "  expected 2i")
print("|z/w|, Arg(z/w)  =", cmath.polar(quotient), "  expected (2, pi/2 =", math.pi / 2, ")")

w = 3 - 4j
print("\n1/w              =", 1 / w)
print("conj(w) / |w|^2  =", w.conjugate() / abs(w) ** 2)
print("|1/w| = 1/|w|    :", math.isclose(abs(1 / w), 1 / abs(w)))
print("Arg(1/w) = -Arg w:", math.isclose(cmath.phase(1 / w), -cmath.phase(w)))
```

### 4.5 Rotation and the matrix view

A rotation by $\beta$ can be done by multiplying by $\cos\beta + i\sin\beta$ or by applying the matrix $\begin{pmatrix} c & -d \\ d & c \end{pmatrix}$; they agree.

```python
import numpy as np

beta = np.pi / 3
u = np.cos(beta) + 1j * np.sin(beta)
z = 3 + 1j

as_complex = u * z
matrix = np.array([[u.real, -u.imag],
                   [u.imag, u.real]])
as_matrix = matrix @ np.array([z.real, z.imag])

print("u * z          =", as_complex)
print("matrix @ (x,y) =", as_matrix)
print("expected       =", ((3 - np.sqrt(3)) / 2, (1 + 3 * np.sqrt(3)) / 2))
print("agree          =", np.allclose([as_complex.real, as_complex.imag], as_matrix))
print("|uz| = |z|     =", np.isclose(abs(as_complex), abs(z)))


def rotate_about(z, centre, angle):
    return centre + np.exp(1j * angle) * (z - centre)


print("\nrotate 2+i by 90 deg about 1:", np.round(rotate_about(2 + 1j, 1, np.pi / 2), 12))
```

(We used `np.exp(1j * angle)` as shorthand for $\cos\beta + i\sin\beta$. Notebook 03 explains why that is legitimate — it is Euler's formula.)

### 4.6 Multiplying a whole shape

Represent a polygon by the array of its vertices; multiplying the array by $w$ transforms every vertex at once. The side lengths all scale by $|w|$ and every interior angle is unchanged.

```python
import numpy as np

square = np.array([1 + 0j, 2 + 0j, 2 + 1j, 1 + 1j])
w = 1 + 1j
image = w * square


def side_lengths(polygon):
    return np.abs(np.roll(polygon, -1) - polygon)


def interior_angles_deg(polygon):
    previous_edge = np.roll(polygon, 1) - polygon
    next_edge = np.roll(polygon, -1) - polygon
    return np.degrees(np.abs(np.angle(previous_edge / next_edge)))


print("image vertices         :", image)
print("side lengths before    :", side_lengths(square))
print("side lengths after     :", np.round(side_lengths(image), 12), " ratio", np.round(side_lengths(image) / side_lengths(square), 12))
print("interior angles before :", interior_angles_deg(square))
print("interior angles after  :", np.round(interior_angles_deg(image), 10))
print("|w| =", abs(w), "  Arg w in degrees =", np.degrees(np.angle(w)))
```

## 5. Visualizations

### 5.1 The product of two numbers

The arrows for $z$, $w$ and $zw$, with angle arcs. The arc for $zw$ is the arc of $z$ followed by the arc of $w$; the length of $zw$ is the product of the lengths.

```python
import matplotlib.pyplot as plt
import numpy as np

z = 1 + 1j
w = np.sqrt(3) + 1j
product = z * w
items = [("$z = 1 + i$", z, "tab:blue", 0.5, (-1.1, 0.25)), ("$w = \\sqrt{3} + i$", w, "tab:orange", 0.8, (0.1, -0.35)),
         ("$zw$", product, "tab:green", 1.1, (0.1, 0.1))]

fig, ax = plt.subplots(figsize=(7, 7))
for label, number, color, arc_radius, (dx, dy) in items:
    ax.annotate("", xy=(number.real, number.imag), xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color=color, linewidth=2.2))
    arc = arc_radius * np.exp(1j * np.linspace(0, np.angle(number), 60))
    ax.plot(arc.real, arc.imag, color=color, linewidth=1.5)
    ax.text(number.real + dx, number.imag + dy,
            f"{label}\n$|\\cdot| = {abs(number):.3f}$, $\\theta = {np.degrees(np.angle(number)):.0f}°$",
            color=color, fontsize=10)

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-0.5, 3.5)
ax.set_ylim(-0.5, 3.5)
ax.set_aspect("equal")
ax.set_title("$|zw| = |z||w| = \\sqrt{2}\\cdot 2$,  $\\arg zw = 45° + 30° = 75°$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
plt.show()
```

### 5.2 Similar triangles

A classical picture: the triangle with vertices $0, 1, w$ is similar to the triangle $0, z, zw$ — the second is the first multiplied by $z$, i.e. rotated by $\operatorname{Arg} z$ and scaled by $|z|$. This gives a ruler-and-compass construction of the product.

```python
import matplotlib.pyplot as plt
import numpy as np

z = 1.2 + 0.9j
w = 0.8 + 1.1j
triangle = np.array([0, 1, w, 0])
image = z * triangle

fig, ax = plt.subplots(figsize=(7, 7))
ax.fill(triangle.real, triangle.imag, alpha=0.3, color="tab:blue", label="triangle $0, 1, w$")
ax.fill(image.real, image.imag, alpha=0.3, color="tab:orange", label="$z\\times$ triangle $= 0, z, zw$")
ax.plot(triangle.real, triangle.imag, color="tab:blue")
ax.plot(image.real, image.imag, color="tab:orange")
for point, label in [(1, "$1$"), (w, "$w$"), (z, "$z$"), (z * w, "$zw$")]:
    ax.scatter([point.real], [point.imag], color="black", zorder=3)
    ax.text(point.real + 0.06, point.imag + 0.06, label, fontsize=13)

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-0.8, 2.0)
ax.set_ylim(-0.4, 2.4)
ax.set_aspect("equal")
ax.set_title("Multiplying by $z$: triangle $(0, 1, w)$ becomes the similar $(0, z, zw)$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower right")
plt.show()
```

### 5.3 Multiplying a whole shape by $w$

A small "house" shape, multiplied by four different $w$. The shape is never distorted: it is rotated about the origin by $\operatorname{Arg} w$ and scaled by $|w|$.

```python
import matplotlib.pyplot as plt
import numpy as np

house = np.array([1 + 0j, 2 + 0j, 2 + 1j, 1.5 + 1.6j, 1 + 1j, 1 + 0j])
multipliers = [("$w = 1$ (original)", 1), ("$w = i$", 1j), ("$w = -1$", -1),
               ("$w = 0.5 - 0.5i$", 0.5 - 0.5j), ("$w = 1 + i$", 1 + 1j)]

fig, ax = plt.subplots(figsize=(8, 8))
for label, w in multipliers:
    image = w * house
    ax.fill(image.real, image.imag, alpha=0.35, label=f"{label}: rotate {np.degrees(np.angle(w)):.0f}°, scale {abs(w):.3f}")
    ax.plot(image.real, image.imag, linewidth=1)

ax.scatter([0], [0], color="black", zorder=3)
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-3.2, 3.2)
ax.set_ylim(-3.2, 3.2)
ax.set_aspect("equal")
ax.set_title("$z \\mapsto wz$: a rotation about 0 combined with a scaling")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower left", fontsize=9)
plt.show()
```

### 5.4 Multiplication by $i$ rotates the whole grid

Apply $z \mapsto iz$ to a grid of points coloured by position. Every point moves a quarter turn counter-clockwise; applying it twice gives $z \mapsto -z$.

```python
import matplotlib.pyplot as plt
import numpy as np

xs, ys = np.meshgrid(np.linspace(0.3, 2, 8), np.linspace(0.2, 1.2, 5))
points = (xs + 1j * ys).ravel()
colors = np.arange(points.size)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for ax, power in zip(axes, [0, 1, 2]):
    image = (1j ** power) * points
    ax.scatter(points.real, points.imag, c="lightgray", s=20)
    ax.scatter(image.real, image.imag, c=colors, cmap="viridis", s=30)
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(-2.3, 2.3)
    ax.set_ylim(-2.3, 2.3)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
titles = ["$z$ (original)", "$iz$: quarter turn", "$i^2 z = -z$: half turn"]
for ax, title in zip(axes, titles):
    ax.set_title(title)
plt.show()
```

1. Moduli multiply, arguments add: $6(\cos 90° + i\sin 90°) = 6i$.

4. Both arrows are in the first quadrant, at about $27°$ and $63°$; the angles add to about $90°$, so $zw$ should lie on (or very near) the positive imaginary axis, with length $\sqrt5 \cdot \sqrt5 = 5$. Compute: $(2 + i)(1 + 2i) = 2 + 4i + i + 2i^2 = 5i$. Indeed $\arctan\frac12 + \arctan 2 = \frac{\pi}{2}$ exactly, and $|5i| = 5 = \sqrt5\sqrt5$ ✓.
5. $(1 + i)(1 + \sqrt3 i) = 1 + \sqrt3 i + i + \sqrt3 i^2 = (1 - \sqrt3) + (1 + \sqrt3)i$, and in polar form it is $\sqrt2 \cdot 2\,(\cos(\frac{\pi}{4} + \frac{\pi}{3}) + i\sin(\ldots)) = 2\sqrt2(\cos\frac{7\pi}{12} + i\sin\frac{7\pi}{12})$. Hence $\cos\frac{7\pi}{12} = \frac{1 - \sqrt3}{2\sqrt2} = \frac{\sqrt2 - \sqrt6}{4}$ and $\sin\frac{7\pi}{12} = \frac{1 + \sqrt3}{2\sqrt2} = \frac{\sqrt2 + \sqrt6}{4}$.
6. $w = \frac{-10i}{3 + 4i} = \frac{-10i(3 - 4i)}{25} = \frac{-30i + 40i^2}{25} = -\frac{8}{5} - \frac{6}{5}i$. Scale factor $|w| = \frac{10}{5} = 2$; angle $\operatorname{Arg} w = \operatorname{Arg}(-10i) - \operatorname{Arg}(3 + 4i) = -\frac{\pi}{2} - \arctan\frac43 \approx -2.498$ rad $\approx -143.13°$ (already in $(-\pi, \pi]$).
7. $c + u(z - c)$ with $c = 1 + i$, $u = -i$: $z - c = 2 + i$, $-i(2 + i) = -2i - i^2 = 1 - 2i$, so the image is $(1 + i) + (1 - 2i) = 2 - i$.

## 8. Conclusion

- In polar form, $r(\cos\alpha + i\sin\alpha)\cdot s(\cos\beta + i\sin\beta) = rs(\cos(\alpha + \beta) + i\sin(\alpha + \beta))$: **moduli multiply, arguments add**. The proof is exactly the angle-addition identities.
- Division divides moduli and subtracts arguments; $1/w$ has length $1/|w|$ and angle $-\arg w$.
- The map $z \mapsto wz$ rotates the plane about $0$ by $\operatorname{Arg} w$ and scales it by $|w|$. Shapes are preserved; sizes scale by $|w|$. Rotation about another centre $c$ is $z \mapsto c + u(z - c)$.
- Multiplication by $i$ is a quarter turn, which is why $i^2 = -1$.
- For **principal** values, $\operatorname{Arg}(zw)$ may differ from $\operatorname{Arg} z + \operatorname{Arg} w$ by $\pm 2\pi$; the identity holds for the sets $\arg$.
