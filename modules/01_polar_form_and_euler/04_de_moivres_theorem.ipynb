# De Moivre's Theorem

$$
(\cos\theta + i\sin\theta)^n = \cos n\theta + i\sin n\theta
$$

Raising a complex number to a power multiplies its argument and raises its modulus to that power. This notebook proves De Moivre's theorem (by induction and via Euler's formula), uses it to compute large powers in one line, and turns it into a machine for **multiple-angle identities** — expanding the left-hand side with the binomial theorem gives formulas for $\cos 3\theta$, $\sin 5\theta$ and so on, which we derive and verify with sympy. Geometrically, the powers $z, z^2, z^3, \dots$ trace a **spiral**.

This is notebook 04 of module 01 (**Polar Form and Euler's Formula**) in the **Learning Complex Numbers** series.

## 1. Introduction

Computing $(1 + i)^{20}$ by expanding $20$ brackets is hopeless. In polar form it is immediate: $1 + i = \sqrt2 e^{i\pi/4}$, so $(1 + i)^{20} = 2^{10}e^{5\pi i} = -1024$.

The same theorem, read the other way round, produces trigonometric identities that would otherwise need a chain of addition formulas. Abraham de Moivre found it around 1707 — decades before Euler's formula — and it is the bridge to the next notebook, where we run it backwards to find **roots**.

Powers of complex numbers also appear wherever something is repeated: iterating $z \mapsto wz$ in a discrete dynamical system, the poles of a digital filter, the eigenvalues of a rotation matrix raised to a power (module 04), and the $n$-th powers $\omega^k$ that build the discrete Fourier transform.

## 2. Theory

### 2.1 The theorem

**De Moivre's theorem.** For every real $\theta$ and every integer $n$,

$$
(\cos\theta + i\sin\theta)^n = \cos n\theta + i\sin n\theta.
$$

**Proof by induction ($n \ge 0$).** For $n = 0$ both sides are $1$. If it holds for $n$, then by the multiplication rule of notebook 02 (arguments add):

$$
(\cos\theta + i\sin\theta)^{n+1} = (\cos n\theta + i\sin n\theta)(\cos\theta + i\sin\theta) = \cos(n + 1)\theta + i\sin(n + 1)\theta.
$$

**Negative $n$.** $(\cos\theta + i\sin\theta)^{-1} = \cos\theta - i\sin\theta = \cos(-\theta) + i\sin(-\theta)$, because the product with $\cos\theta + i\sin\theta$ is $\cos^2\theta + \sin^2\theta = 1$. Then $(\cos\theta + i\sin\theta)^{-m} = \left[\cos(-\theta) + i\sin(-\theta)\right]^m = \cos(-m\theta) + i\sin(-m\theta)$ by the positive case.

**Proof via Euler.** In one line: $(e^{i\theta})^n = e^{in\theta}$. (The induction proof above is what justifies this law of exponents for integer $n$.)

### 2.2 Powers of any complex number

For $z = re^{i\theta}$ with $r > 0$:

$$
\boxed{\, z^n = r^n e^{in\theta} = r^n(\cos n\theta + i\sin n\theta) \,}
$$

so $|z^n| = |z|^n$ and $n\theta$ is an argument of $z^n$. As with products, $n\operatorname{Arg} z$ need not be the *principal* argument: reduce it modulo $2\pi$ into $(-\pi, \pi]$.

> **Caution — non-integer $n$.** The theorem is for **integer** $n$. For $n = \tfrac12$ the left side has two candidate values (two square roots) and the formula silently picks one; for $n = \tfrac13$ there are three. Handling all of them is exactly the subject of notebook 05.

### 2.3 Multiple-angle identities for free

Expand the left side with the binomial theorem, writing $c = \cos\theta$, $s = \sin\theta$:

$$
\cos n\theta + i\sin n\theta = (c + is)^n = \sum_{k=0}^{n}\binom{n}{k}c^{n-k}(is)^k.
$$

Since $i^k$ cycles $1, i, -1, -i$, the **even** $k$ give real terms and the **odd** $k$ give imaginary terms:

$$
\cos n\theta = \sum_{k \text{ even}} (-1)^{k/2}\binom{n}{k}c^{n-k}s^{k}, \qquad
\sin n\theta = \sum_{k \text{ odd}} (-1)^{(k-1)/2}\binom{n}{k}c^{n-k}s^{k}.
$$

For $n = 3$: $(c + is)^3 = c^3 + 3ic^2 s - 3cs^2 - is^3$, so

$$
\cos 3\theta = c^3 - 3cs^2 = 4\cos^3\theta - 3\cos\theta, \qquad
\sin 3\theta = 3c^2 s - s^3 = 3\sin\theta - 4\sin^3\theta,
$$

using $s^2 = 1 - c^2$ and $c^2 = 1 - s^2$.

Replacing every $s^2$ by $1 - c^2$ always works for $\cos n\theta$ (only even powers of $s$ appear), giving a polynomial in $\cos\theta$ of degree $n$: the **Chebyshev polynomial** $T_n$, with $\cos n\theta = T_n(\cos\theta)$.

### 2.4 The reverse direction: powers into multiple angles

Combining De Moivre with $\cos\theta = \frac{e^{i\theta} + e^{-i\theta}}{2}$ (notebook 03) turns *powers* of $\cos$ and $\sin$ into *sums* of $\cos k\theta$ and $\sin k\theta$ — the form needed to integrate them:

$$
\cos^3\theta = \frac{(e^{i\theta} + e^{-i\theta})^3}{8} = \frac{e^{3i\theta} + 3e^{i\theta} + 3e^{-i\theta} + e^{-3i\theta}}{8}
= \frac{\cos 3\theta + 3\cos\theta}{4}.
$$

### 2.5 The geometry of $z^n$: spirals

The sequence $1, z, z^2, z^3, \dots$ turns by the same angle $\theta$ each step while its length is multiplied by $r$ each step. Its points lie on the **logarithmic spiral** $\rho = r^{\varphi/\theta}$ (in polar coordinates $(\rho, \varphi)$):

| $\lvert z\rvert$ | Behaviour of $z^n$ |
|---|---|
| $r > 1$ | spirals **outwards** to $\infty$ |
| $r = 1$ | walks around the unit circle; returns to $1$ exactly when $\theta/2\pi$ is rational ($z$ a root of unity) |
| $r < 1$ | spirals **inwards** to $0$ |

## 3. Worked Examples

### Example 1 — $(1 + i)^{20}$

$1 + i = \sqrt2 e^{i\pi/4}$, so $(1 + i)^{20} = (\sqrt2)^{20} e^{20\pi i/4} = 2^{10} e^{5\pi i} = 1024\, e^{i\pi} = -1024$.

### Example 2 — $(1 - \sqrt3 i)^{7}$

$1 - \sqrt3 i = 2e^{-i\pi/3}$, so the power is $2^7 e^{-7\pi i/3}$. Reduce: $-\frac{7\pi}{3} + 2\pi = -\frac{\pi}{3}$. Hence

$$
(1 - \sqrt3 i)^7 = 128\, e^{-i\pi/3} = 128\left(\tfrac12 - \tfrac{\sqrt3}{2}i\right) = 64 - 64\sqrt3\, i.
$$

### Example 3 — a negative power

$(\sqrt3 + i)^{-4}$: $\sqrt3 + i = 2e^{i\pi/6}$, so $(\sqrt3 + i)^{-4} = 2^{-4}e^{-4\pi i/6} = \frac{1}{16}e^{-2\pi i/3} = \frac{1}{16}\left(-\frac12 - \frac{\sqrt3}{2}i\right) = -\frac{1}{32} - \frac{\sqrt3}{32}i$.

### Example 4 — $\cos 4\theta$ and $\sin 4\theta$

$(c + is)^4 = c^4 + 4ic^3s - 6c^2s^2 - 4ics^3 + s^4$. Real and imaginary parts:

$$
\cos 4\theta = c^4 - 6c^2s^2 + s^4 = 8\cos^4\theta - 8\cos^2\theta + 1, \qquad
\sin 4\theta = 4c^3 s - 4cs^3 = 4\sin\theta\cos\theta(\cos^2\theta - \sin^2\theta).
$$

(For the second form of $\cos 4\theta$, substitute $s^2 = 1 - c^2$: $c^4 - 6c^2(1 - c^2) + (1 - c^2)^2 = 8c^4 - 8c^2 + 1$.)

### Example 5 — $\sin^4\theta$ as a sum of cosines

$$
\sin^4\theta = \left(\frac{e^{i\theta} - e^{-i\theta}}{2i}\right)^4 = \frac{e^{4i\theta} - 4e^{2i\theta} + 6 - 4e^{-2i\theta} + e^{-4i\theta}}{16}
= \frac{\cos 4\theta - 4\cos 2\theta + 3}{8}.
$$

Consequence: $\int_0^{2\pi}\sin^4\theta\, d\theta = \frac{3}{8}\cdot 2\pi = \frac{3\pi}{4}$, since the cosine terms integrate to zero over a full period.

### Example 6 — when is $z^n$ real?

$z = 1 + i$: $z^n = 2^{n/2}e^{in\pi/4}$ is real exactly when $n\pi/4$ is a multiple of $\pi$, i.e. when $4 \mid n$. Positive real when $8 \mid n$, negative real when $n \equiv 4 \pmod 8$.

## 4. Python Examples

### 4.1 Powers via De Moivre versus repeated multiplication

```python
import cmath
import math


def power_de_moivre(z, n):
    r, theta = cmath.polar(z)
    return cmath.rect(r ** n, n * theta)


for z, n in [(1 + 1j, 20), (1 - math.sqrt(3) * 1j, 7), (math.sqrt(3) + 1j, -4), (0.9 + 0.3j, 50)]:
    via_polar = power_de_moivre(z, n)
    via_python = z ** n
    print(f"({z:.4f})^{n:>3}:  De Moivre {via_polar:.6f}   z**n {via_python:.6f}   "
          f"agree {abs(via_polar - via_python) < 1e-9 * max(1, abs(via_python))}")

print("\nexpected Ex 2:", complex(64, -64 * math.sqrt(3)), "  Ex 3:", complex(-1 / 32, -math.sqrt(3) / 32))
```

The tolerance is relative (scaled by $|z^n|$): $(1 + i)^{20}$ has size $1024$, so absolute errors of order $10^{-13}$ are just rounding.

### 4.2 Exact powers with sympy

```python
import sympy as sp

print("(1 + i)^20       =", sp.expand((1 + sp.I) ** 20))
print("(1 - sqrt3 i)^7  =", sp.expand((1 - sp.sqrt(3) * sp.I) ** 7))
print("(sqrt3 + i)^-4   =", sp.simplify(sp.expand(sp.radsimp((sp.sqrt(3) + sp.I) ** -4))))

theta = sp.symbols("theta", real=True)
n = sp.symbols("n", integer=True)
print("\n(e^{i theta})^n  =", sp.powsimp(sp.exp(sp.I * theta) ** n))
```

### 4.3 Multiple-angle identities with sympy

Expand $(\cos\theta + i\sin\theta)^n$, take real and imaginary parts, substitute $\sin^2 = 1 - \cos^2$, and compare with sympy's own `expand_trig`.

```python
import sympy as sp

theta = sp.symbols("theta", real=True)
c, s = sp.symbols("c s", real=True)   # stand-ins for cos(theta), sin(theta)

for n in [2, 3, 4, 5]:
    expansion = sp.expand((c + sp.I * s) ** n)
    cos_n = sp.re(expansion)
    sin_n = sp.im(expansion)
    cos_n_in_c = sp.expand(cos_n.subs(s**2, 1 - c**2))
    # verify: rewrite the difference in exponentials, where it must cancel to exactly 0
    cos_difference = cos_n.subs({c: sp.cos(theta), s: sp.sin(theta)}) - sp.cos(n * theta)
    sin_difference = sin_n.subs({c: sp.cos(theta), s: sp.sin(theta)}) - sp.sin(n * theta)
    cos_ok = sp.expand(cos_difference.rewrite(sp.exp)) == 0
    sin_ok = sp.expand(sin_difference.rewrite(sp.exp)) == 0
    print(f"n = {n}")
    print(f"   cos {n}θ = {cos_n}   =   {cos_n_in_c}   (verified: {cos_ok})")
    print(f"   sin {n}θ = {sin_n}   (verified: {sin_ok})")
```

### 4.4 Chebyshev polynomials

The polynomial $T_n$ with $\cos n\theta = T_n(\cos\theta)$ is built into sympy as `chebyshevt`. Our De Moivre derivation reproduces it.

```python
import sympy as sp

c, s = sp.symbols("c s", real=True)
for n in range(1, 7):
    from_de_moivre = sp.expand(sp.re(sp.expand((c + sp.I * s) ** n)).subs(s**2, 1 - c**2))
    chebyshev = sp.expand(sp.chebyshevt(n, c))
    print(f"T_{n}(c) = {str(chebyshev):<36} matches De Moivre: {sp.simplify(from_de_moivre - chebyshev) == 0}")
```

`subs(s**2, 1 - c**2)` also catches $s^4 = (s^2)^2$ and $s^6$, because sympy matches powers of the pattern.

### 4.5 Powers of $\cos$ and $\sin$ as sums of multiple angles

Sympy's trigonometric simplification module `fu` has the two rules we need: `TR8` turns products of sines and cosines into sums, and `TR7` lowers a square, $\cos^2 x \to \frac{1 + \cos 2x}{2}$. Applying them until nothing changes "linearises" any power — the same result as rewriting in exponentials by hand.

```python
import sympy as sp
from sympy.simplify.fu import TR7, TR8


def linearise(expression):
    previous = None
    while previous != expression:
        previous = expression
        expression = sp.expand(TR8(TR7(expression)))
    return expression


theta = sp.symbols("theta", real=True)
for expression in [sp.cos(theta) ** 3, sp.sin(theta) ** 4, sp.cos(theta) ** 5, sp.sin(theta) ** 2 * sp.cos(theta) ** 2]:
    linear = linearise(expression)
    integral = sp.integrate(expression, (theta, 0, 2 * sp.pi))
    print(f"{str(expression):<28} = {str(linear):<52}  integral over [0, 2pi] = {integral}")
```

## 5. Visualizations

### 5.1 Powers as spirals

Plot $z^0, z^1, \dots, z^{N}$ for three choices of $z$ with the same argument $\frac{\pi}{8}$ and moduli $1.08$, $1$ and $0.92$. Each step turns $22.5°$; the modulus decides whether the points spiral out, stay on the circle or spiral in.

```python
import matplotlib.pyplot as plt
import numpy as np

angle = np.pi / 8
cases = [(1.08, "$|z| = 1.08$: outward"), (1.0, "$|z| = 1$: on the circle"), (0.92, "$|z| = 0.92$: inward")]
n_values = np.arange(0, 33)

fig, axes = plt.subplots(1, 3, figsize=(17, 5.5))
fig.subplots_adjust(wspace=0.3)
for ax, (modulus, title) in zip(axes, cases):
    z = modulus * np.exp(1j * angle)
    powers = z ** n_values
    smooth_n = np.linspace(0, n_values[-1], 800)
    spiral = z ** smooth_n   # principal power: r^t e^{i t angle}, the logarithmic spiral through the points
    ax.plot(spiral.real, spiral.imag, color="lightgray", linewidth=1)
    points = ax.scatter(powers.real, powers.imag, c=n_values, cmap="viridis", s=30, zorder=3)
    if modulus <= 1:   # in the outward spiral the first powers are too crowded to label
        for n in [0, 1, 2, 3]:
            ax.text(powers[n].real + 0.08, powers[n].imag + 0.08, f"$z^{n}$", fontsize=10)
    limit = max(1.3, 1.1 * np.max(np.abs(powers)))
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(-limit, limit)
    ax.set_ylim(-limit, limit)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
fig.colorbar(points, ax=axes, shrink=0.8, label="power n")
plt.show()
```

### 5.2 $(1 + i)^n$: one eighth of a turn per step

$|1 + i| = \sqrt2$ and $\operatorname{Arg}(1 + i) = 45°$. The arrows to $(1 + i)^n$ for $n = 0, \dots, 8$ turn $45°$ each and grow by $\sqrt2$; $(1 + i)^4 = -4$ and $(1 + i)^8 = 16$ land on the real axis (Example 6).

```python
import matplotlib.pyplot as plt
import numpy as np

z = 1 + 1j
fig, ax = plt.subplots(figsize=(8, 8))
colors = plt.cm.plasma(np.linspace(0, 0.85, 9))
for n in range(9):
    power = z ** n
    ax.annotate("", xy=(power.real, power.imag), xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color=colors[n], linewidth=2))
    offset = 0.8 * power / abs(power)
    ax.text(power.real + offset.real, power.imag + offset.imag, f"$n={n}$",
            color=colors[n], fontsize=10, ha="center", va="center")
    print(f"(1+i)^{n} = {complex(round(power.real), round(power.imag))}")

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-6, 18)
ax.set_ylim(-10, 6)
ax.set_aspect("equal")
ax.set_title("Powers of $1 + i$: rotate $45°$, scale by $\\sqrt{2}$ each step")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
plt.show()
```

### 5.3 $\cos 3\theta$ as a polynomial in $\cos\theta$

Plot $\cos 3\theta$ against $x = \cos\theta$ for $\theta \in [0, \pi]$: the points fall exactly on the cubic $T_3(x) = 4x^3 - 3x$. The same for $n = 4, 5$.

```python
import matplotlib.pyplot as plt
import numpy as np

theta = np.linspace(0, np.pi, 40)
x = np.linspace(-1, 1, 400)
chebyshev = {3: 4 * x**3 - 3 * x, 4: 8 * x**4 - 8 * x**2 + 1, 5: 16 * x**5 - 20 * x**3 + 5 * x}

plt.figure(figsize=(8, 5))
for n, polynomial in chebyshev.items():
    line, = plt.plot(x, polynomial, label=f"$T_{n}(x)$")
    plt.scatter(np.cos(theta), np.cos(n * theta), s=14, color=line.get_color(), zorder=3)
plt.axhline(0, color="black", linewidth=0.8)
plt.xlabel("$x = \\cos\\theta$")
plt.ylabel("$\\cos n\\theta$")
plt.title("Dots: $(\\cos\\theta, \\cos n\\theta)$.  Curves: $T_3 = 4x^3 - 3x$, $T_4$, $T_5$")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

### 5.4 The unit-circle picture of $(\cos\theta + i\sin\theta)^3$

On the unit circle, cubing moves the point at angle $\theta$ to angle $3\theta$. Below, a short arc of angles $[0, \frac{\pi}{2}]$ (blue) is mapped by $z \mapsto z^3$ to the arc $[0, \frac{3\pi}{2}]$ (orange): every angle is tripled, and the horizontal coordinate of the image is $\cos 3\theta$. The marked points A, B, C at $20°, 45°, 80°$ land at $60°, 135°, 240°$.

```python
import matplotlib.pyplot as plt
import numpy as np

thetas = np.linspace(0, np.pi / 2, 200)
arc = np.exp(1j * thetas)
cubed = arc ** 3
marked = np.exp(1j * np.array([np.pi / 9, np.pi / 4, 4 * np.pi / 9]))

fig, ax = plt.subplots(figsize=(7, 7))
circle = np.exp(1j * np.linspace(0, 2 * np.pi, 400))
ax.plot(circle.real, circle.imag, color="lightgray")
ax.plot(arc.real, arc.imag, color="tab:blue", linewidth=5, alpha=0.6, label="$\\theta \\in [0, \\pi/2]$")
ax.plot(cubed.real, cubed.imag, color="tab:orange", linewidth=2.5, label="image under $z^3$: angles $3\\theta$")
for point, symbol in zip(marked, ["A", "B", "C"]):
    image = point ** 3
    degrees = round(np.degrees(np.angle(point)))
    ax.scatter([point.real, image.real], [point.imag, image.imag], color=["tab:blue", "tab:orange"], zorder=3)
    ax.text(1.12 * point.real, 1.12 * point.imag, f"{symbol}: {degrees}°", color="tab:blue", ha="center", va="center")
    ax.text(1.18 * image.real, 1.18 * image.imag, f"{symbol}$^3$: {3 * degrees}°", color="tab:orange", ha="center", va="center")

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-1.4, 1.4)
ax.set_ylim(-1.4, 1.4)
ax.set_aspect("equal")
ax.set_title("$(\\cos\\theta + i\\sin\\theta)^3 = \\cos 3\\theta + i \\sin 3\\theta$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.legend(loc="lower left")
plt.show()
```

## 7. Solutions

1. $\sqrt3 + i = 2e^{i\pi/6}$, so $(\sqrt3 + i)^6 = 64e^{i\pi} = -64$. $\ -1 + i = \sqrt2 e^{3\pi i/4}$, so $(-1 + i)^{10} = 32e^{30\pi i/4} = 32e^{15\pi i/2} = 32e^{-i\pi/2} = -32i$ (reduce $\frac{15\pi}{2} - 8\pi = -\frac{\pi}{2}$).
2. $e^{-6\pi i/12} = e^{-i\pi/2} = -i$.
3. $(c + is)^2 = c^2 - s^2 + 2ics$; comparing with $\cos 2\theta + i\sin 2\theta$ gives both identities.
4. $z^5 = 0.95^5 e^{i\pi} \approx -0.774$: on the **negative real axis**, inside the unit circle. $z^{10} = 0.95^{10}e^{2\pi i} \approx 0.599$: on the **positive real axis**, further inside. (Each step turns $36°$, so five steps make a half-turn.)
5. $(c + is)^5$ has imaginary part $5c^4s - 10c^2s^3 + s^5$. Substitute $c^2 = 1 - s^2$: $5(1 - s^2)^2 s - 10(1 - s^2)s^3 + s^5 = 5s - 10s^3 + 5s^5 - 10s^3 + 10s^5 + s^5$, so $\sin 5\theta = 16\sin^5\theta - 20\sin^3\theta + 5\sin\theta$.
6. $\cos^4\theta = \frac{(e^{i\theta} + e^{-i\theta})^4}{16} = \frac{e^{4i\theta} + 4e^{2i\theta} + 6 + 4e^{-2i\theta} + e^{-4i\theta}}{16} = \frac{\cos 4\theta + 4\cos 2\theta + 3}{8}$. Over $[0, \pi]$ the cosine terms integrate to $0$, so $\int_0^\pi\cos^4\theta\, d\theta = \frac{3\pi}{8}$.
7. $(\sqrt3 + i)^n = 2^n e^{in\pi/6}$. (a) Real iff $n\pi/6 \in \pi\mathbb{Z}$, i.e. $6 \mid n$. (b) Purely imaginary iff $n\pi/6 \in \frac{\pi}{2} + \pi\mathbb{Z}$, i.e. $n \equiv 3 \pmod 6$. Positive real needs $n\pi/6 \in 2\pi\mathbb{Z}$, so $12 \mid n$; the smallest is $n = 12$, giving $2^{12} = 4096$.
8. $\cos 3\theta + i\sin 3\theta = (c + is)^3 = (c^3 - 3cs^2) + i(3c^2s - s^3)$. So $\tan 3\theta = \frac{3c^2s - s^3}{c^3 - 3cs^2}$. Divide numerator and denominator by $c^3$ and write $t = s/c$: $\tan 3\theta = \frac{3t - t^3}{1 - 3t^2}$.
9. $1 + e^{i\theta} = e^{i\theta/2}(e^{-i\theta/2} + e^{i\theta/2}) = 2\cos\frac{\theta}{2}\, e^{i\theta/2}$. So $(1 + e^{i\theta})^n = 2^n\cos^n\frac{\theta}{2}\, e^{in\theta/2}$, whose real part is $2^n\cos^n\frac{\theta}{2}\cos\frac{n\theta}{2}$. On the other hand, by the binomial theorem $(1 + e^{i\theta})^n = \sum_k\binom{n}{k}e^{ik\theta}$, whose real part is $\sum_k\binom{n}{k}\cos k\theta$.

## 8. Conclusion

- **De Moivre:** $(\cos\theta + i\sin\theta)^n = \cos n\theta + i\sin n\theta$ for every integer $n$; in exponential form, $\left(re^{i\theta}\right)^n = r^n e^{in\theta}$.
- Large powers become one-liners: raise the modulus to the $n$, multiply the angle by $n$, reduce the angle to $(-\pi, \pi]$.
- Expanding $(\cos\theta + i\sin\theta)^n$ binomially and splitting real and imaginary parts gives $\cos n\theta$ and $\sin n\theta$ as polynomials in $\cos\theta$, $\sin\theta$ (Chebyshev polynomials $T_n$).
- In the other direction, $\cos\theta = \frac{e^{i\theta} + e^{-i\theta}}{2}$ turns powers $\cos^n\theta$, $\sin^n\theta$ into sums of $\cos k\theta$ — ideal for integration.
- The sequence $z^n$ is a logarithmic spiral: outward if $|z| > 1$, inward if $|z| < 1$, around the unit circle if $|z| = 1$.
- The theorem is for **integer** exponents. Fractional exponents have several values, handled next.
