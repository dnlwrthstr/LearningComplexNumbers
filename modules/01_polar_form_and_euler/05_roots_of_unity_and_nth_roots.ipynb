# Roots of Unity and $n$-th Roots

Over the real numbers, $x^3 = 8$ has one solution and $x^4 = 16$ has two. Over the complex numbers, **every** nonzero number has exactly $n$ distinct $n$-th roots, and they sit at the vertices of a regular $n$-gon centred at the origin. This notebook derives that by running De Moivre's theorem backwards, studies the $n$-th **roots of unity** (the solutions of $z^n = 1$), shows that they sum to zero, identifies the **primitive** ones, and finally returns to Bombelli's cubic from notebook 00/01: the three cube roots of $2 + 11i$ produce all three real roots of $x^3 = 15x + 4$.

This is notebook 05 of module 01 (**Polar Form and Euler's Formula**) in the **Learning Complex Numbers** series.

## 1. Introduction

Notebook 04 raised complex numbers to integer powers: $\left(re^{i\theta}\right)^n = r^n e^{in\theta}$. Going backwards — given $w$, find all $z$ with $z^n = w$ — is where the $2\pi k$ ambiguity of the argument (notebook 01) finally matters. Angles that differ by a full turn describe the same number $w$, but after dividing by $n$ they describe *different* numbers $z$. That is why there are $n$ roots instead of one.

Roots of unity are everywhere:

- the discrete Fourier transform is built from powers of $e^{-2\pi i/N}$ (notebook 04/05, module 07);
- they factor $z^n - 1$ and appear in number theory and cryptography (cyclotomic polynomials);
- they describe $n$-fold rotational symmetry — regular polygons, crystal lattices, the $n$ equally-spaced phases of a three-phase power supply;
- they complete Cardano's formula: without the other cube roots, the cubic formula finds only one of three roots.

## 2. Theory

### 2.1 The $n$-th roots of unity

Solve $z^n = 1$. Write $z = \rho e^{i\varphi}$. Then $z^n = \rho^n e^{in\varphi}$ and $1 = 1\cdot e^{i\cdot 0}$. By the equality rule of notebook 01 (moduli equal, arguments equal **up to whole turns**):

$$
\rho^n = 1, \qquad n\varphi = 2\pi k \ \text{ for some } k \in \mathbb{Z}.
$$

Since $\rho > 0$ is real, $\rho = 1$, and $\varphi = \frac{2\pi k}{n}$. Values of $k$ that differ by a multiple of $n$ give the same point, so exactly $n$ of them are distinct:

$$
\boxed{\, \omega_k = e^{2\pi i k/n} = \cos\frac{2\pi k}{n} + i\sin\frac{2\pi k}{n}, \qquad k = 0, 1, \dots, n - 1. \,}
$$

These are the **$n$-th roots of unity**. They lie on the unit circle, equally spaced at angle $\frac{2\pi}{n}$, starting at $1$: the vertices of a **regular $n$-gon**.

Writing $\omega = \omega_1 = e^{2\pi i/n}$, every root is a power of it: $\omega_k = \omega^k$. Multiplying by $\omega$ rotates the polygon onto itself.

Examples: $n = 2$: $\pm 1$. $\ n = 3$: $1, \ -\frac12 \pm \frac{\sqrt3}{2}i$. $\ n = 4$: $1, i, -1, -i$. $\ n = 6$: the cube roots of unity together with their negatives.

### 2.2 They sum to zero

For $n \ge 2$,

$$
\sum_{k=0}^{n-1}\omega^k = 0.
$$

**Algebraic proof.** Geometric series with ratio $\omega \ne 1$: $\sum_{k=0}^{n-1}\omega^k = \frac{\omega^n - 1}{\omega - 1} = \frac{1 - 1}{\omega - 1} = 0$.

**Symmetry proof.** Let $S$ be the sum. Multiplying by $\omega$ just permutes the roots ($\omega\cdot\omega^k = \omega^{k+1}$, and $\omega^n = 1 = \omega^0$), so $\omega S = S$. Since $\omega \ne 1$, $S = 0$. Geometrically: the polygon's vertices balance at its centre — the centre of mass of a regular polygon is the origin.

**Polynomial proof.** The roots of unity are the roots of $z^n - 1 = z^n + 0\cdot z^{n-1} + \dots - 1$. The sum of the roots of a monic polynomial is minus the coefficient of $z^{n-1}$, which is $0$.

Taking real and imaginary parts gives two trigonometric facts for free: $\sum_{k=0}^{n-1}\cos\frac{2\pi k}{n} = 0$ and $\sum_{k=0}^{n-1}\sin\frac{2\pi k}{n} = 0$.

### 2.3 Primitive roots

$\omega^k$ is a **primitive** $n$-th root of unity if its powers produce *all* $n$ roots — equivalently, if $n$ is the smallest positive exponent with $(\omega^k)^n = 1$. This happens exactly when

$$
\gcd(k, n) = 1.
$$

If $d = \gcd(k, n) > 1$, then $(\omega^k)^{n/d} = \omega^{n\cdot(k/d)} = 1$ already, so $\omega^k$ is really an $(n/d)$-th root of unity in disguise. The number of primitive $n$-th roots is Euler's totient $\varphi(n)$. For $n = 6$: $\omega^1, \omega^5$ are primitive; $\omega^2, \omega^4$ are primitive cube roots; $\omega^3 = -1$ is a square root; $\omega^0 = 1$.

The polynomial whose roots are exactly the primitive $n$-th roots is the **cyclotomic polynomial** $\Phi_n(z)$, and $z^n - 1 = \prod_{d \mid n}\Phi_d(z)$.

### 2.4 The $n$-th roots of any $w \ne 0$

Write $w = Re^{i\phi}$. The same argument as in 2.1, now with $\rho^n = R$ and $n\varphi = \phi + 2\pi k$, gives

$$
\boxed{\, z_k = R^{1/n}\, e^{i(\phi + 2\pi k)/n}, \qquad k = 0, 1, \dots, n - 1, \,}
$$

where $R^{1/n}$ is the ordinary positive real root. All $n$ roots have the same modulus $R^{1/n}$ and are spaced by $\frac{2\pi}{n}$: a regular $n$-gon of radius $R^{1/n}$, rotated so that one vertex is at angle $\frac{\phi}{n}$.

Compactly: if $z_0$ is any one $n$-th root of $w$, then all of them are

$$
z_0, \ z_0\omega, \ z_0\omega^2, \ \dots, \ z_0\omega^{n-1}.
$$

The root with $k = 0$ and $\phi = \operatorname{Arg} w$ is the **principal $n$-th root** $\sqrt[n]{w} = R^{1/n}e^{i\operatorname{Arg}(w)/n}$; it is what Python's `w ** (1/n)` and `cmath.sqrt` return. Its argument lies in $(-\frac{\pi}{n}, \frac{\pi}{n}]$.

### 2.5 Back to Bombelli

Notebook 00/01 used Cardano's formula on $x^3 - 15x - 4 = 0$ and found $x = u + v$ with $u^3 = 2 + 11i$, $v^3 = 2 - 11i$, and the pairing condition $uv = -\frac{p}{3} = 5$. Bombelli spotted one cube root, $u = 2 + i$. Now we know there are **three**:

$$
u_k = (2 + i)\,\omega^k, \qquad \omega = e^{2\pi i/3} = -\tfrac12 + \tfrac{\sqrt3}{2}i, \qquad k = 0, 1, 2.
$$

The pairing $uv = 5 = |2 + i|^2$ forces $v_k = 5/u_k = \overline{u_k}$ (since $|u_k|^2 = 5$). So $x_k = u_k + \overline{u_k} = 2\operatorname{Re} u_k$, always **real**:

$$
\begin{aligned}
k = 0:&\quad u_0 = 2 + i, & x_0 &= 4, \\
k = 1:&\quad u_1 = (2 + i)\left(-\tfrac12 + \tfrac{\sqrt3}{2}i\right) = \left(-1 - \tfrac{\sqrt3}{2}\right) + \left(\sqrt3 - \tfrac12\right)i, & x_1 &= -2 - \sqrt3, \\
k = 2:&\quad u_2 = (2 + i)\left(-\tfrac12 - \tfrac{\sqrt3}{2}i\right) = \left(-1 + \tfrac{\sqrt3}{2}\right) + \left(-\sqrt3 - \tfrac12\right)i, & x_2 &= -2 + \sqrt3.
\end{aligned}
$$

These are exactly the three real roots $4, -2 \pm \sqrt3$ found by factoring in notebook 00/01. Cardano's formula, read with *all* cube roots, gives *all* roots — and each one passes through the complex plane.

## 3. Worked Examples

### Example 1 — the cube roots of unity

$n = 3$: $\omega_k = e^{2\pi ik/3}$ for $k = 0, 1, 2$:

$$
1, \qquad -\tfrac12 + \tfrac{\sqrt3}{2}i, \qquad -\tfrac12 - \tfrac{\sqrt3}{2}i.
$$

Sum: $1 - \frac12 - \frac12 + \left(\frac{\sqrt3}{2} - \frac{\sqrt3}{2}\right)i = 0$ ✓. Equilateral triangle on the unit circle. Note $\omega_2 = \overline{\omega_1} = \omega_1^2$, and the factorisation $z^3 - 1 = (z - 1)(z^2 + z + 1)$: the two non-real cube roots solve $z^2 + z + 1 = 0$.

### Example 2 — the fourth roots of $-16$

$-16 = 16e^{i\pi}$, so $R^{1/4} = 2$ and the arguments are $\frac{\pi + 2\pi k}{4} = \frac{\pi}{4}, \frac{3\pi}{4}, \frac{5\pi}{4}, \frac{7\pi}{4}$:

$$
z = 2e^{i\pi/4} = \sqrt2 + \sqrt2 i, \quad -\sqrt2 + \sqrt2 i, \quad -\sqrt2 - \sqrt2 i, \quad \sqrt2 - \sqrt2 i.
$$

A square of radius $2$, rotated $45°$. Check: $(\sqrt2 + \sqrt2 i)^2 = 2 + 4i - 2 = 4i$, and $(4i)^2 = -16$ ✓.

### Example 3 — the square roots of $i$

$i = e^{i\pi/2}$, so $\sqrt{i} = \pm e^{i\pi/4} = \pm\frac{\sqrt2}{2}(1 + i)$. The principal one is $\frac{\sqrt2}{2}(1 + i)$ (argument $\frac{\pi}{4} \in (-\frac{\pi}{2}, \frac{\pi}{2}]$).

### Example 4 — the cube roots of $8i$

$8i = 8e^{i\pi/2}$, so $R^{1/3} = 2$ and the arguments are $\frac{\pi/2 + 2\pi k}{3} = \frac{\pi}{6}, \frac{5\pi}{6}, \frac{3\pi}{2}$:

$$
z = \sqrt3 + i, \qquad -\sqrt3 + i, \qquad -2i.
$$

Check the last: $(-2i)^3 = -8i^3 = 8i$ ✓.

### Example 5 — primitive sixth roots

$n = 6$: $\gcd(k, 6) = 1$ for $k = 1, 5$ only, so the primitive sixth roots are $e^{\pm i\pi/3} = \frac12 \pm \frac{\sqrt3}{2}i$. Their minimal polynomial is $\Phi_6(z) = z^2 - z + 1$. Also $z^6 - 1 = \Phi_1\Phi_2\Phi_3\Phi_6 = (z - 1)(z + 1)(z^2 + z + 1)(z^2 - z + 1)$.

### Example 6 — the other cube roots of $2 + 11i$

Exactly as in §2.5: $u_1 = (2 + i)\omega \approx -1.866 + 1.232i$ and $u_2 = (2 + i)\omega^2 \approx -0.134 - 2.232i$. Cube $u_1$ to check: its modulus is $\sqrt5$ (so $|u_1^3| = 5\sqrt5 = |2 + 11i|$ ✓) and its argument is $\operatorname{Arg}(2 + i) + \frac{2\pi}{3}$, which tripled gives $3\operatorname{Arg}(2 + i) + 2\pi$ — the argument of $2 + 11i$ plus a full turn ✓.

## 4. Python Examples

### 4.1 Roots of unity with numpy

```python
import numpy as np


def roots_of_unity(n):
    k = np.arange(n)
    return np.exp(2j * np.pi * k / n)


for n in [2, 3, 4, 6]:
    roots = roots_of_unity(n)
    print(f"n = {n}: {np.round(roots, 6)}")
    print(f"        max |z^n - 1| = {np.max(np.abs(roots ** n - 1)):.1e},   |sum| = {abs(roots.sum()):.1e}")
```

The sums are zero up to rounding ($\sim 10^{-16}$), and $z^n = 1$ holds to the same accuracy — again, compare with a tolerance, never `==`.

### 4.2 All $n$-th roots of any $w$, versus Python's principal root

```python
import cmath

import numpy as np


def nth_roots(w, n):
    R, phi = cmath.polar(w)
    k = np.arange(n)
    return R ** (1 / n) * np.exp(1j * (phi + 2 * np.pi * k) / n)


for w, n in [(-16, 4), (1j, 2), (8j, 3), (2 + 11j, 3)]:
    roots = nth_roots(complex(w), n)
    print(f"{n} roots z with z^{n} = {w}:")
    for root in roots:
        print(f"    {root:.6f}    root^{n} = {root ** n:.6f}")
    principal = complex(w) ** (1 / n)
    print(f"    Python's w ** (1/{n}) = {principal:.6f}  (= the k = 0 root: {np.isclose(principal, roots[0])})\n")
```

### 4.3 Polynomial root-finders agree

The $n$-th roots of $w$ are the roots of the polynomial $z^n - w$; `np.roots` finds them numerically and `sympy` exactly.

```python
import numpy as np
import sympy as sp

coefficients = [1, 0, 0, -(2 + 11j)]   # z^3 - (2 + 11i)
print("np.roots(z^3 - (2+11i)) =", np.round(np.sort_complex(np.roots(coefficients)), 6))

z = sp.symbols("z")
exact = sp.solve(z**3 - (2 + 11 * sp.I), z)
print("sympy solve:")
for root in exact:
    simplified = sp.nsimplify(sp.expand(sp.simplify(root)))
    print("   ", simplified, "  cubed:", sp.expand(simplified**3))
```

### 4.4 Primitive roots and cyclotomic polynomials

```python
import math

import numpy as np
import sympy as sp


def multiplicative_order(root, n, tolerance=1e-9):
    for order in range(1, n + 1):
        if abs(root ** order - 1) < tolerance:
            return order


n = 12
omega = np.exp(2j * np.pi / n)
primitive = [k for k in range(n) if math.gcd(k, n) == 1]
print(f"n = {n}: primitive exponents k = {primitive}; count = {len(primitive)} = totient({n}) = {sp.totient(n)}")
for k in range(n):
    print(f"   omega^{k:<2d}: order {multiplicative_order(omega ** k, n):2d}   gcd(k, n) = {math.gcd(k, n)}")

z = sp.symbols("z")
print("\nz^12 - 1 =", sp.factor(z**12 - 1))
print("Phi_12(z) =", sp.cyclotomic_poly(12, z))
```

The order of $\omega^k$ is $n / \gcd(k, n)$, and the order-$12$ elements are exactly the primitive ones. `sp.factor` splits $z^{12} - 1$ over the rationals into the cyclotomic polynomials $\Phi_d$ for the divisors $d = 1, 2, 3, 4, 6, 12$.

### 4.5 Bombelli, completed

Apply Cardano's formula with each of the three cube roots $u_k$ of $2 + 11i$, pair it with $v_k = -p/(3u_k)$, and recover all three roots of $x^3 - 15x - 4 = 0$.

```python
import cmath

import numpy as np

p, q = -15, -4
radicand = -q / 2 + cmath.sqrt(q * q / 4 + p**3 / 27)       # 2 + 11i
print("u^3 must equal", radicand)

R, phi = cmath.polar(radicand)
cube_roots = [R ** (1 / 3) * cmath.exp(1j * (phi + 2 * np.pi * k) / 3) for k in range(3)]

found = []
for k, u in enumerate(cube_roots):
    v = -p / (3 * u)
    x = u + v
    found.append(x)
    print(f"k = {k}: u = {u:.6f}, v = {v:.6f} (conj(u)? {np.isclose(v, u.conjugate())}), x = u + v = {x:.6f}")

expected = np.array([4, -2 - np.sqrt(3), -2 + np.sqrt(3)])
print("\nexpected roots 4, -2 - sqrt3, -2 + sqrt3 =", np.round(expected, 6))
print("all match:", np.allclose(np.array(found), expected))
print("residuals x^3 - 15x - 4:", [f"{abs(x**3 - 15 * x - 4):.1e}" for x in found])
```

## 5. Visualizations

### 5.1 Roots of unity are regular polygons

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(2, 3, figsize=(13, 9))
circle = np.exp(1j * np.linspace(0, 2 * np.pi, 400))
for ax, n in zip(axes.ravel(), [3, 4, 5, 6, 8, 12]):
    roots = np.exp(2j * np.pi * np.arange(n) / n)
    polygon = np.append(roots, roots[0])
    ax.plot(circle.real, circle.imag, color="lightgray")
    ax.fill(polygon.real, polygon.imag, alpha=0.15, color="tab:blue")
    ax.plot(polygon.real, polygon.imag, color="tab:blue", linewidth=1)
    for root in roots:
        ax.annotate("", xy=(root.real, root.imag), xytext=(0, 0),
                    arrowprops=dict(arrowstyle="->", color="tab:blue", linewidth=1))
    ax.scatter(roots.real, roots.imag, color="crimson", zorder=3)
    ax.text(1.18 * roots[1].real, 1.18 * roots[1].imag, "$\\omega$", fontsize=12, ha="center", va="center")
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(-1.4, 1.4)
    ax.set_ylim(-1.4, 1.4)
    ax.set_aspect("equal")
    ax.set_title(f"$z^{{{n}}} = 1$:  {n} roots")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
fig.tight_layout()
plt.show()
```

### 5.2 Why the sum is zero: arrows head to tail

Place the arrows $1, \omega, \omega^2, \dots, \omega^{n-1}$ head to tail. Each turns by $\frac{2\pi}{n}$ relative to the previous one, so the chain is a closed regular polygon and returns to its starting point: the total displacement is $0$.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for ax, n in zip(axes, [3, 5, 8]):
    roots = np.exp(2j * np.pi * np.arange(n) / n)
    positions = np.concatenate([[0], np.cumsum(roots)])
    colors = plt.cm.viridis(np.linspace(0, 0.9, n))
    for k in range(n):
        start, end = positions[k], positions[k + 1]
        ax.annotate("", xy=(end.real, end.imag), xytext=(start.real, start.imag),
                    arrowprops=dict(arrowstyle="->", color=colors[k], linewidth=2))
        middle = (start + end) / 2
        ax.text(middle.real, middle.imag + 0.12, f"$\\omega^{k}$", color=colors[k], ha="center")
    ax.scatter([0], [0], color="crimson", zorder=3, s=50)
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(positions.real.min() - 0.6, positions.real.max() + 0.6)
    ax.set_ylim(positions.imag.min() - 0.6, positions.imag.max() + 0.6)
    ax.set_aspect("equal")
    ax.set_title(f"n = {n}: the chain closes, sum = 0")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

### 5.3 Primitive roots: who generates the whole polygon?

For $n = 12$, step around the polygon by $\omega^k$ (join $1 \to \omega^k \to \omega^{2k} \to \cdots$). If $\gcd(k, 12) = 1$ the path visits all twelve vertices (a "star polygon"); otherwise it closes early on a smaller polygon.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

n = 12
vertices = np.exp(2j * np.pi * np.arange(n) / n)
fig, axes = plt.subplots(2, 3, figsize=(13, 9))
for ax, k in zip(axes.ravel(), [1, 5, 7, 2, 3, 4]):
    steps = np.exp(2j * np.pi * k * np.arange(n + 1) / n)
    primitive = math.gcd(k, n) == 1
    ax.scatter(vertices.real, vertices.imag, color="lightgray", s=40, zorder=2)
    ax.plot(steps.real, steps.imag, color="tab:green" if primitive else "tab:red", linewidth=1.8)
    ax.scatter(steps.real, steps.imag, color="black", s=15, zorder=3)
    order = n // math.gcd(k, n)
    ax.set_title(f"k = {k}: gcd = {math.gcd(k, n)}, order {order}" + ("  (primitive)" if primitive else ""))
    ax.axhline(0, color="black", linewidth=0.5)
    ax.axvline(0, color="black", linewidth=0.5)
    ax.set_xlim(-1.3, 1.3)
    ax.set_ylim(-1.3, 1.3)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
fig.suptitle("Powers of $\\omega^k$ for $n = 12$: green = primitive (visits all 12 roots)")
fig.tight_layout()
plt.show()
```

### 5.4 The cube roots of $2 + 11i$ and Bombelli's real roots

Left: the three cube roots $u_k$ of $2 + 11i$ (an equilateral triangle of radius $\sqrt5$) and their conjugates $v_k = \bar u_k$, the cube roots of $2 - 11i$. Each pair $u_k + \bar u_k = 2\operatorname{Re} u_k$ lands on the real axis. Right: those three real numbers are exactly where $x^3 - 15x - 4$ crosses zero.

```python
import cmath

import matplotlib.pyplot as plt
import numpy as np

w = 2 + 11j
R, phi = cmath.polar(w)
u = R ** (1 / 3) * np.exp(1j * (phi + 2 * np.pi * np.arange(3)) / 3)
v = np.conj(u)
x = (u + v).real
colors = ["tab:blue", "tab:orange", "tab:green"]

fig, (ax_plane, ax_graph) = plt.subplots(1, 2, figsize=(14, 6.5))
circle = np.sqrt(5) * np.exp(1j * np.linspace(0, 2 * np.pi, 400))
ax_plane.plot(circle.real, circle.imag, color="lightgray", label="$|z| = \\sqrt{5}$")
triangle = np.append(u, u[0])
ax_plane.plot(triangle.real, triangle.imag, color="gray", linewidth=0.8)
for k in range(3):
    ax_plane.scatter([u[k].real], [u[k].imag], color=colors[k], s=70, zorder=3)
    ax_plane.scatter([v[k].real], [v[k].imag], color=colors[k], s=70, marker="s", zorder=3)
    ax_plane.plot([u[k].real, v[k].real], [u[k].imag, v[k].imag], ":", color=colors[k])
    ax_plane.scatter([x[k]], [0], color=colors[k], s=120, marker="*", zorder=4,
                     label=f"$u_{k} + \\bar u_{k} = {x[k]:.4f}$")
    ax_plane.text(u[k].real + 0.15, u[k].imag + 0.15, f"$u_{k}$", color=colors[k], fontsize=12)
    ax_plane.text(v[k].real + 0.15, v[k].imag - 0.35, f"$\\bar u_{k}$", color=colors[k], fontsize=12)
ax_plane.axhline(0, color="black", linewidth=0.8)
ax_plane.axvline(0, color="black", linewidth=0.8)
ax_plane.set_xlim(-4.5, 4.8)
ax_plane.set_ylim(-3.2, 3.2)
ax_plane.set_aspect("equal")
ax_plane.set_title("Cube roots of $2 \\pm 11i$ (circles $u_k$, squares $\\bar u_k$)")
ax_plane.set_xlabel("Re")
ax_plane.set_ylabel("Im")
ax_plane.grid(True, alpha=0.3)
ax_plane.legend(loc="upper center", bbox_to_anchor=(0.5, -0.12), ncol=2, fontsize=9)

xs = np.linspace(-4.5, 4.8, 400)
ax_graph.plot(xs, xs**3 - 15 * xs - 4, color="black", label="$x^3 - 15x - 4$")
for k in range(3):
    ax_graph.scatter([x[k]], [0], color=colors[k], s=120, marker="*", zorder=3)
ax_graph.axhline(0, color="black", linewidth=0.8)
ax_graph.set_xlabel("x")
ax_graph.set_ylabel("y")
ax_graph.set_title("The same three numbers are the real roots")
ax_graph.grid(True, alpha=0.3)
ax_graph.legend(loc="upper left")
plt.show()
```

## 7. Solutions

1. $1, i, -1, -i$. Sum: $(1 - 1) + (i - i) = 0$. Product: $1\cdot i\cdot(-1)\cdot(-i) = i\cdot i = -1$.

5. $\gcd(k, 8) = 1$ for $k = 1, 3, 5, 7$: $e^{i\pi/4}, e^{3\pi i/4}, e^{5\pi i/4}, e^{7\pi i/4} = \frac{\sqrt2}{2}(\pm 1 \pm i)$. $\Phi_8(z) = \frac{z^8 - 1}{z^4 - 1} = z^4 + 1$.
6. Real part of the sum: $1 + \cos\frac{2\pi}{5} + \cos\frac{4\pi}{5} + \cos\frac{6\pi}{5} + \cos\frac{8\pi}{5} = 0$. Since $\cos\frac{8\pi}{5} = \cos\frac{2\pi}{5}$ and $\cos\frac{6\pi}{5} = \cos\frac{4\pi}{5}$, this is $1 + 2\left(\cos\frac{2\pi}{5} + \cos\frac{4\pi}{5}\right) = 0$.
7. $z^n - 1 = \prod_k(z - \omega^k)$. Comparing constant terms: $-1 = (-1)^n\prod_k\omega^k$, so $\prod_k\omega^k = (-1)^{n+1}$. (Alternatively, $\prod_k\omega^k = \omega^{0 + 1 + \dots + (n-1)} = e^{2\pi i\cdot\frac{n(n-1)}{2}/n} = e^{i\pi(n-1)} = (-1)^{n-1}$.)
8. $p = -6$, $q = -4$: $u^3 = 2 + 2i = 2\sqrt2 e^{i\pi/4}$, so $u_k = \sqrt2 e^{i(\pi/4 + 2\pi k)/3}$. One of them is $-1 + i = \sqrt2 e^{3\pi i/4}$ ($k = 1$). Pairing $uv = 2 = |u|^2$ again gives $v = \bar u$, so $x_k = 2\operatorname{Re} u_k = 2\sqrt2\cos\frac{\pi/4 + 2\pi k}{3}$: $2\sqrt2\cos\frac{\pi}{12} = 1 + \sqrt3$, $\ 2\sqrt2\cos\frac{3\pi}{4} = -2$, $\ 2\sqrt2\cos\frac{17\pi}{12} = 1 - \sqrt3$.
9. Let $q = \omega^m$. If $n \mid m$ then $q = 1$ and each term is $1$, sum $n$. Otherwise $q \ne 1$ and $q^n = (\omega^n)^m = 1$, so the geometric series gives $\frac{q^n - 1}{q - 1} = 0$.

Numerical checks:

```python
import cmath
import math

import numpy as np
import sympy as sp


def nth_roots(w, n):
    R, phi = cmath.polar(complex(w))
    return R ** (1 / n) * np.exp(1j * (phi + 2 * np.pi * np.arange(n)) / n)


fourth = nth_roots(1, 4)
print("Ex 1:", np.round(fourth, 12), " |sum| =", f"{abs(fourth.sum()):.1e}", " product =", np.round(np.prod(fourth).real, 12) + 0.0)
print("Ex 2:", np.round(nth_roots(-27, 3), 6))
print("Ex 3:", np.round(nth_roots(-2j, 2), 12), " cmath.sqrt(-2j) =", cmath.sqrt(-2j))
fifth = nth_roots(32j, 5)
print("Ex 4:", np.round(fifth, 6), " angles (deg):", np.round(np.degrees(np.angle(fifth)), 3))

z = sp.symbols("z")
primitive_8 = [np.exp(2j * np.pi * k / 8) for k in range(8) if math.gcd(k, 8) == 1]
print("Ex 5:", np.round(primitive_8, 6), " Phi_8 =", sp.cyclotomic_poly(8, z))
print("Ex 6:", math.cos(2 * math.pi / 5) + math.cos(4 * math.pi / 5))
print("Ex 7: products for n = 1..6:", [round(np.prod(nth_roots(1, n)).real, 12) + 0.0 for n in range(1, 7)])

u_values = nth_roots(2 + 2j, 3)
x_values = [u + (-(-6) / (3 * u)) for u in u_values]
print("Ex 8:", np.round(x_values, 12), " expected", sorted([-2, 1 + math.sqrt(3), 1 - math.sqrt(3)]))

n = 6
omega = np.exp(2j * np.pi / n)
print("Ex 9: sums for m = 0..7 (n = 6):", [round(sum(omega ** (k * m) for k in range(n)).real, 12) + 0.0 for m in range(0, 8)])
```

## 8. Conclusion

- $z^n = 1$ has exactly $n$ solutions $\omega^k = e^{2\pi ik/n}$, $k = 0, \dots, n - 1$: the vertices of a regular $n$-gon on the unit circle, generated by powers of $\omega = e^{2\pi i/n}$.
- For $n \ge 2$ they sum to $0$ — by the geometric series, by rotational symmetry, or because the coefficient of $z^{n-1}$ in $z^n - 1$ is zero.
- $\omega^k$ is **primitive** (generates all $n$ roots) iff $\gcd(k, n) = 1$; there are $\varphi(n)$ of them, the roots of the cyclotomic polynomial $\Phi_n$.
- Every $w = Re^{i\phi} \ne 0$ has $n$ distinct $n$-th roots $R^{1/n}e^{i(\phi + 2\pi k)/n}$ — one root times all the roots of unity. Python's `w ** (1/n)` and `cmath.sqrt` return only the principal one.
- With all three cube roots of $2 + 11i$, Cardano's formula yields all three real roots $4, -2 \pm \sqrt3$ of Bombelli's cubic — closing the story begun in notebook 00/01.
