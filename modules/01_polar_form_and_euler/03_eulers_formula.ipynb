# Euler's Formula

$$
e^{i\theta} = \cos\theta + i\sin\theta
$$

This notebook derives Euler's formula twice — once from the Taylor series of $e^x$, $\cos x$ and $\sin x$, and once from the differential equation $f' = if$ — and watches the partial sums of the series curl onto the unit circle. It then rewrites polar form as $z = re^{i\theta}$, meets Euler's identity $e^{i\pi} + 1 = 0$, and inverts the formula to express $\cos$ and $\sin$ through exponentials.

This is notebook 03 of module 01 (**Polar Form and Euler's Formula**) in the **Learning Complex Numbers** series.

## 1. Introduction

Notebook 02 showed that the unit vectors $u(\theta) = \cos\theta + i\sin\theta$ obey

$$
u(\alpha)\,u(\beta) = u(\alpha + \beta), \qquad u(0) = 1.
$$

A function that turns addition into multiplication is the defining behaviour of an **exponential**: $a^{x}a^{y} = a^{x + y}$. Euler's formula says $u(\theta)$ is not just *like* an exponential — it *is* one, $u(\theta) = e^{i\theta}$.

The payoff is enormous. Polar form becomes $re^{i\theta}$, and all the rules of exponents apply. Trigonometric identities turn into algebra. Rotations, oscillations and waves become exponentials: an AC current, a pendulum, a quantum state's phase, a Fourier mode $e^{2\pi i k x}$. Nearly every later notebook in this series writes complex numbers this way.

## 2. Theory

### 2.1 What could $e^{i\theta}$ even mean?

"$e$ multiplied by itself $i\theta$ times" is meaningless. We need a *definition* of $e^z$ for complex $z$ that agrees with the real exponential when $z$ is real. The power series is the natural candidate, because it uses only addition and multiplication, which already work for complex numbers:

$$
e^{z} = \sum_{n=0}^{\infty} \frac{z^n}{n!} = 1 + z + \frac{z^2}{2!} + \frac{z^3}{3!} + \cdots
$$

This series converges for every complex $z$ (the factorials beat any power; module 06 makes this precise), and for real $z$ it is the familiar $e^x$.

### 2.2 Derivation 1: Taylor series

Put $z = i\theta$. The powers of $i$ cycle with period four, $i^0, i^1, i^2, i^3 = 1, i, -1, -i$, so

$$
\begin{aligned}
e^{i\theta} &= 1 + i\theta + \frac{(i\theta)^2}{2!} + \frac{(i\theta)^3}{3!} + \frac{(i\theta)^4}{4!} + \frac{(i\theta)^5}{5!} + \cdots \\
&= 1 + i\theta - \frac{\theta^2}{2!} - i\frac{\theta^3}{3!} + \frac{\theta^4}{4!} + i\frac{\theta^5}{5!} - \cdots
\end{aligned}
$$

Collect the real terms (even powers) and the imaginary terms (odd powers). (Rearranging is allowed because the series converges absolutely.)

$$
e^{i\theta} = \underbrace{\left(1 - \frac{\theta^2}{2!} + \frac{\theta^4}{4!} - \cdots\right)}_{\cos\theta}
+ i\underbrace{\left(\theta - \frac{\theta^3}{3!} + \frac{\theta^5}{5!} - \cdots\right)}_{\sin\theta}.
$$

The two brackets are exactly the Taylor series of $\cos\theta$ and $\sin\theta$. Hence

$$
\boxed{\, e^{i\theta} = \cos\theta + i\sin\theta \,}
$$

### 2.3 Derivation 2: the differential equation $f' = if$

The real exponential is characterised by $f' = af$, $f(0) = 1$, with solution $f(t) = e^{at}$. Ask for the solution with $a = i$:

$$
f'(t) = i\,f(t), \qquad f(0) = 1.
$$

**Geometric reading.** Think of $f(t)$ as a moving point and $f'(t)$ as its velocity. Multiplying by $i$ rotates by $90°$, so the velocity is always **perpendicular** to the position vector and has the **same length** as it. Motion perpendicular to the radius does not change the distance from the origin: the point moves on a circle. Starting at $1$, the circle is the unit circle, the speed is $|f| = 1$, so after time $t$ the point has travelled arc length $t$ — it sits at angle $t$, i.e. at $\cos t + i\sin t$.

**Algebraic check.** $g(t) = \cos t + i\sin t$ satisfies $g'(t) = -\sin t + i\cos t = i(\cos t + i\sin t) = i\,g(t)$ and $g(0) = 1$. To see that it is the *only* solution, let $f$ be any solution and look at $h(t) = f(t)\,g(-t)$, where $g(-t) = \cos t - i\sin t$. Then

$$
h'(t) = f'(t)g(-t) - f(t)g'(-t) = i f(t) g(-t) - f(t)\, i\, g(-t) = 0,
$$

so $h$ is constant, $h(t) = h(0) = 1$. Since $g(t)g(-t) = \cos^2 t + \sin^2 t = 1$, we get $f(t) = g(t)$. The solution of $f' = if$, $f(0) = 1$ — which deserves the name $e^{it}$ — is $\cos t + i\sin t$.

### 2.4 Euler's identity

At $\theta = \pi$: $e^{i\pi} = \cos\pi + i\sin\pi = -1$, so

$$
e^{i\pi} + 1 = 0.
$$

Five fundamental constants ($0, 1, e, i, \pi$) and three operations in one line. Geometrically it just says: **walk half-way around the unit circle and you reach $-1$**. Also useful: $e^{i\pi/2} = i$, $e^{2\pi i} = 1$, and $e^{i(\theta + 2\pi)} = e^{i\theta}$ — the complex exponential is periodic with period $2\pi i$.

### 2.5 Exponential polar form

Every nonzero complex number becomes

$$
z = r(\cos\theta + i\sin\theta) = re^{i\theta}, \qquad r = |z|,\ \theta \in \arg z.
$$

The rules of notebook 02 are now the laws of exponents:

$$
re^{i\alpha}\cdot se^{i\beta} = rs\,e^{i(\alpha + \beta)}, \qquad
\frac{re^{i\alpha}}{se^{i\beta}} = \frac{r}{s}e^{i(\alpha - \beta)}, \qquad
\overline{re^{i\theta}} = re^{-i\theta}, \qquad
\left|e^{i\theta}\right| = 1 \ \ (\theta \text{ real}).
$$

For a general exponent $z = x + iy$, $e^{z} = e^{x}e^{iy}$: modulus $e^{x}$, argument $y$. (Notebook 03/02 studies this map in depth.)

### 2.6 Cosine and sine from exponentials

Write the formula for $\theta$ and $-\theta$, using that $\cos$ is even and $\sin$ is odd:

$$
e^{i\theta} = \cos\theta + i\sin\theta, \qquad e^{-i\theta} = \cos\theta - i\sin\theta.
$$

Add and subtract:

$$
\boxed{\, \cos\theta = \frac{e^{i\theta} + e^{-i\theta}}{2}, \qquad \sin\theta = \frac{e^{i\theta} - e^{-i\theta}}{2i} \,}
$$

Geometrically: $e^{i\theta}$ and $e^{-i\theta}$ are mirror images in the real axis; their average is the real point $\cos\theta$, and half their difference is $i\sin\theta$. A real oscillation $\cos\theta$ is the sum of **two counter-rotating** complex exponentials — the starting point of Fourier analysis (module 07).

## 3. Worked Examples

### Example 1 — evaluate exponentials

- $e^{i\pi/2} = \cos\frac{\pi}{2} + i\sin\frac{\pi}{2} = i$.
- $e^{i\pi/3} = \frac12 + \frac{\sqrt3}{2}i$.
- $e^{-i\pi/4} = \frac{\sqrt2}{2} - \frac{\sqrt2}{2}i$.
- $e^{2 + i\pi} = e^2 e^{i\pi} = -e^2 \approx -7.389$.
- $3e^{i\cdot 5\pi/2} = 3e^{i\pi/2} = 3i$ (subtract one full turn).

### Example 2 — convert to exponential form

$z = -1 + \sqrt3 i$ has $r = 2$, $\operatorname{Arg} z = \frac{2\pi}{3}$ (notebook 01), so $z = 2e^{2\pi i/3}$.
$z = -2 - 2i = 2\sqrt2\, e^{-3\pi i/4}$.

### Example 3 — multiply and divide in exponential form

$$
\frac{(1 + i)^{6}}{(\sqrt3 + i)^{2}} = \frac{\left(\sqrt2 e^{i\pi/4}\right)^{6}}{\left(2e^{i\pi/6}\right)^{2}}
= \frac{8\,e^{3\pi i/2}}{4\,e^{i\pi/3}} = 2e^{i(3\pi/2 - \pi/3)} = 2e^{7\pi i/6} = -\sqrt3 - i.
$$

### Example 4 — an identity from $\cos\theta = \frac{e^{i\theta} + e^{-i\theta}}{2}$

$$
\cos^2\theta = \frac{\left(e^{i\theta} + e^{-i\theta}\right)^2}{4} = \frac{e^{2i\theta} + 2 + e^{-2i\theta}}{4}
= \frac12 + \frac12\cdot\frac{e^{2i\theta} + e^{-2i\theta}}{2} = \frac{1 + \cos 2\theta}{2}.
$$

The half-angle identity drops out of pure algebra.

### Example 5 — the angle-addition formula, recovered

$$
\cos(\alpha + \beta) + i\sin(\alpha + \beta) = e^{i(\alpha + \beta)} = e^{i\alpha}e^{i\beta}
= (\cos\alpha + i\sin\alpha)(\cos\beta + i\sin\beta).
$$

Expanding the right side and matching real and imaginary parts gives both addition formulas — the argument of notebook 02 run backwards.

### Example 6 — $i^i$ is real

$i = e^{i\pi/2}$, so (taking this principal representation) $i^i = \left(e^{i\pi/2}\right)^i = e^{i \cdot i\pi/2} = e^{-\pi/2} \approx 0.2079$. A real number! Other choices $i = e^{i(\pi/2 + 2\pi k)}$ give other real values $e^{-\pi/2 - 2\pi k}$; notebook 03/04 sorts out complex powers properly.

## 4. Python Examples

### 4.1 Partial sums of the series

Sum $\sum_{n=0}^{N} (i\theta)^n / n!$ for increasing $N$ and compare with $\cos\theta + i\sin\theta$.

```python
import cmath
import math


def exp_partial_sum(z, n_terms):
    total = 0
    term = 1 + 0j            # z^0 / 0!
    for n in range(n_terms):
        total += term
        term = term * z / (n + 1)   # next term: z^(n+1) / (n+1)!
    return total


theta = 2.0
target = complex(math.cos(theta), math.sin(theta))
print(f"cos(2) + i sin(2) = {target:.12f}\n")
for n_terms in [1, 2, 3, 4, 6, 8, 12, 16, 20]:
    approximation = exp_partial_sum(1j * theta, n_terms)
    print(f"N = {n_terms:2d} terms: {approximation:.12f}   error = {abs(approximation - target):.2e}")
```

The error shrinks faster than geometrically once $N$ exceeds $\theta$ — the $n!$ in the denominator wins. For large $\theta$ more terms are needed before that happens.

### 4.2 `cmath.exp` and `np.exp` agree with $\cos + i\sin$

```python
import cmath

import numpy as np

thetas = np.linspace(-10, 10, 2001)
lhs = np.exp(1j * thetas)
rhs = np.cos(thetas) + 1j * np.sin(thetas)
print("max |e^{i theta} - (cos + i sin)| over [-10, 10]:", np.max(np.abs(lhs - rhs)))
print("max | |e^{i theta}| - 1 |                      :", np.max(np.abs(np.abs(lhs) - 1)))

print("\ncmath.exp(i pi/2)  =", cmath.exp(1j * cmath.pi / 2))
print("cmath.exp(i pi/3)  =", cmath.exp(1j * cmath.pi / 3))
print("cmath.exp(2 + i pi)=", cmath.exp(2 + 1j * cmath.pi), "  -e^2 =", -cmath.e ** 2)
```

### 4.3 Euler's identity, numerically and exactly

Floating point leaves a residue of about $10^{-16}$ in the imaginary part (because $\pi$ itself is rounded), which is why we compare with a tolerance. Sympy, working with the exact symbol $\pi$, gets exactly $-1$.

```python
import cmath

import sympy as sp

value = cmath.exp(1j * cmath.pi)
print("cmath.exp(i*pi)        =", value)
print("e^(i pi) + 1           =", value + 1)
print("== -1 ?                ", value == -1, " <- never compare complex floats with ==")
print("within 1e-12 of -1 ?   ", abs(value + 1) < 1e-12)
print("sympy exp(I*pi) + 1    =", sp.exp(sp.I * sp.pi) + 1)
print("sympy exp(I*pi/2)      =", sp.exp(sp.I * sp.pi / 2))
print("i^i (principal)        =", (1j) ** 1j, " = e^(-pi/2) =", cmath.exp(-cmath.pi / 2))
```

### 4.4 Symbolic derivation with sympy

Sympy can expand $e^{i\theta}$ into its Taylor series, split it into real and imaginary parts, and compare with the series for $\cos$ and $\sin$. `expand(complex=True)` and `rewrite` show the formula directly.

```python
import sympy as sp

theta = sp.symbols("theta", real=True)
order = 10

series_exp = sp.series(sp.exp(sp.I * theta), theta, 0, order).removeO()
real_part = sp.re(sp.expand(series_exp))
imag_part = sp.im(sp.expand(series_exp))

print("series of e^{i theta}:", sp.expand(series_exp))
print("real part            :", real_part)
print("series of cos        :", sp.series(sp.cos(theta), theta, 0, order).removeO())
print("imaginary part       :", imag_part)
print("series of sin        :", sp.series(sp.sin(theta), theta, 0, order).removeO())
print()
print("exp(I theta), expand(complex=True):", sp.exp(sp.I * theta).expand(complex=True))
print("cos(theta).rewrite(exp)           :", sp.cos(theta).rewrite(sp.exp))
print("sin(theta).rewrite(exp)           :", sp.sin(theta).rewrite(sp.exp))
```

### 4.5 Solving $f' = if$ numerically

Integrate $f' = if$, $f(0) = 1$ with a simple step-by-step method and compare with $e^{it}$. The midpoint method below (a second-order Runge–Kutta scheme) stays close to the unit circle; the crude forward Euler step $f \leftarrow f + h\,if$ multiplies $|f|$ by $|1 + ih| = \sqrt{1 + h^2} > 1$ every step and spirals outwards.

```python
import numpy as np


def integrate(step_function, t_end, n_steps):
    h = t_end / n_steps
    f = 1 + 0j
    for _ in range(n_steps):
        f = step_function(f, h)
    return f


def forward_euler_step(f, h):
    return f + h * (1j * f)


def midpoint_step(f, h):
    half = f + 0.5 * h * (1j * f)
    return f + h * (1j * half)


t_end = 2 * np.pi
exact = np.exp(1j * t_end)
for n_steps in [20, 200, 2000]:
    euler = integrate(forward_euler_step, t_end, n_steps)
    midpoint = integrate(midpoint_step, t_end, n_steps)
    print(f"{n_steps:5d} steps:  forward Euler |f| = {abs(euler):.6f}, error {abs(euler - exact):.2e}   "
          f"midpoint |f| = {abs(midpoint):.8f}, error {abs(midpoint - exact):.2e}")
```

### 4.6 Cosine and sine from exponentials

Check $\cos\theta = \frac{e^{i\theta} + e^{-i\theta}}{2}$, $\sin\theta = \frac{e^{i\theta} - e^{-i\theta}}{2i}$ numerically, then use sympy to prove Example 4 and the triple-product identity $\sin^3\theta = \frac{3\sin\theta - \sin 3\theta}{4}$ by rewriting in exponentials.

```python
import numpy as np
import sympy as sp

theta_values = np.linspace(-6, 6, 1001)
cos_from_exp = (np.exp(1j * theta_values) + np.exp(-1j * theta_values)) / 2
sin_from_exp = (np.exp(1j * theta_values) - np.exp(-1j * theta_values)) / (2j)
print("cos formula max error:", np.max(np.abs(cos_from_exp - np.cos(theta_values))))
print("sin formula max error:", np.max(np.abs(sin_from_exp - np.sin(theta_values))))

theta = sp.symbols("theta", real=True)
for lhs, rhs in [(sp.cos(theta) ** 2, (1 + sp.cos(2 * theta)) / 2),
                 (sp.sin(theta) ** 3, (3 * sp.sin(theta) - sp.sin(3 * theta)) / 4)]:
    difference = sp.expand((lhs - rhs).rewrite(sp.exp))
    print(f"{lhs} = {rhs}:  difference in exponential form simplifies to {sp.simplify(difference)}")
```

## 5. Visualizations

### 5.1 Partial sums spiral onto the unit circle

For $\theta = 2$, the partial sums $S_N = \sum_{n=0}^{N} (i\theta)^n/n!$ are drawn as a chain of arrows: each new term $(i\theta)^n/n!$ is turned a further $90°$ from the previous one (another factor of $i$) and, once $n > \theta$, is shorter. The chain spirals in on the point $e^{2i}$ on the unit circle.

```python
import matplotlib.pyplot as plt
import numpy as np

theta = 2.0
n_terms = 12
terms = [(1j * theta) ** n / np.prod(np.arange(1, n + 1)) for n in range(n_terms)]
partial_sums = np.concatenate([[0], np.cumsum(terms)])

fig, ax = plt.subplots(figsize=(7, 7))
circle = np.exp(1j * np.linspace(0, 2 * np.pi, 400))
ax.plot(circle.real, circle.imag, color="lightgray", label="unit circle")
colors = plt.cm.viridis(np.linspace(0, 0.9, n_terms))
for n in range(n_terms):
    start, end = partial_sums[n], partial_sums[n + 1]
    ax.annotate("", xy=(end.real, end.imag), xytext=(start.real, start.imag),
                arrowprops=dict(arrowstyle="->", color=colors[n], linewidth=2))
    if n < 6:
        ax.text(end.real + 0.06, end.imag + 0.06, f"$S_{n}$", color=colors[n], fontsize=11)

target = np.exp(1j * theta)
ax.scatter([target.real], [target.imag], color="crimson", s=90, zorder=4, label="$e^{2i} = \\cos 2 + i \\sin 2$")
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-2.6, 1.6)
ax.set_ylim(-0.9, 3.0)
ax.set_aspect("equal")
ax.set_title("Partial sums of $\\sum (2i)^n / n!$ converge to $e^{2i}$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower left")
plt.show()
```

### 5.2 Polynomial approximations of the whole circle

Now let $\theta$ run over $[0, 2\pi]$ and draw the curve $\theta \mapsto S_N(i\theta)$ for several $N$. Low-order truncations wander off; by $N \approx 20$ the polynomial traces the unit circle to plotting accuracy.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

thetas = np.linspace(0, 2 * np.pi, 600)
fig, ax = plt.subplots(figsize=(7, 7))
for n_terms in [2, 4, 6, 8, 12, 20]:
    curve = sum((1j * thetas) ** n / math.factorial(n) for n in range(n_terms))
    ax.plot(curve.real, curve.imag, label=f"$N = {n_terms}$ terms")
circle = np.exp(1j * thetas)
ax.plot(circle.real, circle.imag, "k--", linewidth=1, label="$e^{i\\theta}$")

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.set_aspect("equal")
ax.set_title("$\\theta \\mapsto \\sum_{n<N} (i\\theta)^n/n!$  for  $\\theta \\in [0, 2\\pi]$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower left")
plt.show()
```

### 5.3 Velocity perpendicular to position: the ODE picture

Along $f(t) = e^{it}$, the velocity $f'(t) = if(t)$ is the position rotated by $90°$. Red arrows: positions; blue arrows: velocities, attached at the moving point. They are always tangent to the circle and always of length 1.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(7, 7))
circle = np.exp(1j * np.linspace(0, 2 * np.pi, 400))
ax.plot(circle.real, circle.imag, color="lightgray")

for t in np.linspace(0, 2 * np.pi, 12, endpoint=False):
    position = np.exp(1j * t)
    velocity = 1j * position
    ax.annotate("", xy=(position.real, position.imag), xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color="tab:red", linewidth=1.5))
    ax.annotate("", xy=(position.real + 0.45 * velocity.real, position.imag + 0.45 * velocity.imag),
                xytext=(position.real, position.imag),
                arrowprops=dict(arrowstyle="->", color="tab:blue", linewidth=1.5))

ax.plot([], [], color="tab:red", label="position $f(t) = e^{it}$")
ax.plot([], [], color="tab:blue", label="velocity $f'(t) = i f(t)$ (scaled 0.45)")
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-1.7, 1.7)
ax.set_ylim(-1.7, 1.7)
ax.set_aspect("equal")
ax.set_title("$f' = if$: velocity is position turned by $90°$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.legend(loc="lower left")
plt.show()
```

### 5.4 Cosine as the average of two counter-rotating exponentials

$e^{i\theta}$ turns counter-clockwise and $e^{-i\theta}$ clockwise. Their sum (the diagonal of the rhombus) always lies on the real axis and equals $2\cos\theta$. The right panel shows the real and imaginary parts of $e^{i\theta}$ against $\theta$: a cosine and a sine wave.

```python
import matplotlib.pyplot as plt
import numpy as np

theta = 1.0
forward = np.exp(1j * theta)
backward = np.exp(-1j * theta)

fig, (ax_plane, ax_wave) = plt.subplots(1, 2, figsize=(13, 5.5))
circle = np.exp(1j * np.linspace(0, 2 * np.pi, 400))
ax_plane.plot(circle.real, circle.imag, color="lightgray")
for number, color, label in [(forward, "tab:blue", "$e^{i\\theta}$"), (backward, "tab:orange", "$e^{-i\\theta}$"),
                             (forward + backward, "tab:green", "$e^{i\\theta} + e^{-i\\theta} = 2\\cos\\theta$")]:
    ax_plane.annotate("", xy=(number.real, number.imag), xytext=(0, 0),
                      arrowprops=dict(arrowstyle="->", color=color, linewidth=2))
    ax_plane.plot([], [], color=color, label=label)
ax_plane.plot([forward.real, (forward + backward).real], [forward.imag, 0], ":", color="gray")
ax_plane.plot([backward.real, (forward + backward).real], [backward.imag, 0], ":", color="gray")
ax_plane.axhline(0, color="black", linewidth=0.8)
ax_plane.axvline(0, color="black", linewidth=0.8)
ax_plane.set_xlim(-1.3, 1.6)
ax_plane.set_ylim(-1.3, 1.3)
ax_plane.set_aspect("equal")
ax_plane.set_title("$\\theta = 1$: the imaginary parts cancel")
ax_plane.set_xlabel("Re")
ax_plane.set_ylabel("Im")
ax_plane.legend(loc="lower left", fontsize=9)

thetas = np.linspace(0, 4 * np.pi, 500)
ax_wave.plot(thetas, np.exp(1j * thetas).real, label="$\\mathrm{Re}\\, e^{i\\theta} = \\cos\\theta$")
ax_wave.plot(thetas, np.exp(1j * thetas).imag, label="$\\mathrm{Im}\\, e^{i\\theta} = \\sin\\theta$")
ax_wave.axhline(0, color="black", linewidth=0.8)
ax_wave.set_xlabel("$\\theta$")
ax_wave.set_title("Projections of uniform circular motion")
ax_wave.grid(True, alpha=0.3)
ax_wave.set_ylim(-1.15, 1.6)
ax_wave.legend(loc="upper center", ncol=2)
plt.show()
```

## 7. Solutions

1. $e^{i\pi/6} = \frac{\sqrt3}{2} + \frac12 i$. $\ 4e^{-i\pi/2} = -4i$. $\ e^{1 + i\pi/4} = e\cdot e^{i\pi/4} = \frac{e\sqrt2}{2}(1 + i) \approx 1.922 + 1.922i$.
2. $-3 = 3e^{i\pi}$. $\ 1 - i = \sqrt2 e^{-i\pi/4}$. $\ -\sqrt3 + i = 2e^{5\pi i/6}$.
3. $|e^{i\theta}|^2 = \cos^2\theta + \sin^2\theta = 1$. $\overline{e^{i\theta}} = \cos\theta - i\sin\theta = \cos(-\theta) + i\sin(-\theta) = e^{-i\theta}$, and $e^{i\theta}e^{-i\theta} = e^{0} = 1$.
4. The eight points $e^{ik\pi/4}$ are the vertices of a regular octagon starting at $1$. Since $\pi \approx 3.14$, the angle $3$ rad is just short of the half-turn, so $e^{3i}$ is closest to $-1$: $|e^{3i} + 1| = 2\sin\frac{\pi - 3}{2} \approx 0.1415$, versus about $1.08$ for $e^{2i}$ and $1.76$ for $e^{i}$.
5. $\frac{1 + i}{\sqrt2} = e^{i\pi/4}$, so the power is $e^{100\pi i/4} = e^{25\pi i} = e^{i\pi} = -1$ (25 half-turns).
6. $\sin^2\theta = \frac{(e^{i\theta} - e^{-i\theta})^2}{(2i)^2} = \frac{e^{2i\theta} - 2 + e^{-2i\theta}}{-4} = \frac12 - \frac12\cos 2\theta$. For the product: $2\sin\alpha\cos\beta = 2\cdot\frac{e^{i\alpha} - e^{-i\alpha}}{2i}\cdot\frac{e^{i\beta} + e^{-i\beta}}{2} = \frac{e^{i(\alpha + \beta)} - e^{-i(\alpha + \beta)}}{2i} + \frac{e^{i(\alpha - \beta)} - e^{-i(\alpha - \beta)}}{2i}$, which is $\sin(\alpha + \beta) + \sin(\alpha - \beta)$.
7. $\frac{d}{d\theta}\sum_{n\ge0}\frac{(i\theta)^n}{n!} = \sum_{n\ge1}\frac{n\, i^n\theta^{n-1}}{n!} = i\sum_{n\ge1}\frac{(i\theta)^{n-1}}{(n-1)!} = i\sum_{m\ge0}\frac{(i\theta)^m}{m!} = ie^{i\theta}$. (Term-by-term differentiation of a power series is justified inside its radius of convergence; module 06.)
8. See the check cell: $N = 14$ for $\theta = 1$ and $N = 44$ for $\theta = 10$; for $\theta = 30$ the target is **never** reached in double precision. The terms $\theta^n/n!$ *grow* until $n \approx \theta$ before they shrink — for $\theta = 30$ the largest is $30^{30}/30! \approx 8 \times 10^{11}$ — so many terms are needed, and those huge terms must cancel to leave a result of size $1$. Each carries a rounding error of about $10^{-16}$ relative to its size, so roughly $10^{-4}$ of absolute error survives and the error stalls there. The fix: reduce $\theta$ modulo $2\pi$ first (allowed by periodicity), so you only ever sum the series for $|\theta| \le \pi$.
9. With $q = e^{i\theta} \ne 1$: $\sum_{k=0}^{n-1} q^k = \frac{q^n - 1}{q - 1}$. Factor half-angles: $q^n - 1 = e^{in\theta/2}\cdot 2i\sin\frac{n\theta}{2}$ and $q - 1 = e^{i\theta/2}\cdot 2i\sin\frac{\theta}{2}$, so the sum is $e^{i(n-1)\theta/2}\frac{\sin(n\theta/2)}{\sin(\theta/2)}$. Taking the real part: $\displaystyle\sum_{k=0}^{n-1}\cos k\theta = \frac{\sin(n\theta/2)}{\sin(\theta/2)}\cos\frac{(n-1)\theta}{2}$.

## 8. Conclusion

- The exponential series $\sum z^n/n!$ makes sense for complex $z$. At $z = i\theta$, the period-4 cycle of $i^n$ splits it into the cosine and sine series: $e^{i\theta} = \cos\theta + i\sin\theta$.
- Equivalently, $e^{it}$ is the unique solution of $f' = if$, $f(0) = 1$: velocity perpendicular to position, so uniform motion around the unit circle.
- Euler's identity $e^{i\pi} + 1 = 0$ says half a turn around the unit circle lands on $-1$. $e^{i\theta}$ is $2\pi$-periodic.
- Polar form becomes $z = re^{i\theta}$; multiplying, dividing and conjugating follow the laws of exponents.
- $\cos\theta = \frac{e^{i\theta} + e^{-i\theta}}{2}$ and $\sin\theta = \frac{e^{i\theta} - e^{-i\theta}}{2i}$ turn trigonometric identities into algebra.
- Numerically, use `cmath.exp` / `np.exp` and compare with tolerances; symbolically, `sp.exp(sp.I*theta)`, `.rewrite(sp.exp)` and `expand(complex=True)`.
