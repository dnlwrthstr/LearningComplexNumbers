# Cauchy's Integral Formula

Cauchy's theorem says closed integrals of holomorphic functions vanish. Put a single singularity in on purpose — divide by $z - a$ — and the integral no longer vanishes; instead it **returns the value $f(a)$**:

$$
f(a) = \frac{1}{2\pi i}\oint_\gamma \frac{f(z)}{z - a}\,dz .
$$

The values of a holomorphic function on a circle determine every value inside. Nothing like this is true for real differentiable functions, and its consequences are dramatic: holomorphic functions have derivatives of every order, bounded entire functions are constant (Liouville), the fundamental theorem of algebra follows in two lines, and $|f|$ can never have a maximum inside a region (maximum modulus principle).

This is notebook 03 of module 06 (**Complex integration**) in the **Learning Complex Numbers** series.

## 1. Introduction

Think of a thin metal plate in equilibrium: the temperature on its edge determines the temperature everywhere inside. Holomorphic functions behave the same way — their real and imaginary parts are harmonic (notebook 05/05) — and Cauchy's integral formula is the exact recipe for "inside from the boundary".

The formula is also a computational tool. Integrals like $\oint_{|z|=2} \frac{e^z}{z - 1}\,dz$ or $\oint_{|z|=2}\frac{\cos z}{z^3}\,dz$ become one-line evaluations, and derivatives of any order can be computed as integrals — numerically stable integrals, unlike finite differences.

## 2. Theory

### 2.1 The formula

> **Cauchy's integral formula.** Let $f$ be holomorphic on a simply connected region $D$, let $\gamma$ be a positively oriented simple closed curve in $D$, and let $a$ be a point inside $\gamma$. Then
> $$f(a) = \frac{1}{2\pi i}\oint_\gamma \frac{f(z)}{z - a}\,dz.$$

**Proof.** The integrand $\frac{f(z)}{z-a}$ is holomorphic except at $a$. By deformation (notebook 02) we may replace $\gamma$ by the small circle $C_r: |z - a| = r$. Split

$$
\oint_{C_r} \frac{f(z)}{z - a}\,dz = f(a)\oint_{C_r} \frac{dz}{z - a} + \oint_{C_r} \frac{f(z) - f(a)}{z - a}\,dz = 2\pi i\,f(a) + \oint_{C_r} \frac{f(z) - f(a)}{z - a}\,dz.
$$

The left side does not depend on $r$. In the last integral the integrand tends to $f'(a)$ as $z \to a$, so it is bounded by some $K$ near $a$; the ML-inequality gives $\left|\oint_{C_r}\right| \le K \cdot 2\pi r \to 0$. Hence the left side equals $2\pi i\,f(a)$. $\blacksquare$

If $a$ lies **outside** $\gamma$, the integrand is holomorphic inside and the integral is $0$ (Cauchy's theorem).

### 2.2 The mean-value property

Take $\gamma$ the circle $z = a + re^{it}$. Then $dz = i r e^{it}\,dt$ and $z - a = re^{it}$, so

$$
f(a) = \frac{1}{2\pi i}\int_0^{2\pi} \frac{f(a + re^{it})}{re^{it}}\,i r e^{it}\,dt = \frac{1}{2\pi}\int_0^{2\pi} f(a + r e^{it})\,dt.
$$

**The value at the centre is the average of the values on any circle around it.**

### 2.3 Derivatives of every order

Differentiate the formula with respect to $a$ under the integral sign (allowed, the integrand is smooth in $a$ away from the curve). Each derivative of $\frac{1}{z-a}$ brings a factor:

$$
\frac{d}{da}\frac{1}{z-a} = \frac{1}{(z-a)^2}, \qquad \frac{d^n}{da^n}\frac{1}{z-a} = \frac{n!}{(z-a)^{n+1}}.
$$

> **Cauchy's formula for derivatives.**
> $$f^{(n)}(a) = \frac{n!}{2\pi i}\oint_\gamma \frac{f(z)}{(z-a)^{n+1}}\,dz, \qquad n = 0, 1, 2, \dots$$

The right side exists for every $n$, so **a holomorphic function is infinitely differentiable**, and every derivative is again holomorphic. Compare real analysis: $g(x) = x|x|$ has $g'(x) = 2|x|$, which has no derivative at $0$. Being once complex-differentiable is a far stronger condition than being once real-differentiable.

### 2.4 Cauchy's estimates

If $|f| \le M$ on the circle $|z - a| = r$, the ML-inequality applied to the derivative formula gives

$$
|f^{(n)}(a)| \le \frac{n!}{2\pi} \cdot \frac{M}{r^{n+1}} \cdot 2\pi r = \frac{n!\,M}{r^n}.
$$

### 2.5 Liouville's theorem and the fundamental theorem of algebra

> **Liouville.** A bounded entire function is constant.

If $|f| \le M$ on all of $\mathbb{C}$, Cauchy's estimate with $n = 1$ holds for **every** radius: $|f'(a)| \le M/r \to 0$ as $r \to \infty$. So $f' \equiv 0$ and $f$ is constant. Consequences:

- $\sin z$ and $\cos z$ are entire and not constant, so they are **unbounded** on $\mathbb{C}$ (indeed $|\sin(iy)| = |\sinh y| \to \infty$), unlike on $\mathbb{R}$.
- **Fundamental theorem of algebra.** If a non-constant polynomial $p$ had no zero, $1/p$ would be entire; and since $|p(z)| \to \infty$ as $|z| \to \infty$, $1/p$ would be bounded. Liouville makes $1/p$ — hence $p$ — constant: a contradiction.

### 2.6 The maximum modulus principle

> **Maximum modulus.** If $f$ is holomorphic and non-constant on a connected open region $D$, then $|f|$ has no maximum inside $D$. On a closed bounded region, $\max|f|$ is attained on the boundary.

Sketch: suppose $|f(a)| \ge |f(z)|$ for all $z$ near $a$. The mean-value property gives $|f(a)| \le \frac{1}{2\pi}\int_0^{2\pi} |f(a + re^{it})|\,dt \le |f(a)|$, so equality holds throughout, which forces $|f| = |f(a)|$ on every small circle. A holomorphic function of constant modulus is constant (Cauchy–Riemann), and so $f$ is constant near $a$ — and then everywhere on $D$.

Geometrically: the graph of $|f|$ over the plane has no peaks, only saddles and the edges of the region.

## 3. Worked Examples

### Example 1 — $\oint_{|z|=2} \frac{e^z}{z - 1}\,dz$

$f(z) = e^z$ is entire and $a = 1$ is inside $|z| = 2$:

$$
\oint_{|z|=2} \frac{e^z}{z - 1}\,dz = 2\pi i\,e^1 = 2\pi e\,i \approx 17.08i.
$$

### Example 2 — $\oint_{|z|=2} \frac{\cos z}{z^3}\,dz$

This is the derivative formula with $n = 2$, $a = 0$, $f = \cos$: $\oint \frac{f(z)}{z^{3}}\,dz = \frac{2\pi i}{2!}f''(0)$. Since $f''(z) = -\cos z$, $f''(0) = -1$:

$$
\oint_{|z|=2} \frac{\cos z}{z^3}\,dz = \frac{2\pi i}{2}\cdot(-1) = -\pi i.
$$

### Example 3 — $\oint_{|z - i| = 1} \frac{dz}{z^2 + 1}$

Factor $z^2 + 1 = (z - i)(z + i)$. Inside $|z - i| = 1$ only $z = i$ is a singularity, so take $f(z) = \frac{1}{z + i}$, which is holomorphic there:

$$
\oint_{|z-i|=1} \frac{f(z)}{z - i}\,dz = 2\pi i\,f(i) = 2\pi i\cdot\frac{1}{2i} = \pi.
$$

With $|z| = 3$, both $\pm i$ are inside; the two contributions are $\pi$ and $-\pi$, total $0$.

### Example 4 — $\oint_{|z|=1} \frac{\sin z}{z^4}\,dz$

$n = 3$: $\frac{2\pi i}{3!}\sin'''(0) = \frac{2\pi i}{6}\cdot(-\cos 0) = -\frac{\pi i}{3}$.

### Example 5 — mean value

$f(z) = z^2 + 3$, circle $|z - 1| = 2$: the average of $f$ over the circle must be $f(1) = 4$. Check: $f(1 + 2e^{it}) = 1 + 4e^{it} + 4e^{2it} + 3$, and the oscillating terms average to $0$, leaving $4$. ✓

### Example 6 — a Cauchy estimate

$f(z) = e^z$ on $|z| = r$: $M = e^r$, so $|f^{(n)}(0)| = 1 \le \frac{n!\,e^r}{r^n}$ for every $r$. The best bound comes from minimising $e^r/r^n$: at $r = n$ it is $\frac{n!\,e^n}{n^n} \approx \sqrt{2\pi n}$ (Stirling). The estimate is never far off.

### Example 7 — maximum modulus on the unit disc

$f(z) = z^2 + z + 1$ on $|z| \le 1$. On the boundary $z = e^{it}$, $|f| \le 3$ with equality at $z = 1$. By maximum modulus, $|f| < 3$ at every interior point, e.g. $|f(0)| = 1$ and $|f(0.5)| = 1.75$.

## 4. Python Examples

### 4.1 Cauchy's integral formula, numerically

Sample $f$ on a circle, integrate $\frac{f(z)}{z-a}$, divide by $2\pi i$. For points inside we recover $f(a)$; for points outside we get $0$.

```python
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def cauchy_value(f, a, center=0, radius=1, n_points=2001):
    # (1 / 2 pi i) * closed integral of f(z) / (z - a) over the circle |z - center| = radius
    t = np.linspace(0, 2 * np.pi, n_points)
    z = center + radius * np.exp(1j * t)
    dz = 1j * radius * np.exp(1j * t)
    return np.trapezoid(f(z) / (z - a) * dz, t) / (2j * np.pi)


for name, f in [("exp", np.exp), ("sin", np.sin), ("z^3 - z", lambda z: z**3 - z)]:
    for a in [0.3, -0.5 + 0.4j, 1.7]:
        recovered = cauchy_value(f, a, radius=1)
        inside = abs(a) < 1
        expected = f(a) if inside else 0
        print(f"f = {name:>8}, a = {a!s:>12} ({'inside ' if inside else 'outside'}):  integral = {recovered:.10f}   expected = {expected:.10f}")
```

### 4.2 The worked examples

```python
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def circle_integral(g, center, radius, n_points=4001):
    t = np.linspace(0, 2 * np.pi, n_points)
    z = center + radius * np.exp(1j * t)
    return np.trapezoid(g(z) * 1j * radius * np.exp(1j * t), t)


print("Ex 1:", circle_integral(lambda z: np.exp(z) / (z - 1), 0, 2), "  expected", 2j * np.pi * np.e)
print("Ex 2:", circle_integral(lambda z: np.cos(z) / z**3, 0, 2), "  expected", -1j * np.pi)
print("Ex 3:", circle_integral(lambda z: 1 / (z**2 + 1), 1j, 1), "  expected", np.pi)
print("     with |z| = 3:", circle_integral(lambda z: 1 / (z**2 + 1), 0, 3))
print("Ex 4:", circle_integral(lambda z: np.sin(z) / z**4, 0, 1), "  expected", -1j * np.pi / 3)

t = np.linspace(0, 2 * np.pi, 4001)
values = (1 + 2 * np.exp(1j * t)) ** 2 + 3
print("Ex 5: mean of f over |z - 1| = 2:", np.trapezoid(values, t) / (2 * np.pi))
```

### 4.3 Derivatives as integrals

The derivative formula with the trapezoidal rule gives $f^{(n)}(a)$ accurately even for large $n$, where finite differences would drown in rounding error. The factor $n!/r^n$ amplifies rounding errors in the integral, so for high orders a larger radius is better (here $r = 3$; the function is entire, so any radius is allowed). We compare against sympy's exact derivatives of $f(z) = e^{z}\cos z$ at $a = 0.2 + 0.1i$.

```python
import math

import numpy as np
import sympy as sp

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def cauchy_derivative(f, a, order, radius=3.0, n_points=256):
    # f^(n)(a) = n! / (2 pi i) * closed integral of f(z) / (z - a)^(n + 1) on |z - a| = radius
    t = np.linspace(0, 2 * np.pi, n_points + 1)
    z = a + radius * np.exp(1j * t)
    dz = 1j * radius * np.exp(1j * t)
    return math.factorial(order) * np.trapezoid(f(z) / (z - a) ** (order + 1) * dz, t) / (2j * np.pi)


z = sp.symbols("z")
expression = sp.exp(z) * sp.cos(z)
a = 0.2 + 0.1j
for order in [0, 1, 2, 5, 10, 15]:
    exact = complex(sp.diff(expression, z, order).subs(z, sp.Float(a.real) + sp.I * sp.Float(a.imag)).evalf())
    numeric = cauchy_derivative(lambda w: np.exp(w) * np.cos(w), a, order)
    print(f"order {order:2d}:  integral = {numeric:.8e}   sympy = {exact:.8e}   relative error = {abs(numeric - exact) / abs(exact):.1e}")
```

## 5. Visualizations

### 5.1 The boundary determines the inside

Sample $f(z) = e^{z}\sin(2z)$ at only 64 points on the unit circle. Cauchy's formula then reconstructs $f$ on a grid filling the disc. Left: $|f|$ reconstructed; right: the error, which is tiny in the interior and grows only very close to the circle (where $\frac{1}{z-a}$ is nearly singular and 64 samples are not enough).

```python
import matplotlib.pyplot as plt
import numpy as np

n_samples = 64
t = 2 * np.pi * np.arange(n_samples) / n_samples   # periodic trapezoidal rule = plain average
boundary = np.exp(1j * t)
f = lambda z: np.exp(z) * np.sin(2 * z)
boundary_values = f(boundary)

axis = np.linspace(-1, 1, 301)
grid = axis[None, :] + 1j * axis[:, None]
inside = np.abs(grid) < 0.995

# f(a) = (1 / 2 pi) * integral of f(z) z / (z - a) dt over t, because dz = i z dt
reconstructed = np.full(grid.shape, np.nan, dtype=complex)
points = grid[inside]
kernel = boundary[None, :] / (boundary[None, :] - points[:, None])
reconstructed[inside] = (kernel * boundary_values[None, :]).mean(axis=1)

error = np.abs(reconstructed - f(grid))

fig, (ax_value, ax_error) = plt.subplots(1, 2, figsize=(13, 5.5))
true_maximum = np.abs(f(points)).max()
image = ax_value.imshow(np.abs(reconstructed), extent=[-1, 1, -1, 1], origin="lower", cmap="viridis", vmin=0, vmax=true_maximum)
fig.colorbar(image, ax=ax_value, label="|f| reconstructed")
ax_value.scatter(boundary.real, boundary.imag, s=10, color="red", label="64 boundary samples")
ax_value.set_title(r"$|e^z \sin 2z|$ rebuilt from the boundary")
ax_value.legend(loc="lower left", fontsize=9)
image = ax_error.imshow(np.log10(error + 1e-17), extent=[-1, 1, -1, 1], origin="lower", cmap="magma")
fig.colorbar(image, ax=ax_error, label="log10 error")
ax_error.set_title("Reconstruction error")
for ax in (ax_value, ax_error):
    ax.set_xlim(-1.1, 1.1)
    ax.set_ylim(-1.1, 1.1)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
plt.tight_layout()
plt.show()

print("max error for |a| < 0.8 :", np.nanmax(np.where(np.abs(grid) < 0.8, error, np.nan)))
print("max error for |a| < 0.995:", np.nanmax(error))
```

### 5.2 The mean-value property

The values of $f(z) = z^2 + z + 1$ on the circle $|z - 0.5i| = 1$ form a closed curve in the $w$-plane (right). Their average — the centre of mass of that curve, parametrized by angle — is exactly $f(0.5i)$.

```python
import matplotlib.pyplot as plt
import numpy as np

center, radius = 0.5j, 1.0
f = lambda z: z**2 + z + 1
t = np.linspace(0, 2 * np.pi, 400, endpoint=False)
circle = center + radius * np.exp(1j * t)
image_curve = f(circle)
average = image_curve.mean()

fig, (ax_z, ax_w) = plt.subplots(1, 2, figsize=(12, 5.5))
ax_z.plot(np.append(circle, circle[0]).real, np.append(circle, circle[0]).imag, color="tab:blue")
ax_z.scatter(center.real, center.imag, color="black", zorder=3, label="centre a = 0.5i")
ax_z.set_title(r"Circle $|z - 0.5i| = 1$ in the $z$-plane")
closed_image = np.append(image_curve, image_curve[0])
ax_w.plot(closed_image.real, closed_image.imag, color="tab:blue", label="f(circle)")
ax_w.scatter(average.real, average.imag, s=120, marker="o", facecolor="none", edgecolor="red", linewidth=2, label=f"average = {average:.3f}")
ax_w.scatter(f(center).real, f(center).imag, marker="x", color="black", s=60, zorder=3, label=f"f(a) = {f(center):.3f}")
ax_w.set_title(r"Its image under $f(z) = z^2 + z + 1$")
for ax in (ax_z, ax_w):
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="lower right", fontsize=9)
plt.tight_layout()
plt.show()
```

### 5.3 Maximum modulus: no peaks inside

The landscape $|f(z)|$ for $f(z) = z^2 + z + \tfrac12$ on the closed unit disc. The maximum (white star) sits on the boundary circle at $z = 1$; inside there are only valleys (the zeros $\frac{-1 \pm i}{2}$ of $f$, marked ×), slopes and a saddle between the valleys. (The polynomial of Example 7 would work too, but its zeros $e^{\pm 2\pi i/3}$ lie exactly on the boundary.)

```python
import matplotlib.pyplot as plt
import numpy as np

f = lambda z: z**2 + z + 0.5
axis = np.linspace(-1, 1, 401)
grid = axis[None, :] + 1j * axis[:, None]
modulus = np.where(np.abs(grid) <= 1, np.abs(f(grid)), np.nan)

index = np.nanargmax(modulus)
location = grid.flat[index]
zeros = np.roots([1, 1, 0.5])

fig, ax = plt.subplots(figsize=(7, 6))
image = ax.contourf(axis, axis, modulus, levels=30, cmap="viridis")
fig.colorbar(image, ax=ax, label="|f(z)|")
ax.contour(axis, axis, modulus, levels=15, colors="white", linewidths=0.4, alpha=0.6)
t = np.linspace(0, 2 * np.pi, 300)
ax.plot(np.cos(t), np.sin(t), color="black", linewidth=1)
ax.scatter(location.real, location.imag, marker="*", s=300, color="white", edgecolor="black", zorder=3,
           label=f"max |f| = {np.nanmax(modulus):.3f} at z = {location.real:.2f}{location.imag:+.2f}i")
ax.scatter(zeros.real, zeros.imag, marker="x", color="red", s=80, zorder=3, label="zeros of f")
ax.set_aspect("equal")
ax.set_title(r"$|z^2 + z + 1/2|$ on the closed unit disc")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.legend(loc="lower left", fontsize=9)
plt.show()
```

### 5.4 Liouville: entire functions cannot stay bounded

$|\sin z|$ over a large square. Bounded by $1$ on the real axis, it grows like $\frac12 e^{|\mathrm{Im}\,z|}$ away from it.

```python
import matplotlib.pyplot as plt
import numpy as np

axis_x = np.linspace(-8, 8, 400)   # even sizes keep the grid off the zeros of sin at k pi
axis_y = np.linspace(-3, 3, 200)
grid = axis_x[None, :] + 1j * axis_y[:, None]

fig, ax = plt.subplots(figsize=(10, 4.5))
image = ax.imshow(np.log10(np.abs(np.sin(grid))), extent=[-8, 8, -3, 3], origin="lower", cmap="viridis", vmin=-2)
fig.colorbar(image, ax=ax, label=r"$\log_{10} |\sin z|$")
ax.axhline(0, color="white", linewidth=0.8)
ax.axvline(0, color="white", linewidth=0.8)
ax.set_aspect("equal")
ax.set_title(r"$|\sin z|$: bounded on the real axis, exponential growth off it")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
plt.show()
```

## 7. Solutions

4. $\frac{1}{z(z-1)} = \frac{1}{z - 1} - \frac{1}{z}$. Both $0$ and $1$ are inside: $2\pi i\,e^{1} - 2\pi i\,e^{0} = 2\pi i(e - 1)$.
5. The circle runs from $-0.5$ to $2.5$ on the real axis and up to $1.5i$, so $0$ is inside; $3$ is outside; $\pm 2i$ are not singularities of this integrand at all (they are distractors). 6. On $|z| = 2$, $|z - 3| \ge 1$, so $M = 1$ and $|f'''(0)| \le \frac{3!\cdot 1}{2^3} = 0.75$. Exact: $f'''(z) = -\frac{6}{(z - 3)^4}$, $|f'''(0)| = \frac{6}{81} \approx 0.074$.
7. On $|z - a| = r$ we have $|z| \le |a| + r$, so $M \le 5(|a| + r)^2$ and $|f'''(a)| \le \frac{6\cdot 5(|a| + r)^2}{r^3} \to 0$ as $r \to \infty$. 8. $|e^z| = e^{\operatorname{Re} z}$, maximal ($= e$) on the whole right edge $\operatorname{Re} z = 1$ — on the boundary, as the principle requires. The minimum ($= 1$) is on the left edge. Since $e^z$ has no zeros, $1/e^z = e^{-z}$ is holomorphic too, and maximum modulus for $e^{-z}$ is a minimum principle for $e^z$. (A function with a zero inside has its minimum $0$ there.)

## 8. Conclusion

- **Cauchy's integral formula:** $f(a) = \frac{1}{2\pi i}\oint_\gamma \frac{f(z)}{z-a}\,dz$ for $a$ inside $\gamma$; the boundary values determine the function inside.
- On a circle it becomes the **mean-value property**: $f(a)$ is the average of $f$ over any circle centred at $a$.
- **Derivatives:** $f^{(n)}(a) = \frac{n!}{2\pi i}\oint \frac{f(z)}{(z-a)^{n+1}}\,dz$, so holomorphic functions are infinitely differentiable. Numerically this is an excellent way to compute high derivatives.
- **Cauchy's estimates** $|f^{(n)}(a)| \le n!M/r^n$ lead to **Liouville** (bounded entire $\Rightarrow$ constant) and from there to the fundamental theorem of algebra; $\sin$ and $\cos$ must be unbounded on $\mathbb{C}$.
- **Maximum modulus:** a non-constant holomorphic function has no interior maximum of $|f|$.
