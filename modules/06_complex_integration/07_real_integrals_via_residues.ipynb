# Real Integrals via Residues

The residue theorem computes integrals around closed curves. Real integrals run along the real line or over $[0, 2\pi]$ — not closed curves. The trick is to **close them**: add a large semicircle whose contribution vanishes as it grows, or recognise $[0, 2\pi]$ as the unit circle in disguise. Then a real integral becomes $2\pi i$ times a sum of residues.

Three families are covered here: rational functions on $\mathbb{R}$, such as $\int_{-\infty}^{\infty}\frac{dx}{1+x^4} = \frac{\pi}{\sqrt2}$; trigonometric integrals over a period, such as $\int_0^{2\pi}\frac{d\theta}{2 + \cos\theta} = \frac{2\pi}{\sqrt3}$; and Fourier-type integrals, such as $\int_{-\infty}^{\infty}\frac{\cos x}{1+x^2}\,dx = \frac{\pi}{e}$, which have no elementary antiderivative at all. Every result is compared with numerical quadrature.

This is notebook 07 of module 06 (**Complex integration**) in the **Learning Complex Numbers** series.

## 1. Introduction

In real calculus $\int_{-\infty}^{\infty}\frac{dx}{1+x^2} = \pi$ comes from the antiderivative $\arctan x$, found with the substitution $x = \tan u$. For $\frac{1}{1+x^4}$ the same approach needs a painful partial-fraction decomposition, and for $\frac{\cos x}{1+x^2}$ there is no elementary antiderivative — yet the answer $\pi/e$ is as clean as can be. Residues explain the cleanliness: the integral only "sees" the pole at $z = i$, and $e^{iz}$ at $z = i$ is $e^{-1}$.

These integrals are not just tricks. The Fourier-type integral $\int \frac{e^{ikx}}{x^2 + a^2}dx = \frac{\pi}{a}e^{-k a}$ is the Fourier transform of a Lorentzian, central in spectroscopy and signal processing, and the semicircle argument is how engineers invert Laplace and $z$-transforms.

## 2. Theory

### 2.1 Rational functions on the real line

Let $f = \frac{P}{Q}$ with polynomials $P, Q$, where

- $Q$ has no real zeros, and
- $\deg Q \ge \deg P + 2$ (so $|f(x)| \le K/x^2$ for large $|x|$ and $\int_{-\infty}^{\infty} f$ converges).

Take $R$ larger than every $|a|$ with $Q(a) = 0$, and the closed contour $\Gamma_R$: the segment $[-R, R]$ followed by the upper semicircle $C_R: z = Re^{it}$, $t \in [0, \pi]$. It encloses exactly the poles in the upper half-plane, so

$$
\int_{-R}^{R} f(x)\,dx + \int_{C_R} f(z)\,dz = 2\pi i \sum_{\operatorname{Im} a > 0} \operatorname{Res}(f, a).
$$

On $C_R$, $|f(z)| \le \frac{K}{R^2}$ for large $R$, and $C_R$ has length $\pi R$. By the ML-inequality

$$
\left|\int_{C_R} f(z)\,dz\right| \le \pi R \cdot \frac{K}{R^2} = \frac{\pi K}{R} \xrightarrow{R\to\infty} 0.
$$

Letting $R \to \infty$:

$$
\boxed{\int_{-\infty}^{\infty} \frac{P(x)}{Q(x)}\,dx = 2\pi i \sum_{\operatorname{Im} a > 0} \operatorname{Res}\left(\frac{P}{Q}, a\right).}
$$

(Closing in the lower half-plane works too: the contour is then clockwise, giving $-2\pi i\sum_{\operatorname{Im} a < 0}$ — the same number, since all residues sum to $0$ when $\deg Q \ge \deg P + 2$.)

### 2.2 Trigonometric integrals over a period

For $\int_0^{2\pi} R(\cos\theta, \sin\theta)\,d\theta$ with $R$ a rational function, put $z = e^{i\theta}$. As $\theta$ runs over $[0, 2\pi]$, $z$ runs once counter-clockwise around the unit circle, and

$$
\cos\theta = \frac{1}{2}\left(z + \frac1z\right), \qquad \sin\theta = \frac{1}{2i}\left(z - \frac1z\right), \qquad d\theta = \frac{dz}{iz}.
$$

The integral becomes $\oint_{|z|=1}(\text{rational function of } z)\,dz = 2\pi i\sum_{|a| < 1}\operatorname{Res}$. No limits, no estimates.

### 2.3 Fourier-type integrals and Jordan's lemma

To evaluate $\int_{-\infty}^{\infty} f(x)\cos(kx)\,dx$ with $k > 0$, we would like to close in the upper half-plane. **Do not use $\cos(kz)$**: with $z = x + iy$,

$$
\cos(kz) = \frac{e^{ikz} + e^{-ikz}}{2}, \qquad |e^{ikz}| = e^{-ky}, \quad |e^{-ikz}| = e^{ky},
$$

so $|\cos(kz)|$ grows like $\frac12 e^{kR}$ at the top of the semicircle. Instead integrate $f(z)\,e^{ikz}$, which is **small** in the upper half-plane ($|e^{ikz}| = e^{-ky} \le 1$), and take the real part at the end (for real $f$):

$$
\int_{-\infty}^{\infty} f(x)\cos(kx)\,dx = \operatorname{Re}\int_{-\infty}^{\infty} f(x)\,e^{ikx}\,dx, \qquad \int_{-\infty}^{\infty} f(x)\sin(kx)\,dx = \operatorname{Im}\int_{-\infty}^{\infty} f(x)\,e^{ikx}\,dx.
$$

If $\deg Q \ge \deg P + 2$, the ML-bound of 2.1 still kills the semicircle, since $|e^{ikz}| \le 1$ there. When $\deg Q = \deg P + 1$ (e.g. $\frac{x}{x^2+1}$), the ML bound is only $\pi R\cdot\frac{K}{R} = \pi K$ — useless. The oscillation must be exploited:

> **Jordan's lemma.** If $|g(z)| \le M_R$ on $C_R$ and $k > 0$, then
> $$\left|\int_{C_R} g(z)\,e^{ikz}\,dz\right| \le \frac{\pi M_R}{k}.$$

*Proof.* On $C_R$, $|e^{ikz}| = e^{-kR\sin t}$ and $|dz| = R\,dt$, so the integral is at most $M_R R\int_0^\pi e^{-kR\sin t}dt = 2M_R R\int_0^{\pi/2} e^{-kR\sin t}dt$. On $[0, \frac{\pi}{2}]$, $\sin t \ge \frac{2t}{\pi}$ (the chord lies below the concave arc), so the integral is at most $2M_R R\int_0^{\infty} e^{-2kRt/\pi}dt = 2M_R R\cdot\frac{\pi}{2kR} = \frac{\pi M_R}{k}$. $\blacksquare$

The length $\pi R$ of the arc has disappeared: $e^{ikz}$ is exponentially small on all of it except near the two ends. So $M_R \to 0$ suffices, i.e. $\deg Q \ge \deg P + 1$.

## 3. Worked Examples

### Example 1 — $\int_{-\infty}^{\infty} \frac{dx}{1+x^2}$

Poles $\pm i$; upper: $i$, with $\operatorname{Res} = \frac{1}{2z}\big|_{z=i} = \frac{1}{2i}$. So the integral is $2\pi i \cdot \frac{1}{2i} = \pi$.

Real-calculus check: with $x = \tan u$, $dx = \sec^2 u\,du$ and $1 + \tan^2 u = \sec^2 u$, so $\int_{-\infty}^{\infty}\frac{dx}{1+x^2} = \int_{-\pi/2}^{\pi/2} du = \pi$. ✓

### Example 2 — $\int_{-\infty}^{\infty} \frac{dx}{1+x^4}$

Poles: $a^4 = -1$, $a = e^{i\pi/4}, e^{3i\pi/4}, e^{5i\pi/4}, e^{7i\pi/4}$. Upper half-plane: $a_1 = e^{i\pi/4} = \frac{1 + i}{\sqrt2}$ and $a_2 = e^{3i\pi/4} = \frac{-1 + i}{\sqrt2}$. Residues $\frac{1}{4a^3} = \frac{a}{4a^4} = -\frac{a}{4}$. Sum: $-\frac{a_1 + a_2}{4} = -\frac{1}{4}\cdot\frac{2i}{\sqrt2} = -\frac{i}{2\sqrt2}$. Then

$$
\int_{-\infty}^{\infty} \frac{dx}{1+x^4} = 2\pi i\cdot\left(-\frac{i}{2\sqrt2}\right) = \frac{\pi}{\sqrt2} \approx 2.2214.
$$

### Example 3 — $\int_{-\infty}^{\infty} \frac{dx}{(x^2+1)^2}$

Double pole at $i$: $f = \frac{1}{(z - i)^2(z + i)^2}$, $\operatorname{Res} = \frac{d}{dz}(z + i)^{-2}\big|_{i} = -2(2i)^{-3} = \frac{-2}{-8i} = \frac{1}{4i}$. Integral: $2\pi i\cdot\frac{1}{4i} = \frac{\pi}{2}$.

### Example 4 — $\int_0^{2\pi} \frac{d\theta}{2 + \cos\theta}$

Substitute: $2 + \frac{z + z^{-1}}{2} = \frac{z^2 + 4z + 1}{2z}$, so

$$
\int_0^{2\pi}\frac{d\theta}{2 + \cos\theta} = \oint_{|z|=1} \frac{2z}{z^2 + 4z + 1}\cdot\frac{dz}{iz} = \oint_{|z|=1}\frac{2}{i(z^2 + 4z + 1)}\,dz.
$$

Roots $z = -2 \pm \sqrt3$: $\alpha = -2 + \sqrt3 \approx -0.268$ (inside), $\beta = -2 - \sqrt3 \approx -3.73$ (outside). $\operatorname{Res}$ at $\alpha$: $\frac{2}{i(\alpha - \beta)} = \frac{2}{2\sqrt3\,i} = \frac{1}{\sqrt3\,i}$. Result: $2\pi i\cdot\frac{1}{\sqrt3\,i} = \frac{2\pi}{\sqrt3} \approx 3.6276$.

### Example 5 — $\int_0^{2\pi} \frac{d\theta}{5 - 4\sin\theta}$

$5 - 4\cdot\frac{z - z^{-1}}{2i} = 5 + 2i\left(z - \frac1z\right) = \frac{2iz^2 + 5z - 2i}{z}$. With $d\theta = \frac{dz}{iz}$:

$$
\oint_{|z|=1}\frac{dz}{i(2iz^2 + 5z - 2i)} = \oint_{|z|=1}\frac{-dz}{2z^2 - 5iz - 2}.
$$

$2z^2 - 5iz - 2 = 2(z - 2i)(z - \tfrac{i}{2})$; inside: $\frac{i}{2}$. $\operatorname{Res} = \frac{-1}{2(\frac{i}{2} - 2i)} = \frac{-1}{-3i} = \frac{1}{3i}$. Result: $2\pi i\cdot\frac{1}{3i} = \frac{2\pi}{3}$.

### Example 6 — $\int_{-\infty}^{\infty} \frac{\cos x}{1+x^2}\,dx$

Integrate $\frac{e^{iz}}{1 + z^2}$: the only upper pole is $i$, $\operatorname{Res} = \frac{e^{i\cdot i}}{2i} = \frac{e^{-1}}{2i}$. Semicircle: $\left|\frac{e^{iz}}{1+z^2}\right| \le \frac{1}{R^2 - 1}$, so ML (or Jordan) gives $0$. Hence

$$
\int_{-\infty}^{\infty}\frac{e^{ix}}{1+x^2}\,dx = 2\pi i\cdot\frac{e^{-1}}{2i} = \frac{\pi}{e}, \qquad \int_{-\infty}^{\infty}\frac{\cos x}{1+x^2}\,dx = \frac{\pi}{e} \approx 1.1557,
$$

and the sine integral is $0$ (as it must be: $\frac{\sin x}{1+x^2}$ is odd).

### Example 7 — the general Lorentzian

The same steps with $\frac{e^{ikz}}{z^2 + a^2}$ ($a, k > 0$), pole $ia$, residue $\frac{e^{-ka}}{2ia}$, give

$$
\int_{-\infty}^{\infty}\frac{\cos(kx)}{x^2 + a^2}\,dx = \frac{\pi}{a}e^{-ka}.
$$

The Fourier transform of a Lorentzian is a decaying exponential.

## 4. Python Examples

### 4.1 Residues for any rational function

`np.roots` finds the poles; for simple poles $\operatorname{Res} = \frac{P(a)}{Q'(a)}$. Summing over the upper half-plane gives the integral.

For the numerical comparison we use the **trigonometric substitution** $x = \tan u$ from the prerequisite notebook: $\int_{-\infty}^{\infty} f(x)\,dx = \int_{-\pi/2}^{\pi/2} f(\tan u)\sec^2 u\,du$ turns an infinite interval into a finite one. For rational $f$ with $\deg Q \ge \deg P + 2$ the new integrand is bounded, so a simple midpoint rule works (it never evaluates the endpoints $u = \pm\frac{\pi}{2}$, where $\tan u$ is infinite).

```python
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def real_line_integral_by_residues(numerator, denominator):
    # Integral over R of P/Q (simple poles, deg Q >= deg P + 2, no real poles) = 2 pi i * sum of upper residues
    P, Q = np.poly1d(numerator), np.poly1d(denominator)
    poles = Q.roots
    upper = poles[poles.imag > 0]
    residues = P(upper) / Q.deriv()(upper)
    return 2j * np.pi * residues.sum()


def real_line_integral_by_quadrature(f, n_intervals=200000):
    # Substitute x = tan(u): integral of f(tan u) / cos(u)^2 over (-pi/2, pi/2).
    # Midpoint rule: it never evaluates the endpoints, where tan(u) is infinite.
    step = np.pi / n_intervals
    u = -np.pi / 2 + (np.arange(n_intervals) + 0.5) * step
    return np.sum(f(np.tan(u)) / np.cos(u) ** 2) * step


cases = {
    "1/(1+x^2)": ([1], [1, 0, 1], np.pi),
    "1/(1+x^4)": ([1], [1, 0, 0, 0, 1], np.pi / np.sqrt(2)),
    "1/(x^2+2x+2)": ([1], [1, 2, 2], np.pi),
    "x^2/(x^4+5x^2+4)": ([1, 0, 0], [1, 0, 5, 0, 4], np.pi / 3),
}
for name, (numerator, denominator, exact) in cases.items():
    by_residues = real_line_integral_by_residues(numerator, denominator)
    by_quadrature = real_line_integral_by_quadrature(lambda x: np.polyval(numerator, x) / np.polyval(denominator, x))
    print(f"{name:>18}:  residues = {by_residues.real:.10f} (imag {by_residues.imag:+.0e})   quadrature = {by_quadrature:.10f}   exact = {exact:.10f}")
```

The double pole of $\frac{1}{(x^2+1)^2}$ is not covered by $\frac{P(a)}{Q'(a)}$ ($Q'(i) = 0$); sympy handles that case, and the whole chain symbolically:

```python
import sympy as sp

x = sp.symbols("x", real=True)
z = sp.symbols("z")
theta = sp.symbols("theta", real=True)

double_pole = 2 * sp.pi * sp.I * sp.residue(1 / (z**2 + 1) ** 2, z, sp.I)
print("1/(x^2+1)^2   residue method:", sp.simplify(double_pole), "   sympy integrate:", sp.integrate(1 / (x**2 + 1) ** 2, (x, -sp.oo, sp.oo)))

lorentzian = 2 * sp.pi * sp.I * sp.residue(sp.exp(sp.I * z) / (1 + z**2), z, sp.I)
print("cos x/(1+x^2) residue method:", sp.simplify(sp.re(lorentzian)), "   sympy integrate:",
      sp.simplify(sp.integrate(sp.cos(x) / (1 + x**2), (x, -sp.oo, sp.oo))))

print("1/(2+cos t)   sympy integrate:", sp.integrate(1 / (2 + sp.cos(theta)), (theta, 0, 2 * sp.pi)), "  = 2 pi / sqrt(3) =", sp.N(2 * sp.pi / sp.sqrt(3)))
```

### 4.2 Trigonometric integrals: substitution and residues, automated

`sympy` performs the substitution $z = e^{i\theta}$, finds the poles inside the unit circle and sums residues; the periodic trapezoidal rule (very accurate for smooth periodic integrands) checks it.

```python
import numpy as np
import sympy as sp

z = sp.symbols("z")
cos_theta = (z + 1 / z) / 2
sin_theta = (z - 1 / z) / (2 * sp.I)


def trig_integral_by_residues(integrand_in_z):
    # integrand_in_z(cos, sin) -> expression; integral over [0, 2 pi] = closed integral over |z| = 1 with d theta = dz / (i z)
    expression = sp.cancel(integrand_in_z(cos_theta, sin_theta) / (sp.I * z))
    numerator, denominator = sp.fraction(expression)
    poles = sp.roots(sp.Poly(denominator, z))
    inside = [pole for pole in poles if abs(complex(pole)) < 1]
    return sp.nsimplify(sp.simplify(2 * sp.pi * sp.I * sum(sp.residue(expression, z, pole) for pole in inside))), inside


theta = 2 * np.pi * np.arange(4096) / 4096
cases = {
    "1/(2 + cos)": (lambda c, s: 1 / (2 + c), lambda t: 1 / (2 + np.cos(t))),
    "1/(5 - 4 sin)": (lambda c, s: 1 / (5 - 4 * s), lambda t: 1 / (5 - 4 * np.sin(t))),
    "sin^2/(5 + 4 cos)": (lambda c, s: s**2 / (5 + 4 * c), lambda t: np.sin(t) ** 2 / (5 + 4 * np.cos(t))),
}
for name, (symbolic, numeric) in cases.items():
    value, inside = trig_integral_by_residues(symbolic)
    quadrature = 2 * np.pi * numeric(theta).mean()
    print(f"{name:>18}: residues = {value} = {float(value):.10f}   quadrature = {quadrature:.10f}   poles inside: {inside}")
```

### 4.3 Fourier-type integrals: truncated quadrature

For $\int_{-\infty}^{\infty}\frac{\cos(kx)}{x^2+a^2}dx$ the integrand decays like $\frac{1}{x^2}$, so truncating to $[-L, L]$ costs at most $\frac{2}{L}$ (and much less, because the tail oscillates). A fine trapezoidal grid then matches $\frac{\pi}{a}e^{-ka}$.

```python
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

half_length = 2000.0
x = np.linspace(-half_length, half_length, 2_000_001)
for k, a in [(1, 1), (2, 1), (1, 3), (5, 0.5)]:
    quadrature = np.trapezoid(np.cos(k * x) / (x**2 + a**2), x)
    exact = np.pi / a * np.exp(-k * a)
    print(f"k = {k}, a = {a}:  quadrature = {quadrature:.8f}   pi/a e^(-ka) = {exact:.8f}   difference = {abs(quadrature - exact):.1e}")
```

### 5.2 How fast does the arc integral vanish?

The size of $\int_{C_R}$ against $R$ for three integrands. For $\frac{1}{1+z^2}$ the arc integral falls off like $\frac{1}{R}$, just under the ML bound $\frac{\pi R}{R^2 - 1}$ (dashed); with the extra factor $e^{iz}$ it falls off even faster, because $e^{iz}$ is exponentially small away from the ends of the arc. For $\frac{z\,e^{iz}}{1+z^2}$ the ML bound $\frac{\pi R^2}{R^2-1}$ tends to $\pi$ and proves nothing, but the integral itself still goes to $0$ — Jordan's lemma, bound $\frac{\pi R}{R^2 - 1}$ with $M_R = \frac{R}{R^2-1}$, $k = 1$.

```python
import matplotlib.pyplot as plt
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def arc_integral(f, radius, n_points=40001):
    t = np.linspace(0, np.pi, n_points)
    z = radius * np.exp(1j * t)
    return np.trapezoid(f(z) * 1j * z, t)


radii = np.logspace(0.4, 2.5, 40)
integrands = {
    r"$1/(1+z^2)$": lambda z: 1 / (1 + z**2),
    r"$e^{iz}/(1+z^2)$": lambda z: np.exp(1j * z) / (1 + z**2),
    r"$z\, e^{iz}/(1+z^2)$": lambda z: z * np.exp(1j * z) / (1 + z**2),
}

plt.figure(figsize=(9, 5.5))
for (label, f), color in zip(integrands.items(), ["tab:blue", "tab:orange", "tab:green"]):
    sizes = [abs(arc_integral(f, r)) for r in radii]
    plt.loglog(radii, sizes, "o", markersize=4, color=color, label=label)
plt.loglog(radii, np.pi * radii / (radii**2 - 1), "--", color="black", label=r"$\pi R/(R^2-1)$ (ML, and Jordan for $z e^{iz}$)")
plt.loglog(radii, np.pi * radii**2 / (radii**2 - 1), ":", color="tab:green", label=r"ML bound for $z e^{iz}/(1+z^2)$: $\to \pi$")
plt.title(r"Size of the arc integral $\int_{C_R}$")
plt.xlabel("R")
plt.ylabel(r"$|\int_{C_R} f\, dz|$")
plt.grid(True, which="both", alpha=0.3)
plt.legend(fontsize=9)
plt.show()
```

### 5.3 Why $e^{iz}$ and not $\cos z$

Along the semicircle $z = Re^{it}$ with $R = 10$: $|e^{iz}| = e^{-R\sin t}$ is at most $1$ and tiny in the middle of the arc, whereas $|\cos z|$ reaches $\cosh R \approx 1.1\times 10^4$ at the top. Closing with $\cos z$ would add an enormous arc contribution instead of a vanishing one.

```python
import matplotlib.pyplot as plt
import numpy as np

radius = 10
t = np.linspace(0, np.pi, 500)
z = radius * np.exp(1j * t)

plt.figure(figsize=(9, 5))
plt.semilogy(t, np.abs(np.exp(1j * z)), linewidth=2, label=r"$|e^{iz}| = e^{-R \sin t}$")
plt.semilogy(t, np.abs(np.cos(z)), linewidth=2, label=r"$|\cos z|$")
plt.axhline(1, color="gray", linestyle=":")
plt.title(r"Along the upper semicircle $z = 10 e^{it}$")
plt.xlabel("t (angle along the arc)")
plt.ylabel("modulus (log scale)")
plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.show()
```

### 5.4 The trigonometric substitution in the plane

For $\int_0^{2\pi}\frac{d\theta}{2 + \cos\theta}$, the interval $[0, 2\pi]$ becomes the unit circle. Only the root $-2 + \sqrt3$ of $z^2 + 4z + 1$ lies inside; its partner $-2 - \sqrt3$ (outside) does not contribute. The two roots are reciprocals of each other ($\alpha\beta = 1$), which is typical of these integrals: $z \mapsto 1/z$ swaps inside and outside.

```python
import matplotlib.pyplot as plt
import numpy as np

roots = np.roots([1, 4, 1])
t = np.linspace(0, 2 * np.pi, 300)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(np.cos(t), np.sin(t), linewidth=2, label=r"$z = e^{i\theta}$, $\theta \in [0, 2\pi]$")
ax.annotate("", xy=(np.cos(0.5), np.sin(0.5)), xytext=(np.cos(0.3), np.sin(0.3)), arrowprops=dict(arrowstyle="->", linewidth=2, color="tab:blue"))
for root in roots:
    inside = abs(root) < 1
    ax.scatter(root.real, root.imag, marker="x", s=120, linewidth=3, color="red" if inside else "gray", zorder=3)
    ax.annotate(f"{root.real:.3f}" + (" (inside)" if inside else " (outside)"), (root.real, root.imag), xytext=(0, 12),
                textcoords="offset points", ha="center", color="red" if inside else "gray")
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-4.3, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect("equal")
ax.set_title(r"Poles of $2 / (i(z^2 + 4z + 1))$ and the unit circle")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower left")
plt.show()
```

## 7. Solutions

1. Upper pole $2i$, $\operatorname{Res} = \frac{1}{4i}$, integral $2\pi i\cdot\frac{1}{4i} = \frac{\pi}{2}$. With $x = 2\tan u$: $dx = 2\sec^2u\,du$, $x^2 + 4 = 4\sec^2 u$, so $\int_{-\pi/2}^{\pi/2}\frac{du}{2} = \frac{\pi}{2}$. ✓
2. $3 + \frac{z + z^{-1}}{2} \cdot 2 = \frac{z^2 + 3z + 1}{z}$, so the integrand becomes $\frac{1}{i(z^2 + 3z + 1)}$. Roots $\frac{-3 \pm\sqrt5}{2}$; inside: $\alpha = \frac{-3 + \sqrt5}{2} \approx -0.382$. $\operatorname{Res} = \frac{1}{i(\alpha - \beta)} = \frac{1}{i\sqrt5}$. Integral $\frac{2\pi}{\sqrt5}$.
3. Poles $-1 \pm i$; upper $-1 + i$, $\operatorname{Res} = \frac{1}{2(-1 + i) + 2} = \frac{1}{2i}$. Integral $\pi$. (Equivalently, shift $x \to x - 1$ to get $\int\frac{dx}{x^2 + 1}$.)
4. Poles $\pm i, \pm 2i$, all on the imaginary axis; the semicircle (for $R > 2$) encloses $i$ and $2i$. $\operatorname{Res}_i = \frac{1}{2i\cdot(4 - 1)} = \frac{1}{6i}$; $\operatorname{Res}_{2i} = \frac{1}{(1 - 4)\cdot 4i} = -\frac{1}{12i}$. Sum $\frac{1}{12i}$; integral $\frac{\pi}{6}$.
5. $\operatorname{Re}\,2\pi i\cdot\frac{e^{2i\cdot i}}{2i} = \pi e^{-2} \approx 0.4252$. In the lower half-plane $|e^{2iz}| = e^{-2y}$ is **large** ($y < 0$): the arc contribution blows up instead of vanishing. (For $\cos(2x)$ one could use $e^{-2iz}$ and the lower half-plane instead — that works, with the clockwise sign.)
6. $\int\frac{e^{iz}}{z^2 + 2z + 2}$: $\operatorname{Res}_{-1+i} = \frac{e^{i(-1 + i)}}{2i} = \frac{e^{-1}e^{-i}}{2i}$; times $2\pi i$: $\frac{\pi}{e}(\cos 1 - i\sin 1)$. Imaginary part: $-\frac{\pi\sin 1}{e} \approx -0.9725$ (and the cosine integral is $\frac{\pi\cos 1}{e}$).
7. $\sin^2\theta = -\frac{(z^2 - 1)^2}{4z^2}$, $5 + 4\cos\theta = \frac{(2z + 1)(z + 2)}{z}$, $d\theta = \frac{dz}{iz}$: the integrand is $g(z) = -\frac{(z^2-1)^2}{4i\,z^2(2z+1)(z + 2)}$. Inside: $z = 0$ (double) and $z = -\frac12$. $\operatorname{Res}_{-1/2} = -\frac{(1/4 - 1)^2}{4i\cdot\frac14\cdot 2\cdot\frac32} = -\frac{3}{16i}$. $\operatorname{Res}_0 = \frac{d}{dz}\left[-\frac{(z^2-1)^2}{4i(2z^2 + 5z + 2)}\right]_{z=0} = -\frac{1}{4i}\cdot\frac{0\cdot 2 - 1\cdot 5}{4} = \frac{5}{16i}$. Sum $\frac{2}{16i} = \frac{1}{8i}$, integral $\frac{\pi}{4}$.
8. $\frac{z}{z^2 + 1}$ decays only like $\frac1R$, so ML gives $\pi R\cdot\frac{R}{R^2 - 1} \to \pi$, not $0$. Jordan with $M_R = \frac{R}{R^2 - 1} \to 0$ gives $\frac{\pi R}{R^2-1} \to 0$. Then $\int\frac{x e^{ix}}{x^2+1} = 2\pi i\cdot\frac{i e^{-1}}{2i} = \frac{i\pi}{e}$, whose imaginary part is $\frac{\pi}{e}$.
9. $\operatorname{Res}_{a_k}\frac{1}{1 + z^{2n}} = \frac{1}{2n a_k^{2n-1}} = \frac{a_k}{2n a_k^{2n}} = -\frac{a_k}{2n}$. With $\omega = e^{i\pi/n}$, $a_k = e^{i\pi/(2n)}\omega^k$, and $\sum_{k=0}^{n-1}\omega^k = \frac{1 - \omega^n}{1 - \omega} = \frac{2}{1 - \omega}$. So the integral is $2\pi i\cdot\left(-\frac{e^{i\pi/(2n)}}{2n}\cdot\frac{2}{1 - e^{i\pi/n}}\right) = \frac{-2\pi i}{n\left(e^{-i\pi/(2n)} - e^{i\pi/(2n)}\right)} = \frac{-2\pi i}{n\cdot(-2i\sin\frac{\pi}{2n})} = \frac{\pi}{n\sin\frac{\pi}{2n}}$.

Numerical checks:

```python
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def real_line_quadrature(f, n_intervals=200000):
    # x = tan(u) and the midpoint rule, as in section 4.1
    step = np.pi / n_intervals
    u = -np.pi / 2 + (np.arange(n_intervals) + 0.5) * step
    return np.sum(f(np.tan(u)) / np.cos(u) ** 2) * step


theta = 2 * np.pi * np.arange(4096) / 4096
print("Ex 1:", real_line_quadrature(lambda x: 1 / (x**2 + 4)), " vs", np.pi / 2)
print("Ex 2:", 2 * np.pi * np.mean(1 / (3 + 2 * np.cos(theta))), " vs", 2 * np.pi / np.sqrt(5))
print("Ex 3:", real_line_quadrature(lambda x: 1 / (x**2 + 2 * x + 2)), " vs", np.pi)
print("Ex 4:", real_line_quadrature(lambda x: 1 / ((x**2 + 1) * (x**2 + 4))), " vs", np.pi / 6)

x = np.linspace(-2000, 2000, 2_000_001)
print("Ex 5:", np.trapezoid(np.cos(2 * x) / (x**2 + 1), x), " vs", np.pi * np.exp(-2))
print("Ex 6:", np.trapezoid(np.sin(x) / (x**2 + 2 * x + 2), x), " vs", -np.pi * np.sin(1) / np.e)
print("Ex 7:", 2 * np.pi * np.mean(np.sin(theta) ** 2 / (5 + 4 * np.cos(theta))), " vs", np.pi / 4)

half_length = 2000 * np.pi + np.pi / 2
x = np.linspace(-half_length, half_length, 4_000_001)
print("Ex 8:", np.trapezoid(x * np.sin(x) / (x**2 + 1), x), " vs", np.pi / np.e)

for n in [1, 2, 3, 10]:
    value = real_line_quadrature(lambda x: 1 / (1 + x ** (2 * n)))
    print(f"Ex 9, n = {n:2d}:", value, " vs", np.pi / (n * np.sin(np.pi / (2 * n))))
```

## 8. Conclusion

- **Rational functions on $\mathbb{R}$** ($\deg Q \ge \deg P + 2$, no real poles): close with the upper semicircle; the ML-inequality kills the arc, and $\int_{-\infty}^{\infty}\frac{P}{Q}dx = 2\pi i\sum_{\operatorname{Im}a > 0}\operatorname{Res}$.
- **Trigonometric integrals over $[0, 2\pi]$:** $z = e^{i\theta}$, $\cos\theta = \frac{z + z^{-1}}{2}$, $\sin\theta = \frac{z - z^{-1}}{2i}$, $d\theta = \frac{dz}{iz}$; sum the residues inside $|z| = 1$.
- **Fourier-type integrals:** use $e^{ikz}$ (small in the upper half-plane for $k > 0$), never $\cos(kz)$ (exponentially large there); take real or imaginary parts at the end. **Jordan's lemma** handles integrands decaying only like $\frac1x$.
- Results such as $\int\frac{dx}{1+x^4} = \frac{\pi}{\sqrt2}$, $\int_0^{2\pi}\frac{d\theta}{2+\cos\theta} = \frac{2\pi}{\sqrt3}$ and $\int\frac{\cos x}{1+x^2}dx = \frac{\pi}{e}$ agree with numerical quadrature (with $x = \tan u$ for infinite ranges, periodic trapezoid for periods).
