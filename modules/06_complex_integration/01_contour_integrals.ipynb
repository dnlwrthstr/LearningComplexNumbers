# Contour Integrals

A real integral $\int_a^b f(x)\,dx$ runs along a piece of the real line. In the complex plane there is no single "line from $a$ to $b$": there are infinitely many paths, and we integrate **along a chosen curve**. This notebook defines the contour integral $\int_\gamma f(z)\,dz$, reduces it to an ordinary integral over a parameter $t$, and computes the single most important integral in complex analysis, $\oint dz/z = 2\pi i$.

We also meet the first sign that holomorphic functions are special: for $f(z) = z^2$ the answer does not depend on the path, while for $f(z) = \bar z$ it does. Finally, the **ML-inequality** lets us bound an integral we cannot compute — the workhorse estimate for the rest of the module.

This is notebook 01 of module 06 (**Complex integration**) in the **Learning Complex Numbers** series.

## 1. Introduction

Why integrate along curves at all? Because the payoff is enormous. By the end of this module, contour integrals will

- recover every value of a holomorphic function from its values on a circle (Cauchy's integral formula),
- produce power series and their radius of convergence,
- evaluate real integrals such as $\int_{-\infty}^{\infty} \frac{\cos x}{1 + x^2}\,dx = \pi/e$ that resist every trick from real calculus.

All of that rests on one definition and one computation, both in this notebook.

- Bound an integral with the ML-inequality $\left|\int_\gamma f\,dz\right| \le M L$.

## 2. Theory

### 2.1 Curves and parametrizations

A **curve** (or path) is a continuous, piecewise continuously differentiable map

$$
\gamma : [a, b] \to \mathbb{C}, \qquad t \mapsto \gamma(t) = x(t) + i\,y(t).
$$

It starts at $\gamma(a)$, ends at $\gamma(b)$, and is **closed** if $\gamma(a) = \gamma(b)$. The direction of increasing $t$ is the **orientation**. Its derivative $\gamma'(t) = x'(t) + i\,y'(t)$ is the velocity, a complex number pointing along the curve.

The three curves you will use constantly:

| Curve | Parametrization | $\gamma'(t)$ |
|---|---|---|
| segment from $z_0$ to $z_1$ | $\gamma(t) = z_0 + t(z_1 - z_0)$, $t \in [0, 1]$ | $z_1 - z_0$ |
| circle $\lvert z - c\rvert = r$, counter-clockwise | $\gamma(t) = c + r e^{it}$, $t \in [0, 2\pi]$ | $i r e^{it}$ |
| arc of that circle from angle $\alpha$ to $\beta$ | same, $t \in [\alpha, \beta]$ | $i r e^{it}$ |

The convention for closed curves is **counter-clockwise = positive** (the region enclosed is on your left). The symbol $\oint$ just means "the curve is closed".

### 2.2 The definition

Chop $\gamma$ at points $z_0, z_1, \dots, z_N$ along the curve and form the Riemann sum

$$
\sum_{k=1}^{N} f(z_k)\,(z_k - z_{k-1}) = \sum_{k} f(z_k)\,\Delta z_k .
$$

Each term multiplies two complex numbers: the value of $f$ and the small **step** $\Delta z_k$ along the curve (a tiny arrow). As the steps shrink, $\Delta z_k \approx \gamma'(t_k)\,\Delta t$, and the sum becomes

$$
\boxed{\int_\gamma f(z)\,dz = \int_a^b f(\gamma(t))\,\gamma'(t)\,dt.}
$$

The right-hand side is an ordinary integral of a complex-valued function of the real variable $t$: integrate the real and imaginary parts separately.

### 2.3 Properties

- **Linear**: $\int_\gamma (\alpha f + \beta g)\,dz = \alpha\int_\gamma f\,dz + \beta\int_\gamma g\,dz$.
- **Reversal**: running the curve backwards ($-\gamma$) flips the sign, $\int_{-\gamma} f\,dz = -\int_\gamma f\,dz$.
- **Concatenation**: if $\gamma$ is $\gamma_1$ followed by $\gamma_2$, the integrals add.
- **Reparametrization invariance**: any other parametrization tracing the same curve in the same direction gives the same value (the chain rule absorbs the change of variable).

### 2.4 The fundamental integral

Let $\gamma(t) = a + r e^{it}$, $t \in [0, 2\pi]$, and $n$ an integer. Then $(\gamma(t) - a)^n = r^n e^{int}$ and $\gamma'(t) = i r e^{it}$, so

$$
\oint_{|z-a|=r} (z-a)^n\,dz = \int_0^{2\pi} r^n e^{int}\, i r e^{it}\,dt = i\,r^{n+1}\int_0^{2\pi} e^{i(n+1)t}\,dt.
$$

If $n + 1 \neq 0$, the integral of $e^{i(n+1)t}$ over a full period is $0$. If $n = -1$, the integrand is $e^{0} = 1$ and the integral is $2\pi$. Hence

$$
\boxed{\oint_{|z-a|=r} (z-a)^n\,dz = \begin{cases} 2\pi i, & n = -1,\\ 0, & n \neq -1.\end{cases}}
$$

The answer does not depend on the radius $r$. For $n = -1$ the reason is visible in the integrand: $\frac{1}{\gamma}\,\gamma' = \frac{i r e^{it}}{r e^{it}} = i$ is **constant** — at every point the step $dz$ is perpendicular to $z$, and dividing by $z$ rotates it to point straight "up" ($i$). This single fact powers residues, Cauchy's formula and Laurent series.

### 2.5 Antiderivatives and path independence

If $F$ is holomorphic with $F' = f$ along $\gamma$, then $\frac{d}{dt}F(\gamma(t)) = f(\gamma(t))\gamma'(t)$ by the chain rule, so

$$
\int_\gamma f(z)\,dz = F(\gamma(b)) - F(\gamma(a)).
$$

The integral depends **only on the endpoints**, and over any closed curve it is $0$. So $z^n$ with $n \ne -1$ has the antiderivative $\frac{z^{n+1}}{n+1}$ (on $\mathbb{C}$, or $\mathbb{C}\setminus\{0\}$ for negative $n$) and its closed integrals vanish — matching 2.4.

$1/z$ is the exception: its "antiderivative" $\log z$ is multivalued and jumps by $2\pi i$ once around the origin. That jump is exactly the $2\pi i$.

$f(z) = \bar z$ has **no** antiderivative at all (it is not holomorphic), and its integral between two points does depend on the path — see Example 3.

### 2.6 Length and the ML-inequality

The **length** of $\gamma$ is $L = \int_a^b |\gamma'(t)|\,dt$ (a circle of radius $r$ has $L = 2\pi r$). If $|f(z)| \le M$ for every $z$ on $\gamma$, then

$$
\boxed{\left|\int_\gamma f(z)\,dz\right| \le \int_a^b |f(\gamma(t))|\,|\gamma'(t)|\,dt \le M L.}
$$

The first step is the triangle inequality for integrals (a sum of arrows is at most as long as the arrows laid end to end). The ML-inequality is crude, but it is all you need to show that an integral **tends to zero**, which is how contours are "pushed to infinity" in notebook 07.

## 3. Worked Examples

### Example 1 — $\int_\gamma z\,dz$ along the segment from $0$ to $1 + i$

$\gamma(t) = (1+i)t$, $\gamma'(t) = 1 + i$, $t \in [0, 1]$:

$$
\int_0^1 (1+i)t\,(1+i)\,dt = (1+i)^2 \int_0^1 t\,dt = 2i \cdot \tfrac12 = i.
$$

Check with the antiderivative $F(z) = z^2/2$: $F(1+i) - F(0) = \frac{2i}{2} = i$. ✓

### Example 2 — $\oint_{|z|=1} \frac{dz}{z}$

$\gamma(t) = e^{it}$, $\gamma'(t) = i e^{it}$:

$$
\int_0^{2\pi} \frac{i e^{it}}{e^{it}}\,dt = \int_0^{2\pi} i\,dt = 2\pi i.
$$

Clockwise ($\gamma(t) = e^{-it}$) gives $-2\pi i$; twice around ($t \in [0, 4\pi]$) gives $4\pi i$. The integral **counts turns around $0$**, times $2\pi i$.

### Example 3 — $\int \bar z\,dz$ from $0$ to $1 + i$ along two paths

*Path A, straight:* $\gamma(t) = (1+i)t$, $\bar\gamma = (1-i)t$:

$$
\int_0^1 (1-i)t\,(1+i)\,dt = 2\int_0^1 t\,dt = 1.
$$

*Path B, along the real axis to $1$, then up to $1 + i$:*

- leg 1: $\gamma(t) = t$, $\gamma' = 1$: $\int_0^1 t\,dt = \tfrac12$;
- leg 2: $\gamma(t) = 1 + it$, $\gamma' = i$, $\bar\gamma = 1 - it$: $\int_0^1 (1 - it)\,i\,dt = \int_0^1 (i + t)\,dt = i + \tfrac12$.

Total: $1 + i$. **Different from path A** — same endpoints, different answers. Path B followed by path A reversed is a closed loop around the triangle with vertices $0, 1, 1+i$, and $\oint \bar z\,dz = (1 + i) - 1 = i = 2i \times \text{Area}$ (area $\tfrac12$). Notebook 02 explains the $2i \times$ area.

### Example 4 — $\oint_{|z|=2} z^3\,dz$ and $\oint_{|z-1|=1} \frac{dz}{(z-1)^2}$

Both are covered by 2.4: $n = 3$ and $n = -2$ give $0$. Only the power $-1$ survives.

### Example 5 — an ML bound

Bound $\left|\oint_{|z|=2} \frac{dz}{z^2 + 1}\right|$.

- $L = 2\pi \cdot 2 = 4\pi$.
- On $|z| = 2$, the reverse triangle inequality gives $|z^2 + 1| \ge |z|^2 - 1 = 3$, so $M = \frac13$.

Hence $\left|\oint\right| \le \frac{4\pi}{3} \approx 4.19$. (The true value turns out to be $0$ — notebook 06 — so the bound is honest but far from tight.)

### Example 6 — the bound that goes to zero

On the upper half-circle $C_R$: $z = R e^{it}$, $t \in [0, \pi]$, with $R > 1$: $L = \pi R$ and $\left|\frac{1}{z^2 + 1}\right| \le \frac{1}{R^2 - 1}$, so

$$
\left|\int_{C_R} \frac{dz}{z^2 + 1}\right| \le \frac{\pi R}{R^2 - 1} \xrightarrow{R \to \infty} 0.
$$

This is the first step of evaluating $\int_{-\infty}^{\infty} \frac{dx}{1+x^2}$ with residues (notebook 07).

## 4. Python Examples

### 4.1 A contour-integral helper

The definition is directly computable: sample $t$, evaluate $f(\gamma(t))\gamma'(t)$, integrate with the trapezoidal rule. For closed smooth curves on a uniform grid the trapezoidal rule is spectacularly accurate (the error falls faster than any power of the step), so a few thousand points give near machine precision.

```python
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def contour_integral(f, gamma, dgamma, t_start, t_end, n_points=4001):
    # Integral of f along gamma: integral of f(gamma(t)) * gamma'(t) dt with the trapezoidal rule
    t = np.linspace(t_start, t_end, n_points)
    return np.trapezoid(f(gamma(t)) * dgamma(t), t)


def circle(center, radius):
    # Counter-clockwise circle as (gamma, gamma') for t in [0, 2 pi]
    return (lambda t: center + radius * np.exp(1j * t),
            lambda t: 1j * radius * np.exp(1j * t))


def segment(z_start, z_end):
    # Straight segment as (gamma, gamma') for t in [0, 1]
    return (lambda t: z_start + t * (z_end - z_start),
            lambda t: (z_end - z_start) * np.ones_like(t))


gamma, dgamma = circle(0, 1)
value = contour_integral(lambda z: 1 / z, gamma, dgamma, 0, 2 * np.pi)
print("closed integral of 1/z on |z| = 1 :", value)
print("2 pi i                             :", 2j * np.pi)
print("agree to 1e-12:", abs(value - 2j * np.pi) < 1e-12)
```

### 4.2 $\oint (z - a)^n\,dz$ for many $n$ and several circles

The table confirms 2.4: only $n = -1$ survives, and the radius and centre do not matter.

```python
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def contour_integral(f, gamma, dgamma, t_start, t_end, n_points=4001):
    t = np.linspace(t_start, t_end, n_points)
    return np.trapezoid(f(gamma(t)) * dgamma(t), t)


for center, radius in [(0, 1), (2 - 1j, 0.5), (-1 + 3j, 4)]:
    gamma = lambda t: center + radius * np.exp(1j * t)
    dgamma = lambda t: 1j * radius * np.exp(1j * t)
    print(f"circle |z - ({center})| = {radius}")
    for n in range(-3, 4):
        value = contour_integral(lambda z: (z - center) ** n, gamma, dgamma, 0, 2 * np.pi)
        value = complex(round(value.real, 10) + 0.0, round(value.imag, 10) + 0.0)  # + 0.0 turns -0.0 into 0.0
        print(f"   n = {n:2d}:  {value}")
```

### 4.3 The worked examples, numerically and exactly

Sympy can do the $t$-integral exactly once the parametrization is substituted in.

```python
import numpy as np
import sympy as sp

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def contour_integral(f, gamma, dgamma, t_start, t_end, n_points=4001):
    t = np.linspace(t_start, t_end, n_points)
    return np.trapezoid(f(gamma(t)) * dgamma(t), t)


t = sp.symbols("t", real=True)


def exact_integral(f_expr, gamma_expr, t_start, t_end):
    # f_expr is a function of the symbol z; substitute gamma(t) and integrate over t exactly
    integrand = f_expr(gamma_expr) * sp.diff(gamma_expr, t)
    return sp.simplify(sp.integrate(sp.expand(integrand), (t, t_start, t_end)))


# Example 1: z along the segment 0 -> 1 + i
print("Ex 1 exact  :", exact_integral(lambda z: z, (1 + sp.I) * t, 0, 1))
# Example 3, path A and path B for conj(z)
path_a = exact_integral(lambda z: sp.conjugate(z), (1 + sp.I) * t, 0, 1)
path_b = exact_integral(lambda z: sp.conjugate(z), t, 0, 1) + exact_integral(lambda z: sp.conjugate(z), 1 + sp.I * t, 0, 1)
print("Ex 3 path A :", path_a)
print("Ex 3 path B :", path_b)

# The same for z^2 shows path independence (antiderivative z^3 / 3)
z2_a = exact_integral(lambda z: z**2, (1 + sp.I) * t, 0, 1)
z2_b = exact_integral(lambda z: z**2, t, 0, 1) + exact_integral(lambda z: z**2, 1 + sp.I * t, 0, 1)
print("z^2 path A  :", sp.expand(z2_a), "  path B:", sp.expand(z2_b), "  F(1+i) - F(0):", sp.expand((1 + sp.I) ** 3 / 3))

# Numerical check of path A for conj(z)
value = contour_integral(np.conj, lambda s: (1 + 1j) * s, lambda s: (1 + 1j) * np.ones_like(s), 0, 1)
print("Ex 3 path A numerically:", value)
```

### 4.4 Orientation and winding

Running the circle clockwise, or twice around, scales $\oint dz/z$ by $-1$ or $2$.

```python
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

t = np.linspace(0, 1, 8001)
for turns in [1, -1, 2, 3]:
    gamma = np.exp(2j * np.pi * turns * t)
    dgamma = 2j * np.pi * turns * gamma
    value = np.trapezoid(dgamma / gamma, t)
    print(f"{turns:2d} turn(s): integral / (2 pi i) = {value / (2j * np.pi):.6f}")
```

### 4.5 Checking the ML-inequality

We compare the true size of $\int_{C_R} \frac{dz}{z^2+1}$ on the upper half-circle with the bound $\frac{\pi R}{R^2 - 1}$ from Example 6.

```python
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

t = np.linspace(0, np.pi, 20001)
for radius in [2, 5, 10, 100]:
    z = radius * np.exp(1j * t)
    dz = 1j * z
    value = np.trapezoid(dz / (z**2 + 1), t)
    bound = np.pi * radius / (radius**2 - 1)
    print(f"R = {radius:4d}:  |integral| = {abs(value):.6f}   ML bound = {bound:.6f}   bound holds: {abs(value) <= bound}")
```

## 5. Visualizations

### 5.1 A contour integral is a sum of products of arrows

Below, a quarter of the unit circle is chopped into steps $\Delta z_k$ (black arrows). For $f(z) = 1/z$ each product $f(z_k)\,\Delta z_k$ (right panel) points almost exactly in the $+i$ direction: dividing by $z_k$ rotates the tangent step back by $\arg z_k$. Laying them head to tail climbs straight up the imaginary axis.

```python
import matplotlib.pyplot as plt
import numpy as np

n_steps = 8
angles = np.linspace(0, np.pi / 2, n_steps + 1)
points = np.exp(1j * angles)
steps = np.diff(points)
products = (1 / points[1:]) * steps

fig, (ax_left, ax_right) = plt.subplots(1, 2, figsize=(12, 5.5))

circle = np.exp(1j * np.linspace(0, 2 * np.pi, 300))
ax_left.plot(circle.real, circle.imag, color="lightgray")
for start, step in zip(points[:-1], steps):
    ax_left.annotate("", xy=(start.real + step.real, start.imag + step.imag), xytext=(start.real, start.imag),
                     arrowprops=dict(arrowstyle="->", color="black", linewidth=1.5))
ax_left.scatter(points.real, points.imag, color="tab:blue", zorder=3, label="$z_k$")
ax_left.set_title(r"Steps $\Delta z_k$ along a quarter circle")

running = 0
for product in products:
    ax_right.annotate("", xy=((running + product).real, (running + product).imag), xytext=(running.real, running.imag),
                      arrowprops=dict(arrowstyle="->", color="tab:red", linewidth=1.5))
    running = running + product
ax_right.scatter([0, running.real], [0, running.imag], color="tab:red", zorder=3)
ax_right.text(running.real + 0.05, running.imag, f"8-step sum = {running:.3f}\nexact $i\\pi/2$ = {np.pi / 2:.3f}i", va="center")
ax_right.set_title(r"Products $f(z_k)\,\Delta z_k$ for $f = 1/z$, head to tail")

for ax, limits in [(ax_left, (-1.3, 1.3)), (ax_right, (-0.8, 1.8))]:
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(limits)
    ax.set_ylim(limits)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
ax_left.legend(loc="lower left")
plt.tight_layout()
plt.show()
```

### 5.2 The running integral as a path

Plot $I(s) = \int_0^s f(\gamma(t))\gamma'(t)\,dt$ in the complex plane as $s$ goes from $0$ to $2\pi$ on the unit circle. For $1/z$ the running integral walks straight up to $2\pi i$ and never comes back. For $z^2$ and $1/z^2$ it makes loops and returns to $0$ — their antiderivatives are single-valued.

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 2 * np.pi, 2001)
z = np.exp(1j * t)
dz = 1j * z
dt = t[1] - t[0]

integrands = {
    "$f = 1/z$": 1 / z,
    "$f = z^2$": z**2,
    "$f = 1/z^2$": 1 / z**2,
}

fig, ax = plt.subplots(figsize=(7, 7))
for label, values in integrands.items():
    products = values * dz
    running = np.concatenate([[0], np.cumsum((products[1:] + products[:-1]) / 2 * dt)])
    ax.plot(running.real, running.imag, linewidth=2, label=label)
    ax.scatter(running[-1].real, running[-1].imag, s=60, zorder=3)

ax.scatter(0, 2 * np.pi, marker="x", color="black", s=80, label=r"$2\pi i$")
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-3.5, 3.5)
ax.set_ylim(-1, 7)
ax.set_aspect("equal")
ax.set_title(r"Running integral $\int_0^s f(e^{it})\, i e^{it}\, dt$ for $s \in [0, 2\pi]$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend()
plt.show()
```

### 5.3 Path independence versus path dependence

Three paths from $0$ to $1 + i$. For $z^2$ (holomorphic, antiderivative $z^3/3$) all three give the same value; for $\bar z$ they disagree.

```python
import matplotlib.pyplot as plt
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

s = np.linspace(0, 1, 4001)
end = 1 + 1j


def concatenate(*legs):
    # Join legs (each sampled on s in [0, 1]) into one path with parameter in [0, number of legs]
    points = np.concatenate([leg if k == 0 else leg[1:] for k, leg in enumerate(legs)])
    parameter = np.linspace(0, len(legs), len(points))
    return points, parameter


paths = {
    "straight": concatenate(end * s),
    "right then up": concatenate(s + 0j, 1 + 1j * s),
    "parabola $y = x^2$": concatenate(s + 1j * s**2),
}

fig, ax = plt.subplots(figsize=(6.5, 6.5))
for name, (points, parameter) in paths.items():
    derivative = np.gradient(points, parameter)
    value_z2 = np.trapezoid(points**2 * derivative, parameter)
    value_conj = np.trapezoid(np.conj(points) * derivative, parameter)
    label = f"{name}\n  $z^2$: {value_z2:.3f}\n  $\\bar z$: {value_conj:.3f}"
    ax.plot(points.real, points.imag, linewidth=2, label=label)

ax.scatter([0, 1], [0, 1], color="black", zorder=3)
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-0.2, 1.6)
ax.set_ylim(-0.2, 1.6)
ax.set_aspect("equal")
ax.set_title("Integrals from 0 to 1 + i along three paths")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left", fontsize=9)
plt.show()
```

Every path gives $z^2 \to \frac{(1+i)^3}{3} = -\frac23 + \frac23 i$, while $\bar z$ gives $1$, $1 + i$ and something else again (exercise 7 asks you to compute it by hand).

### 5.4 The ML-bound shrinking to zero

On a log–log plot, both the true integral over the half-circle and its ML bound fall like $1/R$.

```python
import matplotlib.pyplot as plt
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

t = np.linspace(0, np.pi, 20001)
radii = np.logspace(0.2, 3, 40)
actual = []
for radius in radii:
    z = radius * np.exp(1j * t)
    actual.append(abs(np.trapezoid(1j * z / (z**2 + 1), t)))
bound = np.pi * radii / (radii**2 - 1)

plt.figure(figsize=(8, 5))
plt.loglog(radii, bound, label=r"ML bound $\pi R / (R^2 - 1)$")
plt.loglog(radii, actual, "o", markersize=4, label=r"$|\int_{C_R} dz/(z^2+1)|$")
plt.title("Half-circle integral and its ML bound")
plt.xlabel("radius R")
plt.ylabel("size")
plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.show()
```

## 7. Solutions

1. $\gamma(t) = 2i + 3e^{it}$, $t \in [0, 2\pi]$; $\gamma'(t) = 3i e^{it}$.
2. $\gamma(t) = 1 + t(i - 1)$, $\gamma' = i - 1$. $\int_0^1 (1 + t(i-1))(i-1)\,dt = (i - 1) + \frac{(i-1)^2}{2} = (i - 1) + \frac{-2i}{2} = -1$. Antiderivative: $\frac{i^2 - 1^2}{2} = \frac{-1 - 1}{2} = -1$. ✓
3. By 2.4 with $a = 3$: $2\pi i$, $0$, $0$.
4. $\oint \bar z\,dz = \oint \frac{dz}{z} = 2\pi i$. Area of the unit disk is $\pi$, and $2i\pi = 2\pi i$. ✓
5. On the arc, each $f\,\Delta z$ points in the $+i$ direction (as in 5.1), and the arc covers half a turn, so the guess is $\pi i$. Indeed $\int_0^\pi \frac{i e^{it}}{e^{it}}\,dt = \pi i$. The segment from $1$ to $-1$ passes **through $0$**, where $1/z$ is undefined: the integral diverges (it is not even defined). The lower half circle would give $-\pi i$; the two halves differ by $2\pi i$, the full loop.
6. $L = |1 + i| = \sqrt2$. On the segment $z = (1+i)t$, $|e^z| = e^{\operatorname{Re} z} = e^{t} \le e$, so $M = e$ and $|\int| \le e\sqrt2 \approx 3.84$. Exact: $e^{1+i} - 1 = e\cos 1 - 1 + i e \sin 1 \approx 0.469 + 2.287i$, modulus $\approx 2.33$. ✓
7. $\gamma = s + is^2$, $\gamma' = 1 + 2is$, $\bar\gamma = s - is^2$. Product: $(s - is^2)(1 + 2is) = s + 2is^2 - is^2 + 2s^3 = s + 2s^3 + i s^2$. Integrate over $[0, 1]$: $\frac12 + \frac12 + \frac{i}{3} = 1 + \frac{i}{3}$.
8. Legs (counter-clockwise): $0 \to 1$: $z = t$, $\operatorname{Re} z = t$, $dz = dt$: $\frac12$. $1 \to 1+i$: $z = 1 + it$, $\operatorname{Re} = 1$, $dz = i\,dt$: $i$. $1 + i \to i$: $z = 1 + i - t$, $\operatorname{Re} = 1 - t$, $dz = -dt$: $-\frac12$. $i \to 0$: $z = i(1 - t)$, $\operatorname{Re} = 0$: $0$. Sum $= i$, and the area is $1$. ✓ (In general $\oint x\,dz = \oint x\,dx + i\oint x\,dy = 0 + i\cdot\text{Area}$ by Green's theorem.)
9. By the chain rule $(\gamma\circ\phi)'(s) = \gamma'(\phi(s))\,\phi'(s)$. Substitute $t = \phi(s)$, $dt = \phi'(s)\,ds$: the left side becomes $\int_a^b f(\gamma(t))\gamma'(t)\,dt$. (A decreasing $\phi$ swaps the limits and produces the minus sign of the reversal rule.)

Numerical checks:

```python
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def contour_integral(f, gamma, dgamma, t_start, t_end, n_points=20001):
    t = np.linspace(t_start, t_end, n_points)
    return np.trapezoid(f(gamma(t)) * dgamma(t), t)


ones = np.ones_like
print("Ex 2:", contour_integral(lambda z: z, lambda t: 1 + t * (1j - 1), lambda t: (1j - 1) * ones(t), 0, 1))
for n in [-1, 4, -5]:
    value = contour_integral(lambda z: (z - 3) ** n, lambda t: 3 + 2 * np.exp(1j * t), lambda t: 2j * np.exp(1j * t), 0, 2 * np.pi)
    print(f"Ex 3, n = {n:2d}:", np.round(value, 10) + 0.0)
print("Ex 4:", contour_integral(np.conj, lambda t: np.exp(1j * t), lambda t: 1j * np.exp(1j * t), 0, 2 * np.pi), " vs 2 pi i =", 2j * np.pi)
print("Ex 5:", contour_integral(lambda z: 1 / z, lambda t: np.exp(1j * t), lambda t: 1j * np.exp(1j * t), 0, np.pi), " vs pi i")
value = contour_integral(np.exp, lambda t: (1 + 1j) * t, lambda t: (1 + 1j) * ones(t), 0, 1)
print("Ex 6:", value, " |value| =", abs(value), " bound =", np.e * np.sqrt(2))
print("Ex 7:", contour_integral(np.conj, lambda s: s + 1j * s**2, lambda s: 1 + 2j * s, 0, 1), " vs 1 + i/3")
legs = [(0, 1), (1, 1 + 1j), (1 + 1j, 1j), (1j, 0)]
total = sum(contour_integral(np.real, lambda t, a=a, b=b: a + t * (b - a), lambda t, a=a, b=b: (b - a) * ones(t), 0, 1) for a, b in legs)
print("Ex 8:", total)
```

## 8. Conclusion

- A contour integral is defined through a parametrization: $\int_\gamma f\,dz = \int_a^b f(\gamma(t))\gamma'(t)\,dt$ — a limit of sums of products $f(z_k)\,\Delta z_k$.
- It is linear, flips sign under reversal, adds over concatenated paths and ignores the choice of parametrization.
- The fundamental integral: $\oint_{|z-a|=r}(z-a)^n\,dz = 2\pi i$ if $n = -1$ and $0$ otherwise, for every radius. $\oint dz/z$ counts turns around the origin.
- If $f$ has an antiderivative $F$, the integral is $F(\text{end}) - F(\text{start})$: path-independent, and $0$ around closed curves. $\bar z$ has none, and its integrals depend on the path.
- The ML-inequality $\left|\int_\gamma f\,dz\right| \le ML$ bounds what we cannot compute, and shows that integrals over large arcs vanish.
- Numerically, `np.trapezoid` on $f(\gamma(t))\gamma'(t)$ is a reliable contour integrator, and extremely accurate on closed curves.
