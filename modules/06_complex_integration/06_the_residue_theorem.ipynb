# The Residue Theorem

Everything in this module comes together here. Around an isolated singularity $a$, a function has a Laurent series, and integrating it term by term around a small circle kills every term except $\frac{a_{-1}}{z - a}$. That one coefficient is the **residue** $\operatorname{Res}(f, a)$. Deforming any closed contour onto small circles around the singularities inside it gives the **residue theorem**:

$$
\oint_\gamma f(z)\,dz = 2\pi i \sum_{a \text{ inside } \gamma} \operatorname{Res}(f, a).
$$

A contour integral becomes a finite sum of numbers, each computed by a short formula. This notebook develops those formulas, checks the theorem numerically, and uses sympy's `residue`.

This is notebook 06 of module 06 (**Complex integration**) in the **Learning Complex Numbers** series.

## 1. Introduction

Cauchy's theorem says integrals around regions **without** singularities vanish; Cauchy's integral formula handles one simple singularity of the form $\frac{f(z)}{z - a}$. The residue theorem handles **any finite number of isolated singularities of any kind**, and in practice it is the form of complex integration people use: in engineering (inverse Laplace and $z$-transforms), physics (Green's functions, dispersion relations), probability (characteristic functions), and for evaluating real integrals, which is notebook 07.

## 2. Theory

### 2.1 The residue

Let $a$ be an isolated singularity of $f$, with Laurent series $f(z) = \sum_{n=-\infty}^{\infty} a_n (z - a)^n$ on $0 < |z - a| < R$. The **residue** of $f$ at $a$ is

$$
\operatorname{Res}(f, a) = a_{-1}.
$$

On a small circle $C_\varepsilon$ around $a$, integrate term by term. By the fundamental integral of notebook 01, $\oint_{C_\varepsilon} (z - a)^n\,dz = 0$ for $n \ne -1$ and $2\pi i$ for $n = -1$, so

$$
\oint_{C_\varepsilon} f(z)\,dz = 2\pi i\,a_{-1} = 2\pi i\,\operatorname{Res}(f, a).
$$

The residue is literally "what remains" after integrating.

### 2.2 The residue theorem

> **Residue theorem.** Let $\gamma$ be a positively oriented simple closed curve, and let $f$ be holomorphic inside and on $\gamma$ except at finitely many points $a_1, \dots, a_k$ inside $\gamma$. Then
> $$\oint_\gamma f(z)\,dz = 2\pi i \sum_{j=1}^{k} \operatorname{Res}(f, a_j).$$

*Proof.* Deform $\gamma$ onto small circles around each $a_j$ (notebook 02, 2.4), then apply 2.1 to each. $\blacksquare$

For a closed curve that winds around $a_j$ several times or clockwise, each residue is weighted by the **winding number** $n(\gamma, a_j) = \frac{1}{2\pi i}\oint_\gamma \frac{dz}{z - a_j}$.

Cauchy's integral formula is the special case $\frac{f(z)}{z - a}$ with $f$ holomorphic: $\operatorname{Res} = f(a)$.

### 2.3 Computing residues

**Simple pole** ($m = 1$): $f(z) = \frac{a_{-1}}{z - a} + a_0 + \cdots$, so

$$
\operatorname{Res}(f, a) = \lim_{z \to a} (z - a) f(z).
$$

**Quotient with a simple zero:** if $f = \frac{g}{h}$ with $g(a) \ne 0$, $h(a) = 0$, $h'(a) \ne 0$, then $(z - a)\frac{g}{h} = \frac{g(z)}{(h(z) - h(a))/(z - a)} \to \frac{g(a)}{h'(a)}$:

$$
\operatorname{Res}\left(\frac{g}{h}, a\right) = \frac{g(a)}{h'(a)}.
$$

This is the formula you will use most. Example: $\operatorname{Res}\left(\frac{1}{z^4 + 1}, a\right) = \frac{1}{4a^3}$ at each root of $a^4 = -1$.

**Pole of order $m$:** $(z - a)^m f(z) = a_{-m} + \cdots + a_{-1}(z - a)^{m-1} + a_0 (z - a)^m + \cdots$ is holomorphic, and $a_{-1}$ is its coefficient of $(z - a)^{m-1}$, i.e. its $(m-1)$-th Taylor coefficient:

$$
\operatorname{Res}(f, a) = \frac{1}{(m-1)!}\lim_{z \to a} \frac{d^{m-1}}{dz^{m-1}}\Big[(z - a)^m f(z)\Big].
$$

Using a larger $m$ than the true order is harmless; using a smaller one is wrong.

**Essential singularity:** no shortcut — read $a_{-1}$ off the Laurent series. $e^{1/z} = 1 + \frac1z + \cdots$ has residue $1$ at $0$.

**Warning:** a pole with zero residue is possible, e.g. $\frac{1}{z^2}$ or $\frac{1 - \cos z}{z^4}$ (notebook 05). The order of a pole and its residue are independent pieces of information.

### 2.4 The residue at infinity

If $f$ is holomorphic for $|z| > R$, expand $f(z) = \sum b_n z^n$ there and define $\operatorname{Res}(f, \infty) = -b_{-1}$. A large circle $|z| = \rho > R$ gives $\oint_{|z|=\rho} f\,dz = 2\pi i\,b_{-1}$, and it also encloses every finite singularity. Hence

$$
\sum_{\text{all finite } a} \operatorname{Res}(f, a) + \operatorname{Res}(f, \infty) = 0.
$$

This saves work when many poles are inside: for $\frac{1}{z^4 + 1} = z^{-4} - z^{-8} + \cdots$ on $|z| > 1$, $b_{-1} = 0$, so the four finite residues sum to $0$. For a rational function $\frac{p}{q}$ with $\deg q \ge \deg p + 2$ the same is always true.

## 3. Worked Examples

### Example 1 — $\oint_{|z|=2} \frac{z}{z^2 + 1}\,dz$

Simple poles at $\pm i$, both inside. With $g = z$, $h = z^2 + 1$, $h' = 2z$: $\operatorname{Res} = \frac{\pm i}{\pm 2i} = \frac12$ at each.

$$
\oint = 2\pi i\left(\tfrac12 + \tfrac12\right) = 2\pi i.
$$

### Example 2 — $\oint_{|z|=2} \frac{e^z}{(z - 1)^2}\,dz$

Pole of order $2$ at $1$: $\operatorname{Res} = \frac{1}{1!}\frac{d}{dz}\left[e^z\right]_{z=1} = e$. So $\oint = 2\pi i\,e$.

### Example 3 — $\oint_{|z|=3} \frac{z^2}{(z - 1)^3}\,dz$

Pole of order $3$ at $1$: $\operatorname{Res} = \frac{1}{2!}\frac{d^2}{dz^2}\left[z^2\right] = \frac{2}{2} = 1$. So $\oint = 2\pi i$.

### Example 4 — $\oint_{|z|=1} e^{1/z}\,dz$

Essential singularity at $0$; from $e^{1/z} = 1 + \frac1z + \frac{1}{2z^2} + \cdots$, $\operatorname{Res} = 1$. So $\oint = 2\pi i$.

### Example 5 — $\oint_{|z|=2} \tan z\,dz$

$\tan z = \frac{\sin z}{\cos z}$ has simple poles where $\cos z = 0$: $z = \frac{\pi}{2} + k\pi$. Inside $|z| = 2$: $\pm\frac{\pi}{2}$ ($\approx \pm 1.571$). $\operatorname{Res} = \frac{\sin a}{-\sin a} = -1$ at each, so

$$
\oint_{|z|=2} \tan z\,dz = 2\pi i(-1 - 1) = -4\pi i.
$$

### Example 6 — $\oint_{|z|=2} \frac{dz}{z^4 + 1}$

Four simple poles $a_k = e^{i(\pi/4 + k\pi/2)}$, all on $|z| = 1$, so inside. $\operatorname{Res} = \frac{1}{4a_k^3} = \frac{a_k}{4a_k^4} = -\frac{a_k}{4}$. The four roots sum to $0$ (they are symmetric about the origin), so $\oint = 0$ — in agreement with 2.4 ($\operatorname{Res}(f, \infty) = 0$).

### Example 7 — a double pole with zero residue

$\oint_{|z|=1} \frac{1 - \cos z}{z^4}\,dz = 2\pi i \cdot 0 = 0$, since the Laurent series $\frac{1}{2z^2} - \frac{1}{24} + \cdots$ has no $\frac1z$ term.

### Example 8 — only some poles inside

$\oint \frac{dz}{z(z - 1)(z - 3)}$ around a circle centred at $1$: poles at $0$, $1$, $3$. The circle $|z - 1| = 1$ is a trap — $|0 - 1| = 1$ puts the pole $0$ **on** the contour, and the integral is not defined. On $|z - 1| = \frac12$ only $z = 1$ is inside, with $\operatorname{Res} = \frac{1}{1\cdot(1 - 3)} = -\frac12$, so $\oint = -\pi i$. On $|z - 1| = \frac32$, the poles $0$ and $1$ are inside: $\operatorname{Res}(f, 0) = \frac{1}{(-1)(-3)} = \frac13$, total $2\pi i\left(\frac13 - \frac12\right) = -\frac{\pi i}{3}$. Always check where the poles sit relative to the contour.

## 4. Python Examples

### 4.1 Residue formulas in sympy

We implement the order-$m$ formula and compare with `sp.residue`, which works from the Laurent series.

```python
import sympy as sp

z = sp.symbols("z")


def residue_at_pole(expression, point, order):
    # Res = 1/(m-1)! * lim_{z -> a} d^{m-1}/dz^{m-1} [(z - a)^m f(z)]
    regular_part = sp.simplify((z - point) ** order * expression)
    derivative = sp.diff(regular_part, z, order - 1)
    return sp.simplify(sp.limit(derivative, z, point) / sp.factorial(order - 1))


cases = [
    (z / (z**2 + 1), sp.I, 1),
    (sp.exp(z) / (z - 1) ** 2, 1, 2),
    (z**2 / (z - 1) ** 3, 1, 3),
    (sp.tan(z), sp.pi / 2, 1),
    ((1 - sp.cos(z)) / z**4, 0, 2),
    (1 / (z**2 + 1) ** 2, sp.I, 2),
]
for expression, point, order in cases:
    by_formula = residue_at_pole(expression, point, order)
    by_sympy = sp.residue(expression, z, point)
    print(f"Res({expression}, {point}) = {sp.nsimplify(by_formula)}   sympy.residue: {sp.simplify(by_sympy)}")

# The quotient rule g(a) / h'(a) at a root of z^4 + 1
a = sp.exp(sp.I * sp.pi / 4)
print("Res(1/(z^4+1), e^(i pi/4)): g/h' =", sp.N(1 / (4 * a**3), 6), "  sympy.residue:", sp.N(sp.residue(1 / (z**4 + 1), z, a), 6),
      "  -a/4 =", sp.N(-a / 4, 6))
```

`sp.residue` needs a Laurent series it can compute. For $e^{1/z}$ it raises `NotImplementedError` (in current sympy), so we read $a_{-1}$ from the series of $e^w$ with $w = 1/z$:

```python
import sympy as sp

z, w = sp.symbols("z w")
try:
    print("sp.residue(exp(1/z), z, 0) =", sp.residue(sp.exp(1 / z), z, 0))
except NotImplementedError as error:
    print("sp.residue failed:", type(error).__name__)
laurent = sp.series(sp.exp(w), w, 0, 5).removeO().subs(w, 1 / z)
print("Laurent series:", laurent)
print("coefficient of 1/z:", sp.expand(laurent).coeff(z, -1))
```

### 4.2 Numerical contour integrals versus $2\pi i \sum \operatorname{Res}$

```python
import numpy as np


def circle_integral(f, center, radius, n_points=4000):
    # Periodic trapezoidal rule on |z - center| = radius
    t = 2 * np.pi * np.arange(n_points) / n_points
    z = center + radius * np.exp(1j * t)
    return np.mean(f(z) * 1j * radius * np.exp(1j * t)) * 2 * np.pi


examples = [
    ("z/(z^2+1), |z|=2", lambda z: z / (z**2 + 1), 0, 2, 2j * np.pi),
    ("e^z/(z-1)^2, |z|=2", lambda z: np.exp(z) / (z - 1) ** 2, 0, 2, 2j * np.pi * np.e),
    ("z^2/(z-1)^3, |z|=3", lambda z: z**2 / (z - 1) ** 3, 0, 3, 2j * np.pi),
    ("e^(1/z), |z|=1", lambda z: np.exp(1 / z), 0, 1, 2j * np.pi),
    ("tan z, |z|=2", np.tan, 0, 2, -4j * np.pi),
    ("1/(z^4+1), |z|=2", lambda z: 1 / (z**4 + 1), 0, 2, 0),
    ("(1-cos z)/z^4, |z|=1", lambda z: (1 - np.cos(z)) / z**4, 0, 1, 0),
    ("1/(z(z-1)(z-3)), |z-1|=1/2", lambda z: 1 / (z * (z - 1) * (z - 3)), 1, 0.5, -1j * np.pi),
    ("1/(z(z-1)(z-3)), |z-1|=3/2", lambda z: 1 / (z * (z - 1) * (z - 3)), 1, 1.5, -1j * np.pi / 3),
]
for name, f, center, radius, predicted in examples:
    value = circle_integral(f, center, radius)
    value = complex(round(value.real, 10) + 0.0, round(value.imag, 10) + 0.0)   # + 0.0 turns -0.0 into 0.0
    print(f"{name:>28}:  numerical = {value.real:+.10f}{value.imag:+.10f}j   2 pi i * sum Res = {complex(predicted).imag:+.10f}j")
```

### 4.3 A general recipe for rational functions

For $f = p/q$ with simple poles, `np.roots` finds the poles and $\frac{p(a)}{q'(a)}$ the residues. We integrate over a contour that encloses only some of the poles and compare.

```python
import numpy as np


def circle_integral(f, center, radius, n_points=4000):
    t = 2 * np.pi * np.arange(n_points) / n_points
    z = center + radius * np.exp(1j * t)
    return np.mean(f(z) * 1j * radius * np.exp(1j * t)) * 2 * np.pi


numerator = np.poly1d([1, 0, 2])                 # z^2 + 2
denominator = np.poly1d(np.poly([1, -1, 2j, 0.5 - 1.5j, -2 + 0.5j]))   # poles chosen by hand
poles = denominator.roots
residues = numerator(poles) / denominator.deriv()(poles)

center, radius = 0.3 + 0.2j, 1.6
inside = np.abs(poles - center) < radius
prediction = 2j * np.pi * residues[inside].sum()
value = circle_integral(lambda z: numerator(z) / denominator(z), center, radius)

for pole, residue, is_inside in zip(poles, residues, inside):
    print(f"pole {pole:+.3f}: residue {residue:+.4f}  {'inside' if is_inside else 'outside'}")
print("2 pi i * sum of residues inside:", prediction)
print("numerical contour integral     :", value)
print("all residues sum to (deg q >= deg p + 2):", residues.sum())
```

### 4.4 Winding numbers

A curve may wind around a pole more than once. The limaçon $z(t) = (0.5 + \cos t)\,e^{it}$ has an inner loop: the point $0.25$ is encircled **twice**, the point $1$ once. For $f(z) = \frac{1}{(z - 0.25)(z - 1)}$, with residues $-\frac43$ at $0.25$ and $+\frac43$ at $1$, the theorem predicts $2\pi i\left(2\cdot(-\tfrac43) + 1\cdot\tfrac43\right) = -\tfrac83\pi i$.

```python
import matplotlib.pyplot as plt
import numpy as np


def closed_integral(f, points):
    values = f(points)
    return np.sum((values[1:] + values[:-1]) / 2 * np.diff(points))


t = np.linspace(0, 2 * np.pi, 40001)
limacon = (0.5 + np.cos(t)) * np.exp(1j * t)
poles = {0.25: -4 / 3, 1.0: 4 / 3}

prediction = 0
for pole, residue in poles.items():
    winding = closed_integral(lambda z: 1 / (z - pole), limacon) / (2j * np.pi)
    prediction += 2j * np.pi * round(winding.real) * residue
    print(f"pole {pole}: winding number = {winding.real:.6f}, residue = {residue:+.4f}")
value = closed_integral(lambda z: 1 / ((z - 0.25) * (z - 1)), limacon)
print("numerical integral          :", value)
print("2 pi i * sum(winding * Res) :", prediction, "  (-8/3) pi i =", -8j * np.pi / 3)

fig, ax = plt.subplots(figsize=(6, 5))
ax.plot(limacon.real, limacon.imag, linewidth=2)
for k in [2000, 15000, 22000]:
    ax.annotate("", xy=(limacon[k + 300].real, limacon[k + 300].imag), xytext=(limacon[k].real, limacon[k].imag),
                arrowprops=dict(arrowstyle="->", linewidth=2, color="tab:blue"))
ax.scatter([0.25, 1.0], [0, 0], marker="x", color="red", s=100, linewidth=3, zorder=3)
ax.text(0.12, -0.3, "0.25 (twice)", color="red")
ax.text(0.95, 0.1, "1 (once)", color="red")
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_aspect("equal")
ax.set_title(r"Limaçon $(0.5 + \cos t)\, e^{it}$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
plt.show()
```

### 5.2 A big contour equals a sum of small circles

Deformation (notebook 02) in a picture: the integral around the big curve equals the sum of the integrals around tiny circles at each enclosed pole, and each tiny circle contributes $2\pi i \times$ its residue.

```python
import matplotlib.pyplot as plt
import numpy as np


def closed_integral(f, points):
    values = f(points)
    return np.sum((values[1:] + values[:-1]) / 2 * np.diff(points))


f = lambda z: np.exp(z) / (z * (z - 1) * (z + 1j))
poles = np.array([0, 1, -1j])
t = np.linspace(0, 2 * np.pi, 20001)
big_curve = (1.9 + 0.3 * np.cos(3 * t)) * np.exp(1j * t) + 0.2

big_value = closed_integral(f, big_curve)
small_values = [closed_integral(f, pole + 0.15 * np.exp(1j * t)) for pole in poles]

fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(big_curve.real, big_curve.imag, linewidth=2, label=f"big curve: {big_value:.4f}")
for pole, name, value in zip(poles, ["0", "1", "-i"], small_values):
    small_circle = pole + 0.15 * np.exp(1j * t)
    ax.plot(small_circle.real, small_circle.imag, linewidth=2, label=f"small circle at {name}: {value:.4f}")
    ax.scatter(pole.real, pole.imag, marker="x", color="red", s=80, zorder=3)
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_aspect("equal")
ax.set_title(r"$\oint e^z / (z(z-1)(z+i))\, dz$: big curve = sum of small circles")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left", fontsize=9)
plt.show()

print("sum of the small circles:", sum(small_values))
print("big curve               :", big_value)
```

### 5.3 The integral jumps when the contour swallows a pole

$\frac{1}{2\pi i}\oint_{|z| = r} \tan z\,dz$ as a function of the radius $r$. It is constant while no pole crosses the circle, and drops by $2$ each time $r$ passes $\frac{\pi}{2}, \frac{3\pi}{2}, \dots$, where two poles ($\pm$) with residue $-1$ each enter. (Radii that land exactly on a pole are skipped.)

```python
import matplotlib.pyplot as plt
import numpy as np


def circle_integral(f, center, radius, n_points=4000):
    t = 2 * np.pi * np.arange(n_points) / n_points
    z = center + radius * np.exp(1j * t)
    return np.mean(f(z) * 1j * radius * np.exp(1j * t)) * 2 * np.pi


radii = np.linspace(0.2, 8.5, 400)
pole_distances = np.pi / 2 + np.pi * np.arange(3)
radii = radii[np.min(np.abs(radii[:, None] - pole_distances[None, :]), axis=1) > 0.03]
values = np.array([circle_integral(np.tan, 0, r, n_points=6000) for r in radii]) / (2j * np.pi)

plt.figure(figsize=(9, 4.8))
plt.plot(radii, values.real, ".", markersize=4, label=r"$\frac{1}{2\pi i}\oint_{|z|=r} \tan z\, dz$")
for distance in pole_distances:
    plt.axvline(distance, color="red", linestyle=":", linewidth=1)
plt.text(pole_distances[0] + 0.05, -5.3, r"poles at $\pm\pi/2, \pm 3\pi/2, \pm 5\pi/2$", color="red")
plt.title("Sum of enclosed residues of tan z as the circle grows")
plt.xlabel("radius r")
plt.ylabel("value")
plt.ylim(-6.5, 1)
plt.grid(True, alpha=0.3)
plt.legend(loc="upper right")
plt.show()
```

2. $\frac{e^z}{z^4} = \frac{1}{z^4}\left(1 + z + \frac{z^2}{2} + \frac{z^3}{6} + \cdots\right)$: the $\frac1z$ coefficient is $\frac16$. (Or: pole of order 4, $\frac{1}{3!}\frac{d^3}{dz^3}e^z\big|_0 = \frac16$.)
3. Residues: at $0$, $\frac{3\cdot 0 + 1}{0 - 1} = -1$; at $1$, $\frac{3 + 1}{1} = 4$. Both inside: $2\pi i(-1 + 4) = 6\pi i$.
4. Poles $\pm 1$, $\pm i$. Distances from $i$: to $i$, $0$; to $\pm 1$, $\sqrt2 \approx 1.41 < 1.5$; to $-i$, $2 > 1.5$. Inside: $1, -1, i$. Residues $\frac{1}{4a^3} = \frac{a}{4a^4} = \frac{a}{4}$: sum $\frac{1 - 1 + i}{4} = \frac{i}{4}$. Integral $2\pi i\cdot\frac{i}{4} = -\frac{\pi}{2}$.
5. $z\sin z = z^2 - \frac{z^4}{6} + \cdots$ has a zero of order $2$: a pole of order $2$. $\frac{1}{z\sin z} = \frac{1}{z^2}\left(1 + \frac{z^2}{6} + \cdots\right) = \frac{1}{z^2} + \frac16 + \cdots$: residue $0$.
6. $z^2\sin\frac1z = z^2\left(\frac1z - \frac{1}{6z^3} + \cdots\right) = z - \frac{1}{6z} + \cdots$: residue $-\frac16$ (essential singularity), integral $-\frac{\pi i}{3}$.
7. For $|z| > 1$: $\frac{z^5}{z^6 - 1} = \frac1z\cdot\frac{1}{1 - z^{-6}} = \frac1z + \frac{1}{z^7} + \cdots$, so $b_{-1} = 1$ and the integral is $2\pi i$. Directly: six simple poles at the sixth roots of unity, residues $\frac{a^5}{6a^5} = \frac16$ each, sum $1$. ✓
8. On $|z| = 1$, $|3z| = 3 > 2 \ge |z^5 + 1|$, so (Rouché) the zero count matches that of $3z$: one zero. The integral below returns $1$.
9. $\pi\cot\pi z = \frac{\pi\cos\pi z}{\sin\pi z}$; at $z = n$, $\frac{g}{h'} = \frac{\pi\cos\pi n}{\pi\cos\pi n} = 1$. Near $0$: $\pi\cot\pi z = \frac1z - \frac{\pi^2 z}{3} - \cdots$, so $\frac{\pi\cot\pi z}{z^2} = \frac{1}{z^3} - \frac{\pi^2}{3z} - \cdots$: residue $-\frac{\pi^2}{3}$. At $n \ne 0$ the residue of $\frac{\pi\cot\pi z}{z^2}$ is $\frac{1}{n^2}$. The residue theorem on $|z| = N + \frac12$: $\frac{1}{2\pi i}\oint = -\frac{\pi^2}{3} + 2\sum_{n=1}^{N}\frac{1}{n^2} \to 0$, so $\sum \frac{1}{n^2} = \frac{\pi^2}{6}$.

Numerical checks:

```python
import numpy as np
import sympy as sp


def circle_integral(f, center, radius, n_points=4000):
    t = 2 * np.pi * np.arange(n_points) / n_points
    z = center + radius * np.exp(1j * t)
    return np.mean(f(z) * 1j * radius * np.exp(1j * t)) * 2 * np.pi


z = sp.symbols("z")
print("Ex 1:", sp.residue(1 / (z**2 + 4), z, 2 * sp.I), "  Ex 2:", sp.residue(sp.exp(z) / z**4, z, 0),
      "  Ex 5:", sp.residue(1 / (z * sp.sin(z)), z, 0))
print("Ex 3:", circle_integral(lambda w: (3 * w + 1) / (w * (w - 1)), 0, 3), " vs", 6j * np.pi)
print("Ex 4:", circle_integral(lambda w: 1 / (w**4 - 1), 1j, 1.5), " vs", -np.pi / 2)
print("Ex 6:", circle_integral(lambda w: w**2 * np.sin(1 / w), 0, 1), " vs", -1j * np.pi / 3)
print("Ex 7:", circle_integral(lambda w: w**5 / (w**6 - 1), 0, 2), " vs", 2j * np.pi)
count = circle_integral(lambda w: (5 * w**4 + 3) / (w**5 + 3 * w + 1), 0, 1) / (2j * np.pi)
print("Ex 8: zeros in |z| < 1 by the argument principle:", count, "  by np.roots:", int(np.sum(np.abs(np.roots([1, 0, 0, 0, 3, 1])) < 1)))
for n_terms in [5, 20, 100]:
    integral = circle_integral(lambda w: np.pi / np.tan(np.pi * w) / w**2, 0, n_terms + 0.5, n_points=20000)
    print(f"Ex 9: N = {n_terms:3d}: |integral| = {abs(integral):.2e}   sum 1/n^2 = {np.sum(1 / np.arange(1, n_terms + 1) ** 2):.6f}   pi^2/6 = {np.pi**2 / 6:.6f}")
```

## 8. Conclusion

- The **residue** $\operatorname{Res}(f, a) = a_{-1}$ is the only Laurent coefficient that survives integration around $a$: $\oint_{C_\varepsilon} f\,dz = 2\pi i\,\operatorname{Res}(f, a)$.
- **Residue theorem:** $\oint_\gamma f\,dz = 2\pi i\sum_{\text{inside}}\operatorname{Res}(f, a_j)$ (weighted by winding numbers in general).
- Formulas: simple pole $\lim (z - a)f$; quotient $\frac{g(a)}{h'(a)}$; order $m$: $\frac{1}{(m-1)!}\lim\frac{d^{m-1}}{dz^{m-1}}[(z - a)^m f]$; essential: read the Laurent series.
- The residues of all finite singularities plus the residue at infinity sum to $0$ — useful when many poles are enclosed.
- `sympy.residue` handles poles; numerical integration with the periodic trapezoidal rule is a reliable independent check. The integral jumps exactly when a contour crosses a pole.
