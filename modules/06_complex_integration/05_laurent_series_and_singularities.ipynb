# Laurent Series and Singularities

A Taylor series lives on a disc and stops at the first singularity. To describe a function **around** a singularity we need negative powers too:

$$
f(z) = \sum_{n=-\infty}^{\infty} a_n (z - c)^n = \cdots + \frac{a_{-2}}{(z-c)^2} + \frac{a_{-1}}{z - c} + a_0 + a_1(z - c) + \cdots,
$$

a **Laurent series**, which converges on an **annulus**. The negative-power part, the *principal part*, is a complete fingerprint of the singularity: none means removable, finitely many means a pole, infinitely many means an **essential singularity** — where, by the Casorati–Weierstrass theorem, the function comes arbitrarily close to every complex value in every neighbourhood. We will see that last fact in colour.

This is notebook 05 of module 06 (**Complex integration**) in the **Learning Complex Numbers** series.

## 1. Introduction

Real calculus has no good language for what happens near $x = 0$ in $\frac{\sin x}{x}$, $\frac{1}{x^2}$ and $e^{-1/x^2}$. In the complex plane they are three different species of isolated singularity, and Laurent series tell them apart at a glance:

| function | Laurent series at $0$ | type |
|---|---|---|
| $\frac{\sin z}{z}$ | $1 - \frac{z^2}{6} + \cdots$ | removable |
| $\frac{1}{z^2}$ | $\frac{1}{z^2}$ | pole of order 2 |
| $e^{1/z}$ | $1 + \frac1z + \frac{1}{2z^2} + \frac{1}{6z^3} + \cdots$ | essential |

The single coefficient $a_{-1}$, the **residue**, is what survives contour integration ($\oint (z - c)^n\,dz = 0$ unless $n = -1$). That is the bridge to the residue theorem in notebook 06.

## 2. Theory

### 2.1 Annuli and Laurent's theorem

An **annulus** centred at $c$ is $A = \{z : r < |z - c| < R\}$ with $0 \le r < R \le \infty$. With $r = 0$ it is a **punctured disc** $0 < |z - c| < R$.

> **Laurent's theorem.** If $f$ is holomorphic on the annulus $A$, then
> $$f(z) = \sum_{n=-\infty}^{\infty} a_n (z - c)^n, \qquad a_n = \frac{1}{2\pi i}\oint_{|w - c| = \rho} \frac{f(w)}{(w - c)^{n+1}}\,dw,$$
> for every $z \in A$ and any $\rho$ with $r < \rho < R$. The series converges absolutely, and the expansion is unique.

*Proof sketch.* For $z$ in the annulus, take circles $C_1$ (radius $\rho_1$, near $r$) and $C_2$ (radius $\rho_2$, near $R$) on either side of $z$. Cauchy's formula on the region between them (deformation, notebook 02) gives

$$
f(z) = \frac{1}{2\pi i}\oint_{C_2} \frac{f(w)}{w - z}\,dw - \frac{1}{2\pi i}\oint_{C_1} \frac{f(w)}{w - z}\,dw.
$$

On $C_2$, $|z - c| < |w - c|$, so expand $\frac{1}{w - z}$ in powers of $\frac{z - c}{w - c}$, exactly as for Taylor series: this gives the non-negative powers. On $C_1$, $|w - c| < |z - c|$, so expand instead in powers of $\frac{w - c}{z - c}$:

$$
-\frac{1}{w - z} = \frac{1}{(z - c) - (w - c)} = \sum_{k=0}^{\infty} \frac{(w - c)^k}{(z - c)^{k+1}},
$$

which gives the negative powers. $\blacksquare$

Notice that $a_n$ is **not** $f^{(n)}(c)/n!$ any more — $f$ need not even be defined at $c$.

### 2.2 The same function, different annuli

$f(z) = \frac{1}{z(z - 1)}$ is holomorphic on two annuli centred at $0$: $0 < |z| < 1$ and $1 < |z| < \infty$. Using $\frac{1}{z(z-1)} = \frac{1}{z}\cdot\frac{1}{z - 1}$:

- $0 < |z| < 1$: $\frac{1}{z - 1} = -\sum_{n \ge 0} z^n$, so $f(z) = -\frac1z - 1 - z - z^2 - \cdots$.
- $|z| > 1$: $\frac{1}{z - 1} = \frac{1}{z}\cdot\frac{1}{1 - 1/z} = \sum_{n \ge 1} z^{-n}$, so $f(z) = \frac{1}{z^2} + \frac{1}{z^3} + \frac{1}{z^4} + \cdots$.

Different annuli, different series — uniqueness only holds on a fixed annulus.

### 2.3 Classifying isolated singularities

$c$ is an **isolated singularity** of $f$ if $f$ is holomorphic on a punctured disc $0 < |z - c| < R$. Write the Laurent series there. The **principal part** is $\sum_{n < 0} a_n (z - c)^n$.

| Type | Principal part | Behaviour as $z \to c$ | Example at $0$ |
|---|---|---|---|
| **removable** | zero | $f(z)$ has a finite limit | $\frac{\sin z}{z} \to 1$ |
| **pole of order $m$** | finitely many terms, $a_{-m} \ne 0$ | $\lvert f(z)\rvert \to \infty$; $(z - c)^m f(z) \to a_{-m} \ne 0$ | $\frac{e^z}{z^3}$, $m = 3$ |
| **essential** | infinitely many terms | no limit, finite or infinite | $e^{1/z}$ |

Useful tests:

- **Riemann's removable singularity theorem:** if $f$ is **bounded** near $c$, the singularity is removable. (Cauchy estimates show $a_{-n} = 0$ for $n \ge 1$.)
- **Poles from zeros:** if $g$ has a zero of order $m$ at $c$ (i.e. $g(c) = \cdots = g^{(m-1)}(c) = 0 \ne g^{(m)}(c)$), then $\frac{1}{g}$ has a pole of order $m$ there. More generally, for $\frac{h}{g}$ subtract the zero orders: $\frac{1 - \cos z}{z^4}$ has a zero of order $2$ over a zero of order $4$, a pole of order $2$.
- A **simple pole** is a pole of order $1$.

### 2.4 Casorati–Weierstrass

> **Casorati–Weierstrass.** If $c$ is an essential singularity of $f$, then for every $w \in \mathbb{C}$, every $\varepsilon > 0$ and every $\delta > 0$ there is a $z$ with $0 < |z - c| < \delta$ and $|f(z) - w| < \varepsilon$.

In words: the image of **every** punctured neighbourhood of $c$ is dense in $\mathbb{C}$.

*Proof.* If not, some disc $|f(z) - w| < \varepsilon$ is missed near $c$. Then $g(z) = \frac{1}{f(z) - w}$ is bounded by $1/\varepsilon$ there, so by Riemann's theorem $g$ has a removable singularity at $c$. Then $f = w + \frac1g$ is either holomorphic at $c$ (if $g(c) \ne 0$) or has a pole (if $g(c) = 0$) — not an essential singularity. $\blacksquare$

**Picard's great theorem** says much more: near an essential singularity $f$ takes every complex value, with at most one exception, **infinitely often**. For $e^{1/z}$ the exception is $0$, and any other $w$ is hit at

$$
e^{1/z} = w \iff \frac1z = \operatorname{Log} w + 2\pi i k \iff z_k = \frac{1}{\operatorname{Log} w + 2\pi i k}, \quad k \in \mathbb{Z},
$$

points that pile up at $0$ as $|k| \to \infty$.

## 3. Worked Examples

### Example 1 — three annuli for $\frac{1}{z(z - 1)}$

At centre $0$ see 2.2. At centre $1$, on $0 < |z - 1| < 1$, write $u = z - 1$:

$$
\frac{1}{z(z - 1)} = \frac{1}{u}\cdot\frac{1}{1 + u} = \frac1u\sum_{n\ge0}(-1)^n u^n = \frac{1}{z - 1} - 1 + (z - 1) - (z - 1)^2 + \cdots.
$$

A simple pole at $1$ with $a_{-1} = 1$.

### Example 2 — $\frac{\sin z}{z}$ at $0$

$\frac{\sin z}{z} = \frac{1}{z}\left(z - \frac{z^3}{6} + \frac{z^5}{120} - \cdots\right) = 1 - \frac{z^2}{6} + \frac{z^4}{120} - \cdots$. No negative powers: removable; defining the value at $0$ to be $1$ makes it entire.

### Example 3 — $\frac{e^z}{z^3}$ at $0$

$$
\frac{e^z}{z^3} = \frac{1}{z^3}\left(1 + z + \frac{z^2}{2} + \frac{z^3}{6} + \cdots\right) = \frac{1}{z^3} + \frac{1}{z^2} + \frac{1}{2z} + \frac16 + \frac{z}{24} + \cdots.
$$

Pole of order $3$; the coefficient of $\frac1z$ is $a_{-1} = \frac12$.

### Example 4 — $\frac{1 - \cos z}{z^4}$ at $0$

$1 - \cos z = \frac{z^2}{2} - \frac{z^4}{24} + \frac{z^6}{720} - \cdots$, so

$$
\frac{1 - \cos z}{z^4} = \frac{1}{2z^2} - \frac{1}{24} + \frac{z^2}{720} - \cdots.
$$

A pole of order $2$, not $4$: the numerator's zero of order $2$ cancels two powers. Here $a_{-1} = 0$.

### Example 5 — $e^{1/z}$ at $0$

Substitute $w = 1/z$ into $e^{w} = \sum \frac{w^n}{n!}$:

$$
e^{1/z} = \sum_{n=0}^{\infty} \frac{1}{n!\,z^n} = 1 + \frac1z + \frac{1}{2z^2} + \frac{1}{6z^3} + \cdots, \qquad 0 < |z| < \infty.
$$

Infinitely many negative powers: essential. Along the positive real axis $e^{1/x} \to \infty$; along the negative real axis $e^{1/x} \to 0$; along the imaginary axis $|e^{1/(iy)}| = |e^{-i/y}| = 1$. Three different behaviours, so no limit.

### Example 6 — hitting $w = 1 + i$ arbitrarily close to $0$

$\operatorname{Log}(1 + i) = \ln\sqrt2 + i\frac{\pi}{4}$, so $z_k = \frac{1}{\ln\sqrt2 + i\left(\frac{\pi}{4} + 2\pi k\right)}$. For large $k$, $|z_k| \approx \frac{1}{2\pi k}$: there are solutions within $0.001$ of the origin as soon as $k \ge 160$.

### Example 7 — orders from zeros

$\frac{1}{\sin z}$ at $0$: $\sin z$ has a simple zero ($\sin'(0) = 1 \neq 0$), so a simple pole. $\frac{z}{\sin z}$: zero of order 1 over zero of order 1, removable (limit $1$). $\frac{1}{z^2(z - 3)^3}$: pole of order $2$ at $0$ and of order $3$ at $3$.

## 4. Python Examples

### 4.1 Laurent series with sympy

`sp.series` handles negative powers directly at finite points, and `sp.series(..., z, sp.oo)` expands for large $|z|$ (the outer annulus). For $e^{1/z}$ we substitute into the series of $e^w$. (Sympy may mix the variables in its output: around $z = 1$ it prints $-2 + z$ for $-1 + (z - 1)$.)

```python
import sympy as sp

z, w = sp.symbols("z w")
f = 1 / (z * (z - 1))
print("1/(z(z-1)), 0 < |z| < 1     :", sp.series(f, z, 0, 4))
print("1/(z(z-1)), |z| > 1         :", sp.series(f, z, sp.oo, 6))
print("1/(z(z-1)), 0 < |z - 1| < 1 :", sp.series(f, z, 1, 3))
print()
print("sin(z)/z                    :", sp.series(sp.sin(z) / z, z, 0, 6))
print("exp(z)/z^3                  :", sp.series(sp.exp(z) / z**3, z, 0, 2))
print("(1 - cos z)/z^4             :", sp.series((1 - sp.cos(z)) / z**4, z, 0, 3))
print("exp(1/z), first terms       :", sp.series(sp.exp(w), w, 0, 6).removeO().subs(w, 1 / z))
```

### 4.2 Laurent coefficients as contour integrals

$a_n = \frac{1}{2\pi i}\oint_{|w - c| = \rho}\frac{f(w)}{(w - c)^{n+1}}\,dw$ for positive **and negative** $n$. Integrating $\frac{1}{z(z-1)}$ on $|z| = 0.5$ and on $|z| = 2$ produces the two different series of 2.2 — the circle picks the annulus.

```python
import math

import numpy as np


def laurent_coefficients(f, center, radius, n_values, n_points=512):
    # a_n = (1 / (2 pi rho^n)) * integral of f(c + rho e^{it}) e^{-int} dt  (periodic trapezoidal rule)
    t = 2 * np.pi * np.arange(n_points) / n_points
    samples = f(center + radius * np.exp(1j * t))
    return {n: (samples * np.exp(-1j * n * t)).mean() / radius**n for n in n_values}


f = lambda z: 1 / (z * (z - 1))
for radius in [0.5, 2.0]:
    coefficients = laurent_coefficients(f, 0, radius, range(-4, 3))
    printable = ", ".join(f"a_{n} = {round(value.real, 3) + 0.0:+.3f}" for n, value in coefficients.items())
    print(f"circle |z| = {radius}:  {printable}")

coefficients = laurent_coefficients(lambda z: np.exp(1 / z), 0, 1.0, range(-6, 2))
print()
print("exp(1/z):", ", ".join(f"a_{n} = {round(value.real, 5) + 0.0:.5f}" for n, value in coefficients.items()))
print("1/|n|! for n = -6..0:", [round(1 / math.factorial(k), 5) for k in range(6, -1, -1)], " (and a_1 = 0)")
```

### 4.3 Classifying singularities numerically

A pole of order $m$ is detected by $(z - c)^m f(z)$ staying bounded and nonzero as $z \to c$; an essential singularity gives different limits along different directions. We approach $0$ along four directions.

```python
import numpy as np

functions = {
    "sin(z)/z": lambda z: np.sin(z) / z,
    "exp(z)/z^3": lambda z: np.exp(z) / z**3,
    "(1-cos z)/z^4": lambda z: (1 - np.cos(z)) / z**4,
    "exp(1/z)": lambda z: np.exp(1 / z),
}
directions = {"from +1": 1, "from -1": -1, "from +i": 1j, "from -i": -1j}
distance = 0.05

for name, f in functions.items():
    values = {label: f(distance * direction) for label, direction in directions.items()}
    print(name)
    for label, value in values.items():
        print(f"   {label}: |f| = {abs(value):.4e}   |z^2 f| = {abs((distance) ** 2 * value):.4e}   |z^3 f| = {abs((distance) ** 3 * value):.4e}")
```

Reading the table: $\frac{\sin z}{z}$ stays near $1$ (removable). $|z^3 f| \approx 1$ for $\frac{e^z}{z^3}$ (order 3) and $|z^2 f| \approx \frac12$ for $\frac{1-\cos z}{z^4}$ (order 2), the same from every direction. For $e^{1/z}$ the modulus is $e^{20} \approx 5 \times 10^8$ from the right, $e^{-20} \approx 2 \times 10^{-9}$ from the left and exactly $1$ from above and below: an essential singularity.

### 4.4 Picard in action: solving $e^{1/z} = 1 + i$ near $0$

```python
import cmath

target = 1 + 1j
for k in [0, 1, 10, 100, 1000]:
    z_k = 1 / (cmath.log(target) + 2j * cmath.pi * k)
    value = cmath.exp(1 / z_k)
    print(f"k = {k:5d}:  z_k = {z_k:.3e}   |z_k| = {abs(z_k):.2e}   exp(1/z_k) = {value:.10f}")
```

## 5. Visualizations

### 5.1 The annuli of $\frac{1}{z(z-1)}$

Coloured rings show where each Laurent expansion holds, with the partial sums' error on a test circle in each region. The dividing circle $|z| = 1$ passes through the singularity at $1$.

```python
import matplotlib.pyplot as plt
import numpy as np

f = lambda z: 1 / (z * (z - 1))
inner_series = lambda z, terms: -sum(z ** (n - 1) for n in range(terms))      # 0 < |z| < 1
outer_series = lambda z, terms: sum(z ** (-n) for n in range(2, terms + 2))   # |z| > 1

axis = np.linspace(-2.5, 2.5, 500)
grid = axis[None, :] + 1j * axis[:, None]
modulus = np.abs(grid)
error = np.where(modulus < 1, np.abs(inner_series(grid, 30) - f(grid)), np.abs(outer_series(grid, 30) - f(grid)))

fig, ax = plt.subplots(figsize=(7.5, 7))
image = ax.imshow(np.log10(error + 1e-16), extent=[-2.5, 2.5, -2.5, 2.5], origin="lower", cmap="magma", vmin=-16, vmax=0)
fig.colorbar(image, ax=ax, label="log10 error of 30-term Laurent sum")
t = np.linspace(0, 2 * np.pi, 300)
ax.plot(np.cos(t), np.sin(t), color="cyan", linewidth=1.5)
ax.scatter([0, 1], [0, 0], marker="x", color="cyan", s=120, linewidth=3, zorder=3)
ax.text(-0.45, 0.35, r"$-\frac{1}{z} - 1 - z - \cdots$", color="white", fontsize=12)
ax.text(-2.3, 1.9, r"$\frac{1}{z^2} + \frac{1}{z^3} + \cdots$", color="white", fontsize=12)
ax.set_aspect("equal")
ax.set_title(r"Two Laurent series of $1/(z(z-1))$ centred at 0")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
plt.show()
```

Both series fail near the circle $|z| = 1$ (bright ring), and they fail worst near $z = 1$, the singularity that separates the two annuli.

### 5.2 Three kinds of singularity in domain colouring

Domain colouring paints $z$ by the value $f(z)$: **hue** = argument of $f(z)$, and **brightness bands** repeat each time $|f(z)|$ doubles. To avoid overflow for $e^{1/z}$ (which reaches $e^{1/|z|}$) we colour from $\log f$ directly: hue from $\operatorname{Im}\log f$, bands from $\operatorname{Re}\log f = \ln|f|$.

- removable ($\frac{\sin z}{z}$): smooth colours through $0$, nothing special happens;
- pole of order 3 ($\frac{1}{z^3}$): hues cycle **3 times** around $0$ (clockwise as you go counter-clockwise), bands crowd in as $|f| \to \infty$;
- essential ($e^{1/z}$): every colour, infinitely often, crammed into every neighbourhood of $0$.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import hsv_to_rgb


def colour_from_log(log_values):
    # Hue from the argument, brightness bands that repeat each time the modulus doubles
    hue = (log_values.imag / (2 * np.pi)) % 1.0
    bands = (log_values.real / np.log(2)) % 1.0
    value = 0.55 + 0.45 * bands
    return hsv_to_rgb(np.stack([hue, np.full_like(hue, 0.9), value], axis=-1))


half_width = 1.0
axis = np.linspace(-half_width, half_width, 600)   # even count: no grid point at exactly 0
grid = axis[None, :] + 1j * axis[:, None]

panels = {
    r"removable: $\sin z / z$": np.log(np.sin(grid) / grid),
    r"pole of order 3: $1/z^3$": -3 * np.log(grid),
    r"essential: $e^{1/z}$": 1 / grid,   # log of exp(1/z), computed without overflow
}

fig, axes = plt.subplots(1, 3, figsize=(15, 5.2))
for ax, (title, log_values) in zip(axes, panels.items()):
    ax.imshow(colour_from_log(log_values), extent=[-half_width, half_width, -half_width, half_width], origin="lower")
    ax.set_title(title)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
plt.tight_layout()
plt.show()
```

### 5.3 Casorati–Weierstrass: zooming in on $e^{1/z}$

Zoom into $0$ by factors of $5$. A removable singularity or a pole would look *simpler* as we zoom in (a single colour, or a fixed number of hue cycles). $e^{1/z}$ looks the same at every scale — the full colour wheel and every brightness, in every window. The white dots are solutions of $e^{1/z} = 1 + i$ inside each window. (At the very centre of each panel the colours change faster than one per pixel, so the image there is a blur — the function itself only gets wilder.)

```python
import cmath

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import hsv_to_rgb


def colour_from_log(log_values):
    hue = (log_values.imag / (2 * np.pi)) % 1.0
    bands = (log_values.real / np.log(2)) % 1.0
    value = 0.55 + 0.45 * bands
    return hsv_to_rgb(np.stack([hue, np.full_like(hue, 0.9), value], axis=-1))


solutions = np.array([1 / (cmath.log(1 + 1j) + 2j * cmath.pi * k) for k in range(-2000, 2001)])

fig, axes = plt.subplots(1, 3, figsize=(15, 5.2))
for ax, half_width in zip(axes, [0.5, 0.1, 0.02]):
    axis = np.linspace(-half_width, half_width, 600)
    grid = axis[None, :] + 1j * axis[:, None]
    ax.imshow(colour_from_log(1 / grid), extent=[-half_width, half_width, -half_width, half_width], origin="lower")
    visible = solutions[(np.abs(solutions.real) < half_width) & (np.abs(solutions.imag) < half_width)]
    ax.scatter(visible.real, visible.imag, s=12, color="white", edgecolor="black", linewidth=0.5, zorder=3)
    ax.set_title(f"$e^{{1/z}}$ on $|\\mathrm{{Re}}|, |\\mathrm{{Im}}| < {half_width}$  ({len(visible)} solutions shown)")
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
plt.tight_layout()
plt.show()
```

The solutions $z_k = \frac{1}{\operatorname{Log}(1+i) + 2\pi i k}$ are the images under $w \mapsto 1/w$ of the points $\ln\sqrt2 + i(\frac{\pi}{4} + 2\pi k)$ on a vertical line. That line maps to a circle through $0$ of radius $\frac{1}{2\ln\sqrt2} \approx 1.44$, tangent to the imaginary axis; near $0$ the circle hugs the imaginary axis, so the solutions march into the origin almost vertically. Every window, however small, contains infinitely many of them; only $k \le 2000$ are drawn.

### 5.4 No limit at an essential singularity

$\log_{10}|f|$ as $z \to 0$ along three rays. A pole shoots up at the same rate from every direction; $e^{1/z}$ explodes along one ray, collapses along another and stays at $|f| = 1$ along the third.

```python
import matplotlib.pyplot as plt
import numpy as np

r = np.logspace(-2, 0, 300)
rays = {"positive real axis": 1, "negative real axis": -1, "imaginary axis": 1j}

fig, (ax_pole, ax_essential) = plt.subplots(1, 2, figsize=(13, 5))
for (label, direction), style in zip(rays.items(), ["-", "--", ":"]):
    z = r * direction
    ax_pole.plot(r, np.log10(np.abs(1 / z**3)), style, label=label, linewidth=3)   # the three curves coincide
    # log10 |exp(1/z)| = Re(1/z) / ln 10, computed without overflow
    ax_essential.plot(r, (1 / z).real / np.log(10), label=label, linewidth=2)

ax_pole.set_title(r"Pole: $\log_{10} |1/z^3|$, the same on every ray")
ax_essential.set_title(r"Essential: $\log_{10} |e^{1/z}|$ depends on the ray")
for ax in (ax_pole, ax_essential):
    ax.set_xscale("log")
    ax.set_xlabel("distance |z| from 0")
    ax.set_ylabel(r"$\log_{10} |f(z)|$")
    ax.grid(True, alpha=0.3)
    ax.legend()
plt.tight_layout()
plt.show()
```

## 7. Solutions

1. $e^{2/z} = \sum_{n\ge0} \frac{2^n}{n!\,z^n} = 1 + \frac2z + \frac{2}{z^2} + \frac{4}{3z^3} + \cdots$: infinitely many negative powers, essential.
2. (a) $\frac{e^z - 1}{z} = 1 + \frac{z}{2} + \cdots$: removable (value $1$). (b) Pole of order $5$ (the principal part is just $\frac{1}{z^5}$). (c) $\cos\frac1z = 1 - \frac{1}{2z^2} + \frac{1}{24 z^4} - \cdots$: essential. (d) $1 - \cos z$ has a zero of order $2$, $z$ of order $1$: a simple pole, $\frac{z}{1 - \cos z} = \frac{2}{z} + \frac{z}{6} + \cdots$.
3. Order $2$ at $0$, order $3$ at $3$ (the numerator does not vanish at either; at $-1$ there is a zero, not a pole).
4. Poles at $1$ and $2$ split the plane into a disc, an annulus and an outer region. Partial fractions: $f = \frac{1}{z - 2} - \frac{1}{z - 1}$. On $1 < |z| < 2$: $\frac{1}{z - 2} = -\frac12\cdot\frac{1}{1 - z/2} = -\sum_{n\ge0} \frac{z^n}{2^{n+1}}$ (since $|z/2| < 1$) and $-\frac{1}{z - 1} = -\frac1z\cdot\frac{1}{1 - 1/z} = -\sum_{n\ge1} z^{-n}$ (since $|1/z| < 1$). So
$$f(z) = \cdots - \frac{1}{z^2} - \frac{1}{z} - \frac12 - \frac{z}{4} - \frac{z^2}{8} - \cdots.$$
5. $\frac{1}{\sin z} = \frac{1}{z}\left(1 - \frac{z^2}{6} + \cdots\right)^{-1} = \frac1z + \frac{z}{6} + \frac{7z^3}{360} + \cdots$, so $f(z) = -\frac{z}{6} - \cdots$. No negative powers: removable, with $f(0) = 0$.
6. $e^{1/z} = -1 \iff \frac1z = i\pi(2k + 1)$, so $z = \frac{-i}{\pi(2k + 1)}$. $|z| < 0.01$ needs $2k + 1 > \frac{100}{\pi} \approx 31.8$, e.g. $k = 16$: $z = -\frac{i}{33\pi} \approx -0.00965i$. $e^{1/z} = 0$ never: the exponential has no zeros.
7. See the code below: $a_{-3} = a_{-2} = a_{-1} = -1$, $a_0 = -\frac12$, $a_1 = -\frac14$, $a_2 = -\frac18$.
8. Write $f(z) = (z - c)^{-m} g(z)$ with $g$ holomorphic and $g(c) \ne 0$. Then $\frac{f'}{f} = \frac{-m}{z - c} + \frac{g'}{g}$, and $\frac{g'}{g}$ is holomorphic near $c$. So $a_{-1} = -m$. For a zero of order $m$, $f = (z - c)^m g$ and $a_{-1} = +m$. For $\frac{z^3}{(z-1)^2}$: $3$ at $0$, $-2$ at $1$. (Integrating $\frac{f'}{f}$ around a contour therefore counts zeros minus poles: the argument principle.)
9. For $w \ne 0$ the solutions $z_k = \frac{1}{\operatorname{Log} w + 2\pi i k}$ satisfy $|z_k| \le \frac{1}{|2\pi k| - |\operatorname{Log} w|} \to 0$, so infinitely many lie in any punctured disc. $e^{1/z} = 0$ is impossible because $e^{u} \neq 0$ for every $u$.

Numerical checks:

```python
import cmath

import numpy as np
import sympy as sp


def laurent_coefficients(f, center, radius, n_values, n_points=512):
    t = 2 * np.pi * np.arange(n_points) / n_points
    samples = f(center + radius * np.exp(1j * t))
    return {n: (samples * np.exp(-1j * n * t)).mean() / radius**n for n in n_values}


z = sp.symbols("z")
print("Ex 2:", sp.series((sp.exp(z) - 1) / z, z, 0, 2), "|", sp.series(z / (1 - sp.cos(z)), z, 0, 3))
print("Ex 5:", sp.series(1 / z - 1 / sp.sin(z), z, 0, 4))
z_16 = complex(0, -1 / (33 * np.pi))
print("Ex 6:", z_16, abs(z_16), cmath.exp(1 / z_16))

coefficients = laurent_coefficients(lambda w: 1 / ((w - 1) * (w - 2)), 0, 1.5, range(-3, 3))
print("Ex 7:", {n: round(value.real, 10) for n, value in coefficients.items()})

f = lambda w: w**3 / (w - 1) ** 2
f_prime = lambda w: (3 * w**2 * (w - 1) ** 2 - 2 * w**3 * (w - 1)) / (w - 1) ** 4
log_derivative = lambda w: f_prime(w) / f(w)
print("Ex 8: a_-1 of f'/f at 0:", laurent_coefficients(log_derivative, 0, 0.5, [-1])[-1].real,
      "  at 1:", laurent_coefficients(log_derivative, 1, 0.5, [-1])[-1].real)
```

## 8. Conclusion

- On an annulus $r < |z - c| < R$, a holomorphic function has a unique **Laurent series** $\sum_{n=-\infty}^{\infty} a_n (z - c)^n$ with $a_n = \frac{1}{2\pi i}\oint \frac{f(w)}{(w - c)^{n+1}}dw$. Different annuli give different series.
- In practice: geometric series, known Taylor series, substitution ($w = 1/z$), or `sp.series` (including `sp.oo` for the outer region).
- Isolated singularities: **removable** (no principal part; bounded), **pole of order $m$** (principal part ends at $(z - c)^{-m}$; $|f| \to \infty$), **essential** (infinite principal part; no limit).
- **Casorati–Weierstrass:** near an essential singularity $f$ comes arbitrarily close to every value; Picard: it hits all but at most one value infinitely often. $e^{1/z}$ hits every $w \ne 0$ at $z_k = 1/(\operatorname{Log} w + 2\pi i k) \to 0$.
- The coefficient $a_{-1}$ is the only one that survives a contour integral.
