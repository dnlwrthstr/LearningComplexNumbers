# Cauchy's Theorem

In notebook 01 the integral of $z^2$ between two points ignored the path, while the integral of $\bar z$ did not. **Cauchy's theorem** says this is no accident: the integral of a holomorphic function around any closed curve in a region without holes is **zero**. The proof is two lines once you combine Green's theorem with the Cauchy–Riemann equations.

The theorem's real power is a consequence: contours can be **deformed** freely through regions where $f$ is holomorphic without changing the integral. That is why $\oint dz/z = 2\pi i$ for every curve that winds once around $0$ — circle, square or squiggle — and it is the principle behind everything in the rest of this module.

This is notebook 02 of module 06 (**Complex integration**) in the **Learning Complex Numbers** series.

## 1. Introduction

In real calculus, $\int_a^b f(x)\,dx$ has only one path. In the plane, a line integral $\int (P\,dx + Q\,dy)$ is path-independent only for special vector fields (gradients, "conservative" fields). Cauchy's theorem says every holomorphic function behaves like a conservative field — the Cauchy–Riemann equations are precisely the condition that makes the curl vanish.

The theorem has a hypothesis that matters: the region must have **no holes**. $1/z$ is holomorphic everywhere except $0$, yet $\oint_{|z|=1} dz/z = 2\pi i \ne 0$. The hole at the origin is what lets the integral be nonzero, and deforming around holes is how we will compute integrals from now on.

## 2. Theory

### 2.1 Statement

> **Cauchy's theorem (Cauchy–Goursat).** Let $D \subseteq \mathbb{C}$ be open and **simply connected**, and let $f$ be holomorphic on $D$. Then for every closed curve $\gamma$ in $D$,
> $$\oint_\gamma f(z)\,dz = 0.$$

A region is **simply connected** if it is connected and every closed curve in it can be shrunk to a point without leaving the region — informally, it has no holes. Discs, half-planes, the whole plane, and the plane with a ray removed (a slit plane) are simply connected. An annulus and the punctured plane $\mathbb{C} \setminus \{0\}$ are not.

### 2.2 Proof sketch: Green meets Cauchy–Riemann

Write $f = u + iv$ and $dz = dx + i\,dy$. Multiplying out,

$$
\oint_\gamma f\,dz = \oint_\gamma (u + iv)(dx + i\,dy) = \oint_\gamma (u\,dx - v\,dy) + i\oint_\gamma (v\,dx + u\,dy).
$$

**Green's theorem** turns a line integral around the boundary of a region $\Omega$ into a double integral over it: $\oint_{\partial\Omega} (P\,dx + Q\,dy) = \iint_\Omega (Q_x - P_y)\,dA$. Applied to both pieces:

$$
\oint_\gamma f\,dz = \iint_\Omega (-v_x - u_y)\,dA + i\iint_\Omega (u_x - v_y)\,dA.
$$

The Cauchy–Riemann equations say $u_y = -v_x$ and $u_x = v_y$, so **both integrands are identically zero**. $\blacksquare$

(This sketch assumes $f'$ is continuous, which Green's theorem needs. Goursat proved the theorem without that assumption by subdividing triangles; we will see in notebook 03 that $f'$ is automatically continuous anyway.)

The same computation explains $\bar z$. With $f = x - iy$: $u = x$, $v = -y$, so $-v_x - u_y = 0$ and $u_x - v_y = 2$. Hence

$$
\oint_\gamma \bar z\,dz = 2i \cdot \text{Area}(\Omega),
$$

the "$2i \times$ area" seen in notebook 01. The failure of Cauchy–Riemann ($u_x - v_y = 2 \neq 0$) is measured exactly by the integral.

A compact way to say the same thing uses the Wirtinger derivative $\partial f/\partial\bar z = \frac12(f_x + i f_y)$: Green's theorem becomes $\oint_\gamma f\,dz = 2i\iint_\Omega \frac{\partial f}{\partial \bar z}\,dA$, and Cauchy–Riemann is $\partial f/\partial \bar z = 0$.

### 2.3 Path independence and antiderivatives

If closed integrals vanish, then two paths $\gamma_1, \gamma_2$ from $z_0$ to $z_1$ give the same integral ($\gamma_1$ followed by $\gamma_2$ reversed is closed). So on a simply connected $D$ we can define

$$
F(z) = \int_{z_0}^{z} f(w)\,dw \quad (\text{along any path in } D),
$$

and one checks $F'(z) = f(z)$. **Every holomorphic function on a simply connected region has an antiderivative.** Example: on the slit plane $\mathbb{C} \setminus (-\infty, 0]$, the antiderivative of $1/z$ with $F(1) = 0$ is the principal logarithm $\operatorname{Log} z$. On the punctured plane (not simply connected) no antiderivative exists — otherwise $\oint dz/z$ would be $0$.

### 2.4 Deforming contours

Suppose $\gamma_1$ and $\gamma_2$ are two counter-clockwise closed curves, $\gamma_2$ inside $\gamma_1$, and $f$ is holomorphic on the ring-shaped region between them (and on the curves). Cut the ring with two close parallel segments: you get a closed curve that runs around $\gamma_1$, along a cut, backwards around $\gamma_2$, and back along the cut. It encloses no singularity, so its integral is $0$; the two cut segments cancel. Therefore

$$
\oint_{\gamma_1} f\,dz = \oint_{\gamma_2} f\,dz.
$$

**A contour can be slid, stretched and bent anywhere, as long as it does not cross a singularity.** More generally, if $\gamma$ encloses isolated singularities $a_1, \dots, a_k$, then

$$
\oint_\gamma f\,dz = \sum_{j=1}^{k} \oint_{|z - a_j| = \varepsilon} f\,dz,
$$

a sum of integrals over tiny circles — the seed of the residue theorem (notebook 06).

### 2.5 Consequences in one table

| $f$ | Region | $\oint_\gamma f\,dz$ for $\gamma$ around $0$ once |
|---|---|---|
| polynomial, $e^z$, $\sin z$ (entire) | $\mathbb{C}$ | $0$ |
| $\frac{1}{z-2}$, $\gamma$ inside $\lvert z\rvert < 2$ | disc $\lvert z\rvert<2$ | $0$ |
| $\frac{1}{z}$ | $\mathbb{C}\setminus\{0\}$, not simply connected | $2\pi i$ (any shape, by deformation) |
| $\frac{1}{z^2}$ | $\mathbb{C}\setminus\{0\}$ | $0$ (it has the antiderivative $-1/z$) |
| $\bar z$ | not holomorphic | $2i \times$ area |

Note the $1/z^2$ line: Cauchy's theorem gives *sufficient* conditions for a zero integral, not necessary ones.

## 3. Worked Examples

### Example 1 — $\oint_{|z|=1} e^{z}\cos z\,dz$

$e^z \cos z$ is entire (products of entire functions are entire), and $\mathbb{C}$ is simply connected: the integral is $0$. No parametrization needed.

### Example 2 — $\oint_{|z|=1} \frac{dz}{z - 2}$

The only singularity, $z = 2$, lies **outside** the unit circle. The disc $|z| < 3/2$ is simply connected, contains the curve and avoids $z = 2$, so the integral is $0$.

### Example 3 — $\oint_{\text{square}} \frac{dz}{z}$, square with corners $\pm 1 \pm i$

The square winds once around $0$, the only singularity. By 2.4 it can be deformed into the unit circle: the answer is $2\pi i$. (Doing the four legs by hand gives four integrals of the form $\int \frac{dz}{z}$ along segments, each equal to $\frac{\pi}{2} i$ — try one.)

### Example 4 — $\oint_{|z|=2} \frac{dz}{z^2 - 1}$

Partial fractions: $\frac{1}{z^2 - 1} = \frac12\left(\frac{1}{z - 1} - \frac{1}{z + 1}\right)$. Both $\pm 1$ are inside $|z| = 2$; deform onto small circles around each:

$$
\oint_{|z|=2} \frac{dz}{z^2-1} = \frac12\left(2\pi i - 2\pi i\right) = 0.
$$

With the smaller circle $|z - 1| = 1$, only $z = 1$ is inside, and the answer is $\frac12 \cdot 2\pi i = \pi i$.

### Example 5 — an antiderivative along any path

$\int_\gamma e^{z}\,dz$ from $0$ to $i\pi$: $e^z$ is entire with antiderivative $e^z$, so for **every** path the answer is $e^{i\pi} - e^{0} = -2$.

### Example 6 — $\int_\gamma \frac{dz}{z}$ from $1$ to $i$ in the slit plane

On $\mathbb{C} \setminus (-\infty, 0]$ the antiderivative is $\operatorname{Log} z$, so for any path that avoids the negative real axis,

$$
\int_\gamma \frac{dz}{z} = \operatorname{Log} i - \operatorname{Log} 1 = i\frac{\pi}{2}.
$$

A path that goes the long way round (clockwise through $-1$) crosses the cut and picks up an extra $-2\pi i$: $i\frac{\pi}{2} - 2\pi i = -i\frac{3\pi}{2}$.

### Example 7 — Green's theorem for $\bar z$ on a disc

On $|z - c| = r$, $\oint \bar z\,dz = 2i\cdot\pi r^2$. For $r = 1$, $c = 0$: $2\pi i$ (as computed directly in notebook 01). The centre $c$ does not matter: area is translation-invariant.

## 4. Python Examples

### 4.1 Integrating around arbitrary closed curves

We describe a closed curve by its samples $z_0, \dots, z_N$ with $z_N = z_0$ and use $\int f\,dz \approx \sum_k \frac{f(z_{k-1}) + f(z_k)}{2}(z_k - z_{k-1})$ — the trapezoidal rule written directly with complex steps. This works for any shape, including polygons. Three curves: a circle, a square, and an irregular "blob" $z(t) = (1 + 0.3\cos 3t + 0.2\sin 5t)\,e^{it}$.

```python
import numpy as np


def closed_integral(f, points):
    # Trapezoidal rule on a sampled curve: sum of average f times the complex step
    values = f(points)
    steps = np.diff(points)
    return np.sum((values[1:] + values[:-1]) / 2 * steps)


def make_curves(n_points=8001, shift=0):
    t = np.linspace(0, 2 * np.pi, n_points)
    circle = np.exp(1j * t)
    blob = (1 + 0.3 * np.cos(3 * t) + 0.2 * np.sin(5 * t)) * np.exp(1j * t)
    side = np.linspace(0, 1, n_points // 4)
    corners = [1 + 1j, -1 + 1j, -1 - 1j, 1 - 1j, 1 + 1j]
    square = np.concatenate([a + (b - a) * side[:-1] for a, b in zip(corners[:-1], corners[1:])] + [np.array([corners[0]])])
    return {"circle": circle + shift, "square": square + shift, "blob": blob + shift}


functions = {
    "exp(z) cos(z)": lambda z: np.exp(z) * np.cos(z),
    "z^5 - 3z": lambda z: z**5 - 3 * z,
    "1/(z - 2)": lambda z: 1 / (z - 2),
    "1/z": lambda z: 1 / z,
    "1/z^2": lambda z: 1 / z**2,
}

curves = make_curves()
print(f"{'f':>14} | " + " | ".join(f"{name:>22}" for name in curves))
for name, f in functions.items():
    row = []
    for points in curves.values():
        value = closed_integral(f, points)
        row.append(f"{value.real:+.6f}{value.imag:+.6f}j")
    print(f"{name:>14} | " + " | ".join(f"{entry:>22}" for entry in row))
print("2 pi i =", 2j * np.pi)
```

Every entire function gives $0$ on every curve; $\frac{1}{z-2}$ gives $0$ because its pole is outside all three curves; $\frac1z$ gives $2\pi i$ on all three; $\frac{1}{z^2}$ gives $0$ despite the singularity. (The square uses the trapezoidal rule on straight edges, so its tiny error is larger than for the smooth curves.)

### 4.2 $\bar z$ measures area

For $\bar z$ the integral should equal $2i$ times the enclosed area. We compute the area independently with the **shoelace formula** $\frac12\sum_k (x_k y_{k+1} - x_{k+1} y_k)$.

```python
import numpy as np


def closed_integral(f, points):
    values = f(points)
    steps = np.diff(points)
    return np.sum((values[1:] + values[:-1]) / 2 * steps)


def shoelace_area(points):
    x, y = points.real, points.imag
    return 0.5 * np.sum(x[:-1] * y[1:] - x[1:] * y[:-1])


t = np.linspace(0, 2 * np.pi, 8001)
shapes = {
    "unit circle": np.exp(1j * t),
    "circle r = 2 at 3 + i": 3 + 1j + 2 * np.exp(1j * t),
    "ellipse a = 3, b = 1": 3 * np.cos(t) + 1j * np.sin(t),
    "blob": (1 + 0.3 * np.cos(3 * t) + 0.2 * np.sin(5 * t)) * np.exp(1j * t),
}
for name, points in shapes.items():
    integral = closed_integral(np.conj, points)
    area = shoelace_area(points)
    print(f"{name:>22}:  integral of conj(z) = {integral:.6f}   2i * area = {2j * area:.6f}")
```

### 4.3 Green's theorem, symbolically

Sympy confirms the two integrands $-v_x - u_y$ and $u_x - v_y$ of 2.2 vanish for holomorphic functions and do not for $\bar z$ or $|z|^2$.

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
z = x + sp.I * y

candidates = {
    "z^3": z**3,
    "exp(z)": sp.exp(z),
    "1/(z - 2)": 1 / (z - 2),
    "conj(z)": sp.conjugate(z),
    "|z|^2": z * sp.conjugate(z),
}
for name, expression in candidates.items():
    u = sp.re(sp.expand_complex(expression))
    v = sp.im(sp.expand_complex(expression))
    real_part_integrand = sp.simplify(-sp.diff(v, x) - sp.diff(u, y))
    imag_part_integrand = sp.simplify(sp.diff(u, x) - sp.diff(v, y))
    print(f"{name:>10}:  -v_x - u_y = {real_part_integrand},   u_x - v_y = {imag_part_integrand}")
```

For $|z|^2 = x^2 + y^2$ the integrands are $-2y$ and $2x$, so $\oint |z|^2\,dz = 2i\iint (x + iy)\,dA = 2i \cdot \text{Area} \cdot (\text{centroid})$ — zero for a region centred at $0$, nonzero otherwise. A zero integral on one curve does not make a function holomorphic.

### 4.4 Antiderivatives: the logarithm, and the jump across the cut

We integrate $1/z$ from $1$ to $i$ along two arcs: counter-clockwise (stays in the slit plane) and clockwise (crosses the negative real axis).

```python
import cmath

import numpy as np


def closed_integral(f, points):
    values = f(points)
    steps = np.diff(points)
    return np.sum((values[1:] + values[:-1]) / 2 * steps)


counter_clockwise = np.exp(1j * np.linspace(0, np.pi / 2, 4001))
clockwise = np.exp(1j * np.linspace(0, -3 * np.pi / 2, 4001))

print("counter-clockwise arc :", closed_integral(lambda z: 1 / z, counter_clockwise))
print("clockwise arc         :", closed_integral(lambda z: 1 / z, clockwise))
print("Log(i) - Log(1)       :", cmath.log(1j) - cmath.log(1))
print("difference of the arcs / (2 pi i):",
      (closed_integral(lambda z: 1 / z, counter_clockwise) - closed_integral(lambda z: 1 / z, clockwise)) / (2j * np.pi))
```

### 4.5 Building an antiderivative by integrating

Section 2.3 defines $F(z) = \int_{z_0}^{z} f(w)\,dw$. We build $F$ for $f = \cos$ by integrating along the straight segment from $0$ to each target point, and compare with $\sin z$.

```python
import numpy as np


def closed_integral(f, points):
    values = f(points)
    steps = np.diff(points)
    return np.sum((values[1:] + values[:-1]) / 2 * steps)


def antiderivative(f, z_start, z_end, n_points=4001):
    return closed_integral(f, np.linspace(z_start, z_end, n_points))


for target in [1, 1j, 2 - 1j, -3 + 0.5j]:
    built = antiderivative(np.cos, 0, target)
    print(f"z = {target!s:>10}:  integral of cos from 0 = {built:.8f}   sin z = {np.sin(target):.8f}")
```

## 5. Visualizations

### 5.1 Contours that give the same answer

Every curve below winds once around the singularity at $0$ (red ×) and gives $\oint dz/z = 2\pi i$; the dashed curve does not enclose $0$ and gives $0$.

```python
import matplotlib.pyplot as plt
import numpy as np


def closed_integral(f, points):
    values = f(points)
    steps = np.diff(points)
    return np.sum((values[1:] + values[:-1]) / 2 * steps)


t = np.linspace(0, 2 * np.pi, 8001)
curves = {
    "circle": 0.6 * np.exp(1j * t),
    "blob": (1.4 + 0.4 * np.cos(3 * t) + 0.25 * np.sin(5 * t)) * np.exp(1j * t),
    "shifted ellipse": 0.5 + 2.2 * np.cos(t) + 1j * 1.2 * np.sin(t),
    "away from 0": 2 + 1.2j + 0.6 * np.exp(1j * t),
}

fig, ax = plt.subplots(figsize=(7.5, 7))
for name, points in curves.items():
    value = closed_integral(lambda z: 1 / z, points) / (2j * np.pi)
    style = "--" if name == "away from 0" else "-"
    ax.plot(points.real, points.imag, style, linewidth=2, label=f"{name}: integral = {round(value.real, 3) + 0.0:.3f} x 2 pi i")
    k = len(points) // 8
    ax.annotate("", xy=(points[k + 50].real, points[k + 50].imag), xytext=(points[k].real, points[k].imag),
                arrowprops=dict(arrowstyle="->", linewidth=2))

ax.scatter(0, 0, marker="x", color="red", s=100, zorder=3, label="singularity of 1/z")
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-2.5, 3.2)
ax.set_ylim(-2.2, 2.6)
ax.set_aspect("equal")
ax.set_title(r"$\oint dz/z$ depends only on whether the curve winds around 0")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower right", fontsize=9)
plt.show()
```

### 5.2 Deforming a square into a circle

Interpolate continuously between the square with corners $\pm1\pm i$ and the circle of radius $0.5$: $\gamma_s = (1 - s)\,\gamma_{\text{square}} + s\,\gamma_{\text{circle}}$. For $1/z$ (holomorphic on every intermediate curve) the integral stays at $2\pi i$. For $\bar z$ it follows the shrinking area. For $1/(z - 0.8)$ it jumps when the curves sweep across the singularity at $0.8$ — deformation is only allowed through regions where $f$ is holomorphic.

```python
import matplotlib.pyplot as plt
import numpy as np


def closed_integral(f, points):
    values = f(points)
    steps = np.diff(points)
    return np.sum((values[1:] + values[:-1]) / 2 * steps)


n_points = 8000
t = np.linspace(0, 2 * np.pi, n_points + 1)
circle = 0.5 * np.exp(1j * t)
# The square parametrized by the same angle t: scale each circle point out to the boundary of the square
direction = np.exp(1j * t)
square = direction / np.maximum(np.abs(direction.real), np.abs(direction.imag))

blend = np.linspace(0, 1, 120)  # avoids s = 0.4, where the curve passes exactly through 0.8
results = {"1/z": [], "conj(z)": [], "1/(z - 0.8)": []}
functions = {"1/z": lambda z: 1 / z, "conj(z)": np.conj, "1/(z - 0.8)": lambda z: 1 / (z - 0.8)}
for s in blend:
    curve = (1 - s) * square + s * circle
    for name, f in functions.items():
        results[name].append(closed_integral(f, curve).imag)

fig, (ax_curves, ax_values) = plt.subplots(1, 2, figsize=(13, 5.5))
for s in [0, 0.25, 0.5, 0.75, 1]:
    curve = (1 - s) * square + s * circle
    ax_curves.plot(curve.real, curve.imag, label=f"s = {s}")
ax_curves.scatter([0, 0.8], [0, 0], marker="x", color="red", s=80, zorder=3)
ax_curves.text(0.82, 0.08, "0.8", color="red")
ax_curves.axhline(0, color="black", linewidth=0.8)
ax_curves.axvline(0, color="black", linewidth=0.8)
ax_curves.set_aspect("equal")
ax_curves.set_title("Square deformed into a circle")
ax_curves.set_xlabel("Re")
ax_curves.set_ylabel("Im")
ax_curves.grid(True, alpha=0.3)
ax_curves.legend(fontsize=8, loc="upper left")

for name, values in results.items():
    ax_values.plot(blend, values, linewidth=2, label=f"f = {name}")
ax_values.axhline(2 * np.pi, color="gray", linestyle=":", label=r"$2\pi$")
ax_values.set_title(r"$\mathrm{Im} \oint_{\gamma_s} f\, dz$ during the deformation")
ax_values.set_xlabel("deformation parameter s")
ax_values.set_ylabel("imaginary part of the integral")
ax_values.grid(True, alpha=0.3)
ax_values.legend()
plt.tight_layout()
plt.show()
```

The $1/(z - 0.8)$ curve sits at $2\pi$ while $0.8$ is inside, then drops to $0$ at the moment the shrinking curve passes the point $0.8$. All three integrals have zero real part here, so only the imaginary part is plotted.

### 5.3 Why the cut segments cancel

The keyhole argument of 2.4, drawn: an outer circle, an inner circle traversed backwards, and two segments along a cut traversed in opposite directions. The whole closed path avoids the singularity, so its integral is $0$, which forces outer $=$ inner.

```python
import matplotlib.pyplot as plt
import numpy as np


def closed_integral(f, points):
    values = f(points)
    steps = np.diff(points)
    return np.sum((values[1:] + values[:-1]) / 2 * steps)


gap = 0.08
outer_radius, inner_radius = 2.0, 0.5
outer_angle = np.arcsin(gap / outer_radius)
inner_angle = np.arcsin(gap / inner_radius)

outer_arc = outer_radius * np.exp(1j * np.linspace(outer_angle, 2 * np.pi - outer_angle, 4000))
lower_cut = np.linspace(outer_arc[-1], inner_radius * np.exp(-1j * inner_angle), 400)
inner_arc = inner_radius * np.exp(1j * np.linspace(2 * np.pi - inner_angle, inner_angle, 4000))
upper_cut = np.linspace(inner_arc[-1], outer_arc[0], 400)
keyhole = np.concatenate([outer_arc, lower_cut[1:], inner_arc[1:], upper_cut[1:]])

f = lambda z: np.exp(z) / z
print("keyhole integral of e^z / z           :", closed_integral(f, keyhole))
print("full outer circle (counter-clockwise)  :", closed_integral(f, outer_radius * np.exp(1j * np.linspace(0, 2 * np.pi, 8001))))
print("full inner circle (counter-clockwise)  :", closed_integral(f, inner_radius * np.exp(1j * np.linspace(0, 2 * np.pi, 8001))))

fig, ax = plt.subplots(figsize=(6.5, 6.5))
ax.plot(keyhole.real, keyhole.imag, linewidth=2, color="tab:blue")
for piece, index in [(outer_arc, 1000), (inner_arc, 1000), (lower_cut, 200), (upper_cut, 200)]:
    ax.annotate("", xy=(piece[index + 20].real, piece[index + 20].imag), xytext=(piece[index].real, piece[index].imag),
                arrowprops=dict(arrowstyle="->", color="tab:blue", linewidth=2))
ax.scatter(0, 0, marker="x", color="red", s=100, zorder=3)
ax.axhline(0, color="black", linewidth=0.5)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_aspect("equal")
ax.set_xlim(-2.4, 2.4)
ax.set_ylim(-2.4, 2.4)
ax.set_title("Keyhole contour: outer circle, cut, inner circle backwards, cut")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
plt.show()
```

### 5.4 The logarithm as an integral, and its cut

Define $F(z) = \int_1^z \frac{dw}{w}$ along the straight segment from $1$ to $z$. Such a segment never crosses the negative real axis, so $F$ lives on the slit plane and should equal $\operatorname{Log} z$. Its imaginary part (left) is the argument, jumping by $2\pi$ across the cut; the right panel shows the difference from `np.log` (400 trapezoid steps per segment).

```python
import matplotlib.pyplot as plt
import numpy as np

n_grid, n_steps = 200, 400  # an even grid size keeps every grid point off the real axis (and off the cut)
axis = np.linspace(-2, 2, n_grid)
grid = axis[None, :] + 1j * axis[:, None]
s = np.linspace(0, 1, n_steps + 1)[:, None, None]

# Straight path from 1 to every grid point at once: w(s) = 1 + s (z - 1), dw = (z - 1) ds
paths = 1 + s * (grid[None, :, :] - 1)
integrand = (grid - 1)[None, :, :] / paths
log_by_integration = np.sum((integrand[1:] + integrand[:-1]) / 2, axis=0) / n_steps

error = np.abs(log_by_integration - np.log(grid))

fig, (ax_arg, ax_error) = plt.subplots(1, 2, figsize=(13, 5.5))
image = ax_arg.imshow(log_by_integration.imag, extent=[-2, 2, -2, 2], origin="lower", cmap="twilight")
fig.colorbar(image, ax=ax_arg, label=r"$\mathrm{Im}\, F(z)$")
ax_arg.set_title(r"$\mathrm{Im} \int_1^z dw/w$ along straight segments")
image = ax_error.imshow(np.log10(error + 1e-17), extent=[-2, 2, -2, 2], origin="lower", cmap="viridis")
fig.colorbar(image, ax=ax_error, label=r"$\log_{10} |F(z) - \mathrm{Log}\, z|$")
ax_error.set_title("Difference from the principal logarithm")
for ax in (ax_arg, ax_error):
    ax.plot([-2, 0], [0, 0], color="red", linewidth=2)
    ax.scatter(1, 0, color="white", edgecolor="black", zorder=3)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
plt.tight_layout()
plt.show()
```

The difference is around $10^{-6}$ — the discretisation error of the trapezoidal rule, not a real discrepancy. It is largest close to $0$ and just beside the negative real axis, where the segments from $1$ pass very near the singularity and $1/w$ changes fast; more steps shrink it everywhere.

## 7. Solutions

1. Both are $0$ by Cauchy's theorem: polynomials and $\sin(z^2)$ are entire, and $\mathbb{C}$ is simply connected.
2. $z^2 + 4 = 0$ at $z = \pm 2i$, both outside the unit circle. The disc $|z| < 1.5$ is simply connected and $f$ is holomorphic there, so the integral is $0$.
3. The disc: yes. The annulus: no (a circle around the hole cannot be shrunk). The plane minus $[0, \infty)$: yes (a slit plane). The plane minus $\pm 1$: no.
4. The circle $|z - 1| = 1.5$ runs from $-0.5$ to $2.5$ on the real axis, so $0$ is inside and $3$ is outside. Prediction: $2\pi i + 0 = 2\pi i$.
5. $\frac{z}{z^2 - 1} = \frac12\left(\frac{1}{z-1} + \frac{1}{z+1}\right)$; both poles are inside, so the integral is $\frac12(2\pi i + 2\pi i) = 2\pi i$.
6. $F(z) = \frac12 e^{z^2}$. $(1+i)^2 = 2i$, so the answer is $\frac12\left(e^{2i} - 1\right) = \frac12(\cos 2 - 1) + \frac{i}{2}\sin 2 \approx -0.708 + 0.455i$.
7. $\oint x\,dz = \oint x\,dx + i\oint x\,dy$. Green with $(P, Q) = (x, 0)$: $0$; with $(P, Q) = (0, x)$: $\iint 1\,dA$. So $i \cdot$Area. Similarly $\oint y\,dz = \oint y\,dx + i\oint y\,dy = -\text{Area} + 0$. Unit circle: $i\pi$ and $-\pi$.
8. Cauchy's theorem does not apply (the punctured disc is not simply connected), but $1/z^2$ has the antiderivative $-1/z$ on $\mathbb{C}\setminus\{0\}$, so every closed integral vanishes. Example with $6\pi i$: $f(z) = \frac{3}{z}$ (or $\frac{3}{z} + \frac{1}{z^2} + z$).
9. $e^{-z^2}$ is entire, so the integral around the rectangle is $0$: bottom side ($\int_{-R}^{R} e^{-x^2}dx$) + right side − top side ($\int_{-R}^{R} e^{-(x+ib)^2}dx$) − left side $= 0$. On the vertical sides $|e^{-z^2}| = e^{-(R^2 - y^2)} \le e^{b^2 - R^2} \to 0$, and they have length $b$, so by ML they vanish. Letting $R \to \infty$, top $=$ bottom $= \sqrt\pi$. Shifting a Gaussian integral off the real axis does not change it — this is how the Fourier transform of a Gaussian is computed.

Numerical checks:

```python
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz


def closed_integral(f, points):
    values = f(points)
    steps = np.diff(points)
    return np.sum((values[1:] + values[:-1]) / 2 * steps)


t = np.linspace(0, 2 * np.pi, 8001)
print("Ex 1:", closed_integral(lambda z: z**4 - 2 * z + 7, 3 * np.exp(1j * t)), closed_integral(lambda z: np.sin(z**2), 3 * np.exp(1j * t)))
print("Ex 2:", closed_integral(lambda z: 1 / (z**2 + 4), np.exp(1j * t)))
print("Ex 4:", closed_integral(lambda z: 1 / z + 1 / (z - 3), 1 + 1.5 * np.exp(1j * t)), " vs 2 pi i =", 2j * np.pi)
print("Ex 5:", closed_integral(lambda z: z / (z**2 - 1), 2 * np.exp(1j * t)))
path = np.linspace(0, 1 + 1j, 4001)
print("Ex 6:", closed_integral(lambda z: z * np.exp(z**2), path), " vs", (np.exp(2j) - 1) / 2)
print("Ex 7:", closed_integral(np.real, np.exp(1j * t)), closed_integral(np.imag, np.exp(1j * t)))
x = np.linspace(-12, 12, 200001)
print("Ex 9:", np.trapezoid(np.exp(-(x + 1j) ** 2), x), " vs sqrt(pi) =", np.sqrt(np.pi))
```

## 8. Conclusion

- **Cauchy's theorem:** if $f$ is holomorphic on a simply connected region, $\oint_\gamma f\,dz = 0$ for every closed curve $\gamma$ in it.
- The proof is Green's theorem plus Cauchy–Riemann: $\oint f\,dz = \iint(-v_x - u_y)\,dA + i\iint(u_x - v_y)\,dA$, and both integrands vanish. For $\bar z$ they do not, and $\oint \bar z\,dz = 2i \cdot \text{Area}$.
- Consequences: path independence and antiderivatives on simply connected regions ($\operatorname{Log}$ on the slit plane).
- **Deformation:** a contour may be moved through any region where $f$ is holomorphic. An integral around several singularities equals the sum of integrals around small circles at each of them.
- Holes matter: $\oint dz/z = 2\pi i$ for every curve winding once around $0$, because $0$ is a hole in the domain of $1/z$.
