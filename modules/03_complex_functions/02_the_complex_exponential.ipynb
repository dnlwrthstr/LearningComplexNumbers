# The Complex Exponential

Euler's formula $e^{i\theta} = \cos\theta + i\sin\theta$ told us what $e$ raised to a purely imaginary power means. This notebook combines it with the real exponential to define $e^z$ for **every** complex $z$, and then studies $e^z$ as a map of the plane.

Two facts dominate everything that follows. The exponential **never vanishes**, and it is **periodic** with the imaginary period $2\pi i$. Geometrically it turns vertical lines into circles and horizontal lines into rays, wrapping each horizontal strip of height $2\pi$ once around the origin. That periodicity is exactly why the logarithm in the next notebook is multi-valued.

This is notebook 02 of module 03 (**Complex functions**) in the **Learning Complex Numbers** series.

## 1. Introduction

On the real line, $e^x$ is the function that equals its own derivative, turns sums into products ($e^{a + b} = e^a e^b$) and grows without bound. In the complex plane it keeps all those algebraic properties but its geometry becomes much richer: the real part of $z$ controls the **size** of $e^z$, the imaginary part controls its **direction**.

Why it matters:

- Every oscillation and decay in physics and engineering is written as $e^{(\sigma + i\omega)t}$: $\sigma$ sets growth or damping, $\omega$ sets frequency (module 07: AC circuits, damped oscillators).
- Fourier analysis is the study of the functions $e^{i\omega t}$.
- The polar form $re^{i\theta}$ is itself just $e^{\ln r + i\theta}$.

## 2. Theory

### 2.1 Definition

For $z = x + iy$ define

$$
e^z = e^{x}\,e^{iy} = e^{x}(\cos y + i \sin y).
$$

This is the only sensible choice: if the law $e^{a + b} = e^a e^b$ is to survive, then $e^{x + iy}$ must equal $e^x \cdot e^{iy}$, and Euler's formula fixes $e^{iy}$. It also agrees with the power series used to prove Euler's formula:

$$
e^z = \sum_{n=0}^{\infty} \frac{z^n}{n!} = 1 + z + \frac{z^2}{2!} + \frac{z^3}{3!} + \cdots,
$$

which converges for every complex $z$. Real and imaginary parts:

$$
u(x, y) = e^x \cos y, \qquad v(x, y) = e^x \sin y.
$$

### 2.2 Modulus and argument

The definition is already in polar form, $r e^{i\theta}$ with $r = e^x$ and $\theta = y$:

$$
|e^z| = e^{\mathrm{Re}\, z}, \qquad \arg e^z = \mathrm{Im}\, z \pmod{2\pi}.
$$

So **the real part sets the size, the imaginary part sets the angle.** (The principal argument $\mathrm{Arg}\, e^z$ is $y$ shifted by a multiple of $2\pi$ into $(-\pi, \pi]$.)

### 2.3 Algebraic properties

For all complex $z, w$ and integers $n$:

$$
e^{z + w} = e^z e^w, \qquad e^{-z} = \frac{1}{e^z}, \qquad (e^z)^n = e^{nz}, \qquad \overline{e^z} = e^{\bar z}, \qquad \frac{d}{dz} e^z = e^z.
$$

*Proof of the addition law.* With $z = x + iy$, $w = s + it$:
$e^z e^w = e^{x + s}(\cos y + i \sin y)(\cos t + i \sin t) = e^{x + s}\big(\cos(y + t) + i \sin(y + t)\big) = e^{z + w}$,
using the angle-addition formulas for the middle step (multiplication adds angles). The derivative statement is made precise in module 05.

### 2.4 Never zero

$|e^z| = e^x > 0$ for every real $x$, so

$$
e^z \neq 0 \quad \text{for all } z \in \mathbb{C}.
$$

Every other value is taken (§2.6), but $0$ is never reached: the range of $e^z$ is $\mathbb{C} \setminus \{0\}$.

### 2.5 Periodicity with period $2\pi i$

Since $\cos$ and $\sin$ have period $2\pi$,

$$
e^{z + 2\pi i} = e^z e^{2\pi i} = e^z (\cos 2\pi + i \sin 2\pi) = e^z.
$$

More precisely, $e^z = 1$ exactly when $e^x = 1$ and $y$ is a multiple of $2\pi$, i.e.

$$
e^z = 1 \iff z = 2\pi i k,\ k \in \mathbb{Z}, \qquad\text{hence}\qquad e^z = e^w \iff z - w \in 2\pi i\,\mathbb{Z}.
$$

The real exponential is one-to-one; the complex exponential is **infinitely-many-to-one**. Its period is purely imaginary, which is why we never see it on the real line.

### 2.6 Solving $e^z = w$

Given $w \neq 0$, write $w = |w| e^{i\varphi}$. Then $e^x e^{iy} = |w| e^{i\varphi}$ forces $e^x = |w|$ and $y = \varphi + 2\pi k$:

$$
e^z = w \iff z = \ln|w| + i(\varphi + 2\pi k), \quad k \in \mathbb{Z}.
$$

Infinitely many solutions, stacked vertically $2\pi$ apart. Choosing $k$ so that the imaginary part lies in $(-\pi, \pi]$ gives the **principal logarithm** of the next notebook.

### 2.7 The exponential as a map

Write $w = e^z$ and use §2.2.

| Set in the $z$-plane | Image in the $w$-plane |
|---|---|
| vertical line $x = a$ | circle $\lvert w\rvert = e^a$, wrapped round infinitely often (once per vertical length $2\pi$) |
| horizontal line $y = b$ | open ray from $0$ (not including $0$) at angle $b$ |
| rectangle $a_1 \le x \le a_2$, $b_1 \le y \le b_2$ (with $b_2 - b_1 < 2\pi$) | annular sector $e^{a_1} \le \lvert w\rvert \le e^{a_2}$, $b_1 \le \arg w \le b_2$ |
| horizontal strip $-\pi < y \le \pi$ | all of $\mathbb{C} \setminus \{0\}$, each point hit exactly once |
| strip $0 < y < \pi$ | upper half-plane |
| left half-plane $x < 0$ | punctured unit disc $0 < \lvert w\rvert < 1$ |

The exponential **converts the rectangular grid into the polar grid**: Cartesian coordinates $(x, y)$ in the $z$-plane become polar coordinates $(\ln r, \theta)$ in the $w$-plane. Moving left sends $w$ toward $0$ but never reaches it — that is §2.4 seen geometrically.

## 3. Worked Examples

### Example 1 — evaluating $e^{1 + i\pi/2}$

$e^{1 + i\pi/2} = e^1 \left(\cos\frac\pi2 + i \sin\frac\pi2\right) = e \cdot i \approx 2.718i$.

Geometrically: modulus $e^1$, argument $\pi/2$, a point on the positive imaginary axis at distance $e$.

### Example 2 — modulus and argument of $e^{2 - 3i}$

$|e^{2 - 3i}| = e^2 \approx 7.389$, and an argument is $-3$. Since $-3 \in (-\pi, \pi]$, the principal argument is also $-3$ (about $-172°$, just below the negative real axis). Cartesian: $e^2(\cos 3 - i \sin 3) \approx -7.315 - 1.043i$.

### Example 3 — all solutions of $e^z = -1$

$|-1| = 1$ so $x = \ln 1 = 0$; $\arg(-1) = \pi$, so $y = \pi + 2\pi k$:

$$
z = i\pi(2k + 1) = \ldots,\ -3\pi i,\ -\pi i,\ \pi i,\ 3\pi i,\ \ldots
$$

The case $k = 0$ is Euler's identity $e^{i\pi} = -1$.

### Example 4 — all solutions of $e^z = 1 + i$

$|1 + i| = \sqrt2$, $\arg(1 + i) = \pi/4$, so

$$
z = \ln\sqrt2 + i\left(\frac\pi4 + 2\pi k\right) = \tfrac12 \ln 2 + i\left(\frac\pi4 + 2\pi k\right), \quad k \in \mathbb{Z}.
$$

Check $k = 0$: $e^{\frac12\ln2} = \sqrt2$, and $\sqrt2\, e^{i\pi/4} = \sqrt2\left(\frac{\sqrt2}{2} + \frac{\sqrt2}{2} i\right) = 1 + i$ ✓.

### Example 5 — $e^z$ is never zero, but gets as close as you like

$e^{-10} \approx 4.5 \times 10^{-5}$, $e^{-10 + 7i}$ has the same tiny modulus. Going left along any horizontal line, $|e^z| = e^x \to 0$, yet for every finite $x$ the value is strictly positive. There is no $z$ with $e^z = 0$, because that would need $e^x = 0$.

### Example 6 — the image of a rectangle

The rectangle $0 \le x \le 1$, $0 \le y \le \pi/2$:

- left side $x = 0$ maps to the quarter of the unit circle from $1$ to $i$;
- right side $x = 1$ maps to the quarter circle of radius $e$ from $e$ to $ei$;
- bottom side $y = 0$ maps to the real segment $[1, e]$;
- top side $y = \pi/2$ maps to the imaginary segment $[i, ei]$.

So the image is the **quarter-annulus** $1 \le |w| \le e$, $0 \le \arg w \le \pi/2$. Shifting the rectangle up by $2\pi$ gives exactly the same image.

## 4. Python Examples

### 4.1 The definition matches `cmath.exp` and `np.exp`

```python
import cmath
import math

import numpy as np


def exp_from_definition(z):
    return math.exp(z.real) * complex(math.cos(z.imag), math.sin(z.imag))


for z in [1 + 1j * math.pi / 2, 2 - 3j, 1j * math.pi, -10 + 7j]:
    ours = exp_from_definition(z)
    library = cmath.exp(z)
    print(f"z = {z:.4f}:  e^z = {library:.6f},  |e^z| = {abs(library):.6f} = e^x = {math.exp(z.real):.6f},  "
          f"agrees: {abs(ours - library) < 1e-12}")

print("np.exp on an array:", np.exp(np.array([0, 1j * np.pi / 2, 1 + 1j * np.pi])))
```

### 4.2 Symbolic $u$ and $v$, and the addition law, with sympy

```python
import sympy as sp

x, y, s, t = sp.symbols("x y s t", real=True)
exponential = sp.exp(x + sp.I * y)
expanded = sp.expand_complex(exponential)
print("e^(x+iy) =", expanded)
print("u =", sp.re(expanded), "   v =", sp.im(expanded))
print("|e^(x+iy)| =", sp.simplify(sp.Abs(exponential)))

product = sp.expand_complex(sp.exp(x + sp.I * y) * sp.exp(s + sp.I * t))
target = sp.expand_complex(sp.exp((x + s) + sp.I * (y + t)))
print("e^z e^w - e^(z+w) simplifies to:", sp.simplify(sp.expand_trig(product - target)))

print("e^(2 pi i) =", sp.exp(2 * sp.pi * sp.I), "   e^(i pi) =", sp.exp(sp.I * sp.pi))
```

### 4.3 Periodicity: shifting by $2\pi i$ changes nothing

```python
import cmath

import numpy as np

z = 0.3 + 0.8j
for k in range(-2, 3):
    shifted = z + 2j * np.pi * k
    print(f"k = {k:+d}:  z + 2 pi i k = {shifted:.4f},  e^(...) = {cmath.exp(shifted):.12f}")

print("a real shift of 2 pi is not a period:", cmath.exp(z + 2 * np.pi))
```

### 4.4 Never zero

Search a big grid for the smallest $|e^z|$: it occurs at the left edge and equals $e^{\min x}$, never $0$. Compare with $\sin$, which does have zeros.

```python
import numpy as np

x = np.linspace(-20, 5, 501)
y = np.linspace(-20, 20, 801)
X, Y = np.meshgrid(x, y)
modulus = np.abs(np.exp(X + 1j * Y))

print("smallest |e^z| on the grid:", modulus.min())
print("e^(-20)                     :", np.exp(-20.0))
print("modulus depends only on x   :", np.allclose(modulus, np.exp(X)))
```

### 4.5 Solving $e^z = w$: infinitely many answers

`cmath.log(w)` returns one solution (the principal one); adding $2\pi i k$ gives the rest. Every one of them exponentiates back to $w$.

```python
import cmath
import math


def solutions_of_exp_equals(w, ks=range(-2, 3)):
    principal = cmath.log(w)
    return [principal + 2j * math.pi * k for k in ks]


for w in [-1, 1 + 1j, -2]:
    print(f"e^z = {w}:")
    for z in solutions_of_exp_equals(w):
        print(f"    z = {z.real:+.4f} {z.imag:+.4f}i    e^z = {cmath.exp(z):.6f}")
```

### 4.6 The power series converges to $e^z$

Partial sums $S_N(z) = \sum_{n=0}^{N} z^n/n!$ approach `cmath.exp(z)` quickly, even for complex $z$.

```python
import cmath
import math

z = 1j * math.pi
partial_sum = 0
for n in range(0, 21):
    partial_sum += z**n / math.factorial(n)
    if n in (1, 2, 4, 6, 10, 15, 20):
        print(f"N = {n:2d}:  S_N = {partial_sum:.10f}   error = {abs(partial_sum - cmath.exp(z)):.2e}")
```

## 5. Visualizations

### 5.1 The rectangular grid becomes the polar grid

Vertical lines (blue) become circles, horizontal lines (orange) become rays. The $z$-window spans one full period $-\pi \le y \le \pi$ vertically, so the image covers every direction exactly once.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, (ax_z, ax_w) = plt.subplots(1, 2, figsize=(12, 6.5), gridspec_kw={"width_ratios": [1, 2]})
x_values = np.linspace(-1, 1, 5)
y_values = np.linspace(-np.pi, np.pi, 13)
t_vertical = np.linspace(-np.pi, np.pi, 400)
t_horizontal = np.linspace(-1, 1, 200)

for a in x_values:
    line = a + 1j * t_vertical
    ax_z.plot(line.real, line.imag, color="tab:blue", linewidth=1.2)
    image = np.exp(line)
    ax_w.plot(image.real, image.imag, color="tab:blue", linewidth=1.2)
for b in y_values:
    line = t_horizontal + 1j * b
    ax_z.plot(line.real, line.imag, color="tab:orange", linewidth=1.2)
    image = np.exp(line)
    ax_w.plot(image.real, image.imag, color="tab:orange", linewidth=1.2)

marker = 1 + 1j * np.pi / 2
ax_z.plot(marker.real, marker.imag, "ko")
ax_w.plot(np.exp(marker).real, np.exp(marker).imag, "ko")
ax_z.annotate("$1 + i\\pi/2$", (marker.real, marker.imag), xytext=(0.3, 2.1), arrowprops=dict(arrowstyle="->"))
ax_w.annotate("$e \\cdot i$", (0, np.e), xytext=(0.8, 2.6), arrowprops=dict(arrowstyle="->"))

ax_z.set_title("z-plane: $-1 \\leq x \\leq 1$, $-\\pi \\leq y \\leq \\pi$")
ax_w.set_title("w-plane: circles $|w| = e^{a}$ and rays $\\arg w = b$")
for ax in (ax_z, ax_w):
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
fig.suptitle("$w = e^z$")
fig.tight_layout(rect=(0, 0, 1, 0.95))
plt.show()
```

### 5.2 Periodicity: different rectangles, same image

Three rectangles $0 \le x \le 1$, $b \le y \le b + \pi/2$ stacked $2\pi$ apart map onto **the same** quarter-annulus (Example 6). The fourth rectangle, shifted only by $\pi$, lands on the opposite quarter.

```python
import matplotlib.pyplot as plt
import numpy as np


def rectangle_boundary(x0, x1, y0, y1, n=100):
    bottom = np.linspace(x0, x1, n) + 1j * y0
    right = x1 + 1j * np.linspace(y0, y1, n)
    top = np.linspace(x1, x0, n) + 1j * y1
    left = x0 + 1j * np.linspace(y1, y0, n)
    return np.concatenate([bottom, right, top, left])


fig, (ax_z, ax_w) = plt.subplots(1, 2, figsize=(12, 7), gridspec_kw={"width_ratios": [1, 1.6]})
pieces = [(-2 * np.pi, "tab:blue", "$y \\in [-2\\pi, -3\\pi/2]$"),
          (0, "tab:green", "$y \\in [0, \\pi/2]$"),
          (2 * np.pi, "tab:purple", "$y \\in [2\\pi, 5\\pi/2]$"),
          (np.pi, "tab:red", "$y \\in [\\pi, 3\\pi/2]$")]
for shift, colour, label in pieces:
    boundary = rectangle_boundary(0, 1, shift, shift + np.pi / 2)
    ax_z.fill(boundary.real, boundary.imag, color=colour, alpha=0.4, label=label)
    image = np.exp(boundary)
    ax_w.fill(image.real, image.imag, color=colour, alpha=0.3)
    ax_w.plot(image.real, image.imag, color=colour, linewidth=1.5)

for k in range(-1, 3):
    ax_z.axhline(2 * np.pi * k, color="gray", linestyle=":", linewidth=1)
ax_z.set_xlim(-1.5, 2.5)
ax_z.set_title("z-plane (dotted: $y = 2\\pi k$)")
ax_w.legend(*ax_z.get_legend_handles_labels(), loc="lower right", fontsize=9)
ax_w.set_xlim(-3.2, 3.2)
ax_w.set_ylim(-3.2, 3.2)
ax_w.set_title("w-plane: blue, green, purple coincide")
for ax in (ax_z, ax_w):
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
fig.suptitle("$e^{z + 2\\pi i} = e^z$: the exponential repeats every $2\\pi i$")
fig.tight_layout(rect=(0, 0, 1, 0.95))
plt.show()
```

### 5.3 Domain colouring of $e^z$

Hue depends only on $y$ (it is $\arg e^z = y$), so the colours form horizontal bands that repeat every $2\pi$. Brightness depends only on $x$: dark on the left ($|e^z| = e^x$ small) and white on the right. There is **no black dot** anywhere: no zeros. Compare with the polynomial $z^2 + 1$, whose zeros are clearly visible.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import hsv_to_rgb


def domain_colouring(f, re_range, im_range, resolution=500):
    X, Y = np.meshgrid(np.linspace(*re_range, resolution), np.linspace(*im_range, resolution))
    W = f(X + 1j * Y)
    hue = (np.angle(W) / (2 * np.pi)) % 1.0
    m = (2 / np.pi) * np.arctan(np.abs(W))
    hsv = np.stack([hue, np.clip(2 - 2 * m, 0, 1), np.clip(2 * m, 0, 1)], axis=-1)
    return hsv_to_rgb(hsv)


fig, axes = plt.subplots(1, 2, figsize=(12, 7), gridspec_kw={"width_ratios": [1, 1.3]})
for ax, f, title, re_range, im_range in [
        (axes[0], np.exp, "$e^z$: horizontal bands, period $2\\pi$, no zeros", (-3, 3), (-2 * np.pi, 2 * np.pi)),
        (axes[1], lambda z: z**2 + 1, "$z^2 + 1$: zeros at $\\pm i$ for comparison", (-2.5, 2.5), (-2.5, 2.5))]:
    ax.imshow(domain_colouring(f, re_range, im_range), origin="lower", extent=[*re_range, *im_range])
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.set_title(title)
for k in (-2, -1, 0, 1, 2):
    axes[0].axhline(k * np.pi, color="white", linestyle=":", linewidth=0.8)
axes[0].set_yticks(np.pi * np.arange(-2, 3))
axes[0].set_yticklabels(["$-2\\pi$", "$-\\pi$", "0", "$\\pi$", "$2\\pi$"])
fig.tight_layout()
plt.show()
```

### 5.4 Partial sums of the series spiral into $e^{i\pi} = -1$

Each partial sum $S_N(i\pi)$ adds one term $(i\pi)^n / n!$, which is rotated a further quarter turn from the previous term. The path of partial sums is a shrinking square spiral that homes in on $-1$.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

z = 1j * math.pi
terms = np.array([z**n / math.factorial(n) for n in range(16)])
partial_sums = np.concatenate([[0], np.cumsum(terms)])

fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(partial_sums.real, partial_sums.imag, "o-", color="tab:blue", markersize=4, label="partial sums $S_N(i\\pi)$")
for n in range(1, 5):
    ax.annotate(f"$S_{n - 1}$", (partial_sums[n].real, partial_sums[n].imag), xytext=(5, 5), textcoords="offset points")
ax.plot(-1, 0, "r*", markersize=15, label="$e^{i\\pi} = -1$")
circle = np.exp(1j * np.linspace(0, 2 * np.pi, 200))
ax.plot(circle.real, circle.imag, color="lightgray")
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_aspect("equal")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_title("$\\sum (i\\pi)^n / n!$ converges to $-1$")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower left")
plt.show()
```

### 5.5 Strips and half-planes

Three regions and their images: the strip $0 < y < \pi$ opens onto the upper half-plane, the half-strip $x < 0$, $0 < y < \pi$ onto the upper half-disc, and the half-strip $x < 0$, $-\pi < y < \pi$ onto the punctured unit disc minus the negative real axis. Points are sampled on a grid and coloured by their $y$-value (dark purple for the lowest $y$, yellow for the highest), so you can see where each horizontal line goes: it becomes a ray at angle $y$.

```python
import matplotlib.pyplot as plt
import numpy as np

regions = [
    ("strip $0 < y < \\pi$", (-2.5, 1.2), (0.01, np.pi - 0.01)),
    ("half-strip $x < 0,\\ 0 < y < \\pi$", (-4, -0.01), (0.01, np.pi - 0.01)),
    ("half-strip $x < 0,\\ -\\pi < y < \\pi$", (-4, -0.01), (-np.pi + 0.01, np.pi - 0.01)),
]
fig, axes = plt.subplots(2, 3, figsize=(15, 9))
for column, (title, (x0, x1), (y0, y1)) in enumerate(regions):
    X, Y = np.meshgrid(np.linspace(x0, x1, 120), np.linspace(y0, y1, 120))
    Z = X + 1j * Y
    W = np.exp(Z)
    axes[0, column].scatter(Z.real, Z.imag, c=Y, cmap="viridis", s=1)
    axes[0, column].set_title("z: " + title)
    axes[0, column].set_xlim(-4.2, 1.5)
    axes[0, column].set_ylim(-3.5, 3.5)
    axes[1, column].scatter(W.real, W.imag, c=Y, cmap="viridis", s=1)
    axes[1, column].set_title("w = e^z")
    axes[1, column].set_xlim(-3.5, 3.5)
    axes[1, column].set_ylim(-3.5, 3.5)
for ax in axes.flat:
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
unit_circle = np.exp(1j * np.linspace(0, 2 * np.pi, 200))
for ax in axes[1, 1:]:
    ax.plot(unit_circle.real, unit_circle.imag, "k--", linewidth=1)
fig.tight_layout()
plt.show()
```

9. Show that $e^z$ maps the strip $S = \{-\pi < \mathrm{Im}\, z \le \pi\}$ one-to-one **onto** $\mathbb{C} \setminus \{0\}$. Explain why this means an inverse function ("logarithm") can only be defined after choosing such a strip.

## 7. Solutions

1. $e^{2 + i\pi/3} = e^2\left(\frac12 + \frac{\sqrt3}{2} i\right) = \frac{e^2}{2} + \frac{e^2\sqrt3}{2} i \approx 3.6945 + 6.3991i$.
2. $|e^{-1 + 5i}| = e^{-1} \approx 0.3679$. An argument is $5$, which is not in $(-\pi, \pi]$; subtract $2\pi$: principal argument $5 - 2\pi \approx -1.2832$.
3. $\overline{e^z} = \overline{e^x(\cos y + i \sin y)} = e^x(\cos y - i \sin y) = e^x(\cos(-y) + i \sin(-y)) = e^{x - iy} = e^{\bar z}$.
4. $|-2| = 2$ and $\arg(-2) = \pi$, so $z = \ln 2 + i(\pi + 2\pi k) = \ln 2 + i\pi(2k + 1)$: a vertical column of points at $x = \ln 2 \approx 0.693$, spaced $2\pi$ apart.
5. $2z = \ln 1 + i(\pi/2 + 2\pi k) = i(\pi/2 + 2\pi k)$, so $z = i(\pi/4 + \pi k)$. Now the solutions are only $\pi$ apart, because $e^{2z}$ has period $\pi i$.
6. On $y = x$, $w = e^x e^{ix}$, so $|w| = e^\theta$ where $\theta = x$ is the angle. The image is the **logarithmic spiral** $r = e^{\theta}$, winding out counter-clockwise as $x \to \infty$ and spiralling in toward $0$ (without reaching it) as $x \to -\infty$.
7. The image is the annular sector $e^{-1} \le |w| \le 1$, $\pi/2 \le \arg w \le \pi$: the part of the ring between radii $1/e$ and $1$ in the **second quadrant**.
8. If $z = 2\pi i k$ then $e^z = \cos 2\pi k + i \sin 2\pi k = 1$. Conversely, $e^z = 1$ gives $|e^z| = e^x = 1$, so $x = 0$; then $\cos y + i\sin y = 1$, so $\cos y = 1$ and $\sin y = 0$, so $y = 2\pi k$. For the deduction: $e^z = e^w \iff e^{z - w} = 1$ (divide, allowed since $e^w \neq 0$) $\iff z - w = 2\pi i k$.
9. *Onto:* given $w \neq 0$, take $z = \ln|w| + i\,\mathrm{Arg}\, w$; since $\mathrm{Arg}\, w \in (-\pi, \pi]$, $z \in S$ and $e^z = w$. *One-to-one:* if $z_1, z_2 \in S$ with $e^{z_1} = e^{z_2}$, then $z_1 - z_2 = 2\pi i k$ (Exercise 8); but the imaginary parts both lie in an interval of length $2\pi$ that is open at one end, so their difference is strictly less than $2\pi$ in size, forcing $k = 0$. On all of $\mathbb{C}$ the map is infinitely-many-to-one, so "the" $z$ with $e^z = w$ is undefined until we restrict to one strip; that choice is a **branch** of the logarithm (notebook 03).

## 8. Conclusion

- $e^{x + iy} = e^x(\cos y + i \sin y)$: the real part sets the modulus $e^x$, the imaginary part sets the angle $y$.
- The laws $e^{z + w} = e^z e^w$, $\overline{e^z} = e^{\bar z}$ and $(e^z)' = e^z$ all survive from the real case.
- $e^z \neq 0$ for every $z$; its range is $\mathbb{C} \setminus \{0\}$.
- $e^z$ has period $2\pi i$: $e^z = e^w \iff z - w \in 2\pi i\mathbb{Z}$. So $e^z = w$ has infinitely many solutions $\ln|w| + i(\arg w + 2\pi k)$.
- As a map, $e^z$ turns vertical lines into circles and horizontal lines into rays: the Cartesian grid becomes the polar grid, and each horizontal strip of height $2\pi$ covers $\mathbb{C} \setminus \{0\}$ exactly once.
