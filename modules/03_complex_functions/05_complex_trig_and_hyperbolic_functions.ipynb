# Complex Trig and Hyperbolic Functions

Euler's formula expresses $\cos\theta$ and $\sin\theta$ through $e^{i\theta}$ and $e^{-i\theta}$. Since $e^z$ now makes sense for every complex $z$, those formulas **define** $\cos z$ and $\sin z$ on the whole plane. This notebook explores what happens: every familiar identity survives, but the most familiar inequality, $|\sin x| \le 1$, does not. Along the imaginary axis sine and cosine turn into the hyperbolic functions and grow exponentially.

We will derive the real and imaginary parts of $\sin(x + iy)$, find all the zeros (there are no new ones), solve equations like $\cos z = 2$ that are impossible over the reals, and see $\sin z$ as a map that turns a grid into confocal ellipses and hyperbolas.

This is notebook 05 of module 03 (**Complex functions**) in the **Learning Complex Numbers** series.

## 1. Introduction

On the real line, $\sin$ and $\cos$ are bounded periodic functions, and $\cosh$ and $\sinh$ are unbounded non-periodic ones. They look like completely different families. In the complex plane they are **the same functions viewed along different directions**:

$$
\cos(iy) = \cosh y, \qquad \sin(iy) = i\sinh y.
$$

Turn the input by $90°$ and a cosine becomes a hyperbolic cosine. This one observation explains a lot:

- why hyperbolic identities look exactly like trig identities with a few sign changes (Osborn's rule);
- why the equation $\cos z = 2$ has solutions — they sit off the real axis, at $z = 2\pi k \pm i\ln(2 + \sqrt3)$;
- why waves in lossy media, optical fibres and transmission lines are written with $\sin$ and $\cos$ of complex arguments: the real part of the argument oscillates, the imaginary part decays or grows.

## 2. Theory

### 2.1 Definitions

For every $z \in \mathbb{C}$:

$$
\cos z = \frac{e^{iz} + e^{-iz}}{2}, \qquad \sin z = \frac{e^{iz} - e^{-iz}}{2i}, \qquad
\cosh z = \frac{e^{z} + e^{-z}}{2}, \qquad \sinh z = \frac{e^{z} - e^{-z}}{2},
$$

and $\tan z = \sin z / \cos z$, $\tanh z = \sinh z / \cosh z$ wherever the denominators are nonzero. For real $z = x$, Euler's formula shows the first two agree with the usual $\cos x$ and $\sin x$, and the last two with the real hyperbolic functions. The power series also carry over:

$$
\sin z = z - \frac{z^3}{3!} + \frac{z^5}{5!} - \cdots, \qquad \cos z = 1 - \frac{z^2}{2!} + \frac{z^4}{4!} - \cdots
$$

### 2.2 Trig and hyperbolic are the same functions, rotated

Substituting $iz$ into the definitions:

$$
\cos(iz) = \cosh z, \qquad \sin(iz) = i\sinh z, \qquad \cosh(iz) = \cos z, \qquad \sinh(iz) = i\sin z.
$$

In particular, on the imaginary axis, $\cos(iy) = \cosh y$ is real and $\ge 1$, and $\sin(iy) = i\sinh y$ is purely imaginary.

### 2.3 Real and imaginary parts

Using the addition formula (which still holds, §2.6) with $z = x + iy$:

$$
\sin(x + iy) = \sin x \cos(iy) + \cos x \sin(iy) = \sin x \cosh y + i \cos x \sinh y,
$$

$$
\cos(x + iy) = \cos x \cos(iy) - \sin x \sin(iy) = \cos x \cosh y - i \sin x \sinh y.
$$

Taking moduli and using $\cosh^2 y = 1 + \sinh^2 y$:

$$
|\sin z|^2 = \sin^2 x + \sinh^2 y, \qquad |\cos z|^2 = \cos^2 x + \sinh^2 y.
$$

### 2.4 Unbounded

From §2.3, $|\sin z| \ge |\sinh y|$, and $\sinh y \approx \frac12 e^{|y|}$ for large $|y|$. So

$$
|\sin(x + iy)| \to \infty \quad \text{as } |y| \to \infty,
$$

and the same for $\cos$. The bound $|\sin x| \le 1$ holds only on the real line (and $|\sin z| \le \cosh y$ in general). Complex sine and cosine are **unbounded** — and module 06 will show (Liouville's theorem) that every non-constant function that is differentiable on the whole plane must be unbounded.

### 2.5 Zeros

$\sin z = 0 \iff |\sin z|^2 = \sin^2 x + \sinh^2 y = 0 \iff \sin x = 0 \text{ and } \sinh y = 0 \iff x = k\pi,\ y = 0.$ Likewise for $\cos$:

$$
\sin z = 0 \iff z = k\pi, \qquad \cos z = 0 \iff z = \tfrac{\pi}{2} + k\pi, \qquad k \in \mathbb{Z}.
$$

**No new zeros** appear off the real axis. By §2.2 the hyperbolic functions have their zeros on the imaginary axis: $\sinh z = 0 \iff z = ik\pi$, $\cosh z = 0 \iff z = i(\frac\pi2 + k\pi)$. So $\tan z$ has poles only at $\frac\pi2 + k\pi$.

### 2.6 Identities survive

Every polynomial identity in $\sin$ and $\cos$ that is true for all real $x$ remains true for all complex $z$:

$$
\sin^2 z + \cos^2 z = 1, \qquad \sin(z + w) = \sin z \cos w + \cos z \sin w, \qquad \cos 2z = \cos^2 z - \sin^2 z, \qquad \frac{d}{dz}\sin z = \cos z, \ \ldots
$$

Two ways to see it: (1) direct verification, since each identity becomes an algebraic identity in $e^{iz}$ and $e^{iw}$; (2) the *identity theorem* of module 06: two complex-differentiable functions that agree on a line segment agree everywhere.

Careful: $\sin^2 z + \cos^2 z = 1$ involves $\sin^2 z$, **not** $|\sin z|^2$, so it does not imply $|\sin z| \le 1$. For $z = i$: $\sin^2 i = (i\sinh 1)^2 = -\sinh^2 1$ and $\cos^2 i = \cosh^2 1$, which sum to $1$ while $|\cos i| \approx 1.54$.

### 2.7 Periodicity

$\sin$ and $\cos$ have period $2\pi$ (real); $\sinh$ and $\cosh$ have period $2\pi i$ (imaginary) — the period rotates with the function, by §2.2. There are no other periods.

### 2.8 Solving $\cos z = w$ and $\sin z = w$

Put $u = e^{iz}$. Then $\cos z = w$ becomes $\frac{u + 1/u}{2} = w$, i.e. the quadratic $u^2 - 2wu + 1 = 0$, so $u = w \pm \sqrt{w^2 - 1}$, and

$$
z = -i\log\big(w \pm \sqrt{w^2 - 1}\big).
$$

Every complex $w$ is attained: $\sin$ and $\cos$ map $\mathbb{C}$ **onto** $\mathbb{C}$.

### 2.9 Sine as a map

With $w = u + iv = \sin z$, §2.3 gives $u = \sin x\cosh y$, $v = \cos x \sinh y$.

- A horizontal line $y = b \neq 0$ maps onto the **ellipse** $\dfrac{u^2}{\cosh^2 b} + \dfrac{v^2}{\sinh^2 b} = 1$.
- A vertical line $x = a$ (with $\sin a, \cos a \neq 0$) maps onto a branch of the **hyperbola** $\dfrac{u^2}{\sin^2 a} - \dfrac{v^2}{\cos^2 a} = 1$.

Since $\cosh^2 b - \sinh^2 b = 1 = \sin^2 a + \cos^2 a$, all these conics share the **foci $\pm 1$**. The vertical strip $-\pi/2 \le x \le \pi/2$ is mapped onto the whole plane.

## 3. Worked Examples

### Example 1 — $\cos i$ and $\sin i$

$\cos i = \dfrac{e^{i \cdot i} + e^{-i \cdot i}}{2} = \dfrac{e^{-1} + e}{2} = \cosh 1 \approx 1.5431$ — a real number bigger than $1$.

$\sin i = \dfrac{e^{-1} - e}{2i} = \dfrac{e - e^{-1}}{2}\, i = i\sinh 1 \approx 1.1752i$.

### Example 2 — $\sin(\frac\pi2 + i)$ and $\cos(\frac\pi2 + i)$

By §2.3 with $x = \pi/2$, $y = 1$: $\sin(\frac\pi2 + i) = 1 \cdot \cosh 1 + i \cdot 0 = \cosh 1$, and $\cos(\frac\pi2 + i) = 0 - i \cdot 1 \cdot \sinh 1 = -i\sinh 1$.

### Example 3 — the modulus of $\sin(3 + 4i)$

$|\sin(3 + 4i)|^2 = \sin^2 3 + \sinh^2 4 \approx 0.0199 + 744.0 = 744.03$, so $|\sin(3 + 4i)| \approx 27.28$. Twenty-seven times larger than any real sine.

### Example 4 — solving $\cos z = 2$

$u = e^{iz}$ satisfies $u^2 - 4u + 1 = 0$, so $u = 2 \pm \sqrt3$. Both are positive reals, with $\ln(2 + \sqrt3) = -\ln(2 - \sqrt3)$ because $(2 + \sqrt3)(2 - \sqrt3) = 1$. Then $iz = \log u = \pm\ln(2 + \sqrt3) + 2\pi i k$:

$$
z = 2\pi k \mp i\ln(2 + \sqrt3) \approx 2\pi k \mp 1.3170i, \qquad k \in \mathbb{Z}.
$$

Check with §2.3: $\cos(0 + 1.3170i) = \cosh(1.3170) = 2$ ✓. Geometrically: two rows of solutions, above and below each real point $2\pi k$ where $\cos$ reaches its real maximum $1$.

### Example 5 — $\sin^2 z + \cos^2 z = 1$ at $z = 1 + i$

$\sin(1 + i) = \sin1\cosh1 + i\cos1\sinh1 \approx 1.2985 + 0.6350i$ and $\cos(1 + i) = \cos1\cosh1 - i\sin1\sinh1 \approx 0.8337 - 0.9889i$. Squaring:
$\sin^2 \approx 1.2828 + 1.6491i$, $\cos^2 \approx -0.2828 - 1.6491i$. Sum: $1$ ✓ — although $|\sin(1+i)|^2 + |\cos(1+i)|^2 \approx 3.76$.

### Example 6 — zeros of $\cosh z$

$\cosh z = \cos(iz)$ vanishes when $iz = \frac\pi2 + k\pi$, i.e. $z = -i(\frac\pi2 + k\pi)$, which as $k$ ranges over $\mathbb{Z}$ is the set $\{i(\frac\pi2 + k\pi)\}$: points on the imaginary axis. The real function $\cosh x \ge 1$ has no zeros — they were hiding off the real line.

## 4. Python Examples

### 4.1 `cmath` agrees with the exponential definitions and the Cartesian formulas

```python
import cmath
import math


def sin_from_exponentials(z):
    return (cmath.exp(1j * z) - cmath.exp(-1j * z)) / 2j


def sin_cartesian(z):
    x, y = z.real, z.imag
    return complex(math.sin(x) * math.cosh(y), math.cos(x) * math.sinh(y))


for z in [1j, math.pi / 2 + 1j, 3 + 4j, 1 + 1j]:
    library = cmath.sin(z)
    print(f"sin({z}) = {library:.6f}   exponentials agree: {abs(sin_from_exponentials(z) - library) < 1e-12}"
          f"   Cartesian formula agrees: {abs(sin_cartesian(z) - library) < 1e-9}")

print()
print("cos(i)  =", cmath.cos(1j), "  cosh(1) =", math.cosh(1))
print("sin(i)  =", cmath.sin(1j), "  i sinh(1) =", 1j * math.sinh(1))
print("|sin(3+4i)| =", abs(cmath.sin(3 + 4j)), "  sqrt(sin^2 3 + sinh^2 4) =", math.sqrt(math.sin(3) ** 2 + math.sinh(4) ** 2))
```

### 4.2 Symbolic real and imaginary parts and identities with sympy

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
z, w = sp.symbols("z w")

print("sin(x+iy) =", sp.expand_complex(sp.sin(x + sp.I * y)))
print("cos(x+iy) =", sp.expand_complex(sp.cos(x + sp.I * y)))
modulus_squared = sp.expand_complex(sp.sin(x + sp.I * y) * sp.conjugate(sp.sin(x + sp.I * y)))
print("|sin(x+iy)|^2 =", modulus_squared)
print("   minus (sin^2 x + sinh^2 y) simplifies to", sp.simplify((modulus_squared - sp.sin(x) ** 2 - sp.sinh(y) ** 2).rewrite(sp.exp)))
print()
print("cos(I*z) rewritten:", sp.cos(sp.I * z), "   sin(I*z):", sp.sin(sp.I * z))
print("sin^2 z + cos^2 z         =", sp.simplify(sp.sin(z) ** 2 + sp.cos(z) ** 2))
print("sin(z+w) - (sin z cos w + cos z sin w) =", sp.simplify(sp.expand_trig(sp.sin(z + w)) - (sp.sin(z) * sp.cos(w) + sp.cos(z) * sp.sin(w))))
print("identity via exponentials  :", sp.simplify((sp.sin(z) ** 2 + sp.cos(z) ** 2).rewrite(sp.exp)))
print("d/dz sin z =", sp.diff(sp.sin(z), z), "   d/dz cosh z =", sp.diff(sp.cosh(z), z))
```

### 4.4 Solving $\cos z = 2$ and $\sin z = 2$

```python
import cmath
import math

import sympy as sp

# formula from section 2.8: z = -i log(w +- sqrt(w^2 - 1)), plus the 2 pi k from the many-valued log
w = 2
for sign in (+1, -1):
    u = w + sign * cmath.sqrt(w * w - 1)
    for k in range(-1, 2):
        z = -1j * cmath.log(u) + 2 * math.pi * k
        print(f"cos z = 2:  z = {z.real:+.4f} {z.imag:+.4f}i   cos z = {cmath.cos(z):.10f}")

print("ln(2 + sqrt 3) =", math.log(2 + math.sqrt(3)))
print("cmath.acos(2) =", cmath.acos(2), "   cmath.asin(2) =", cmath.asin(2), "   sin(asin 2) =", cmath.sin(cmath.asin(2)))

z = sp.symbols("z")
print("sympy solve cos z = 2:", sp.solve(sp.cos(z) - 2, z))
```

### 4.5 The only zeros are real

Scan a grid for small values of $|\sin z|$ and $|\cos z|$. Every near-zero lies on the real axis at $k\pi$ (resp. $\frac\pi2 + k\pi$).

```python
import numpy as np

x = np.linspace(-7, 7, 1401)
y = np.linspace(-3, 3, 601)
X, Y = np.meshgrid(x, y)
Z = X + 1j * Y

for name, values in [("sin", np.sin(Z)), ("cos", np.cos(Z))]:
    small = np.abs(values) < 0.02
    near_zeros = Z[small]
    print(f"{name}: {small.sum()} grid points with |{name} z| < 0.02;  max |Im z| among them = {np.abs(near_zeros.imag).max():.3f}")
    print(f"     real parts cluster near multiples of pi/2: {np.unique(np.round(near_zeros.real / (np.pi / 2)))}")
```

## 5. Visualizations

### 5.1 The landscape $|\sin z|$

Contours of $|\sin z|$ over $-2\pi \le x \le 2\pi$, $-2 \le y \le 2$. Along the real axis (dashed) the modulus never exceeds $1$; moving vertically it grows like $\frac12 e^{|y|}$. The only zeros (white dots) are at $k\pi$.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-2 * np.pi, 2 * np.pi, 600)
y = np.linspace(-2, 2, 300)
X, Y = np.meshgrid(x, y)
modulus = np.abs(np.sin(X + 1j * Y))

fig, ax = plt.subplots(figsize=(12, 5))
filled = ax.contourf(X, Y, modulus, levels=np.linspace(0, 3.8, 20), cmap="magma", extend="max")
ax.contour(X, Y, modulus, levels=[1], colors="cyan", linewidths=1.2)
fig.colorbar(filled, ax=ax, label="|sin z|")
zeros = np.pi * np.arange(-2, 3)
ax.plot(zeros, np.zeros_like(zeros), "o", color="white", markersize=7, label="zeros $k\\pi$")
ax.axhline(0, color="white", linestyle="--", linewidth=0.8)
ax.set_xticks(np.pi * np.arange(-2, 3))
ax.set_xticklabels(["$-2\\pi$", "$-\\pi$", "0", "$\\pi$", "$2\\pi$"])
ax.set_aspect("equal")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_title("$|\\sin z|$ (cyan: the level curve $|\\sin z| = 1$)")
ax.legend(loc="upper right")
plt.show()
```

The cyan level curve $|\sin z| = 1$ shows that the region where sine is "as small as on the real line" is a thin band around the real axis that pinches to nothing at $\frac\pi2 + k\pi$, where the real sine already equals $\pm1$.

### 5.2 Bounded on the real axis, exponential off it

$|\sin(x + iy)|$ along horizontal lines $y = 0, 0.5, 1, 2$. Raising the line lifts the whole curve: its minimum is $\sinh|y|$, its maximum $\cosh y$.

### 5.3 Sine as a map: ellipses and hyperbolas

The strip $-\pi/2 \le x \le \pi/2$, $-1.5 \le y \le 1.5$ under $w = \sin z$. Horizontal lines (orange) become confocal ellipses, vertical lines (blue) become confocal hyperbolas; the common foci $\pm 1$ are marked. The real segment $[-\pi/2, \pi/2]$ is squashed onto $[-1, 1]$, but everything above and below it spreads over the whole plane.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, (ax_z, ax_w) = plt.subplots(1, 2, figsize=(13, 6), gridspec_kw={"width_ratios": [1, 1.4]})
x_values = np.linspace(-np.pi / 2, np.pi / 2, 9)
y_values = np.linspace(-1.5, 1.5, 7)
t_vertical = np.linspace(-1.5, 1.5, 300)
t_horizontal = np.linspace(-np.pi / 2, np.pi / 2, 300)

for a in x_values:
    line = a + 1j * t_vertical
    ax_z.plot(line.real, line.imag, color="tab:blue", linewidth=1)
    ax_w.plot(np.sin(line).real, np.sin(line).imag, color="tab:blue", linewidth=1)
for b in y_values:
    line = t_horizontal + 1j * b
    ax_z.plot(line.real, line.imag, color="tab:orange", linewidth=1)
    image = np.sin(line)
    ax_w.plot(image.real, image.imag, color="tab:orange", linewidth=1)

ax_w.plot([-1, 1], [0, 0], "ko", markersize=7, label="foci $\\pm 1$")
marker = 0.5 + 1j
ax_z.plot(marker.real, marker.imag, "rs")
ax_w.plot(np.sin(marker).real, np.sin(marker).imag, "rs", label="$\\sin(0.5 + i)$")
ax_z.set_title("z-plane: $|x| \\leq \\pi/2$, $|y| \\leq 1.5$")
ax_w.set_title("w = sin z: confocal ellipses and hyperbolas")
ax_w.legend(loc="upper right")
for ax in (ax_z, ax_w):
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

Each horizontal segment covers only **half** an ellipse: $y = b > 0$ gives the upper half (since $v = \cos x \sinh b \ge 0$ for $|x| \le \pi/2$) and $y = -b$ the lower half. The red square $0.5 + i$ lands on the ellipse of $y = 1$ and on the hyperbola of $x = 0.5$.

### 5.4 Domain colouring: $\sin z$, $\cos z$, $\tan z$

Zeros of $\sin$ at $k\pi$ and of $\cos$ at $\frac\pi2 + k\pi$ are the black dots on the real axis; nowhere else is black. Above and below, the pictures wash out to white (unbounded). $\tan z$ has zeros at $k\pi$ and poles (white dots, reversed colours) at $\frac\pi2 + k\pi$, and tends to $\pm i$ far from the real axis — the uniform colours at top and bottom.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import hsv_to_rgb


def domain_colouring(f, re_range, im_range, resolution=500):
    X, Y = np.meshgrid(np.linspace(*re_range, 2 * resolution), np.linspace(*im_range, resolution))
    with np.errstate(divide="ignore", invalid="ignore"):
        W = f(X + 1j * Y)
    hue = (np.angle(W) / (2 * np.pi)) % 1.0
    m = (2 / np.pi) * np.arctan(np.abs(W))
    hsv = np.stack([hue, np.clip(2 - 2 * m, 0, 1), np.clip(2 * m, 0, 1)], axis=-1)
    return hsv_to_rgb(hsv)


re_range, im_range = (-2 * np.pi, 2 * np.pi), (-2.5, 2.5)
fig, axes = plt.subplots(3, 1, figsize=(11, 13))
for ax, f, title in [(axes[0], np.sin, "$\\sin z$: zeros at $k\\pi$ only"),
                     (axes[1], np.cos, "$\\cos z$: zeros at $\\pi/2 + k\\pi$ only"),
                     (axes[2], np.tan, "$\\tan z$: zeros at $k\\pi$, poles at $\\pi/2 + k\\pi$, $\\to \\pm i$ away from the axis")]:
    ax.imshow(domain_colouring(f, re_range, im_range), origin="lower", extent=[*re_range, *im_range])
    ax.set_xticks(np.pi * np.arange(-2, 2.5, 0.5))
    ax.set_xticklabels(["$-2\\pi$", "", "$-\\pi$", "", "0", "", "$\\pi$", "", "$2\\pi$"])
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.set_title(title)
fig.tight_layout()
plt.show()
```

### 5.5 cos and cosh: one function, two directions

Plot $\cos$ along the real axis and along the imaginary axis on the same axes. Along the real axis it is the bounded cosine; along the imaginary axis it is real-valued again and equals $\cosh$.

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(-3, 3, 400)
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(t, np.cos(t + 0j).real, label="$\\cos(t)$  (along the real axis)")
ax.plot(t, np.cos(1j * t).real, label="$\\cos(it)$  (along the imaginary axis)")
ax.plot(t, np.cosh(t), "k:", linewidth=2.5, label="$\\cosh t$")
print("Im cos(it) is zero:", np.allclose(np.cos(1j * t).imag, 0), "  cos(it) = cosh t:", np.allclose(np.cos(1j * t), np.cosh(t)))
ax.axhline(0, color="black", linewidth=0.8)
ax.set_xlabel("t")
ax.set_ylabel("value (real in both cases)")
ax.set_title("$\\cos(it) = \\cosh t$")
ax.grid(True, alpha=0.3)
ax.legend()
plt.show()
```

## 7. Solutions

1. $\sin(1 + i) = \sin1\cosh1 + i\cos1\sinh1 \approx 1.2985 + 0.6350i$. $\cos(2i) = \cosh 2 \approx 3.7622$.
2. $\cosh(iy) = \frac{e^{iy} + e^{-iy}}{2} = \cos y$ by Euler. $\sinh(iy) = \frac{e^{iy} - e^{-iy}}{2} = i \cdot \frac{e^{iy} - e^{-iy}}{2i} = i\sin y$.
3. $|\cos(2 + 3i)|^2 = \cos^2 2 + \sinh^2 3 \approx 0.1732 + 100.358 = 100.531$, so $|\cos(2 + 3i)| \approx 10.027$.
4. $u = e^{iz}$: $\frac{u - 1/u}{2i} = 2 \Rightarrow u^2 - 4iu - 1 = 0 \Rightarrow u = i(2 \pm \sqrt3)$. Then $iz = \log u = \ln(2 \pm \sqrt3) + i(\frac\pi2 + 2\pi k)$, so $z = \frac\pi2 + 2\pi k - i\ln(2 \pm \sqrt3) = \frac\pi2 + 2\pi k \pm i\ln(2 + \sqrt3) \approx \frac\pi2 + 2\pi k \pm 1.3170i$. (Above and below each real maximum $\frac\pi2 + 2\pi k$ of the sine.)
5. $\sinh z = -i\sin(iz) = 0 \iff iz = k\pi \iff z = -ik\pi$, i.e. $z \in i\pi\mathbb{Z}$. $\cosh z = \cos(iz) = 0 \iff z \in i(\frac\pi2 + \pi\mathbb{Z})$. $\sin z - \cos z = \sqrt2\sin(z - \frac\pi4)$ (the real identity survives), which is $0$ iff $z = \frac\pi4 + k\pi$: real zeros only.
6. $\cos(x + i) = \cos x\cosh 1 - i\sin x\sinh 1$: $u = \cosh1\cos x$, $v = -\sinh1\sin x$, the ellipse $\frac{u^2}{\cosh^2 1} + \frac{v^2}{\sinh^2 1} = 1$ with semi-axes $\cosh 1 \approx 1.543$ (horizontal) and $\sinh 1 \approx 1.175$ (vertical), foci $\pm1$. It crosses the real axis at $\pm\cosh1$ (for $x = 0, \pm\pi$) and the imaginary axis at $\mp i\sinh1$ (for $x = \pm\pi/2$). Starting at $x = -\pi$ ($w = -\cosh 1$), $v = -\sinh 1\sin x$ is positive for $-\pi < x < 0$: the curve runs through the upper half to $\cosh 1$, then back through the lower half — **clockwise**, once.
7. $\sin z\cos w + \cos z\sin w = \frac{(e^{iz} - e^{-iz})(e^{iw} + e^{-iw}) + (e^{iz} + e^{-iz})(e^{iw} - e^{-iw})}{4i}$. Expanding, the mixed terms $e^{i(z - w)}$ and $e^{-i(z - w)}$ cancel, leaving $\frac{2e^{i(z + w)} - 2e^{-i(z + w)}}{4i} = \sin(z + w)$.
8. $|\sin z|^2 = \sin^2 x + \sinh^2 y$ with $0 \le \sin^2 x \le 1$ gives $\sinh^2 y \le |\sin z|^2 \le 1 + \sinh^2 y = \cosh^2 y$. Lower bound attained where $\sin x = 0$ ($x = k\pi$), upper where $\sin^2 x = 1$ ($x = \frac\pi2 + k\pi$).
9. $\sin z = \sin w \iff \cos\frac{z + w}{2}\sin\frac{z - w}{2} = 0$. By §2.5 (only real zeros), $\sin\frac{z - w}{2} = 0 \iff \frac{z - w}{2} = k\pi \iff w = z - 2k\pi$, and $\cos\frac{z + w}{2} = 0 \iff \frac{z + w}{2} = \frac\pi2 + k\pi \iff w = \pi - z + 2k\pi$. Replacing $k$ by $-k$ in the first case gives the stated form. This is the same as for real numbers — no new solutions appear, which is a direct consequence of "no new zeros".

## 8. Conclusion

- $\cos z = \frac{e^{iz} + e^{-iz}}{2}$ and $\sin z = \frac{e^{iz} - e^{-iz}}{2i}$ extend the real functions to $\mathbb{C}$; the hyperbolic functions are the same functions rotated: $\cos(iz) = \cosh z$, $\sin(iz) = i\sinh z$.
- $\sin(x + iy) = \sin x\cosh y + i\cos x\sinh y$, $\cos(x + iy) = \cos x\cosh y - i\sin x\sinh y$, and $|\sin z|^2 = \sin^2 x + \sinh^2 y$.
- Complex $\sin$ and $\cos$ are **unbounded** (they grow like $\frac12 e^{|y|}$) and **onto** $\mathbb{C}$: $\cos z = 2$ has the solutions $2\pi k \pm i\ln(2 + \sqrt3)$.
- Their zeros are only the real ones; $\sinh$ and $\cosh$ have their zeros on the imaginary axis.
- All trig identities survive ($\sin^2 z + \cos^2 z = 1$, addition formulas, derivatives), but $\sin^2 z$ is not $|\sin z|^2$, so no bound follows.
- As a map, $\sin$ sends horizontal lines to confocal ellipses and vertical lines to confocal hyperbolas with foci $\pm1$.
