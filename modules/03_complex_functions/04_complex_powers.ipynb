# Complex Powers

What is $i^i$? Or $2^i$, or $(-8)^{1/3}$? For whole-number exponents, $z^n$ is just repeated multiplication. For anything else we need a definition, and the only good one goes through the logarithm:

$$
z^w = e^{w \log z}.
$$

Because $\log z$ has infinitely many values, so — in general — does $z^w$. This notebook works out how many values there are (one, finitely many, or infinitely many, depending on $w$), what the **principal value** is and what Python returns, and why exponent laws like $(z^a)^b = z^{ab}$ and $(zw)^a = z^a w^a$ break down.

This is notebook 04 of module 03 (**Complex functions**) in the **Learning Complex Numbers** series.

## 1. Introduction

In real calculus $a^x$ is defined for $a > 0$ as $e^{x \ln a}$, and $(-8)^{1/3}$ is either $-2$ or "not allowed", depending on the textbook. Complex numbers remove the restrictions — any nonzero base, any complex exponent — but the answer becomes a **set**:

- $(-8)^{1/3}$ has three values, $-2$ and $1 \pm i\sqrt3$. Python's `(-8) ** (1/3)` returns $1 + i\sqrt3$, not $-2$.
- $i^i$ has infinitely many values, and every one of them is a **real** number: $e^{-\pi/2} \approx 0.2079$ is the principal one.
- The familiar "proof" $1 = \sqrt{1} = \sqrt{(-1)(-1)} = \sqrt{-1}\sqrt{-1} = i \cdot i = -1$ fails precisely because $(zw)^{1/2} \neq z^{1/2} w^{1/2}$ for principal values.

Complex powers appear in fractional-order calculus, in solutions $x^{\lambda}$ of Euler–Cauchy differential equations with complex $\lambda$, in signal processing (fractional delays, $z^{-1/2}$), and in scaling laws in physics.

## 2. Theory

### 2.1 Definition

For $z \neq 0$ and any $w \in \mathbb{C}$:

$$
z^w = e^{w \log z} = \left\{ e^{w(\mathrm{Log}\, z + 2\pi i k)} : k \in \mathbb{Z} \right\} = \left\{ e^{w\,\mathrm{Log}\, z}\, e^{2\pi i k w} : k \in \mathbb{Z} \right\}.
$$

The **principal value** uses the principal logarithm ($k = 0$):

$$
\boxed{\ z^w = e^{w\,\mathrm{Log}\, z} \quad\text{(principal value)}\ }
$$

This is what Python's `**`, `cmath` and numpy compute for complex operands. Every value differs from the principal one by a factor $e^{2\pi i k w}$.

> **Convention for $e^w$.** Applied literally to the base $e$, the definition would give the set $e^{w(1 + 2\pi i k)}$. By universal convention, $e^w$ always means the single number $\exp(w)$ from notebook 02.

### 2.2 How many values?

Everything depends on the factors $e^{2\pi i k w}$, $k \in \mathbb{Z}$:

| exponent $w$ | distinct values of $e^{2\pi i k w}$ | number of values of $z^w$ |
|---|---|---|
| integer $n$ | $e^{2\pi i k n} = 1$ for all $k$ | **one**: $z^n$ is single-valued |
| rational $p/q$ in lowest terms ($q \ge 1$) | the $q$-th roots of unity $e^{2\pi i k p/q}$ | **exactly $q$** |
| irrational real | $e^{2\pi i k w}$ all different, dense in the unit circle | **infinitely many**, all of the same modulus |
| non-real | moduli $e^{-2\pi k\,\mathrm{Im}\, w}$ all different | **infinitely many**, on a logarithmic spiral |

For $w = p/q$ the factor $e^{2\pi i k p/q}$ depends only on $k \bmod q$, and as $k$ runs over $0, \ldots, q - 1$ it hits all $q$ different $q$-th roots of unity (because $\gcd(p, q) = 1$).

### 2.3 Roots: $z^{1/n}$

With $z = re^{i\theta}$, $\theta = \mathrm{Arg}\, z$:

$$
z^{1/n} = r^{1/n} e^{i(\theta + 2\pi k)/n}, \qquad k = 0, 1, \ldots, n - 1.
$$

These are the $n$ vertices of a **regular $n$-gon** of radius $r^{1/n}$ centred at $0$. The principal value ($k = 0$) has argument $\theta/n \in (-\pi/n, \pi/n]$. For $n = 2$ this is the principal square root, always in the right half-plane (or on the positive imaginary axis).

Consequence: the principal cube root of $-8$ is $2e^{i\pi/3} = 1 + i\sqrt3$, not $-2$. The real cube root $-2$ is the $k = 1$ value. (`numpy.cbrt` is a *real* function and returns $-2$.)

More generally $z^{p/q} = (z^{1/q})^p$: take the $q$ values of the $q$-th root and raise each to the integer power $p$.

### 2.4 Why $i^i$ is real

$\log i = i\left(\frac\pi2 + 2\pi k\right)$, so

$$
i^i = e^{i \cdot i(\pi/2 + 2\pi k)} = e^{-\pi/2 - 2\pi k}, \qquad k \in \mathbb{Z}.
$$

Every value is a positive real number: $\ldots, e^{3\pi/2}, e^{-\pi/2}, e^{-5\pi/2}, \ldots$ The principal value is $e^{-\pi/2} \approx 0.20788$. In the plane, the values sit on the positive real axis, spaced by a constant **ratio** $e^{2\pi} \approx 535$.

### 2.5 Real base, imaginary exponent

For a positive real $a$, the principal value is $a^{iy} = e^{iy\ln a} = \cos(y\ln a) + i\sin(y\ln a)$, a point on the **unit circle**. For example $2^i = \cos(\ln 2) + i\sin(\ln 2) \approx 0.769 + 0.639i$. An imaginary exponent rotates, just as for the base $e$.

### 2.6 Which exponent laws survive?

Let all powers be principal values.

| law | status |
|---|---|
| $z^a z^b = z^{a + b}$ | **true**: both sides are $e^{(a + b)\mathrm{Log}\, z}$ |
| $(z^a)^n = z^{an}$ for an integer $n$ | **true**: $(e^u)^n = e^{nu}$ |
| $(z^a)^b = z^{ab}$ | **false** in general |
| $(zw)^a = z^a w^a$ | **false** in general |
| $z^{-a} = 1/z^a$ | **true** |

Counterexamples:

- $\big((-1)^2\big)^{1/2} = 1^{1/2} = 1$, but $(-1)^{2 \cdot \frac12} = (-1)^1 = -1$.
- $\big(e^{2\pi i}\big)^i = 1^i = e^{i\,\mathrm{Log}\, 1} = 1$, but $e^{2\pi i \cdot i} = e^{-2\pi} \approx 0.0019$.
- $(-1)^{1/2}(-1)^{1/2} = i \cdot i = -1$, but $\big((-1)(-1)\big)^{1/2} = 1^{1/2} = 1$.

The root cause is always the same: $\mathrm{Log}(z^a) \neq a\,\mathrm{Log}\, z$ and $\mathrm{Log}(zw) \neq \mathrm{Log}\, z + \mathrm{Log}\, w$ (notebook 03, §2.6). The laws are off by factors $e^{2\pi i k a}$ — which equal $1$ exactly when $a$ is an integer. As identities between **sets** of values, the laws are restored up to such factors.

### 2.7 The branch cut of $z^w$

The principal value $e^{w\,\mathrm{Log}\, z}$ inherits the cut of $\mathrm{Log}$: crossing the negative real axis multiplies $z^w$ by $e^{\pm 2\pi i w}$. For $w = 1/2$ that factor is $-1$: the principal square root flips sign across the cut, for instance $\sqrt{-4 + 0.001i} \approx 2i$ but $\sqrt{-4 - 0.001i} \approx -2i$. For integer $w$ the factor is $1$ and there is no cut.

## 3. Worked Examples

### Example 1 — all values of $i^i$

From §2.4: $i^i = e^{-\pi/2 - 2\pi k}$. A few of them: $k = 0$: $0.20788$; $k = -1$: $e^{3\pi/2} \approx 111.32$; $k = 1$: $e^{-5\pi/2} \approx 0.000388$.

### Example 2 — the principal value of $(1 + i)^i$

$\mathrm{Log}(1 + i) = \frac12\ln2 + i\frac\pi4$, so

$$
(1 + i)^i = e^{i(\frac12\ln2 + i\frac\pi4)} = e^{-\pi/4}\, e^{i\frac12\ln2} = e^{-\pi/4}\left(\cos\tfrac{\ln2}{2} + i\sin\tfrac{\ln2}{2}\right) \approx 0.4288 + 0.1549i.
$$

Geometrically: modulus $e^{-\pi/4} \approx 0.456$, argument $\frac12\ln2 \approx 0.347$ rad. The real part of $\mathrm{Log}$ became an angle and the angle became a modulus — multiplying by $i$ swaps their roles.

### Example 3 — $2^i$

$2^i = e^{i\ln2} = \cos(\ln2) + i\sin(\ln2) \approx 0.7692 + 0.6390i$, on the unit circle at angle $\ln2 \approx 39.7°$. All values: $e^{i(\ln2 + 2\pi i k)} = e^{-2\pi k}\,e^{i\ln2}$, on the ray at that angle.

### Example 4 — the three cube roots of $-8$

$-8 = 8e^{i\pi}$, so $(-8)^{1/3} = 2e^{i(\pi + 2\pi k)/3}$:

- $k = 0$: $2e^{i\pi/3} = 1 + i\sqrt3$ (principal),
- $k = 1$: $2e^{i\pi} = -2$,
- $k = 2$: $2e^{i5\pi/3} = 1 - i\sqrt3$.

An equilateral triangle of radius $2$.

### Example 5 — $(-8)^{2/3}$

Square the three cube roots: $(1 + i\sqrt3)^2 = -2 + 2i\sqrt3$, $(-2)^2 = 4$, $(1 - i\sqrt3)^2 = -2 - 2i\sqrt3$. These are $4e^{2\pi i/3}$, $4$, $4e^{-2\pi i/3}$ — the three cube roots of $64$. The principal value is $e^{\frac23\mathrm{Log}(-8)} = 4e^{2\pi i/3} = -2 + 2i\sqrt3$. A real-number textbook would say $(-8)^{2/3} = 4$; that is one of the three complex values but **not** the principal one.

### Example 6 — $(z^a)^b \neq z^{ab}$ with $z = -1 + i$, $a = 2$, $b = \frac12$

$z^2 = (-1 + i)^2 = -2i$ and $(z^2)^{1/2} = \sqrt{-2i} = 1 - i$ (principal: argument $-\pi/4$). But $z^{2 \cdot \frac12} = z = -1 + i$. The two answers differ by the factor $-1$: $\sqrt{z^2}$ is always $\pm z$, and the principal root picks the sign that lies in the right half-plane.

## 4. Python Examples

### 4.1 Python's `**` computes principal values

```python
import cmath
import math

print("1j ** 1j        =", 1j ** 1j, "   e^(-pi/2) =", math.exp(-math.pi / 2))
print("(1+1j) ** 1j    =", (1 + 1j) ** 1j)
print("2 ** 1j         =", 2 ** 1j, "   |2^i| =", abs(2 ** 1j))
print("(-8) ** (1/3)   =", (-8) ** (1 / 3), "   (principal, not -2)")
print("(-8) ** (2/3)   =", (-8) ** (2 / 3), "   (-2 + 2 sqrt(3) i)")

z, w = 1 + 1j, 1j
print("e^(w Log z)     =", cmath.exp(w * cmath.log(z)), "  matches z**w:", abs(cmath.exp(w * cmath.log(z)) - z**w) < 1e-12)
```

numpy agrees for complex arrays, but numpy's real functions behave differently: `np.cbrt` is the real cube root, and a fractional power of a negative **float** array is `nan` (with a warning) rather than a complex number.

```python
import numpy as np

print("np.power(-8+0j, 1/3) =", np.power(-8 + 0j, 1 / 3))
print("np.cbrt(-8)          =", np.cbrt(-8.0), "   (real cube root)")
with np.errstate(invalid="ignore"):
    print("np.power(-8.0, 1/3)  =", np.power(-8.0, 1 / 3), "   (real arithmetic: no real answer)")
```

### 4.2 All values of $z^w$

`all_powers(z, w, ks)` returns $e^{w(\mathrm{Log}\, z + 2\pi i k)}$ for each $k$; `distinct` removes numerical duplicates so we can count values.

```python
import cmath
import math


def all_powers(z, w, ks=range(-6, 7)):
    return [cmath.exp(w * (cmath.log(z) + 2j * math.pi * k)) for k in ks]


def distinct(values, tolerance=1e-9):
    unique = []
    for v in values:
        if all(abs(v - u) > tolerance * max(1, abs(u)) for u in unique):
            unique.append(v)
    return unique


z = -8
for label, w in [("3", 3), ("1/3", 1 / 3), ("2/3", 2 / 3), ("4/6", 4 / 6), ("5/4", 5 / 4), ("sqrt(2)", math.sqrt(2))]:
    values = distinct(all_powers(z, w, ks=range(-20, 21)))
    print(f"(-8)^({label:7s}): {len(values):2d} distinct values among k = -20..20")
```

$4/6$ behaves like $2/3$ (three values): only the exponent in lowest terms matters. For $\sqrt2$ every $k$ gives a new value.

### 4.3 $i^i$: infinitely many real values

```python
import cmath
import math

for k in range(-2, 3):
    value = cmath.exp(1j * (cmath.log(1j) + 2j * math.pi * k))
    print(f"k = {k:+d}:  i^i = {value.real:.6g}  (imaginary part {value.imag:.1e}),  e^(-pi/2 - 2 pi k) = {math.exp(-math.pi / 2 - 2 * math.pi * k):.6g}")
```

### 4.4 Exact work with sympy

sympy evaluates principal values exactly, distinguishes the principal root (`sp.root`) from the real root (`sp.real_root`), and — like with logarithms — refuses to apply exponent laws that are not valid without assumptions.

```python
import sympy as sp

print("exp(I*log(I))        =", sp.exp(sp.I * sp.log(sp.I)), "  (i^i, principal)")
print("exp(I*log(2))        =", sp.expand_complex(sp.exp(sp.I * sp.log(2))), "  (2^i)")
print("root(-8, 3)          =", sp.root(-8, 3), "=", sp.expand_complex(sp.root(-8, 3)))
print("real_root(-8, 3)     =", sp.real_root(-8, 3))
x = sp.symbols("x")
print("all cube roots of -8 =", sp.solve(x**3 + 8, x))

z, a, b = sp.symbols("z a b")
p = sp.symbols("p", positive=True)
print("powdenest((z**a)**b)           :", sp.powdenest((z**a) ** b))
print("powdenest((p**a)**b)           :", sp.powdenest((p**a) ** b))
print("sqrt(z**2) for complex z       :", sp.sqrt(z**2))
print("sqrt(p**2) for positive p      :", sp.sqrt(p**2))
```

### 4.5 Testing the exponent laws numerically

```python
import numpy as np

rng = np.random.default_rng(seed=3)
z = rng.normal(size=50_000) + 1j * rng.normal(size=50_000)
w = rng.normal(size=50_000) + 1j * rng.normal(size=50_000)
a, b = 0.7 + 0.2j, 1.3 - 0.5j

checks = {
    "z^a z^b = z^(a+b)": np.isclose(z**a * z**b, z ** (a + b)),
    "(z^a)^3 = z^(3a)": np.isclose((z**a) ** 3, z ** (3 * a)),
    "(z^a)^b = z^(ab), a=0.7+0.2i": np.isclose((z**a) ** b, z ** (a * b)),
    "(z^a)^b = z^(ab), a=3, b=1/2": np.isclose((z**3) ** 0.5, z**1.5),
    "(zw)^(1/2) = z^(1/2) w^(1/2)": np.isclose(np.sqrt(z * w), np.sqrt(z) * np.sqrt(w)),
}
for law, holds in checks.items():
    print(f"{law:30s} holds for {holds.mean():6.1%} of random samples")

print()
print("((-1)^2)^(1/2) =", ((-1 + 0j) ** 2) ** 0.5, "   (-1)^1 =", (-1 + 0j) ** 1)
print("(e^(2 pi i))^i =", np.exp(2j * np.pi) ** 1j, "   e^(2 pi i * i) =", np.exp(2j * np.pi * 1j))
print("sqrt(-1)*sqrt(-1) =", np.sqrt(-1 + 0j) * np.sqrt(-1 + 0j), "   sqrt((-1)(-1)) =", np.sqrt((-1 + 0j) * (-1 + 0j)))
```

The third line is a warning about testing: with $a = 0.7 + 0.2i$ and normally distributed $z$, the number $\mathrm{Im}(a\,\mathrm{Log}\, z)$ almost never leaves $(-\pi, \pi]$, so random tests miss the failure. With $a = 3$ the law $(z^3)^{1/2} = z^{3/2}$ needs $3\,\mathrm{Arg}\, z \in (-\pi, \pi]$, which happens for exactly one third of all directions.

## 5. Visualizations

### 5.1 Roots form regular polygons

The values of $z^{1/n}$ for $z = -8$ and $n = 3, 5, 8$. The principal value (star) is the vertex with the smallest argument in absolute value — the one closest to the positive real axis, taking the counter-clockwise one in a tie.

```python
import cmath
import math

import matplotlib.pyplot as plt
import numpy as np

z = -8
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for ax, n in zip(axes, [3, 5, 8]):
    roots = np.array([abs(z) ** (1 / n) * cmath.exp(1j * (cmath.phase(z) + 2 * math.pi * k) / n) for k in range(n)])
    polygon = np.append(roots, roots[0])
    ax.plot(polygon.real, polygon.imag, "--", color="gray", linewidth=1)
    for k, root in enumerate(roots):
        ax.annotate("", xy=(root.real, root.imag), xytext=(0, 0), arrowprops=dict(arrowstyle="->", color="tab:blue"))
        ax.annotate(f"k={k}", (root.real, root.imag), xytext=(6, 4), textcoords="offset points", fontsize=9)
    principal = complex(z) ** (1 / n)
    ax.plot(principal.real, principal.imag, "*", color="crimson", markersize=16, label="principal value")
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_aspect("equal")
    ax.set_xlim(-2.6, 2.6)
    ax.set_ylim(-2.6, 2.6)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
    ax.set_title(f"$(-8)^{{1/{n}}}$: {n} values")
    ax.legend(loc="lower left", fontsize=9)
fig.tight_layout()
plt.show()
```

### 5.2 Rational, irrational and complex exponents

The values $1^w = e^{2\pi i k w}$ for $k = 0, \ldots, 199$:

- $w = 3/8$: only 8 distinct points (the eighth roots of unity);
- $w = \sqrt2 - 1$: never repeats, and fills the circle densely;
- $w = 0.3 + 0.05i$: the modulus $e^{-2\pi k \cdot 0.05}$ changes with $k$, so the values spiral (plotted for $k = -10, \ldots, 10$ on a logarithmic spiral).

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(1, 3, figsize=(16, 5.5))
k = np.arange(200)
for ax, w, title in [(axes[0], 3 / 8, "$w = 3/8$: 8 values"),
                     (axes[1], np.sqrt(2) - 1, "$w = \\sqrt{2} - 1$: dense on the circle")]:
    values = np.exp(2j * np.pi * k * w)
    ax.plot(values.real, values.imag, "o", markersize=3, color="tab:blue")
    ax.plot(1, 0, "*", color="crimson", markersize=14, label="principal value $1^w = 1$")
    ax.set_title(title)

w = 0.3 + 0.05j
k_spiral = np.arange(-10, 11)
points = np.exp(2j * np.pi * k_spiral * w)
t = np.linspace(-10, 10, 2000)
axes[2].plot(np.exp(2j * np.pi * t * w).real, np.exp(2j * np.pi * t * w).imag, color="lightgray", label="spiral $e^{2\\pi i t w}$")
axes[2].plot(points.real, points.imag, "o", color="tab:blue", markersize=5, label="values, $k = -10..10$")
axes[2].plot(1, 0, "*", color="crimson", markersize=14, label="principal value")
axes[2].set_xlim(-2.5, 2.5)
axes[2].set_ylim(-2.5, 2.5)
axes[2].set_title("$w = 0.3 + 0.05i$: a logarithmic spiral")

for ax in axes:
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="lower left", fontsize=8)
axes[0].set_xlim(-1.4, 1.4)
axes[0].set_ylim(-1.4, 1.4)
axes[1].set_xlim(-1.4, 1.4)
axes[1].set_ylim(-1.4, 1.4)
fig.suptitle("All values of $1^w = e^{2\\pi i k w}$")
fig.tight_layout(rect=(0, 0, 1, 0.95))
plt.show()
```

### 5.3 The values of $i^i$ on a logarithmic axis

All values of $i^i$ lie on the positive real axis, too far apart to draw on an ordinary Argand diagram. On a logarithmic scale they are equally spaced, $2\pi \log_{10} e \approx 2.73$ decades apart.

```python
import math

import matplotlib.pyplot as plt
import numpy as np

k = np.arange(-2, 4)
values = np.exp(-np.pi / 2 - 2 * np.pi * k)

fig, ax = plt.subplots(figsize=(10, 2.8))
ax.semilogx(values, np.zeros_like(values), "o", color="tab:blue", markersize=9)
ax.semilogx(math.exp(-math.pi / 2), 0, "*", color="crimson", markersize=18, label="principal value $e^{-\\pi/2} \\approx 0.2079$")
for kk, v in zip(k, values):
    ax.annotate(f"k={kk}", (v, 0), xytext=(0, 12), textcoords="offset points", ha="center")
ax.set_yticks([])
ax.set_xlabel("Re (logarithmic scale); every value is real and positive")
ax.set_title("All values of $i^i = e^{-\\pi/2 - 2\\pi k}$")
ax.grid(True, which="major", alpha=0.3)
ax.legend(loc="lower right")
fig.tight_layout()
plt.show()
```

### 5.4 The branch cut of the principal square root and cube root

Domain colouring of $\sqrt z$ and $z^{1/3}$ (principal values). Both have a seam along the negative real axis: crossing it multiplies the value by $e^{\pm 2\pi i w}$, i.e. by $-1$ for the square root and by $e^{\pm 2\pi i/3}$ for the cube root. On the right, $z^2$ for comparison: an integer power has no cut.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import hsv_to_rgb


def domain_colouring(f, re_range, im_range, resolution=600):
    X, Y = np.meshgrid(np.linspace(*re_range, resolution), np.linspace(*im_range, resolution))
    W = f(X + 1j * Y)
    hue = (np.angle(W) / (2 * np.pi)) % 1.0
    m = (2 / np.pi) * np.arctan(np.abs(W))
    hsv = np.stack([hue, np.clip(2 - 2 * m, 0, 1), np.clip(2 * m, 0, 1)], axis=-1)
    return hsv_to_rgb(hsv)


fig, axes = plt.subplots(1, 3, figsize=(16, 5.5))
for ax, f, title in [(axes[0], np.sqrt, "principal $\\sqrt{z}$: sign flips across the cut"),
                     (axes[1], lambda z: z ** (1 / 3), "principal $z^{1/3}$"),
                     (axes[2], lambda z: z**2, "$z^2$: integer power, no cut")]:
    ax.imshow(domain_colouring(f, (-2, 2), (-2, 2)), origin="lower", extent=[-2, 2, -2, 2])
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.set_title(title)
fig.tight_layout()
plt.show()
```

### 5.5 Where does $\sqrt{z^2} = z$ hold?

Colour each point green where the principal value $\sqrt{z^2}$ equals $z$ and red where it equals $-z$. The law $(z^a)^b = z^{ab}$ with $a = 2$, $b = \frac12$ holds exactly on the half-plane where the principal square root lands, $-\pi/2 < \mathrm{Arg}\, z \le \pi/2$.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import ListedColormap

x = np.linspace(-2, 2, 400)
X, Y = np.meshgrid(x, x)
Z = X + 1j * Y
holds = np.isclose(np.sqrt(Z**2), Z)

fig, ax = plt.subplots(figsize=(6.5, 6.5))
ax.imshow(holds, origin="lower", extent=[-2, 2, -2, 2], cmap=ListedColormap(["#e8a0a0", "#a0d8a0"]))
ax.text(1, 1, "$\\sqrt{z^2} = z$", ha="center", fontsize=14)
ax.text(-1, 1, "$\\sqrt{z^2} = -z$", ha="center", fontsize=14)
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_aspect("equal")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_title("$(z^2)^{1/2} = z$ only on the right half-plane")
plt.show()
```

## 7. Solutions

1. $3^i = e^{i\ln3} = \cos(\ln3) + i\sin(\ln3) \approx 0.4548 + 0.8906i$, on the unit circle. All values: $e^{i(\ln3 + 2\pi i k)} = e^{-2\pi k} e^{i\ln3}$, the points of the ray at angle $\ln3$ with moduli $e^{-2\pi k}$.
2. $1^{1/4} = e^{2\pi i k/4} \in \{1, i, -1, -i\}$, principal $1$. $(-1)^{1/3} = e^{i(\pi + 2\pi k)/3} \in \{e^{i\pi/3}, -1, e^{-i\pi/3}\}$, principal $e^{i\pi/3} = \frac12 + \frac{\sqrt3}{2} i$.
3. $\log(-1) = i\pi(2k + 1)$, so $(-1)^i = e^{i \cdot i\pi(2k + 1)} = e^{-\pi(2k + 1)}$. All real and positive; the principal value is $e^{-\pi} \approx 0.0432$.
4. $|-32|^{1/5} = 2$, so a regular pentagon of radius $2$. Since $(-2)^5 = -32$, $-2$ is a vertex; the others are at angles $\pm\pi/5$ and $\pm 3\pi/5$. The values are $2e^{i(\pi + 2\pi k)/5}$: $2e^{i\pi/5}$, $2e^{3i\pi/5}$, $-2$, $2e^{-3i\pi/5}$, $2e^{-i\pi/5}$. Principal: $2e^{i\pi/5} \approx 1.618 + 1.176i$, the vertex nearest the positive real axis on the upper side.
5. $1 + i = \sqrt2 e^{i\pi/4}$, so $(1 + i)^{1/2} = \pm 2^{1/4} e^{i\pi/8}$. Principal: $2^{1/4}(\cos\frac\pi8 + i\sin\frac\pi8) \approx 1.0987 + 0.4551i$; the other value is its negative.
6. $6/4 = 3/2$ in lowest terms, so two values: $16^{3/2} = \pm 64$. $(16^{1/4})^6$: the fourth roots $2, 2i, -2, -2i$ raised to the 6th power give $64, -64, 64, -64$: the same set $\{\pm 64\}$. $(16^6)^{1/4}$: the four fourth roots of $16^6 = 2^{24}$ are $64, 64i, -64, -64i$ — **four** values. So $(z^6)^{1/4}$ and $z^{6/4}$ are different sets.
7. $z = w = -1 + i$: $zw = -2i$, principal $\sqrt{-2i} = 1 - i$. But $\sqrt{-1 + i} = 2^{1/4} e^{3i\pi/8}$, so $\sqrt z \sqrt w = \big(2^{1/4} e^{3i\pi/8}\big)^2 = \sqrt2\, e^{3i\pi/4} = -1 + i$. Since $1 - i \neq -1 + i$, the law fails (the two sides differ by the factor $-1$). It holds whenever $\mathrm{Arg}\, z + \mathrm{Arg}\, w \in (-\pi, \pi]$, because then $\mathrm{Log}(zw) = \mathrm{Log}\, z + \mathrm{Log}\, w$ and $(zw)^{1/2} = e^{\frac12\mathrm{Log}(zw)} = e^{\frac12\mathrm{Log}\, z} e^{\frac12\mathrm{Log}\, w}$.
8. The values are $e^{\frac pq\mathrm{Log}\, z} \cdot e^{2\pi i k p/q}$. Two indices $k, k'$ give the same value iff $e^{2\pi i (k - k')p/q} = 1$ iff $q$ divides $(k - k')p$. Since $\gcd(p, q) = 1$, this happens iff $q$ divides $k - k'$. So the values depend only on $k \bmod q$, and the $q$ residues $0, \ldots, q - 1$ give $q$ different values.
9. $(z^a)^n = (e^{a\mathrm{Log}\, z})^n = e^{na\mathrm{Log}\, z} = z^{an}$, because integer powers of $e^u$ need no logarithm: $(e^u)^n$ is $n$-fold multiplication, which equals $e^{nu}$ by the addition law. For non-integer $b$, $(z^a)^b = e^{b\,\mathrm{Log}(z^a)}$, and $\mathrm{Log}(z^a) = \mathrm{Log}(e^{a\mathrm{Log}\, z})$ equals $a\,\mathrm{Log}\, z$ only when $\mathrm{Im}(a\,\mathrm{Log}\, z) \in (-\pi, \pi]$; otherwise it is $a\,\mathrm{Log}\, z + 2\pi i m$ for some integer $m \ne 0$, and the result picks up the factor $e^{2\pi i m b} \neq 1$.

## 8. Conclusion

- $z^w$ is defined as $e^{w \log z}$; its values are $e^{w\,\mathrm{Log}\, z}\, e^{2\pi i k w}$. The principal value $e^{w\,\mathrm{Log}\, z}$ is what Python, numpy and sympy compute.
- Integer exponents give one value; $p/q$ in lowest terms gives exactly $q$ values (a regular $q$-gon for roots); irrational or non-real exponents give infinitely many (dense on a circle, or on a logarithmic spiral).
- $i^i = e^{-\pi/2 - 2\pi k}$: every value is real. The principal one is $e^{-\pi/2} \approx 0.2079$.
- The principal cube root of $-8$ is $1 + i\sqrt3$; `np.cbrt` and `sympy.real_root` give the real root $-2$.
- $z^a z^b = z^{a + b}$ and $(z^a)^n = z^{an}$ hold for principal values; $(z^a)^b = z^{ab}$ and $(zw)^a = z^a w^a$ can fail by a factor $e^{2\pi i k a}$. Non-integer powers inherit the branch cut of $\mathrm{Log}$.
