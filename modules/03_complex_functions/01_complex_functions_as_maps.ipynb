# Complex Functions as Maps

A real function $f:\mathbb{R}\to\mathbb{R}$ is drawn as a curve: one input axis, one output axis, two dimensions in all. A complex function $f:\mathbb{C}\to\mathbb{C}$ has a two-dimensional input and a two-dimensional output, so its "graph" would live in four dimensions. This notebook introduces the two standard ways around that: draw the **input plane and the output plane side by side** and watch how grids, circles and lines are carried across, or **paint the input plane** with colours that encode the output (domain colouring).

Along the way every complex function is split into two real functions of two real variables, $f(x + iy) = u(x, y) + i\,v(x, y)$, which is the language the rest of the course is written in.

This is notebook 01 of module 03 (**Complex functions**) in the **Learning Complex Numbers** series.

## 1. Introduction

In calculus a function is a rule that sends each input to exactly one output. Nothing about that definition needs the inputs to be real. A **complex function** sends each complex number $z$ in some domain to a complex number $w = f(z)$:

$$
f(z) = z^2, \qquad f(z) = \frac{1}{z}, \qquad f(z) = 3z + 2i, \qquad f(z) = \frac{(z - 1)(z + 1)^2}{z - i}.
$$

Because the input and output are both points of a plane, it is more useful to think of $f$ as a **map** (a transformation) from the $z$-plane to the $w$-plane than as a graph. The questions change from "what is the slope?" to "where does this circle go? which way is this region twisted? where is the output zero, where does it blow up?"

This viewpoint pays off throughout the course: the exponential wraps lines into circles (notebook 02), the logarithm needs a cut to be a function at all (notebook 03), Möbius maps send circles to circles (notebook 06), and in module 05 the derivative turns out to be a local rotation and scaling.

## 2. Theory

### 2.1 Complex functions and their real and imaginary parts

A **complex function** $f$ assigns to each $z$ in a set $D \subseteq \mathbb{C}$ (the domain) a single complex number $w = f(z)$. Writing $z = x + iy$ and splitting the output into real and imaginary parts gives two real functions of two real variables:

$$
f(x + iy) = u(x, y) + i\,v(x, y), \qquad u = \mathrm{Re}\, f, \quad v = \mathrm{Im}\, f.
$$

Two examples that we will use repeatedly:

$$
z^2 = (x + iy)^2 = \underbrace{x^2 - y^2}_{u} + i\,\underbrace{2xy}_{v},
\qquad
\frac{1}{z} = \frac{\bar z}{|z|^2} = \underbrace{\frac{x}{x^2 + y^2}}_{u} + i\,\underbrace{\frac{-y}{x^2 + y^2}}_{v}.
$$

So "one complex function" is the same data as "a pair of real functions $(u, v)$", i.e. a map $\mathbb{R}^2 \to \mathbb{R}^2$. What makes complex functions special (module 05) is that the good ones are built from complex arithmetic alone, which forces strong links between $u$ and $v$.

### 2.2 The polar view

Often polar form is clearer. With $z = re^{i\theta}$:

$$
z^2 = r^2 e^{2i\theta}, \qquad z^n = r^n e^{in\theta}, \qquad \frac{1}{z} = \frac{1}{r} e^{-i\theta}.
$$

- $z^2$ **squares the modulus and doubles the argument**. The circle $|z| = r$ goes to the circle $|w| = r^2$, traversed **twice**; the ray $\arg z = \theta$ goes to the ray $\arg w = 2\theta$. The upper half-plane ($0 < \theta < \pi$) is opened out onto the whole plane minus the non-negative real axis.
- $1/z$ **inverts the modulus and negates the argument**: inside of the unit circle goes to outside and vice versa, and the picture is reflected across the real axis. The circle $|z| = r$ goes to $|w| = 1/r$, traversed in the opposite direction.

### 2.3 Linear maps $az + b$

For constants $a = \rho e^{i\alpha} \neq 0$ and $b$, the map $f(z) = az + b$ is

$$
z \;\xrightarrow{\ \times \rho\ }\; \text{scale} \;\xrightarrow{\ \times e^{i\alpha}\ }\; \text{rotate by } \alpha \;\xrightarrow{\ + b\ }\; \text{translate}.
$$

It sends lines to lines, circles to circles, and squares to squares. Every shape keeps its form; only position, size and orientation change.

### 2.4 Images of lines under $z^2$

Take the vertical line $x = a$ (with $a \neq 0$), parametrised by $y$. Then $u = a^2 - y^2$ and $v = 2ay$. Eliminating $y = v/(2a)$:

$$
u = a^2 - \frac{v^2}{4a^2} \qquad \text{(a parabola opening to the left).}
$$

The horizontal line $y = b$ (with $b \neq 0$) gives $u = x^2 - b^2$, $v = 2bx$, so

$$
u = \frac{v^2}{4b^2} - b^2 \qquad \text{(a parabola opening to the right).}
$$

A square grid becomes two families of **confocal parabolas** (all with focus at $0$), and they still cross at right angles. That right angle is no accident; it is the conformality of module 05.

### 2.5 Images of lines under $1/z$

The vertical line $x = a$ (with $a \neq 0$) consists of the points with $\mathrm{Re}\, z = a$. For $w = 1/z$ we have $z = 1/w$, so the condition is $\mathrm{Re}(1/w) = a$:

$$
\frac{u}{u^2 + v^2} = a \iff u^2 + v^2 - \frac{u}{a} = 0 \iff \left(u - \frac{1}{2a}\right)^2 + v^2 = \frac{1}{4a^2}.
$$

A line not through the origin becomes a **circle through the origin**, centred at $\frac{1}{2a}$ with radius $\frac{1}{2|a|}$. (The "point at infinity" on the line goes to $w = 0$.) Notebook 06 shows this is part of a general rule: $1/z$ maps lines and circles to lines and circles.

### 2.6 Domain colouring

Instead of moving shapes, colour every input point $z$ by its output $w = f(z)$:

- **hue** = argument of $w$: red for $\arg w = 0$ (positive real), through yellow, green ($\arg w \approx 2\pi/3$), cyan, blue, magenta and back to red;
- **brightness** = modulus of $w$: dark where $|w|$ is small, washed out to white where $|w|$ is large.

Concretely, with $m = \frac{2}{\pi}\arctan|w| \in [0, 1)$ (so $m = \frac12$ exactly when $|w| = 1$), we use HSV colour with

$$
H = \frac{\arg w}{2\pi} \bmod 1, \qquad S = \min(1,\ 2 - 2m), \qquad V = \min(1,\ 2m).
$$

Then a **zero** of $f$ is a black dot and a **pole** (where $|f| \to \infty$) is a white dot. Near either, every argument occurs, so all the colours meet there. The order is visible too:

- near a zero of order $n$, $f(z) \approx c\,(z - z_0)^n$, so as $z$ circles $z_0$ once counter-clockwise, $\arg f$ increases by $2\pi n$: the colours red → yellow → green → blue → magenta cycle **$n$ times counter-clockwise**;
- near a pole of order $n$, $f(z) \approx c\,(z - z_0)^{-n}$, so the colours cycle $n$ times in the **opposite** order.

## 3. Worked Examples

### Example 1 — $u$ and $v$ for $f(z) = z^2 + z$

$$
(x + iy)^2 + (x + iy) = (x^2 - y^2 + x) + i\,(2xy + y),
$$

so $u = x^2 - y^2 + x$ and $v = 2xy + y = y(2x + 1)$. For instance $f(1 + i) = (1 - 1 + 1) + i(2 + 1) = 1 + 3i$.

### Example 2 — a point under $z^2$, two ways

$z = 1 + i$. Algebraically: $(1 + i)^2 = 1 + 2i + i^2 = 2i$.

Geometrically: $|z| = \sqrt2$ and $\arg z = \pi/4$, so $|z^2| = 2$ and $\arg z^2 = \pi/2$, which is the point $2i$. ✓

### Example 3 — a point under $1/z$, two ways

$z = 1 + i$. Algebraically: $\frac{1}{1 + i} = \frac{1 - i}{(1 + i)(1 - i)} = \frac{1 - i}{2}$.

Geometrically: modulus $1/\sqrt2$ (the reciprocal of $\sqrt2$), argument $-\pi/4$ (the negative of $\pi/4$). ✓

### Example 4 — the line $x = 1$ under $z^2$

From §2.4 with $a = 1$: $u = 1 - v^2/4$. Check with the point $z = 1 + 2i$: $z^2 = 1 + 4i - 4 = -3 + 4i$, and indeed $1 - 4^2/4 = -3$. ✓ The vertex of the parabola is $z = 1 \mapsto w = 1$, and $z = 1 \pm 2i$ both land on it at $v = \pm 4$.

### Example 5 — the line $x = 1$ under $1/z$

From §2.5 with $a = 1$: the image is the circle $|w - \tfrac12| = \tfrac12$. Check the points:

- $z = 1 \mapsto w = 1$, on the circle (distance $\frac12$ from $\frac12$);
- $z = 1 + i \mapsto w = \frac{1 - i}{2}$, distance $|\tfrac12 - \tfrac i2 - \tfrac12| = \tfrac12$ ✓;
- as $y \to \pm\infty$, $w \to 0$, also on the circle.

### Example 6 — reading a domain colouring before drawing it

$f(z) = \dfrac{(z - 1)(z + 1)^2}{z - i}$ has

- a simple zero at $z = 1$: one counter-clockwise colour cycle, black centre;
- a double zero at $z = -1$: two counter-clockwise colour cycles, black centre;
- a simple pole at $z = i$: one clockwise colour cycle, white centre.

Far away $f(z) \approx z^3/z = z^2$, so a large circle shows two counter-clockwise cycles: $1 + 2 - 1 = 2$ (zeros count $+$, poles count $-$). We draw it in §5.

## 4. Python Examples

### 4.1 Complex functions are ordinary Python functions

Because `+`, `*`, `/` and `**` already work on `complex` and on numpy complex arrays, a complex function is just a Python function. It works on one number or on a whole grid at once.

```python
import numpy as np


def square(z):
    return z**2


def reciprocal(z):
    return 1 / z


z = 1 + 1j
print("z^2 at 1+i:", square(z), "   1/z at 1+i:", reciprocal(z))

points = np.array([1, 1j, -1, 1 + 2j])
print("z^2 on an array:", square(points))
print("|z|, arg z      :", np.abs(points), np.angle(points))
print("|z^2|, arg z^2  :", np.abs(square(points)), np.angle(square(points)))
```

The last two lines show the polar rule: moduli are squared ($\sqrt5 \to 5$) and arguments doubled, except that the doubled angle is reported back in $(-\pi, \pi]$ (the argument of $-1$ is $\pi$, and $2\pi$ comes back as $0$).

### 4.2 $u$ and $v$ with sympy

Declare $x$ and $y$ real so sympy can separate real and imaginary parts.

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
z = x + sp.I * y

for name, expression in [("z^2", z**2), ("z^2 + z", z**2 + z), ("1/z", 1 / z), ("z^3", z**3)]:
    u = sp.simplify(sp.re(sp.expand(expression)))
    v = sp.simplify(sp.im(sp.expand(expression)))
    print(f"{name:8s}  u = {u},   v = {v}")
```

### 4.3 Checking the image curves numerically

Push many points of a line through the map and test whether they satisfy the predicted equation.

```python
import numpy as np

y = np.linspace(-5, 5, 1001)
a = 1.0
vertical_line = a + 1j * y

w = vertical_line**2
parabola_error = np.max(np.abs(w.real - (a**2 - w.imag**2 / (4 * a**2))))
print("z^2 of x=1 lies on u = 1 - v^2/4:      max error", parabola_error)

w = 1 / vertical_line
circle_error = np.max(np.abs(np.abs(w - 1 / (2 * a)) - 1 / (2 * a)))
print("1/z of x=1 lies on |w - 1/2| = 1/2:    max error", circle_error)

theta = np.linspace(0, 2 * np.pi, 9)[:-1]
circle_r2 = 2 * np.exp(1j * theta)
print("|z^2| on |z|=2 is 4 everywhere:         ", np.allclose(np.abs(circle_r2**2), 4))
print("|1/z| on |z|=2 is 1/2 everywhere:       ", np.allclose(np.abs(1 / circle_r2), 0.5))
```

### 4.4 A linear map is rotate, scale, translate

$f(z) = (1 + i)z + 2$: here $a = 1 + i = \sqrt2\, e^{i\pi/4}$, so $f$ scales by $\sqrt2$, rotates by $45°$, then shifts by $2$. Distances between image points are exactly $\sqrt2$ times the original distances.

```python
import cmath

import numpy as np

a = 1 + 1j
b = 2
print("|a| =", abs(a), "  arg a (degrees) =", np.degrees(cmath.phase(a)))

square_corners = np.array([0, 1, 1 + 1j, 1j])
images = a * square_corners + b
print("images of the unit square's corners:", images)

side_lengths_before = np.abs(np.diff(np.append(square_corners, square_corners[0])))
side_lengths_after = np.abs(np.diff(np.append(images, images[0])))
print("side length ratios:", side_lengths_after / side_lengths_before)
```

### 4.5 Finding zeros and poles of a rational function

For the function of Example 6, numpy finds the zeros (roots of the numerator) and poles (roots of the denominator). `np.poly` turns a list of roots into polynomial coefficients.

```python
import numpy as np

numerator = np.poly([1, -1, -1])     # (z - 1)(z + 1)^2
denominator = np.poly([1j])          # (z - i)
print("numerator coefficients:", numerator)
print("zeros:", np.roots(numerator))
print("poles:", np.roots(denominator))


def f(z):
    return (z - 1) * (z + 1) ** 2 / (z - 1j)


for z0 in [1, -1, 1j]:
    near = z0 + 1e-6
    print(f"|f| just next to {z0}: {abs(f(near)):.3e}")
```

## 5. Visualizations

### 5.1 A grid and its image, side by side

The helper below draws vertical lines (blue) and horizontal lines (orange) in the $z$-plane on the left, and their images under $f$ in the $w$-plane on the right. A highlighted point and its image help track orientation.

```python
import matplotlib.pyplot as plt
import numpy as np


def plot_grid_and_image(f, x_values, y_values, x_span, y_span, title, w_limits=None, marker=1 + 0.5j):
    fig, (ax_z, ax_w) = plt.subplots(1, 2, figsize=(12, 6))
    t_vertical = np.linspace(*y_span, 400)
    t_horizontal = np.linspace(*x_span, 400)
    for x0 in x_values:
        line = x0 + 1j * t_vertical
        ax_z.plot(line.real, line.imag, color="tab:blue", linewidth=1)
        image = f(line)
        ax_w.plot(image.real, image.imag, color="tab:blue", linewidth=1)
    for y0 in y_values:
        line = t_horizontal + 1j * y0
        ax_z.plot(line.real, line.imag, color="tab:orange", linewidth=1)
        image = f(line)
        ax_w.plot(image.real, image.imag, color="tab:orange", linewidth=1)
    ax_z.plot(marker.real, marker.imag, "ko")
    ax_w.plot(f(marker).real, f(marker).imag, "ko")
    for ax, name in [(ax_z, "z-plane"), (ax_w, "w-plane")]:
        ax.axhline(0, color="black", linewidth=0.8)
        ax.axvline(0, color="black", linewidth=0.8)
        ax.set_aspect("equal")
        ax.set_xlabel("Re")
        ax.set_ylabel("Im")
        ax.grid(True, alpha=0.3)
        ax.set_title(name)
    if w_limits is not None:
        ax_w.set_xlim(*w_limits[0])
        ax_w.set_ylim(*w_limits[1])
    fig.suptitle(title)
    fig.tight_layout(rect=(0, 0, 1, 0.95))
    plt.show()


grid = np.linspace(-2, 2, 9)
grid = grid[np.abs(grid) > 1e-9]   # skip the axes themselves: their images are degenerate
plot_grid_and_image(lambda z: z**2, grid, grid, (-2, 2), (-2, 2),
                    "$w = z^2$: grid lines become confocal parabolas")
```

Each blue line $x = a$ becomes a left-opening parabola and each orange line $y = b$ a right-opening one, exactly as in §2.4. The lines $x = a$ and $x = -a$ land on the **same** parabola, because $z$ and $-z$ have the same square: $z^2$ is two-to-one. The black point $1 + 0.5i$ goes to $0.75 + i$.

### 5.2 A polar grid under $z^2$

Circles and rays are the natural grid for $z^n$. Colouring the rays by angle shows the doubling: rays in the upper half-plane already cover every direction.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, (ax_z, ax_w) = plt.subplots(1, 2, figsize=(12, 6))
theta = np.linspace(0, 2 * np.pi, 400)
for r in [0.5, 1.0, 1.5]:
    circle = r * np.exp(1j * theta)
    ax_z.plot(circle.real, circle.imag, color="gray", linewidth=1)
    ax_w.plot((circle**2).real, (circle**2).imag, color="gray", linewidth=1)

radii = np.linspace(0, 1.5, 100)
ray_angles = np.linspace(0, np.pi, 7)[:-1]
colours = plt.cm.viridis(np.linspace(0, 0.9, len(ray_angles)))
for angle, colour in zip(ray_angles, colours):
    ray = radii * np.exp(1j * angle)
    label = f"{np.degrees(angle):.0f}° → {np.degrees(2 * angle):.0f}°"
    ax_z.plot(ray.real, ray.imag, color=colour, linewidth=2.5, label=label)
    ax_w.plot((ray**2).real, (ray**2).imag, color=colour, linewidth=2.5)

for ax, name in [(ax_z, "z-plane: |z| = 0.5, 1, 1.5 and six rays"), (ax_w, "w-plane: |w| = 0.25, 1, 2.25")]:
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_aspect("equal")
    ax.set_xlim(-2.4, 2.4)
    ax.set_ylim(-2.4, 2.4)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
    ax.set_title(name)
ax_z.legend(loc="lower left", fontsize=9)
fig.suptitle("$w = z^2$ squares the radius and doubles the angle")
fig.tight_layout(rect=(0, 0, 1, 0.95))
plt.show()
```

### 5.3 The grid under $1/z$

Lines not through $0$ become circles through $0$. Lines far from the origin become small circles near $0$, and the region near the origin is flung far away. The axis lines are omitted: they map to themselves.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, (ax_z, ax_w) = plt.subplots(1, 2, figsize=(12, 6))
t = np.linspace(-40, 40, 8001)
values = [-2, -1, -0.5, 0.5, 1, 2]
for a in values:
    for line, colour in [(a + 1j * t, "tab:blue"), (t + 1j * a, "tab:orange")]:
        ax_z.plot(line.real, line.imag, color=colour, linewidth=1)
        image = 1 / line
        ax_w.plot(image.real, image.imag, color=colour, linewidth=1)

unit_circle = np.exp(1j * np.linspace(0, 2 * np.pi, 400))
for ax in (ax_z, ax_w):
    ax.plot(unit_circle.real, unit_circle.imag, "k--", linewidth=1, label="unit circle (maps to itself, flipped)")
ax_z.plot(1, 1, "ko")
ax_w.plot(0.5, -0.5, "ko")

ax_z.set_title("z-plane: lines x = a and y = b")
ax_w.set_title("w-plane: circles through 0")
for ax in (ax_z, ax_w):
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_aspect("equal")
    ax.set_xlim(-2.5, 2.5)
    ax.set_ylim(-2.5, 2.5)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
ax_w.legend(loc="lower left", fontsize=9)
fig.suptitle("$w = 1/z$: lines become circles through the origin; $1 + i \\mapsto (1 - i)/2$")
fig.tight_layout(rect=(0, 0, 1, 0.95))
plt.show()
```

### 5.4 Domain colouring

The function below implements §2.6. It evaluates $f$ on a grid, turns $\arg f$ into hue and $|f|$ into saturation/brightness, and converts HSV to RGB with `matplotlib.colors.hsv_to_rgb`. The first picture is the **colour key**: $f(z) = z$ itself, so the colour at a point is the colour of that complex number.

*The three cells after this one use `domain_colouring` and `show_domain_colouring`, so run this cell first.*

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import hsv_to_rgb


def domain_colouring(f, re_range=(-2, 2), im_range=(-2, 2), resolution=600):
    x = np.linspace(*re_range, resolution)
    y = np.linspace(*im_range, resolution)
    X, Y = np.meshgrid(x, y)
    with np.errstate(divide="ignore", invalid="ignore", over="ignore"):
        W = f(X + 1j * Y)
        hue = (np.angle(W) / (2 * np.pi)) % 1.0
        m = (2 / np.pi) * np.arctan(np.abs(W))
    hue = np.nan_to_num(hue)
    m = np.nan_to_num(m, nan=1.0)              # an exact pole counts as infinitely large
    saturation = np.clip(2 - 2 * m, 0, 1)
    value = np.clip(2 * m, 0, 1)
    return hsv_to_rgb(np.stack([hue, saturation, value], axis=-1))


def show_domain_colouring(ax, f, title, re_range=(-2, 2), im_range=(-2, 2)):
    rgb = domain_colouring(f, re_range, im_range)
    ax.imshow(rgb, origin="lower", extent=[*re_range, *im_range])
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.set_title(title)


fig, ax = plt.subplots(figsize=(6.5, 6.5))
show_domain_colouring(ax, lambda z: z, "Colour key: $f(z) = z$")
for label, point in [("1 (red)", 1.2), ("i", 1.2j), ("-1 (cyan)", -1.2), ("-i", -1.2j)]:
    ax.annotate(label, (point.real, point.imag), color="black", ha="center", va="center", fontsize=11,
                bbox=dict(boxstyle="round", facecolor="white", alpha=0.8))
ax.plot(np.cos(np.linspace(0, 2 * np.pi, 200)), np.sin(np.linspace(0, 2 * np.pi, 200)), "k:", linewidth=1)
plt.show()
```

Black at $0$ (a zero), white far away (large modulus), fully saturated colour on the dotted unit circle $|w| = 1$. Going counter-clockwise around the origin the hue runs red → yellow → green → cyan → blue → magenta → red.

### 5.5 Zeros and poles: $z^2$, $1/z$, $z^3 - 1$

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(16, 5.5))
show_domain_colouring(axes[0], lambda z: z**2, "$z^2$: double zero, colours cycle twice")
show_domain_colouring(axes[1], lambda z: 1 / z, "$1/z$: simple pole, colours reversed")
show_domain_colouring(axes[2], lambda z: z**3 - 1, "$z^3 - 1$: three simple zeros")
fig.tight_layout(rect=(0, 0, 1, 0.95))
plt.show()
```

- **$z^2$**: going once around $0$, every colour appears twice — a zero of order 2.
- **$1/z$**: white centre, and counter-clockwise the hue runs red → magenta → blue → … — the *reverse* order, a pole of order 1. (It is the mirror image of the colour key because $\arg(1/z) = -\arg z$.)
- **$z^3 - 1$**: three black dots at the cube roots of unity $1, e^{\pm 2\pi i/3}$, each with one ordinary colour cycle. A big circle around all three shows three cycles, like $z^3$.

### 5.6 The rational function of Example 6

```python
import matplotlib.pyplot as plt


def f(z):
    return (z - 1) * (z + 1) ** 2 / (z - 1j)


fig, ax = plt.subplots(figsize=(7, 7))
show_domain_colouring(ax, f, "$f(z) = (z-1)(z+1)^2 / (z-i)$", re_range=(-2.5, 2.5), im_range=(-2.5, 2.5))
for label, point in [("simple zero", 1), ("double zero", -1), ("simple pole", 1j)]:
    ax.annotate(label, (point.real, point.imag), xytext=(point.real + 0.15, point.imag - 0.45),
                fontsize=10, bbox=dict(boxstyle="round", facecolor="white", alpha=0.85),
                arrowprops=dict(arrowstyle="->"))
plt.show()
```

### 5.7 A linear map keeps the grid square

A final side-by-side: a linear map. The grid stays a square grid, only rotated, scaled and shifted, matching §4.4.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, (ax_z, ax_w) = plt.subplots(1, 2, figsize=(12, 6))
t = np.linspace(0, 1, 50)
for k in np.linspace(0, 1, 5):
    for line, colour in [(k + 1j * t, "tab:blue"), (t + 1j * k, "tab:orange")]:
        ax_z.plot(line.real, line.imag, color=colour)
        image = (1 + 1j) * line + 2
        ax_w.plot(image.real, image.imag, color=colour)
ax_z.plot(0, 0, "ko")
ax_w.plot(2, 0, "ko")
for ax, name in [(ax_z, "z-plane: unit square"), (ax_w, "w-plane: rotated 45°, scaled by $\\sqrt{2}$, shifted by 2")]:
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(-0.5, 3.5)
    ax.set_ylim(-0.5, 3.5)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
    ax.set_title(name)
fig.suptitle("$w = (1 + i)z + 2$")
fig.tight_layout(rect=(0, 0, 1, 0.95))
plt.show()
```

## 7. Solutions

1. $(x + iy)^3 = x^3 + 3x^2(iy) + 3x(iy)^2 + (iy)^3 = (x^3 - 3xy^2) + i(3x^2 y - y^3)$. So $u = x^3 - 3xy^2$, $v = 3x^2y - y^3$.
2. $\bar z + 2i = x + i(2 - y)$, so $u = x$, $v = 2 - y$. It reflects across the real axis, then shifts up by $2$; equivalently it reflects across the horizontal line $\mathrm{Im}\, z = 1$. (Unlike $az + b$ it reverses orientation; module 05 shows such maps are not complex-differentiable.)
3. Algebraically $(2 - i)^2 = 4 - 4i + i^2 = 3 - 4i$. Geometrically $|2 - i| = \sqrt5$ so $|w| = 5$, and $\arg w = 2\arg(2 - i) = -2\arctan\tfrac12 \approx -0.927$; then $5(\cos(-0.927) + i\sin(-0.927)) = 3 - 4i$ ✓.
4. Corners: $0 \mapsto 0$, $1 \mapsto 1$, $1 + i \mapsto 2i$, $i \mapsto -1$. The side $y = 0$ ($0 \le x \le 1$) maps to the segment $[0, 1]$; the side $x = 0$ maps to $w = -y^2$, the segment $[-1, 0]$. The side $x = 1$ maps to the arc of $u = 1 - v^2/4$ from $1$ to $2i$; the side $y = 1$ to the arc of $u = v^2/4 - 1$ from $-1$ to $2i$. The image is a curved "tent" region in the upper half-plane bounded by $[-1, 1]$ and two parabolic arcs meeting at $2i$. The corner angle at $0$ (a right angle) doubles to a straight angle, while the corners at $1$, $1 + i$, $i$ stay right angles.
5. $|w| = 1/2$: the circle of radius $\tfrac12$. With $z = 2e^{i\theta}$, $w = \tfrac12 e^{-i\theta}$, so as $z$ goes counter-clockwise, $w$ goes **clockwise**.
6. $z = 1/w$ and $\mathrm{Im}(1/w) = \dfrac{-v}{u^2 + v^2}$. Setting this equal to $b$: $u^2 + v^2 + \dfrac{v}{b} = 0$, i.e. $u^2 + \left(v + \frac{1}{2b}\right)^2 = \frac{1}{4b^2}$, a circle through $0$ centred at $-\frac{i}{2b}$.
7. Zeros at $\pm i$ (simple; black dots, one counter-clockwise cycle each). Poles at $\pm 1$ (simple; white dots, one reversed cycle each). Far away $f \to 1$, so the outer region is nearly uniform red with medium brightness; a big circle shows $2 - 2 = 0$ net cycles.
8. Write $z - z_0 = \varepsilon e^{i\phi}$ with small $\varepsilon$. Then $f(z) \approx c\,\varepsilon^n e^{in\phi}$, so $\arg f \approx \arg c + n\phi$. As $\phi$ goes from $0$ to $2\pi$, $\arg f$ increases by $2\pi n$: the hue wheel is traversed $n$ times in the positive direction. For a pole, $f \approx c\,\varepsilon^{-n} e^{-in\phi}$ and $\arg f$ *decreases* by $2\pi n$, so the colours run backwards. The modulus $\varepsilon^{\pm n}$ gives the black or white centre.
9. $x^2 - y^2 = u$ and $2xy = v$, so $x^2 - y^2 = c$ maps to the vertical line $u = c$ and $xy = d$ maps to the horizontal line $v = 2d$. Vertical and horizontal lines meet at right angles, and $z^2$ preserves angles away from $0$ (the theme of module 05: the derivative $2z$ is a rotation-and-scaling), so the preimage hyperbolas must meet at right angles too.

## 8. Conclusion

- A complex function $w = f(z)$ is a map from the $z$-plane to the $w$-plane; equivalently a pair of real functions $u(x, y) = \mathrm{Re}\, f$, $v(x, y) = \mathrm{Im}\, f$.
- $z^2$ squares moduli and doubles arguments: circles go to circles (twice around), rays to rays, a square grid to confocal parabolas. It is two-to-one.
- $1/z$ inverts moduli and negates arguments: lines not through $0$ become circles through $0$.
- $az + b$ scales by $|a|$, rotates by $\arg a$ and translates by $b$; shapes are preserved.
- Domain colouring paints $\arg f$ as hue and $|f|$ as brightness. Zeros are black, poles white; the number of colour cycles around a point is its order (counter-clockwise for zeros, reversed for poles).
