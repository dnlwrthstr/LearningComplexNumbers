# Quadratics and Conjugate Roots

Module 00 showed that a real quadratic with negative discriminant has two complex roots. This notebook takes the quadratic formula all the way: we derive it by **completing the square in $\mathbb{C}$**, so that it works even when the *coefficients* are complex, which in turn requires computing **square roots of complex numbers**. We then prove the **conjugate-root theorem**: the non-real roots of any polynomial with real coefficients come in mirror-image pairs $w, \bar w$. Finally, **Vieta's formulas** link the roots back to the coefficients.

This is notebook 01 of module 02 (**Polynomials and roots**) in the **Learning Complex Numbers** series.

## 1. Introduction

Polynomials are where complex numbers were born, and they remain the place where complex numbers are most obviously *necessary*: eigenvalues of a matrix are the roots of its characteristic polynomial, the poles of a filter are roots of its denominator, and the natural frequencies of a damped oscillator are roots of $m s^2 + c s + k$. In all these applications the coefficients are real, and the roots that show up are either real or come in conjugate pairs $p \pm qi$. That pairing is not a coincidence; it is a theorem, and it is the reason a real system can oscillate (a pair $p \pm qi$ produces $e^{pt}\cos qt$, a real signal).

Two questions drive this notebook:

1. Does the quadratic formula survive when $a, b, c$ are themselves complex? (Yes, once we know how to take $\sqrt{\Delta}$ for complex $\Delta$.)
2. Why do roots of real polynomials always look symmetric across the real axis? (Because conjugation commutes with every real polynomial.)

## 2. Theory

### 2.1 Completing the square in $\mathbb{C}$

Let $a, b, c \in \mathbb{C}$ with $a \neq 0$. Exactly the same algebra as in the real case gives

$$
az^2 + bz + c = a\left[\left(z + \frac{b}{2a}\right)^2 - \frac{b^2 - 4ac}{4a^2}\right].
$$

Every step uses only the field rules (addition, multiplication, division by non-zero numbers), and these hold in $\mathbb{C}$. With the **discriminant** $\Delta = b^2 - 4ac$, the equation $az^2 + bz + c = 0$ becomes

$$
\left(z + \frac{b}{2a}\right)^2 = \frac{\Delta}{4a^2}.
$$

So the problem reduces to: *which numbers $s$ satisfy $s^2 = \Delta$?* If $\delta$ is one of them, then $(-\delta)^2 = \Delta$ as well, and the factorization $s^2 - \delta^2 = (s - \delta)(s + \delta)$ shows there are no others. Hence

$$
z = \frac{-b \pm \delta}{2a}, \qquad \delta^2 = \Delta.
$$

This is the quadratic formula, valid for **all** complex coefficients. The symbol $\pm\sqrt{\Delta}$ simply means "the two square roots of $\Delta$"; which one you call $+$ does not matter.

Consequences: there are two roots, which coincide exactly when $\Delta = 0$; and the classification by "sign of $\Delta$" only makes sense for real coefficients, since a complex $\Delta$ has no sign.

### 2.2 Square roots of a complex number, algebraically

Given $w = u + vi$, look for $\delta = x + yi$ with $x, y$ real and $\delta^2 = w$:

$$
(x + yi)^2 = (x^2 - y^2) + 2xy\, i = u + vi
\quad\Longleftrightarrow\quad
x^2 - y^2 = u, \qquad 2xy = v.
$$

Taking moduli of $\delta^2 = w$ adds a third equation, $x^2 + y^2 = |w| = \sqrt{u^2 + v^2}$. Adding and subtracting it from the first:

$$
x^2 = \frac{|w| + u}{2}, \qquad y^2 = \frac{|w| - u}{2}.
$$

Both right-hand sides are $\ge 0$ because $|w| \ge |u|$. The condition $2xy = v$ fixes the relative sign: $x$ and $y$ have the same sign when $v > 0$ and opposite signs when $v < 0$. So

$$
\delta = \pm\left(\sqrt{\frac{|w| + u}{2}} \;+\; \operatorname{sgn}(v)\, \sqrt{\frac{|w| - u}{2}}\; i\right),
$$

with the convention $\operatorname{sgn}(0) = 1$. In polar form this is the familiar $\delta = \pm\sqrt{r}\,e^{i\theta/2}$ for $w = re^{i\theta}$: square roots **halve the angle** and **take the square root of the length**. The two roots are antipodal, $\delta$ and $-\delta$.

The choice with $x \ge 0$ (right half-plane) is the **principal square root**; it is what `cmath.sqrt` returns, and it corresponds to $\theta = \operatorname{Arg} w \in (-\pi, \pi]$.

### 2.3 The conjugate-root theorem

Recall that conjugation respects arithmetic: $\overline{z + w} = \bar z + \bar w$, $\overline{zw} = \bar z\,\bar w$, hence $\overline{z^k} = \bar z^{\,k}$, and $\bar a = a$ exactly when $a$ is real.

> **Theorem (conjugate roots).** Let $p(z) = a_n z^n + \dots + a_1 z + a_0$ with all $a_k \in \mathbb{R}$. Then for every $z \in \mathbb{C}$
> $$ p(\bar z) = \overline{p(z)}. $$
> In particular, if $p(w) = 0$ then $p(\bar w) = 0$: non-real roots come in conjugate pairs.

*Proof.* Conjugate $p(z)$ term by term:

$$
\overline{p(z)} = \sum_{k=0}^{n} \overline{a_k z^k} = \sum_{k=0}^{n} \overline{a_k}\,\bar z^{\,k} = \sum_{k=0}^{n} a_k\,\bar z^{\,k} = p(\bar z),
$$

using $\overline{a_k} = a_k$ in the third step. If $p(w) = 0$ then $p(\bar w) = \overline{p(w)} = \bar 0 = 0$. $\blacksquare$

The hypothesis "real coefficients" is essential: it is exactly the step $\overline{a_k} = a_k$. For example $z^2 - iz = z(z - i)$ has roots $0$ and $i$, but $\bar i = -i$ is not a root.

**Multiplicities pair up too.** Derivatives of a real polynomial are real polynomials, so $p^{(k)}(\bar w) = \overline{p^{(k)}(w)}$ for all $k$. A root of multiplicity $m$ (where $p, p', \dots, p^{(m-1)}$ vanish but $p^{(m)}$ does not) therefore has a conjugate of the same multiplicity.

**Real quadratic factors.** A conjugate pair $w = p + qi$, $\bar w = p - qi$ multiplies out to a real quadratic:

$$
(z - w)(z - \bar w) = z^2 - (w + \bar w)\, z + w \bar w = z^2 - 2p\, z + (p^2 + q^2) = z^2 - 2\operatorname{Re}(w)\, z + |w|^2.
$$

So once you know one complex root of a real polynomial, you can divide out a whole **real** quadratic factor.

**Odd degree.** Non-real roots of a real polynomial come in pairs, so if the degree is odd (and, as notebook 02 proves, there are exactly $n$ roots counted with multiplicity) at least one root must be real. This also follows from the intermediate value theorem.

### 2.4 Vieta's formulas

If a monic polynomial has roots $z_1, \dots, z_n$ (with multiplicity), it factors as $(z - z_1)\cdots(z - z_n)$. Expanding and comparing coefficients:

$$
(z - z_1)(z - z_2) = z^2 - (z_1 + z_2)\,z + z_1 z_2.
$$

For $az^2 + bz + c$ with roots $z_1, z_2$ (complex coefficients allowed):

$$
z_1 + z_2 = -\frac{b}{a}, \qquad z_1 z_2 = \frac{c}{a}.
$$

In general, for $a_n z^n + a_{n-1} z^{n-1} + \dots + a_0$,

$$
\sum_i z_i = -\frac{a_{n-1}}{a_n}, \qquad \sum_{i<j} z_i z_j = \frac{a_{n-2}}{a_n}, \qquad \dots, \qquad z_1 z_2 \cdots z_n = (-1)^n \frac{a_0}{a_n}.
$$

The $k$-th **elementary symmetric polynomial** of the roots equals $(-1)^k a_{n-k}/a_n$.

Geometrically, for a quadratic the **midpoint** of the two roots is $-\frac{b}{2a}$: the roots are placed symmetrically about it, at $\pm\frac{\delta}{2a}$. Completing the square is exactly the statement "shift the midpoint to the origin".

## 3. Worked Examples

### Example 1: completing the square, real coefficients

Solve $z^2 - 4z + 13 = 0$.

$$
z^2 - 4z + 13 = (z - 2)^2 - 4 + 13 = (z - 2)^2 + 9.
$$

So $(z - 2)^2 = -9 = (3i)^2$, giving $z - 2 = \pm 3i$ and $z = 2 \pm 3i$, a conjugate pair with midpoint $2$.

### Example 2: square root of $3 + 4i$

$u = 3$, $v = 4$, $|w| = 5$. Then $x^2 = \frac{5 + 3}{2} = 4$ and $y^2 = \frac{5 - 3}{2} = 1$. Since $v > 0$, $x$ and $y$ have the same sign: $\delta = \pm(2 + i)$.

Check: $(2 + i)^2 = 4 + 4i + i^2 = 3 + 4i$. ✓

### Example 3: a quadratic with complex coefficients

Solve $z^2 - (3 + i)z + (4 + 3i) = 0$.

1. Discriminant: $\Delta = (3 + i)^2 - 4(4 + 3i) = (9 + 6i - 1) - (16 + 12i) = -8 - 6i$.
2. Square root of $-8 - 6i$: $|w| = 10$, $x^2 = \frac{10 - 8}{2} = 1$, $y^2 = \frac{10 + 8}{2} = 9$. Since $v = -6 < 0$, the signs differ: $\delta = \pm(1 - 3i)$. Check: $(1 - 3i)^2 = 1 - 6i - 9 = -8 - 6i$. ✓
3. Roots: $z = \dfrac{(3 + i) \pm (1 - 3i)}{2}$, so $z_1 = \dfrac{4 - 2i}{2} = 2 - i$ and $z_2 = \dfrac{2 + 4i}{2} = 1 + 2i$.

Note $z_1$ and $z_2$ are **not** conjugates; nothing forces them to be, because the coefficients are not real. Vieta check: $z_1 + z_2 = 3 + i = -b/a$ ✓ and $z_1 z_2 = (2 - i)(1 + 2i) = 2 + 4i - i - 2i^2 = 4 + 3i = c/a$ ✓.

### Example 4: from one complex root to a full factorization

$p(z) = z^4 - 2z^3 + 6z^2 - 8z + 8$ has the root $1 + i$ (check: $(1+i)^2 = 2i$, $(1+i)^3 = -2 + 2i$, $(1+i)^4 = -4$, and $-4 - 2(-2 + 2i) + 6(2i) - 8(1 + i) + 8 = -4 + 4 - 4i + 12i - 8 - 8i + 8 = 0$ ✓).

1. Real coefficients, so $1 - i$ is also a root.
2. The pair gives the real factor $z^2 - 2\operatorname{Re}(1+i)\, z + |1 + i|^2 = z^2 - 2z + 2$.
3. Polynomial division: $p(z) = (z^2 - 2z + 2)(z^2 + 4)$.
4. $z^2 + 4 = 0$ gives $z = \pm 2i$, another conjugate pair.

Over $\mathbb{R}$: $p = (z^2 - 2z + 2)(z^2 + 4)$. Over $\mathbb{C}$: $p = (z - 1 - i)(z - 1 + i)(z - 2i)(z + 2i)$.

### Example 5: Vieta without solving

Let $z_1, z_2$ be the roots of $z^2 - 2z + 5$. Find $z_1^2 + z_2^2$ and $\frac{1}{z_1} + \frac{1}{z_2}$ without solving.

- $z_1 + z_2 = 2$, $z_1 z_2 = 5$.
- $z_1^2 + z_2^2 = (z_1 + z_2)^2 - 2z_1 z_2 = 4 - 10 = -6$.
- $\frac{1}{z_1} + \frac{1}{z_2} = \frac{z_1 + z_2}{z_1 z_2} = \frac{2}{5}$.

(The roots are $1 \pm 2i$; indeed $(1 + 2i)^2 + (1 - 2i)^2 = 2(1 - 4) = -6$.) Both answers are real, as they must be: these expressions are symmetric in $z_1, z_2$, so they do not change under conjugation, which swaps the two roots.

## 4. Python Examples

### 4.1 Square roots from the formula vs `cmath.sqrt`

We implement the algebraic recipe of 2.2 and compare with `cmath.sqrt`, which returns the principal root.

```python
import cmath
import math


def complex_sqrt(w):
    # Principal square root of w via x^2 = (|w| + u)/2, y^2 = (|w| - u)/2, sign(y) = sign(v)
    u, v = w.real, w.imag
    modulus = abs(w)
    x = math.sqrt((modulus + u) / 2)
    y = math.sqrt((modulus - u) / 2)
    if v < 0:
        y = -y
    return complex(x, y)


for w in [3 + 4j, -8 - 6j, -9 + 0j, 2j, -5 + 12j, 1 - 1j]:
    ours = complex_sqrt(w)
    print(f"w = {w!s:>10}   formula: {ours:.6f}   cmath: {cmath.sqrt(w):.6f}   "
          f"square back: {ours**2:.6f}   agree: {abs(ours - cmath.sqrt(w)) < 1e-12}")
```

### 4.2 A quadratic solver for complex coefficients

The formula from 2.1 with `cmath.sqrt` handles every case. We check it on Example 3 and on a random complex quadratic.

```python
import cmath

import numpy as np


def solve_quadratic(a, b, c):
    delta = cmath.sqrt(b * b - 4 * a * c)
    return (-b + delta) / (2 * a), (-b - delta) / (2 * a)


z1, z2 = solve_quadratic(1, -(3 + 1j), 4 + 3j)
print("Example 3 roots:", z1, z2)
print("Vieta: sum =", z1 + z2, " product =", z1 * z2)

rng = np.random.default_rng(0)
a, b, c = rng.normal(size=3) + 1j * rng.normal(size=3)
roots = solve_quadratic(a, b, c)
residuals = [abs(a * z**2 + b * z + c) for z in roots]
print("\nrandom complex quadratic, residuals:", [f"{r:.1e}" for r in residuals])
print("matches np.roots:", np.allclose(sorted(roots, key=lambda z: z.real), sorted(np.roots([a, b, c]), key=lambda z: z.real)))
```

### 4.3 The conjugate-root theorem, symbolically

With sympy we can verify $p(\bar z) = \overline{p(z)}$ for a general cubic with *real* symbolic coefficients and *real* symbols $x, y$ for $z = x + iy$, and see it fail when one coefficient is allowed to be complex.

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
a0, a1, a2, a3 = sp.symbols("a0:4", real=True)
z = x + sp.I * y


def p(t):
    return a3 * t**3 + a2 * t**2 + a1 * t + a0


difference = sp.expand(p(sp.conjugate(z)) - sp.conjugate(p(z)))
print("real coefficients:    p(conj z) - conj(p(z)) =", sp.simplify(difference))

beta = sp.symbols("beta", real=True)


def q(t):
    return t**2 + sp.I * beta * t     # one complex coefficient


difference_q = sp.expand(q(sp.conjugate(z)) - sp.conjugate(q(z)))
print("complex coefficient:  q(conj z) - conj(q(z)) =", sp.simplify(difference_q))
```

The second difference, $2\beta(y + ix) = 2i\beta\,\bar z$, vanishes only when $\beta = 0$ or $z = 0$: the theorem really does need real coefficients.

### 4.4 Factoring from one root (Example 4)

`sp.div` divides out the real quadratic built from the conjugate pair; `sp.factor` with `extension=sp.I` factors fully over $\mathbb{Q}(i)$.

```python
import sympy as sp

z = sp.symbols("z")
p = z**4 - 2 * z**3 + 6 * z**2 - 8 * z + 8
w = 1 + sp.I

print("p(1 + i) =", sp.expand(p.subs(z, w)), "  p(1 - i) =", sp.expand(p.subs(z, sp.conjugate(w))))

pair_factor = sp.expand((z - w) * (z - sp.conjugate(w)))
quotient, remainder = sp.div(p, pair_factor, z)
print("real quadratic from the pair:", pair_factor)
print("quotient:", quotient, "  remainder:", remainder)
print("factor over R (rationals):  ", sp.factor(p))
print("factor over C (Q(i)):       ", sp.factor(p, extension=sp.I))
```

### 4.5 Vieta's formulas for a general cubic

Expand $(z - r_1)(z - r_2)(z - r_3)$ symbolically and read off the coefficients as elementary symmetric polynomials. Then check them numerically on a random cubic with complex coefficients.

```python
import numpy as np
import sympy as sp

z, r1, r2, r3 = sp.symbols("z r1 r2 r3")
cubic = sp.Poly(sp.expand((z - r1) * (z - r2) * (z - r3)), z)
for power, coefficient in zip([3, 2, 1, 0], cubic.all_coeffs()):
    print(f"coefficient of z^{power}: {coefficient}")

rng = np.random.default_rng(1)
coefficients = rng.normal(size=4) + 1j * rng.normal(size=4)
a3, a2, a1, a0 = coefficients
roots = np.roots(coefficients)
e1 = roots.sum()
e2 = roots[0] * roots[1] + roots[0] * roots[2] + roots[1] * roots[2]
e3 = roots.prod()
print("\nnumerical check on a random complex cubic:")
print("e1 = -a2/a3:", np.isclose(e1, -a2 / a3))
print("e2 =  a1/a3:", np.isclose(e2, a1 / a3))
print("e3 = -a0/a3:", np.isclose(e3, -a0 / a3))
```

## 5. Visualizations

### 5.1 The two square roots: half the angle, root of the length

For $w = -5 + 12i$ (modulus $13$), the square roots $\pm(2 + 3i)$ have modulus $\sqrt{13}$ and arguments $\frac12 \operatorname{Arg} w$ and $\frac12 \operatorname{Arg} w + \pi$. They are antipodal: the two solutions of $\delta^2 = w$ always sit on a line through the origin.

```python
import cmath

import matplotlib.pyplot as plt
import numpy as np

w = -5 + 12j
delta = cmath.sqrt(w)
points = {"$w = -5 + 12i$": (w, "tab:blue", (0.4, 0.0)),
          "$\\delta = 2 + 3i$": (delta, "tab:red", (0.4, 0.0)),
          "$-\\delta = -2 - 3i$": (-delta, "tab:orange", (-1.0, -1.0))}

fig, ax = plt.subplots(figsize=(7, 7))
for label, (point, color, offset) in points.items():
    ax.annotate("", xy=(point.real, point.imag), xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color=color, linewidth=2))
    ax.text(point.real + offset[0], point.imag + offset[1], label, color=color, fontsize=12)

# Arcs marking the angle of w and the half angle of delta
for radius, angle, color in [(2.0, cmath.phase(w), "tab:blue"), (1.3, cmath.phase(delta), "tab:red")]:
    t = np.linspace(0, angle, 100)
    ax.plot(radius * np.cos(t), radius * np.sin(t), color=color, linewidth=1)

circle_t = np.linspace(0, 2 * np.pi, 300)
ax.plot(np.sqrt(13) * np.cos(circle_t), np.sqrt(13) * np.sin(circle_t), color="lightgray", linestyle="--",
        label="$|z| = \\sqrt{13}$")
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-8, 6)
ax.set_ylim(-5, 14)
ax.set_aspect("equal")
ax.set_title("Square roots halve the argument\nand are antipodal")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower right")
plt.show()
```

### 5.2 Real coefficients force mirror symmetry

Left: roots of three random polynomials of degree 9 with **real** coefficients; each picture is symmetric across the real axis, and each has at least one real root (odd degree). Right: the same with **complex** coefficients; the symmetry is gone.

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(7)
degree = 9

fig, (ax_real, ax_complex) = plt.subplots(1, 2, figsize=(12, 6))
for trial, color in enumerate(["tab:blue", "tab:red", "tab:green"]):
    real_coefficients = rng.normal(size=degree + 1)
    complex_coefficients = rng.normal(size=degree + 1) + 1j * rng.normal(size=degree + 1)
    for ax, coefficients in [(ax_real, real_coefficients), (ax_complex, complex_coefficients)]:
        roots = np.roots(coefficients)
        ax.scatter(roots.real, roots.imag, color=color, s=40, label=f"polynomial {trial + 1}")

for ax, title in [(ax_real, "Real coefficients: symmetric about Re axis"),
                  (ax_complex, "Complex coefficients: no symmetry")]:
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(-2.2, 2.2)
    ax.set_ylim(-2.2, 2.2)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="lower left", fontsize=9)
plt.tight_layout()
plt.show()
```

### 5.3 Vieta in the plane: midpoint and sum

For Example 3 the roots $z_1 = 2 - i$ and $z_2 = 1 + 2i$ are not conjugate, but they are still symmetric about their **midpoint** $-\frac{b}{2a} = \frac{3 + i}{2}$, at offsets $\pm\frac{\delta}{2} = \pm\frac{1 - 3i}{2}$. Their sum $z_1 + z_2 = 3 + i$ is the far corner of the parallelogram they span.

```python
import matplotlib.pyplot as plt
import numpy as np

z1, z2 = 2 - 1j, 1 + 2j
total = z1 + z2
midpoint = total / 2

fig, ax = plt.subplots(figsize=(7, 7))
for point, label, color in [(z1, "$z_1 = 2 - i$", "tab:red"), (z2, "$z_2 = 1 + 2i$", "tab:red"),
                            (total, "$z_1 + z_2 = 3 + i = -b/a$", "tab:blue")]:
    ax.annotate("", xy=(point.real, point.imag), xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color=color, linewidth=2))
    ax.text(point.real + 0.1, point.imag + 0.1, label, color=color, fontsize=11)

# Parallelogram sides
for start, end in [(z1, total), (z2, total)]:
    ax.plot([start.real, end.real], [start.imag, end.imag], color="gray", linestyle=":")

# Segment between the roots and its midpoint
ax.plot([z1.real, z2.real], [z1.imag, z2.imag], color="tab:red", linestyle="--", linewidth=1)
ax.scatter([midpoint.real], [midpoint.imag], color="black", zorder=3)
ax.text(midpoint.real + 0.1, midpoint.imag - 0.35, "midpoint $-b/2a$", fontsize=11)

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-1, 5.5)
ax.set_ylim(-2, 3)
ax.set_aspect("equal")
ax.set_title("Roots of $z^2 - (3+i)z + (4+3i)$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
plt.show()
```

### 5.4 Roots colliding and splitting

Watch the roots of $z^2 - 2z + c$ as the real parameter $c$ increases from $-3$ to $5$. The midpoint stays at $1$ (Vieta: sum $= 2$). For $c < 1$ ($\Delta > 0$) the two real roots approach each other along the real axis; at $c = 1$ they collide; for $c > 1$ they leave the axis **vertically**, as a conjugate pair $1 \pm i\sqrt{c - 1}$. Real roots can only become complex in pairs, and only by first colliding on the real axis.

```python
import matplotlib.pyplot as plt
import numpy as np

c_values = np.linspace(-3, 5, 41)

fig, ax = plt.subplots(figsize=(8, 6))
scatter = None
for c in c_values:
    roots = np.roots([1, -2, c]).astype(complex)
    scatter = ax.scatter(roots.real, roots.imag, c=[c, c], cmap="viridis", vmin=c_values[0], vmax=c_values[-1], s=25)

ax.scatter([1], [0], marker="x", color="black", s=100, zorder=3, label="collision at $c = 1$")
fig.colorbar(scatter, ax=ax, label="c")
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-2, 4)
ax.set_ylim(-2.5, 2.5)
ax.set_aspect("equal")
ax.set_title("Roots of $z^2 - 2z + c$ as $c$ increases")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper left")
plt.show()
```

## 7. Solutions

1. $z^2 + 6z + 25 = (z + 3)^2 + 16$, so $(z + 3)^2 = -16 = (4i)^2$ and $z = -3 \pm 4i$.
2. $u = -5$, $v = 12$, $|w| = 13$. $x^2 = \frac{13 - 5}{2} = 4$, $y^2 = \frac{13 + 5}{2} = 9$, and $v > 0$ means same signs: $\pm(2 + 3i)$. Check: $(2 + 3i)^2 = 4 + 12i - 9 = -5 + 12i$ ✓.
3. The conjugate $3 + i$ must also be a root, so the quadratic is $z^2 - 2\operatorname{Re}(3 - i)\,z + |3 - i|^2 = z^2 - 6z + 10$.
4. $\Delta = 4 - 4(1 + 2i) = -8i$. Square root: $u = 0$, $v = -8$, $|w| = 8$, $x^2 = y^2 = 4$, opposite signs: $\delta = \pm(2 - 2i)$; check $(2 - 2i)^2 = 4 - 8i - 4 = -8i$ ✓. Roots $z = \frac{2 \pm (2 - 2i)}{2}$: $z_1 = 2 - i$, $z_2 = i$. Vieta: $z_1 + z_2 = 2$ ✓, $z_1 z_2 = (2 - i)i = 2i + 1 = 1 + 2i$ ✓. They are not conjugates, and there is no reason they should be: the constant term $1 + 2i$ is not real.
5. Conjugate-root theorem: $2 - i$ is a root (mirror image across the real axis). Vieta: the three roots sum to $5$, and $(2 + i) + (2 - i) = 4$, so the third root is the real number $1$. Sketch: a pair at $2 \pm i$ and one point at $1$ on the real axis. Factoring confirms: $(z^2 - 4z + 5)(z - 1) = z^3 - 5z^2 + 9z - 5$.
6. With $a_3 = 1, a_2 = -2, a_1 = 3, a_0 = -4$: $e_1 = 2$, $e_2 = 3$, $e_3 = -(-4) = 4$. Then $\sum z_i^2 = e_1^2 - 2e_2 = 4 - 6 = -2$. (A sum of squares can be negative once complex roots are involved.)
7. $\Delta = (i)^2 - 8 = -9$, $\delta = 3i$, $z = \frac{-i \pm 3i}{2}$, so $z_1 = i$ and $z_2 = -2i$. The conjugate of $i$ is $-i$, which is not a root: $(-i)^2 + i(-i) + 2 = -1 + 1 + 2 = 2 \ne 0$. In the proof, the step $\overline{a_1} = a_1$ fails: $a_1 = i$ has $\overline{a_1} = -i$, so $\overline{p(z)} = \bar z^2 - i\bar z + 2$, a *different* polynomial evaluated at $\bar z$.
8. The $n$ roots (with multiplicity) split into real roots and non-real roots. By the conjugate-root theorem (with multiplicities, Exercise 9) the non-real roots pair up as $w, \bar w$ with $w \ne \bar w$, so there is an even number of them. If $n$ is odd, $n$ minus an even number is odd, hence at least $1$: there is a real root.
9. Each derivative $p^{(k)}$ of a real polynomial again has real coefficients, so by the theorem $p^{(k)}(\bar w) = \overline{p^{(k)}(w)}$. Thus $p^{(k)}(\bar w) = 0 \iff p^{(k)}(w) = 0$ for every $k$. The conditions "zero for $k < m$, non-zero for $k = m$" therefore hold at $\bar w$ exactly when they hold at $w$.

## 8. Conclusion

- Completing the square works verbatim in $\mathbb{C}$: $z = \frac{-b \pm \delta}{2a}$ with $\delta^2 = \Delta$, for any complex $a \ne 0, b, c$.
- The square roots of $u + vi$ are $\pm\left(\sqrt{\tfrac{|w| + u}{2}} + \operatorname{sgn}(v)\sqrt{\tfrac{|w| - u}{2}}\, i\right)$: half the angle, root of the length, antipodal.
- For **real** coefficients, $p(\bar z) = \overline{p(z)}$, so non-real roots come in conjugate pairs (with equal multiplicity), mirror images across the real axis. Each pair gives a real quadratic factor $z^2 - 2\operatorname{Re}(w)z + |w|^2$.
- Vieta: the coefficients are, up to sign, the elementary symmetric functions of the roots. For a quadratic the roots are symmetric about $-b/2a$.
- Complex coefficients break the mirror symmetry; Vieta and the quadratic formula still hold.
