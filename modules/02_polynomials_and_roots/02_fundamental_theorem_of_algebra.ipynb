# The Fundamental Theorem of Algebra

Over $\mathbb{R}$, the polynomial $x^2 + 1$ has no roots and $x^3 - 1$ has only one. Over $\mathbb{C}$ the story becomes perfectly regular: **every** non-constant polynomial has a root, so every polynomial of degree $n$ has exactly $n$ roots counted with multiplicity and splits into $n$ linear factors. This is the **fundamental theorem of algebra** (FTA). This notebook states it carefully, derives the factorization over $\mathbb{C}$ and the corresponding factorization over $\mathbb{R}$, and then explains *why* it is true with a picture you can compute: the image of a circle $|z| = R$ under $p$ winds around the origin $n$ times when $R$ is large and $0$ times when $R$ is small, so somewhere in between the curve must pass through $0$.

This is notebook 02 of module 02 (**Polynomials and roots**) in the **Learning Complex Numbers** series.

## 1. Introduction

Each enlargement of the number system was made to solve some equation. One might expect that solving $z^5 + z + 1 = 0$, or a polynomial with complex coefficients such as $z^3 - iz + 2$, would require yet another enlargement. The fundamental theorem of algebra says it never does: $\mathbb{C}$ is **algebraically closed**. Adjoining a single square root of $-1$ is enough to solve every polynomial equation.

This is the reason complex numbers dominate applied mathematics. An $n \times n$ matrix always has $n$ eigenvalues in $\mathbb{C}$, a linear ODE with constant coefficients always has $n$ characteristic roots, a rational transfer function always factors into poles and zeros. None of these statements is true over $\mathbb{R}$.

## 2. Theory

### 2.1 Statement

> **Fundamental theorem of algebra.** Every polynomial $p(z) = a_n z^n + \dots + a_1 z + a_0$ with complex coefficients and degree $n \ge 1$ ($a_n \ne 0$) has at least one root in $\mathbb{C}$.

The coefficients may be complex; the root is guaranteed to exist in $\mathbb{C}$ itself. That is what "algebraically closed" means.

### 2.2 From one root to $n$ roots: the factor theorem

**Factor theorem.** $p(w) = 0$ if and only if $(z - w)$ divides $p(z)$.

*Proof.* Divide $p$ by $z - w$: $p(z) = (z - w)q(z) + r$ with $\deg q = n - 1$ and constant remainder $r$. Setting $z = w$ gives $r = p(w)$. $\blacksquare$

Apply the FTA to $p$ to get a root $z_1$ and write $p(z) = (z - z_1)q_1(z)$; apply it again to $q_1$ (degree $n-1$), and so on. After $n$ steps:

$$
p(z) = a_n (z - z_1)(z - z_2)\cdots(z - z_n).
$$

The $z_k$ need not be distinct. Grouping equal ones,

$$
p(z) = a_n (z - w_1)^{m_1}(z - w_2)^{m_2}\cdots(z - w_r)^{m_r}, \qquad m_1 + m_2 + \dots + m_r = n,
$$

where $m_j$ is the **multiplicity** of the root $w_j$. So: *a polynomial of degree $n$ has exactly $n$ roots counted with multiplicity.* The factorization is unique up to order, and there can be no other roots, since a product of non-zero numbers is non-zero.

A root of multiplicity $m \ge 2$ is also a root of $p'$: differentiating $(z - w)^m g(z)$ leaves a factor $(z - w)^{m-1}$.

### 2.3 Factorization over $\mathbb{R}$

If the coefficients are real, the conjugate-root theorem (notebook 01) pairs each non-real root $w$ with $\bar w$, with equal multiplicity. Each pair multiplies to a real quadratic with negative discriminant:

$$
(z - w)(z - \bar w) = z^2 - 2\operatorname{Re}(w)\, z + |w|^2, \qquad \Delta = 4\operatorname{Re}(w)^2 - 4|w|^2 = -4\operatorname{Im}(w)^2 < 0.
$$

> **Corollary.** Every real polynomial factors over $\mathbb{R}$ into real linear factors and real irreducible quadratic factors.

This is exactly the factorization needed for partial fractions in calculus (notebook 04).

### 2.4 Why it is true: the winding-number picture

Fix $R > 0$ and let $z = Re^{i\theta}$ run once around the circle $|z| = R$ as $\theta$ goes from $0$ to $2\pi$. Then $p(z)$ traces a closed curve $\Gamma_R$ in the plane. If $\Gamma_R$ avoids $0$, its **winding number** around $0$ is

$$
W(R) = \frac{1}{2\pi}\Big(\text{total continuous change of } \arg p(Re^{i\theta}) \text{ as } \theta: 0 \to 2\pi\Big),
$$

an integer, because the curve returns to its starting point. It counts how many times $\Gamma_R$ goes counter-clockwise around $0$.

**Large $R$: $W(R) = n$.** Write $p(z) = a_n z^n\left(1 + \varepsilon(z)\right)$ with $\varepsilon(z) = \frac{a_{n-1}}{a_n z} + \dots + \frac{a_0}{a_n z^n}$. For $|z| = R$ large, $|\varepsilon(z)| < 1$, so $1 + \varepsilon(z)$ stays in the disk of radius 1 around $1$, which lies in the right half-plane and contributes no net winding. The factor $a_n z^n = a_n R^n e^{in\theta}$ winds exactly $n$ times (De Moivre: the argument is multiplied by $n$). Arguments add under multiplication, so $W(R) = n$.

**Small $R$: $W(R) = 0$.** If $p(0) = a_0 = 0$ we already have a root. Otherwise, for $R$ tiny, $p(Re^{i\theta}) \approx a_0$ stays in a small disk around $a_0 \ne 0$, which does not contain $0$, so $W(R) = 0$.

**In between.** As $R$ varies continuously, the curve $\Gamma_R$ deforms continuously. An integer-valued quantity that varies continuously cannot change, *unless* the curve passes through $0$, where the winding number is undefined. Since $W$ goes from $0$ to $n \ge 1$, there must be some $R_0$ for which $\Gamma_{R_0}$ passes through $0$: there is a $z$ with $|z| = R_0$ and $p(z) = 0$.

More is true (the **argument principle**, proved in module 06): if $\Gamma_R$ avoids $0$, then

$$
W(R) = \text{number of roots of } p \text{ in } |z| < R, \text{ counted with multiplicity}.
$$

Each time $R$ passes the modulus of a root of multiplicity $m$, $W$ jumps by $m$. So the step function $R \mapsto W(R)$ is a picture of the moduli of the roots.

### 2.5 Other proofs (without full rigour)

The winding argument above becomes a complete proof once "winding number" is defined rigorously (via a continuous choice of argument, or via the integral $\frac{1}{2\pi i}\oint \frac{p'(z)}{p(z)}\,dz$). Other standard proofs:

- **Minimum modulus (d'Alembert, Argand).** $|p(z)| \to \infty$ as $|z| \to \infty$, so $|p|$ attains a minimum at some $z_0$. If $p(z_0) \ne 0$, expand $p(z_0 + h) = p(z_0) + c_k h^k + (\text{higher powers of } h)$ with $c_k \ne 0$. Choose the direction of $h$ so that $c_k h^k$ points opposite to $p(z_0)$; this uses the existence of $k$-th roots in $\mathbb{C}$. Then $|p(z_0 + h)| < |p(z_0)|$ for small $|h|$, a contradiction. Hence $p(z_0) = 0$. The contour plots in Section 5 show this: $|p|$ has no positive local minima.
- **Liouville's theorem (module 06).** If $p$ had no roots, $1/p$ would be a bounded entire function, hence constant.
- **Algebraic proofs** use the fact that real polynomials of odd degree have a real root (intermediate value theorem) plus Galois theory or induction on the power of $2$ in the degree.

Every proof uses some analysis or topology (completeness of $\mathbb{R}$, continuity). The FTA is not purely an algebraic fact.

## 3. Worked Examples

### Example 1: $z^4 - 1$ over $\mathbb{C}$ and over $\mathbb{R}$

$z^4 - 1 = (z^2 - 1)(z^2 + 1) = (z - 1)(z + 1)(z^2 + 1)$.

- Over $\mathbb{R}$: $(z - 1)(z + 1)(z^2 + 1)$, with $z^2 + 1$ irreducible ($\Delta = -4 < 0$).
- Over $\mathbb{C}$: $(z - 1)(z + 1)(z - i)(z + i)$. Four roots, the fourth roots of unity, at the corners of a square.

### Example 2: counting with multiplicity

$p(z) = z^3 - 3z + 2$. Try $z = 1$: $1 - 3 + 2 = 0$. Divide: $z^3 - 3z + 2 = (z - 1)(z^2 + z - 2) = (z - 1)(z - 1)(z + 2)$. So

$$
p(z) = (z - 1)^2 (z + 2).
$$

Distinct roots: $1$ and $-2$. Counted with multiplicity: $1, 1, -2$, three roots for degree three. Check that $1$ is a double root: $p'(z) = 3z^2 - 3$ and $p'(1) = 0$ ✓.

### Example 3: $z^4 + 4$, a real polynomial with no real roots

$z^4 + 4 > 0$ for real $z$, so all four roots are complex. Add and subtract $4z^2$:

$$
z^4 + 4 = (z^4 + 4z^2 + 4) - 4z^2 = (z^2 + 2)^2 - (2z)^2 = (z^2 - 2z + 2)(z^2 + 2z + 2).
$$

Each quadratic is irreducible over $\mathbb{R}$ ($\Delta = 4 - 8 = -4$). Over $\mathbb{C}$: $z^2 - 2z + 2 = 0$ gives $1 \pm i$ and $z^2 + 2z + 2 = 0$ gives $-1 \pm i$. The four roots $\pm 1 \pm i$ form a square of side 2, as they must: they are the fourth roots of $-4 = 4e^{i\pi}$, with modulus $\sqrt[4]{4} = \sqrt2$ and arguments $\pm\frac{\pi}{4}, \pm\frac{3\pi}{4}$.

### Example 4: a winding count by hand

$p(z) = z^2 + 1$, roots $\pm i$, both on $|z| = 1$.

- On $|z| = 2$: $p(2e^{i\theta}) = 1 + 4e^{2i\theta}$, a circle of radius $4$ centred at $1$, traversed **twice**. It encloses $0$, so $W(2) = 2$: both roots lie inside $|z| < 2$.
- On $|z| = \frac12$: $p(\tfrac12 e^{i\theta}) = 1 + \tfrac14 e^{2i\theta}$, a circle of radius $\frac14$ centred at $1$. It does not enclose $0$, so $W(\tfrac12) = 0$.
- On $|z| = 1$: $1 + e^{2i\theta}$ is the circle of radius $1$ centred at $1$, which **passes through $0$** at $\theta = \pm\frac{\pi}{2}$, that is, at $z = \pm i$. That is where the winding number jumps from $0$ to $2$.

### Example 5: a complex-coefficient example

$p(z) = z^2 - 2iz - 1 = (z - i)^2$. The only root is $i$, with multiplicity $2$. The FTA still says "two roots", namely $i, i$; the conjugate $-i$ is not a root because the coefficients are not real.

## 4. Python Examples

### 4.1 Roots with multiplicity, and factorization over different fields

`sp.roots` returns a dictionary {root: multiplicity}. `sp.factor` factors over $\mathbb{Q}$ by default; `extension=sp.I` allows $i$, and `extension=[sp.I, sp.sqrt(3)]` allows both.

```python
import sympy as sp

z = sp.symbols("z")

for polynomial in [z**4 - 1, z**3 - 3 * z + 2, z**4 + 4, z**2 - 2 * sp.I * z - 1]:
    multiplicities = sp.roots(polynomial, z)
    print(f"p = {polynomial}")
    print(f"   roots with multiplicity: {multiplicities}   total = {sum(multiplicities.values())}"
          f" = degree {sp.degree(polynomial, z)}")
    print(f"   factor over Q:    {sp.factor(polynomial)}")
    print(f"   factor over Q(i): {sp.factor(polynomial, extension=sp.I)}")
    print()

cubic = z**3 + 8
print("z^3 + 8 over Q:               ", sp.factor(cubic))
print("z^3 + 8 over Q(i, sqrt 3):    ", sp.factor(cubic, extension=[sp.I, sp.sqrt(3)]))
```

### 4.2 Rebuilding a polynomial from its roots

`np.roots` finds the roots numerically; `np.poly` multiplies $(z - z_1)\cdots(z - z_n)$ back out. Up to rounding, we recover the coefficients: the factorization $p = a_n \prod (z - z_k)$ in action.

```python
import numpy as np

rng = np.random.default_rng(3)
coefficients = rng.normal(size=8) + 1j * rng.normal(size=8)   # degree 7, complex coefficients
roots = np.roots(coefficients)
rebuilt = coefficients[0] * np.poly(roots)

print("degree:", len(coefficients) - 1, "  number of roots found:", len(roots))
print("max |p(root)|:", np.max(np.abs(np.polyval(coefficients, roots))))
print("coefficients recovered:", np.allclose(rebuilt, coefficients))
```

### 4.3 Computing a winding number

Sample $\theta$ finely, compute $w_k = p(Re^{i\theta_k})$, and add up the small angle changes $\arg(w_{k+1}/w_k) \in (-\pi, \pi]$. Using the ratio avoids the jump of $\operatorname{Arg}$ across the negative real axis: as long as consecutive samples are close, each step is small and the sum tracks the continuous argument.

Our test polynomial has roots at $\frac12$, $\frac52$ and the conjugate pair $-1 \pm \frac32 i$ (modulus $\sqrt{13}/2 \approx 1.80$):

$$
p(z) = \left(z - \tfrac12\right)\left(z - \tfrac52\right)\left(z^2 + 2z + \tfrac{13}{4}\right) = z^4 - z^3 - 1.5z^2 - 7.25z + 4.0625.
$$

```python
import numpy as np

roots = np.array([0.5, 2.5, -1 + 1.5j, -1 - 1.5j])
coefficients = np.poly(roots).real
print("coefficients:", coefficients)
print("moduli of the roots:", np.round(np.abs(roots), 4))


def winding_number(coefficients, radius, samples=4000):
    theta = np.linspace(0, 2 * np.pi, samples + 1)
    values = np.polyval(coefficients, radius * np.exp(1j * theta))
    steps = np.angle(values[1:] / values[:-1])
    return steps.sum() / (2 * np.pi)


print()
for radius in [0.2, 0.4, 1.0, 1.7, 2.0, 2.4, 3.0, 10.0]:
    roots_inside = np.sum(np.abs(roots) < radius)
    print(f"R = {radius:5.1f}:  winding number = {round(winding_number(coefficients, radius), 6) + 0.0:.6f}   "
          f"roots with |z| < R: {roots_inside}")
```

The winding number is an integer up to rounding, and it equals the number of roots inside the circle. It jumps by $2$ at $R = 1.80$, where the conjugate pair crosses together.

### 4.4 The Cauchy bound: how large is "large R"?

All roots satisfy $|z| < 1 + \max_{k < n} |a_k / a_n|$ (Exercise 9). On any circle beyond that radius the winding number is already $n$.

```python
import numpy as np

rng = np.random.default_rng(11)
for trial in range(4):
    degree = rng.integers(3, 9)
    coefficients = rng.normal(size=degree + 1) * 3 + 1j * rng.normal(size=degree + 1)
    bound = 1 + np.max(np.abs(coefficients[1:] / coefficients[0]))
    largest_root = np.max(np.abs(np.roots(coefficients)))
    print(f"degree {degree}: largest |root| = {largest_root:6.3f}  <  Cauchy bound {bound:7.3f}: {largest_root < bound}")
```

## 5. Visualizations

### 5.1 Images of circles under $p$

For our quartic, each panel shows the image $\Gamma_R = p(\{|z| = R\})$ for one radius; the black dot is the starting point $p(R)$ and the arrow the direction of travel. The red cross marks the origin. Count how many times each curve goes around it: $0, 1, 3, 4$.

```python
import matplotlib.pyplot as plt
import numpy as np

roots = np.array([0.5, 2.5, -1 + 1.5j, -1 - 1.5j])
coefficients = np.poly(roots).real
theta = np.linspace(0, 2 * np.pi, 2000)


def winding_number(coefficients, radius, samples=4000):
    t = np.linspace(0, 2 * np.pi, samples + 1)
    values = np.polyval(coefficients, radius * np.exp(1j * t))
    return np.angle(values[1:] / values[:-1]).sum() / (2 * np.pi)


fig, axes = plt.subplots(2, 2, figsize=(11, 11))
for ax, radius in zip(axes.flat, [0.3, 1.2, 2.2, 3.0]):
    image = np.polyval(coefficients, radius * np.exp(1j * theta))
    ax.plot(image.real, image.imag, color="tab:blue", linewidth=1.5)
    # Arrow showing the direction of travel
    k = 150
    ax.annotate("", xy=(image[k + 5].real, image[k + 5].imag), xytext=(image[k].real, image[k].imag),
                arrowprops=dict(arrowstyle="->", color="tab:blue", linewidth=2))
    ax.scatter([0], [0], marker="x", color="red", s=80, zorder=3)
    ax.scatter([image[0].real], [image[0].imag], color="black", s=20, zorder=3)
    ax.axhline(0, color="black", linewidth=0.6)
    ax.axvline(0, color="black", linewidth=0.6)
    ax.set_aspect("equal")
    ax.set_title(f"Image of $|z| = {radius}$:  winding number {round(winding_number(coefficients, radius))}")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

At $R = 0.3$ the image is a small loop near $p(0) = 4.0625$ that does not surround $0$. At $R = 3$ the image looks like a slightly distorted circle traversed four times, the behaviour of $z^4$. In between, loops are gained one or two at a time.

### 5.2 The moment of crossing

Zoom in on what happens as $R$ passes the modulus $\sqrt{13}/2$ of the conjugate pair. The two arcs visible near the origin are the images of the parts of the circle close to $w = -1 + 1.5i$ and to $\bar w$; by the symmetry $p(\bar z) = \overline{p(z)}$ they are mirror images and cross on the real axis. Just before, their crossing point lies to the right of $0$; exactly at $R = |w|$ both arcs pass **through** $0$ (once for $w$, once for $\bar w$); just after, the crossing point has moved to the left and two extra loops now enclose the origin.

```python
import matplotlib.pyplot as plt
import numpy as np

roots = np.array([0.5, 2.5, -1 + 1.5j, -1 - 1.5j])
coefficients = np.poly(roots).real
theta = np.linspace(0, 2 * np.pi, 4000)
critical_radius = abs(roots[2])

fig, axes = plt.subplots(1, 3, figsize=(15, 5.2))
for ax, radius, label in zip(axes, [critical_radius - 0.12, critical_radius, critical_radius + 0.12],
                             ["just inside", "through the pair", "just outside"]):
    image = np.polyval(coefficients, radius * np.exp(1j * theta))
    ax.plot(image.real, image.imag, color="tab:blue", linewidth=1.5)
    ax.scatter([0], [0], marker="x", color="red", s=80, zorder=3)
    ax.axhline(0, color="black", linewidth=0.6)
    ax.axvline(0, color="black", linewidth=0.6)
    ax.set_xlim(-6, 6)
    ax.set_ylim(-6, 6)
    ax.set_aspect("equal")
    ax.set_title(f"$R = {radius:.3f}$ ({label})")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
plt.suptitle("Winding number jumps from 1 to 3 as the circle sweeps over $-1 \\pm 1.5i$")
plt.tight_layout()
plt.show()
```

### 5.3 Winding number as a function of $R$

The step function $W(R)$ jumps exactly at the moduli of the roots, by the multiplicity at that modulus. Reading the plot from left to right is the FTA argument in one picture: start at $0$, end at $n$, and the only way to climb is to cross a root.

```python
import matplotlib.pyplot as plt
import numpy as np

roots = np.array([0.5, 2.5, -1 + 1.5j, -1 - 1.5j])
coefficients = np.poly(roots).real
root_moduli = np.unique(np.round(np.abs(roots), 10))


def winding_number(coefficients, radius, samples=3000):
    t = np.linspace(0, 2 * np.pi, samples + 1)
    values = np.polyval(coefficients, radius * np.exp(1j * t))
    return np.angle(values[1:] / values[:-1]).sum() / (2 * np.pi)


# Avoid radii that hit a root modulus exactly
radii = np.linspace(0.01, 3.5, 351)
radii = radii[np.min(np.abs(radii[:, None] - root_moduli[None, :]), axis=1) > 1e-3]
windings = [round(winding_number(coefficients, radius)) for radius in radii]

plt.figure(figsize=(9, 5))
plt.step(radii, windings, where="post", color="tab:blue", linewidth=2, label="winding number $W(R)$")
for modulus in root_moduli:
    plt.axvline(modulus, color="tab:red", linestyle="--", linewidth=1)
plt.plot([], [], color="tab:red", linestyle="--", label="moduli of roots")
plt.xlabel("radius R")
plt.ylabel("W(R)")
plt.yticks(range(0, 5))
plt.title("Winding number of $p(|z| = R)$ around 0 counts the roots inside")
plt.grid(True, alpha=0.3)
plt.legend(loc="upper left")
plt.show()
```

### 5.4 The modulus landscape: no positive local minima

A contour plot of $\log_{10}|p(z)|$ over the plane. The only "valleys" go down to $-\infty$, at the roots. This is the minimum-modulus argument of 2.5 made visible: wherever $p(z) \ne 0$ there is a downhill direction. The dashed circles are $|z| = 0.5$, $1.80$ and $2.5$, where the winding number jumps.

```python
import matplotlib.pyplot as plt
import numpy as np

roots = np.array([0.5, 2.5, -1 + 1.5j, -1 - 1.5j])
coefficients = np.poly(roots).real

x = np.linspace(-3.2, 3.5, 500)
y = np.linspace(-3.2, 3.2, 500)
X, Y = np.meshgrid(x, y)
log_modulus = np.log10(np.abs(np.polyval(coefficients, X + 1j * Y)) + 1e-12)

fig, ax = plt.subplots(figsize=(8, 7.5))
filled = ax.contourf(X, Y, log_modulus, levels=30, cmap="viridis")
ax.contour(X, Y, log_modulus, levels=30, colors="white", linewidths=0.3)
fig.colorbar(filled, ax=ax, label="$\\log_{10}|p(z)|$")
ax.scatter(roots.real, roots.imag, color="red", s=60, zorder=3, label="roots")

t = np.linspace(0, 2 * np.pi, 300)
for radius in np.unique(np.round(np.abs(roots), 6)):
    ax.plot(radius * np.cos(t), radius * np.sin(t), color="white", linestyle="--", linewidth=1)

ax.axhline(0, color="black", linewidth=0.6)
ax.axvline(0, color="black", linewidth=0.6)
ax.set_aspect("equal")
ax.set_title("$\\log_{10}|p(z)|$: the only minima are the roots")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.legend(loc="upper left")
plt.show()
```

### 5.5 Factorization pictured: $z^4 + 4$

Roots on the circle of radius $\sqrt2$. Each colour is one real irreducible quadratic factor; each factor collects a conjugate pair.

```python
import matplotlib.pyplot as plt
import numpy as np

factor_roots = {
    "$z^2 - 2z + 2$": np.roots([1, -2, 2]),
    "$z^2 + 2z + 2$": np.roots([1, 2, 2]),
}

fig, ax = plt.subplots(figsize=(6.5, 6.5))
t = np.linspace(0, 2 * np.pi, 300)
ax.plot(np.sqrt(2) * np.cos(t), np.sqrt(2) * np.sin(t), color="lightgray", linestyle="--", label="$|z| = \\sqrt{2}$")
for (label, roots), color in zip(factor_roots.items(), ["tab:blue", "tab:orange"]):
    ax.scatter(roots.real, roots.imag, s=100, color=color, zorder=3, label=label)
    ax.plot(roots.real, roots.imag, color=color, linestyle=":", linewidth=1.5)
    for root in roots:
        real_sign = "-" if root.real < 0 else ""
        imag_sign = "-" if root.imag < 0 else "+"
        ax.text(root.real * 1.25, root.imag * 1.25, f"${real_sign}1 {imag_sign} i$", ha="center", va="center", fontsize=12)

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-2.2, 2.2)
ax.set_ylim(-2.2, 2.2)
ax.set_aspect("equal")
ax.set_title("$z^4 + 4 = (z^2 - 2z + 2)(z^2 + 2z + 2)$")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower left", fontsize=9)
plt.show()
```

## 7. Solutions

1. $z^5 - z^3 = z^3(z^2 - 1) = z^3 (z - 1)(z + 1)$. Roots: $0$ (multiplicity 3), $1$ (1), $-1$ (1). Total $3 + 1 + 1 = 5$ ✓.
2. $z = -2$ is a root, so $z^3 + 8 = (z + 2)(z^2 - 2z + 4)$. The quadratic has $\Delta = 4 - 16 = -12$, roots $1 \pm i\sqrt3$. Over $\mathbb{R}$: $(z + 2)(z^2 - 2z + 4)$. Over $\mathbb{C}$: $(z + 2)(z - 1 - i\sqrt3)(z - 1 + i\sqrt3)$.
3. $(z - 1)^2 (z - i) = (z^2 - 2z + 1)(z - i) = z^3 - (2 + i)z^2 + (1 + 2i)z - i$. Not real: it has the root $i$ but not $-i$, which would violate the conjugate-root theorem.
4. $|z| = \frac12$: $p = \frac14 e^{2i\theta} - 1$ is the circle of radius $\frac14$ about $-1$, traversed twice; it does not surround $0$, so $W = 0$ (both roots $\pm1$ lie outside). $|z| = 2$: $p = 4e^{2i\theta} - 1$ is the circle of radius $4$ about $-1$, traversed twice, surrounding $0$, so $W = 2$ (both roots inside).
5. Conjugates $-2i$ and $1 + i$ must also be roots, so the degree is at least 4: $(z^2 + 4)(z^2 - 2z + 2) = z^4 - 2z^3 + 6z^2 - 8z + 8$.
6. Numerically: $1$ root in $|z| < 1$, $3$ in $|z| < 1.3$, $5$ in $|z| < 2$. The moduli of the roots are about $0.332$, $1.263$ (pair) and $1.374$ (pair).
7. On $|z| = 1$: $p(z) = 3z(1 + \varepsilon(z))$ with $\varepsilon(z) = \frac{z^5 + 1}{3z}$ and $|\varepsilon(z)| \le \frac{2}{3} < 1$. So $1 + \varepsilon$ stays in the disk of radius $\frac23$ about $1$, which excludes $0$ and contributes no winding, while $3z$ winds once. Hence $W(1) = 1$ and, by the argument principle, exactly one root lies in $|z| < 1$. (This is a special case of **Rouché's theorem**.)
8. Let $p$ be real of degree $n \ge 3$. By the FTA it has a root $w$. If $w$ is real, $(z - w)$ is a real factor of degree $1 < n$. If $w$ is not real, $\bar w$ is also a root, and $(z - w)(z - \bar w) = z^2 - 2\operatorname{Re}(w)z + |w|^2$ is a real factor of degree $2 < n$; the quotient is real because dividing real polynomials gives real quotients. Either way $p$ is reducible.
9. Let $|z| = R \ge 1 + M$, so $R > 1$. Then
   $$|a_{n-1}z^{n-1} + \dots + a_0| \le |a_n| M (R^{n-1} + \dots + 1) = |a_n| M \frac{R^n - 1}{R - 1} < |a_n| M \frac{R^n}{R - 1} \le |a_n| R^n,$$
   using $M/(R - 1) \le 1$. So $|p(z)| \ge |a_n|R^n - |a_{n-1}z^{n-1} + \dots + a_0| > 0$: no root has $|z| \ge 1 + M$.

Numerical checks:

```python
import numpy as np
import sympy as sp

z = sp.symbols("z")
print("Ex 1:", sp.roots(z**5 - z**3, z))
print("Ex 2:", sp.factor(z**3 + 8), "|", sp.factor(z**3 + 8, extension=[sp.I, sp.sqrt(3)]))
print("Ex 3:", sp.expand((z - 1) ** 2 * (z - sp.I)))


def winding_number(coefficients, radius, samples=4000):
    theta = np.linspace(0, 2 * np.pi, samples + 1)
    values = np.polyval(coefficients, radius * np.exp(1j * theta))
    return np.angle(values[1:] / values[:-1]).sum() / (2 * np.pi)


print("Ex 4: W(1/2) =", round(winding_number([1, 0, -1], 0.5)), " W(2) =", round(winding_number([1, 0, -1], 2)))
print("Ex 5:", sp.expand((z**2 + 4) * (z**2 - 2 * z + 2)))

quintic = [1, 0, 0, 0, 3, 1]
print("Ex 6:", {radius: round(winding_number(quintic, radius)) for radius in [1.0, 1.3, 2.0]},
      " root moduli:", np.round(np.sort(np.abs(np.roots(quintic))), 3))

coefficients = np.array([2, -3, 0, 5, -7, 1])
bound = 1 + np.max(np.abs(coefficients[1:] / coefficients[0]))
print("Ex 9: max |root| =", round(np.max(np.abs(np.roots(coefficients))), 4), " bound =", bound)
```

## 8. Conclusion

- **FTA:** every non-constant complex polynomial has a root in $\mathbb{C}$. By the factor theorem, a degree-$n$ polynomial is $a_n\prod_{k=1}^n (z - z_k)$: exactly $n$ roots counted with multiplicity.
- Over $\mathbb{R}$, conjugate pairs combine into irreducible quadratics: every real polynomial is a product of real linear and real irreducible quadratic factors.
- **Winding picture:** the image of $|z| = R$ under $p$ winds $0$ times for small $R$ and $n$ times for large $R$; the winding number can only change when the image passes through $0$, that is, when the circle passes through a root. More precisely, $W(R)$ counts the roots inside $|z| < R$ (argument principle, module 06).
- Other routes to the FTA: the minimum modulus argument ($|p|$ has no positive local minimum) and Liouville's theorem.
- In Python: `sp.roots` (with multiplicities), `sp.factor(..., extension=...)`, `np.roots` and `np.poly`, and a winding number from summed `np.angle` of consecutive ratios.
