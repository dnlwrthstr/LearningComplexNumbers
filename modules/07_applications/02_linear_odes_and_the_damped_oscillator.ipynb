# Linear ODEs: the Damped Oscillator via $e^{\lambda t}$

A mass on a spring, a pendulum swinging through air, the needle of a measuring instrument, a car's suspension, the RLC circuit of the previous notebook: all obey the same second-order linear differential equation $\ddot x + 2\zeta\omega_0\dot x + \omega_0^2 x = 0$. Guessing a solution of the form $e^{\lambda t}$ turns the differential equation into a **quadratic equation** for $\lambda$ — and when that quadratic has complex roots, the complex exponential $e^{\lambda t}$ quietly produces the oscillation.

This notebook solves the damped oscillator completely, reads its behaviour (underdamped, critically damped, overdamped) off the **position of the roots in the complex plane**, and then treats the driven oscillator and resonance with a single complex amplitude.

This is notebook 02 of module 07 (**Applications**) in the **Learning Complex Numbers** series.

## 1. Introduction

Pull a mass on a spring and let go. If there is no friction it oscillates forever; with a little friction the swings die away; with a lot of friction (imagine the spring immersed in honey) it creeps back to rest without ever overshooting. These three behaviours look qualitatively different, yet they come from one equation with one parameter changed.

The complex numbers explain why. The equation has solutions $e^{\lambda t}$, where $\lambda$ is a root of a quadratic. Real roots give pure exponential decay. Complex roots $\lambda = -\sigma \pm i\omega_d$ give

$$
e^{\lambda t} = e^{-\sigma t}\,e^{\pm i\omega_d t} = e^{-\sigma t}\left(\cos\omega_d t \pm i\sin\omega_d t\right),
$$

an exponential envelope times an oscillation. The **real part of a root is the decay rate, the imaginary part is the frequency**. A single picture of the two roots in the complex plane tells you everything about the motion.

## 2. Theory

### 2.1 The damped oscillator

A mass $m$ on a spring of stiffness $k$ with friction force $-c\,\dot x$ obeys Newton's law $m\ddot x = -kx - c\dot x$, i.e.

$$
m\ddot x + c\dot x + kx = 0 \qquad\Longleftrightarrow\qquad \ddot x + 2\zeta\omega_0\,\dot x + \omega_0^2\,x = 0,
$$

with **natural frequency** $\omega_0 = \sqrt{k/m}$ and dimensionless **damping ratio** $\zeta = \dfrac{c}{2\sqrt{mk}}$. The series RLC circuit $L\ddot q + R\dot q + q/C = 0$ is the same equation with $\omega_0 = 1/\sqrt{LC}$ and $2\zeta\omega_0 = R/L$.

### 2.2 The ansatz $e^{\lambda t}$ and the characteristic equation

The exponential is the function that differentiation only rescales: $\frac{d}{dt}e^{\lambda t} = \lambda e^{\lambda t}$, and this holds for **complex** $\lambda = a + ib$ too, because

$$
\frac{d}{dt}\left(e^{at}(\cos bt + i\sin bt)\right) = e^{at}\big((a\cos bt - b\sin bt) + i(a\sin bt + b\cos bt)\big) = (a + ib)\,e^{(a+ib)t}.
$$

Substituting $x = e^{\lambda t}$ gives $(\lambda^2 + 2\zeta\omega_0\lambda + \omega_0^2)e^{\lambda t} = 0$. Since $e^{\lambda t} \ne 0$, $\lambda$ must solve the **characteristic equation**

$$
p(\lambda) = \lambda^2 + 2\zeta\omega_0\lambda + \omega_0^2 = 0, \qquad \lambda_{1,2} = \omega_0\left(-\zeta \pm \sqrt{\zeta^2 - 1}\right).
$$

The same trick works for any order: $a_n x^{(n)} + \dots + a_1\dot x + a_0 x = 0$ has characteristic polynomial $a_n\lambda^n + \dots + a_0$, which by the fundamental theorem of algebra has $n$ complex roots.

### 2.3 From complex solutions to real ones

The equation is **linear with real coefficients**. Two consequences:

1. Any linear combination of solutions is a solution, so $x(t) = C_1e^{\lambda_1 t} + C_2 e^{\lambda_2 t}$ is the general solution when $\lambda_1 \neq \lambda_2$.
2. If $z(t)$ is a complex solution, then so is $\overline{z(t)}$, hence so are $\mathrm{Re}\,z = \frac{z + \bar z}{2}$ and $\mathrm{Im}\,z = \frac{z - \bar z}{2i}$. Real coefficients also force complex roots to come in **conjugate pairs** $\lambda, \bar\lambda$.

So with a complex root $\lambda = -\sigma + i\omega_d$, every real solution can be written with **one complex constant** $C$:

$$
x(t) = \mathrm{Re}\left(C e^{\lambda t}\right) = |C|\,e^{-\sigma t}\cos(\omega_d t + \arg C).
$$

Matching $x(0) = x_0$ and $\dot x(0) = v_0$: writing $x = e^{-\sigma t}(A\cos\omega_d t + B\sin\omega_d t)$ we need $A = x_0$ and $-\sigma A + \omega_d B = v_0$, and $\mathrm{Re}(Ce^{\lambda t})$ has $A = \mathrm{Re}\,C$, $B = -\mathrm{Im}\,C$, so

$$
C = x_0 - i\,\frac{v_0 + \sigma x_0}{\omega_d}.
$$

### 2.4 Three regimes, one picture

| Regime | Condition | Roots | Real solution |
|---|---|---|---|
| Undamped | $\zeta = 0$ | $\pm i\omega_0$ | $A\cos\omega_0 t + B\sin\omega_0 t$ |
| Underdamped | $0 < \zeta < 1$ | $-\zeta\omega_0 \pm i\omega_d$, $\ \omega_d = \omega_0\sqrt{1 - \zeta^2}$ | $e^{-\zeta\omega_0 t}(A\cos\omega_d t + B\sin\omega_d t)$ |
| Critically damped | $\zeta = 1$ | $-\omega_0$ (double) | $(A + Bt)e^{-\omega_0 t}$ |
| Overdamped | $\zeta > 1$ | two negative reals | $Ae^{\lambda_1 t} + Be^{\lambda_2 t}$ |

**Geometry of the roots.** For $\zeta \le 1$ the roots are conjugates with product $\lambda\bar\lambda = |\lambda|^2 = \omega_0^2$, so they lie on the **circle of radius $\omega_0$**, at angle $\arccos\zeta$ from the negative real axis. As $\zeta$ grows from $0$ to $1$ the two roots slide around the circle from $\pm i\omega_0$ and meet at $-\omega_0$; for $\zeta > 1$ they split apart along the negative real axis, one towards $0$ and one towards $-\infty$.

- **Re** $\lambda$ = decay rate (negative real part means decay, positive means growth).
- **Im** $\lambda$ = angular frequency of oscillation.
- All roots in the open left half-plane $\iff$ every solution decays to $0$: the system is **stable**.

The critically damped case returns to rest fastest without overshooting: the slowest root $-\zeta\omega_0 + \omega_0\sqrt{\zeta^2-1}$ of an overdamped system is closer to $0$ than $-\omega_0$.

### 2.5 The driven oscillator and the complex amplitude

Now push with a periodic force: $\ddot x + 2\zeta\omega_0\dot x + \omega_0^2 x = F\cos\omega t$. Exactly as with phasors, write the force as $\mathrm{Re}(Fe^{i\omega t})$ and look for $x_p = \mathrm{Re}(Xe^{i\omega t})$. Each derivative multiplies by $i\omega$, so

$$
\left(\omega_0^2 - \omega^2 + 2i\zeta\omega_0\omega\right)X = F \qquad\Longrightarrow\qquad X = \frac{F}{p(i\omega)} = \frac{F}{\omega_0^2 - \omega^2 + 2i\zeta\omega_0\omega}.
$$

The **transfer function** $H(i\omega) = 1/p(i\omega)$ is the characteristic polynomial, inverted and evaluated on the imaginary axis. Its modulus is the amplitude gain and minus its argument the **phase lag** $\delta$:

$$
|X| = \frac{F}{\sqrt{(\omega_0^2 - \omega^2)^2 + (2\zeta\omega_0\omega)^2}}, \qquad \delta = \arg\left(\omega_0^2 - \omega^2 + 2i\zeta\omega_0\omega\right) \in (0, \pi).
$$

- Slow driving ($\omega \ll \omega_0$): $X \approx F/\omega_0^2$, in phase.
- At $\omega = \omega_0$: $X = \dfrac{F}{2i\zeta\omega_0^2}$, a lag of exactly $90°$, large when $\zeta$ is small.
- Fast driving ($\omega \gg \omega_0$): $X \approx -F/\omega^2$, small and in antiphase ($180°$ lag).
- The amplitude peaks at $\omega_r = \omega_0\sqrt{1 - 2\zeta^2}$ (when $\zeta < 1/\sqrt2$): **resonance**.

The full solution is $x = x_p + x_h$: steady state plus a **transient** $x_h$ from section 2.4, which dies away at the rate set by the roots.

## 3. Worked Examples

### Example 1 — underdamped: $\ddot x + 2\dot x + 5x = 0$, $x(0) = 1$, $\dot x(0) = 0$

1. Characteristic equation $\lambda^2 + 2\lambda + 5 = 0$: $\lambda = \frac{-2 \pm\sqrt{4 - 20}}{2} = -1 \pm 2i$.
2. So $\sigma = 1$, $\omega_d = 2$ (and $\omega_0 = \sqrt5$, $\zeta = 1/\sqrt5 \approx 0.447$).
3. $C = x_0 - i\frac{v_0 + \sigma x_0}{\omega_d} = 1 - \tfrac12 i$.
4. $x(t) = \mathrm{Re}\left((1 - \tfrac i2)e^{(-1 + 2i)t}\right) = e^{-t}\left(\cos 2t + \tfrac12\sin 2t\right)$.

Check: $x(0) = 1$ and $\dot x(0) = -1 + \tfrac12\cdot 2 = 0$. ✓

### Example 2 — critically damped: $\ddot x + 4\dot x + 4x = 0$, $x(0) = 1$, $\dot x(0) = 0$

$\lambda^2 + 4\lambda + 4 = (\lambda + 2)^2$: double root $-2$. The second solution is $te^{-2t}$, so $x = (A + Bt)e^{-2t}$. Then $A = 1$ and $\dot x(0) = B - 2A = 0$ gives $B = 2$: $x(t) = (1 + 2t)e^{-2t}$.

### Example 3 — overdamped: $\ddot x + 5\dot x + 4x = 0$, $x(0) = 1$, $\dot x(0) = 0$

$\lambda^2 + 5\lambda + 4 = (\lambda + 1)(\lambda + 4)$, roots $-1$ and $-4$. From $A + B = 1$ and $-A - 4B = 0$: $A = \tfrac43$, $B = -\tfrac13$, so $x(t) = \tfrac43 e^{-t} - \tfrac13 e^{-4t}$.

### Example 4 — undamped: $\ddot x + x = 0$

$\lambda^2 + 1 = 0$, so $\lambda = \pm i$: the complex solutions are $e^{it}$ and $e^{-it}$, and their real and imaginary parts are $\cos t$ and $\sin t$. The familiar harmonic oscillator is Euler's formula in disguise.

### Example 5 — driven: $\ddot x + 2\dot x + 5x = 10\cos 2t$

1. With $\omega = 2$: $p(i\omega) = (2i)^2 + 2(2i) + 5 = 1 + 4i$.
2. $X = \dfrac{10}{1 + 4i} = \dfrac{10(1 - 4i)}{17} = 0.588 - 2.353i$, $\ |X| = \dfrac{10}{\sqrt{17}} = 2.425$, $\ \arg X = -\arctan 4 = -1.326$ rad.
3. Steady state: $x_p(t) = 2.425\cos(2t - 1.326)$ — the response lags the force by $76°$.

### Example 6 — where is the resonance peak?

$|X|^2 = F^2/D(u)$ with $u = \omega^2$ and $D(u) = (\omega_0^2 - u)^2 + 4\zeta^2\omega_0^2 u$. Setting $D'(u) = -2(\omega_0^2 - u) + 4\zeta^2\omega_0^2 = 0$ gives $u = \omega_0^2(1 - 2\zeta^2)$, i.e. $\omega_r = \omega_0\sqrt{1 - 2\zeta^2}$, which is a positive frequency only when $\zeta < 1/\sqrt2$. For Example 5 ($\omega_0^2 = 5$, $\zeta^2 = 1/5$): $\omega_r = \sqrt{5 \cdot 0.6} = \sqrt3 \approx 1.732$.

## 4. Python Examples

### 4.1 Characteristic roots and classification

```python
import numpy as np


def classify(c1, c0):
    # Roots and regime of x'' + c1 x' + c0 x = 0
    roots = np.roots([1, c1, c0])
    discriminant = c1**2 - 4 * c0
    if abs(discriminant) < 1e-12:
        regime = "critically damped"
    elif discriminant > 0:
        regime = "overdamped" if c1 > 0 else "real roots"
    else:
        regime = "underdamped" if c1 > 0 else ("undamped" if abs(c1) < 1e-12 else "growing oscillation")
    return roots, regime


for c1, c0 in [(2, 5), (4, 4), (5, 4), (0, 1), (-0.2, 1)]:
    roots, regime = classify(c1, c0)
    print(f"x'' + {c1}x' + {c0}x = 0:  roots = {np.round(roots, 4)}  ->  {regime}")
```

### 4.2 Sympy's `dsolve` agrees with the hand calculations

```python
import sympy as sp

t = sp.symbols("t", real=True)
x = sp.Function("x")

problems = {
    "Ex 1": x(t).diff(t, 2) + 2 * x(t).diff(t) + 5 * x(t),
    "Ex 2": x(t).diff(t, 2) + 4 * x(t).diff(t) + 4 * x(t),
    "Ex 3": x(t).diff(t, 2) + 5 * x(t).diff(t) + 4 * x(t),
}
initial_conditions = {x(0): 1, x(t).diff(t).subs(t, 0): 0}

for name, equation in problems.items():
    solution = sp.dsolve(sp.Eq(equation, 0), x(t), ics=initial_conditions)
    print(f"{name}: x(t) = {sp.simplify(solution.rhs)}")

lam = sp.symbols("lambda")
print("Ex 1 characteristic roots:", sp.solve(lam**2 + 2 * lam + 5, lam))
```

### 4.3 One complex constant fits the initial conditions

The function below builds $x(t) = \mathrm{Re}(Ce^{\lambda t})$ from section 2.3 and compares it with a Runge–Kutta integration of the ODE.

```python
import numpy as np


def underdamped_solution(c1, c0, x0, v0):
    # Real solution of x'' + c1 x' + c0 x = 0 via x = Re(C e^{lambda t}); assumes complex roots
    lam = (-c1 + np.sqrt(complex(c1**2 - 4 * c0))) / 2
    sigma, omega_d = -lam.real, lam.imag
    C = x0 - 1j * (v0 + sigma * x0) / omega_d
    return lambda t: np.real(C * np.exp(lam * t)), lam, C


def rk4_oscillator(c1, c0, x0, v0, t_end, dt, forcing=lambda t: 0.0):
    def rhs(t, y):
        return np.array([y[1], forcing(t) - c1 * y[1] - c0 * y[0]])

    times = np.arange(0, t_end + dt / 2, dt)
    states = np.zeros((len(times), 2))
    states[0] = [x0, v0]
    for k in range(len(times) - 1):
        tk, y = times[k], states[k]
        k1 = rhs(tk, y)
        k2 = rhs(tk + dt / 2, y + dt / 2 * k1)
        k3 = rhs(tk + dt / 2, y + dt / 2 * k2)
        k4 = rhs(tk + dt, y + dt * k3)
        states[k + 1] = y + dt / 6 * (k1 + 2 * k2 + 2 * k3 + k4)
    return times, states[:, 0]


solution, lam, C = underdamped_solution(2, 5, x0=1, v0=0)
times, numeric = rk4_oscillator(2, 5, 1, 0, t_end=8, dt=1e-3)
print("lambda =", lam, "  C =", C)
print("max |Re(C e^{lambda t}) - RK4| =", f"{np.max(np.abs(solution(times) - numeric)):.2e}")
print("closed form e^{-t}(cos 2t + 0.5 sin 2t) matches:",
      np.allclose(solution(times), np.exp(-times) * (np.cos(2 * times) + 0.5 * np.sin(2 * times))))
```

### 4.4 Driven oscillator: complex amplitude, transient and steady state

For Example 5 starting from rest, the full solution is $x = x_p + x_h$. We take $x_p = \mathrm{Re}(Xe^{2it})$ and choose the transient $x_h$ (a homogeneous solution) so that $x_h(0) = -x_p(0)$ and $\dot x_h(0) = -\dot x_p(0)$, using the same complex-constant formula.

```python
import numpy as np

c1, c0, F, omega = 2.0, 5.0, 10.0, 2.0
X = F / ((1j * omega) ** 2 + c1 * (1j * omega) + c0)
print("X =", np.round(X, 4), " |X| =", round(abs(X), 4), " arg X =", round(np.angle(X), 4))

xp0 = X.real                      # x_p(0)
vp0 = (1j * omega * X).real       # x_p'(0)
lam = (-c1 + np.sqrt(complex(c1**2 - 4 * c0))) / 2
sigma, omega_d = -lam.real, lam.imag
h0, hv0 = -xp0, -vp0
C = h0 - 1j * (hv0 + sigma * h0) / omega_d


def full_solution(t):
    return np.real(X * np.exp(1j * omega * t)) + np.real(C * np.exp(lam * t))


dt = 1e-3
times = np.arange(0, 10 + dt / 2, dt)
states = np.zeros((len(times), 2))
for k in range(len(times) - 1):
    def rhs(t, y):
        return np.array([y[1], F * np.cos(omega * t) - c1 * y[1] - c0 * y[0]])
    tk, y = times[k], states[k]
    k1 = rhs(tk, y)
    k2 = rhs(tk + dt / 2, y + dt / 2 * k1)
    k3 = rhs(tk + dt / 2, y + dt / 2 * k2)
    k4 = rhs(tk + dt, y + dt * k3)
    states[k + 1] = y + dt / 6 * (k1 + 2 * k2 + 2 * k3 + k4)

print("max |analytic - RK4| over [0, 10]:", f"{np.max(np.abs(full_solution(times) - states[:, 0])):.2e}")
print("transient amplitude |C| e^{-t} at t = 10:", f"{abs(C) * np.exp(-10):.2e}")
```

### 4.5 Higher order: the characteristic polynomial does all the work

$x''' + x = 0$ has characteristic polynomial $\lambda^3 + 1$, whose roots are the cube roots of $-1$: $-1$ and $e^{\pm i\pi/3} = \tfrac12 \pm \tfrac{\sqrt3}{2}i$. The pair in the right half-plane makes some solutions grow while oscillating.

```python
import numpy as np
import sympy as sp

print("np.roots(lambda^3 + 1):", np.round(np.roots([1, 0, 0, 1]), 6))

t = sp.symbols("t", real=True)
x = sp.Function("x")
print(sp.dsolve(x(t).diff(t, 3) + x(t), x(t)))
```

## 5. Visualizations

### 5.1 Roots in the plane, solutions in time

Left: the characteristic roots for $\omega_0 = 1$ and five damping ratios. Right: the matching solutions with $x(0) = 1$, $\dot x(0) = 0$. Roots further left decay faster; roots higher up oscillate faster; roots on the real axis do not oscillate at all.

```python
import matplotlib.pyplot as plt
import numpy as np

omega_0 = 1.0
zetas = [0.0, 0.15, 0.5, 1.0, 2.0]
colors = ["tab:blue", "tab:orange", "tab:green", "tab:red", "tab:purple"]
t = np.linspace(0, 20, 2000)

fig, (ax_roots, ax_time) = plt.subplots(1, 2, figsize=(13, 5.5), gridspec_kw={"width_ratios": [1, 1.5]})
circle = np.exp(1j * np.linspace(0, 2 * np.pi, 300))
ax_roots.plot(circle.real, circle.imag, color="lightgray", linestyle="--")

for zeta, color in zip(zetas, colors):
    roots = np.roots([1, 2 * zeta * omega_0, omega_0**2])
    ax_roots.scatter(roots.real, roots.imag, s=90, color=color, zorder=3, label=f"$\\zeta$ = {zeta}")

    x = np.zeros_like(t)
    if zeta < 1:
        lam = complex(-zeta * omega_0, omega_0 * np.sqrt(1 - zeta**2))
        C = 1 - 1j * (0 + zeta * omega_0) / lam.imag
        x = np.real(C * np.exp(lam * t))
    elif zeta == 1:
        x = (1 + omega_0 * t) * np.exp(-omega_0 * t)
    else:
        l1, l2 = np.sort(roots.real)
        A = -l2 / (l1 - l2) * 1
        B = 1 - A
        x = A * np.exp(l1 * t) + B * np.exp(l2 * t)
    ax_time.plot(t, x, color=color, label=f"$\\zeta$ = {zeta}")

ax_roots.axhline(0, color="black", linewidth=0.8)
ax_roots.axvline(0, color="black", linewidth=0.8)
ax_roots.set_xlim(-4, 1)
ax_roots.set_ylim(-1.5, 1.5)
ax_roots.set_aspect("equal")
ax_roots.set_xlabel("Re $\\lambda$ (decay rate)")
ax_roots.set_ylabel("Im $\\lambda$ (frequency)")
ax_roots.set_title("Characteristic roots, $\\omega_0 = 1$")
ax_roots.grid(True, alpha=0.3)
ax_roots.legend(loc="lower left", fontsize=9)

ax_time.axhline(0, color="black", linewidth=0.8)
ax_time.set_xlabel("t")
ax_time.set_ylabel("x(t)")
ax_time.set_title("Solutions with x(0) = 1, x'(0) = 0")
ax_time.grid(True, alpha=0.3)
ax_time.legend()
plt.tight_layout()
plt.show()
```

### 5.2 The root locus as damping increases

As $\zeta$ goes from $0$ to $2.5$ the two roots travel along the unit circle until they collide at $-1$ (critical damping), then separate along the real axis. Dots are spaced every $0.1$ in $\zeta$.

```python
import matplotlib.pyplot as plt
import numpy as np

zetas = np.linspace(0, 2.5, 26)
fig, ax = plt.subplots(figsize=(8, 5))
for zeta in zetas:
    roots = np.roots([1, 2 * zeta, 1]).astype(complex)
    color = "tab:blue" if zeta < 1 - 1e-9 else ("tab:red" if zeta < 1 + 1e-9 else "tab:purple")
    ax.scatter(roots.real, roots.imag, color=color, s=25, zorder=3)

arc = np.exp(1j * np.linspace(np.pi / 2, 3 * np.pi / 2, 200))
ax.plot(arc.real, arc.imag, color="lightgray", zorder=1)
ax.annotate("$\\zeta = 0$: $\\pm i$", xy=(0, 1), xytext=(0.15, 1.15))
ax.annotate("$\\zeta = 1$: double root $-1$", xy=(-1, 0), xytext=(-1.6, 0.35),
            arrowprops=dict(arrowstyle="->"))
ax.annotate("$\\zeta > 1$: one root heads to 0 ...", xy=(-0.2, 0), xytext=(-3.1, -1.0),
            arrowprops=dict(arrowstyle="->"))
ax.annotate("... the other to $-\\infty$", xy=(-4.8, 0), xytext=(-4.8, -0.6),
            arrowprops=dict(arrowstyle="->"))
ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-5.2, 0.8)
ax.set_ylim(-1.4, 1.4)
ax.set_aspect("equal")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.set_title("Roots of $\\lambda^2 + 2\\zeta\\lambda + 1$ for $0 \\leq \\zeta \\leq 2.5$")
ax.grid(True, alpha=0.3)
plt.show()
```

### 5.3 The complex solution is a spiral

$z(t) = Ce^{\lambda t}$ with $\lambda = -1 + 2i$ (Example 1) is an arrow that rotates at $2$ rad per unit time while shrinking by $e^{-t}$: a logarithmic spiral into the origin. The real solution $x(t)$ is the spiral's shadow on the real axis, just like a phasor — except that this phasor shrinks.

```python
import matplotlib.pyplot as plt
import numpy as np

lam = -1 + 2j
C = 1 - 0.5j
t = np.linspace(0, 4, 800)
z = C * np.exp(lam * t)
marks = np.array([0.0, 0.4, 0.8, 1.2])

fig, (ax_plane, ax_time) = plt.subplots(1, 2, figsize=(13, 5.5), gridspec_kw={"width_ratios": [1, 1.4]})
ax_plane.plot(z.real, z.imag, color="tab:blue")
for tk in marks:
    zk = C * np.exp(lam * tk)
    ax_plane.annotate("", xy=(zk.real, zk.imag), xytext=(0, 0), arrowprops=dict(arrowstyle="->", color="tab:orange"))
    ax_plane.plot([zk.real, zk.real], [zk.imag, 0], ":", color="gray")
    ax_plane.text(zk.real, zk.imag + 0.06, f"t={tk}", fontsize=9, ha="center")
ax_plane.axhline(0, color="black", linewidth=0.8)
ax_plane.axvline(0, color="black", linewidth=0.8)
ax_plane.set_xlim(-0.7, 1.3)
ax_plane.set_ylim(-0.8, 0.9)
ax_plane.set_aspect("equal")
ax_plane.set_xlabel("Re")
ax_plane.set_ylabel("Im")
ax_plane.set_title("$z(t) = C e^{(-1+2i)t}$, $C = 1 - i/2$")
ax_plane.grid(True, alpha=0.3)

ax_time.plot(t, z.real, color="tab:blue", label="$x(t) = \\mathrm{Re}\\, z(t)$")
ax_time.plot(t, np.abs(z), "--", color="gray", label="envelope $\\pm|C| e^{-t}$")
ax_time.plot(t, -np.abs(z), "--", color="gray")
ax_time.plot(marks, np.real(C * np.exp(lam * marks)), "o", color="tab:orange")
ax_time.axhline(0, color="black", linewidth=0.8)
ax_time.set_xlabel("t")
ax_time.set_title("The real part: a decaying oscillation")
ax_time.grid(True, alpha=0.3)
ax_time.legend()
plt.tight_layout()
plt.show()
```

### 5.4 Resonance curves and phase lag

Gain $|X|\omega_0^2/F$ and phase lag $\delta$ of the driven oscillator against $\omega/\omega_0$. Light damping gives a tall, narrow peak just below $\omega_0$; at $\omega = \omega_0$ every curve lags by exactly $90°$.

```python
import matplotlib.pyplot as plt
import numpy as np

ratio = np.linspace(0.01, 2.5, 2000)       # omega / omega_0, with omega_0 = 1
fig, (ax_gain, ax_phase) = plt.subplots(2, 1, figsize=(9, 7), sharex=True)
for zeta in [0.05, 0.1, 0.25, 0.5, 1.0]:
    X = 1 / (1 - ratio**2 + 2j * zeta * ratio)
    ax_gain.plot(ratio, np.abs(X), label=f"$\\zeta$ = {zeta}")
    ax_phase.plot(ratio, np.degrees(-np.angle(X)))

ax_gain.set_ylim(0, 6)
ax_gain.set_ylabel("$|X|\\,\\omega_0^2 / F$")
ax_gain.set_title("Driven oscillator: gain and phase lag")
ax_gain.legend()
ax_phase.axhline(90, color="gray", linestyle=":")
ax_phase.set_yticks([0, 45, 90, 135, 180])
ax_phase.set_ylabel("phase lag $\\delta$ (degrees)")
ax_phase.set_xlabel("$\\omega / \\omega_0$")
for ax in (ax_gain, ax_phase):
    ax.axvline(1, color="gray", linestyle="--", linewidth=1)
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.5 The complex amplitude $X(\omega)$ in the plane

The same information as one curve: as $\omega$ increases from $0$, $X(\omega) = 1/(1 - \omega^2 + 2i\zeta\omega)$ starts at $1$ on the positive real axis, swings clockwise through the lower half-plane (lag between $0°$ and $180°$), crossing the negative imaginary axis at $\omega = \omega_0$ (lag $90°$), and shrinks to $0$ from the left. The distance from the origin is the gain.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(7, 7))
omega = np.linspace(0, 6, 4000)
for zeta, color in [(0.2, "tab:blue"), (0.5, "tab:green")]:
    X = 1 / (1 - omega**2 + 2j * zeta * omega)
    ax.plot(X.real, X.imag, color=color, label=f"$\\zeta$ = {zeta}")
    for w in [0.5, 1.0, 1.5]:
        point = 1 / (1 - w**2 + 2j * zeta * w)
        ax.plot(point.real, point.imag, "o", color=color)
        ax.text(point.real + 0.06, point.imag, f"$\\omega$={w}", color=color, fontsize=9)

ax.axhline(0, color="black", linewidth=0.8)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlim(-1.6, 1.8)
ax.set_ylim(-2.8, 0.6)
ax.set_aspect("equal")
ax.set_xlabel("Re X")
ax.set_ylabel("Im X")
ax.set_title("Complex amplitude $X(\\omega)$ for $\\omega_0 = 1$, $F = 1$")
ax.grid(True, alpha=0.3)
ax.legend()
plt.show()
```

### 5.6 Transient plus steady state (Example 5)

Starting from rest, the motion is the sum of the steady state $\mathrm{Re}(Xe^{2it})$ and a transient $\mathrm{Re}(Ce^{(-1+2i)t})$ that decays like $e^{-t}$. The RK4 simulation lies on top of the analytic sum.

```python
import matplotlib.pyplot as plt
import numpy as np

c1, c0, F, omega = 2.0, 5.0, 10.0, 2.0
X = F / ((1j * omega) ** 2 + c1 * 1j * omega + c0)
lam = -1 + 2j
h0, hv0 = -X.real, -(1j * omega * X).real
C = h0 - 1j * (hv0 + 1.0 * h0) / 2.0

dt = 1e-3
t = np.arange(0, 10 + dt / 2, dt)
y = np.zeros((len(t), 2))
for k in range(len(t) - 1):
    def rhs(tt, s):
        return np.array([s[1], F * np.cos(omega * tt) - c1 * s[1] - c0 * s[0]])
    k1 = rhs(t[k], y[k])
    k2 = rhs(t[k] + dt / 2, y[k] + dt / 2 * k1)
    k3 = rhs(t[k] + dt / 2, y[k] + dt / 2 * k2)
    k4 = rhs(t[k] + dt, y[k] + dt * k3)
    y[k + 1] = y[k] + dt / 6 * (k1 + 2 * k2 + 2 * k3 + k4)

steady = np.real(X * np.exp(1j * omega * t))
transient = np.real(C * np.exp(lam * t))

plt.figure(figsize=(10, 4.5))
plt.plot(t, y[:, 0], color="black", linewidth=3, alpha=0.35, label="RK4 simulation")
plt.plot(t, steady + transient, color="tab:blue", label="steady + transient")
plt.plot(t, steady, "--", color="tab:orange", label="steady state $\\mathrm{Re}(X e^{2it})$")
plt.plot(t, transient, ":", color="tab:green", label="transient $\\mathrm{Re}(C e^{(-1+2i)t})$")
plt.axhline(0, color="black", linewidth=0.8)
plt.ylim(-3.2, 4.2)
plt.xlabel("t")
plt.ylabel("x(t)")
plt.title("$\\ddot x + 2\\dot x + 5x = 10\\cos 2t$, starting from rest")
plt.grid(True, alpha=0.3)
plt.legend(loc="upper right", ncol=2, fontsize=9)
plt.show()
```

9. With $p(\lambda) = (\lambda - \lambda_0)^2$ the ODE is $\ddot x - 2\lambda_0\dot x + \lambda_0^2 x = 0$. ## 8. Conclusion

- The ansatz $x = e^{\lambda t}$ turns a linear constant-coefficient ODE into its **characteristic polynomial** $p(\lambda) = 0$; this works for complex $\lambda$ because $\frac{d}{dt}e^{\lambda t} = \lambda e^{\lambda t}$ still holds.
- Real coefficients give conjugate root pairs, and real solutions are real parts: $x(t) = \mathrm{Re}(Ce^{\lambda t})$ with one complex constant $C$ fixed by the initial conditions.
- **Re** $\lambda$ is the decay (or growth) rate, **Im** $\lambda$ the oscillation frequency. Underdamped roots sit on the circle $|\lambda| = \omega_0$, collide at $-\omega_0$ for critical damping, and split along the real axis when overdamped. Stability means all roots in the left half-plane.
- A sinusoidal drive is solved by the complex amplitude $X = F/p(i\omega)$: modulus is gain, argument is phase lag; resonance peaks at $\omega_0\sqrt{1 - 2\zeta^2}$ with a $90°$ lag at $\omega_0$.
- The full response is steady state plus a decaying transient — exactly what the RLC simulation of notebook 01 showed.
