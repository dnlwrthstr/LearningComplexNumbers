# Limits and Continuity in $\mathbb{C}$

Calculus starts with limits, and complex calculus is no different. The definition looks exactly like the real one — "$f(z)$ gets close to $L$ whenever $z$ gets close to $z_0$" — but "close" now means *inside a small disc*, not inside a small interval. A disc can be entered from infinitely many directions, along straight lines, spirals or zig-zags, and a limit must give the same answer along **every** one of them. That single change is what later makes complex differentiability so much stronger than real differentiability.

This notebook sets up the $\varepsilon$–$\delta$ language with discs, shows how a limit can fail because different paths disagree (the classic example $\bar z / z$ at $0$), defines continuity, and handles limits at infinity with a first look at the Riemann sphere.

This is notebook 01 of module 05 (**Complex differentiation**) in the **Learning Complex Numbers** series.

## 1. Introduction

On the real line, $x \to 0$ has only two directions of approach: from the left and from the right. A limit exists when the two one-sided limits agree. In the complex plane, $z \to 0$ can happen from any angle, and along any curve. So the bar for "the limit exists" is much higher:

| Setting | Neighbourhood of $z_0$ | Ways to approach |
|---|---|---|
| $\mathbb{R}$ | interval $(x_0 - \delta, x_0 + \delta)$ | 2 (left, right) |
| $\mathbb{C}$ | disc $\lvert z - z_0 \rvert < \delta$ | infinitely many (every direction and curve) |

This matters immediately. In notebook 02 the derivative will be defined as a limit of difference quotients, and demanding that the limit be the same from every direction produces the Cauchy–Riemann equations.

## 2. Theory

### 2.1 Distance and discs

The distance between $z$ and $w$ is $\lvert z - w \rvert = \sqrt{(x_1 - x_2)^2 + (y_1 - y_2)^2}$, the ordinary Euclidean distance in the plane. The **open disc** of radius $r$ about $z_0$ is

$$
D(z_0, r) = \{ z \in \mathbb{C} : \lvert z - z_0 \rvert < r \},
$$

and the **punctured disc** $D^*(z_0, r) = \{ z : 0 < \lvert z - z_0 \rvert < r \}$ leaves out the centre. Discs play the role that intervals play on the real line.

### 2.2 The definition of a limit

Let $f$ be defined on a punctured disc around $z_0$. We write

$$
\lim_{z \to z_0} f(z) = L
$$

if for every $\varepsilon > 0$ there is a $\delta > 0$ such that

$$
0 < \lvert z - z_0 \rvert < \delta \quad\Longrightarrow\quad \lvert f(z) - L \rvert < \varepsilon .
$$

Geometrically: **whatever target disc $D(L, \varepsilon)$ you choose around $L$, there is a disc $D^*(z_0, \delta)$ around $z_0$ that $f$ maps entirely inside it.** The value $f(z_0)$ itself is irrelevant (it need not even be defined).

### 2.3 Limits are path-independent

If the limit exists, then for **every** curve $z(t) \to z_0$ (with $z(t) \neq z_0$) we get $f(z(t)) \to L$, because the curve eventually enters every $\delta$-disc. The contrapositive is the standard tool for showing a limit **does not** exist:

> If $f(z) \to L_1$ along one path and $f(z) \to L_2 \neq L_1$ along another, then $\lim_{z \to z_0} f(z)$ does not exist.

Agreement along a few paths proves nothing, though: all straight lines can agree while a curved path disagrees (exercise 9).

### 2.4 Reduction to real and imaginary parts

Write $f = u + iv$, $z_0 = x_0 + iy_0$, $L = a + ib$. Because

$$
\max(\lvert \operatorname{Re} w \rvert, \lvert \operatorname{Im} w \rvert) \;\le\; \lvert w \rvert \;\le\; \lvert \operatorname{Re} w \rvert + \lvert \operatorname{Im} w \rvert,
$$

we have $f(z) \to L$ exactly when both real two-variable limits hold:

$$
\lim_{(x,y) \to (x_0,y_0)} u(x,y) = a \quad\text{and}\quad \lim_{(x,y) \to (x_0,y_0)} v(x,y) = b .
$$

### 2.5 Limit laws

If $f \to L$ and $g \to M$ then $f \pm g \to L \pm M$, $fg \to LM$, and $f/g \to L/M$ provided $M \neq 0$. The proofs are word-for-word the real ones, using $\lvert zw \rvert = \lvert z \rvert \lvert w \rvert$ and the triangle inequality $\lvert z + w \rvert \le \lvert z \rvert + \lvert w \rvert$.

### 2.6 Continuity

$f$ is **continuous at $z_0$** if $f(z_0)$ is defined and $\lim_{z \to z_0} f(z) = f(z_0)$. From the limit laws:

- polynomials are continuous on all of $\mathbb{C}$;
- rational functions $p/q$ are continuous wherever $q \neq 0$;
- $\operatorname{Re} z$, $\operatorname{Im} z$, $\bar z$ and $\lvert z \rvert$ are continuous everywhere (each changes by at most $\lvert \Delta z \rvert$);
- $e^z$, $\sin z$, $\cos z$ are continuous everywhere (module 03).

The principal argument $\operatorname{Arg} z \in (-\pi, \pi]$ is **not** continuous on the negative real axis: just above $-1$ it is close to $\pi$, just below it is close to $-\pi$. This jump is the branch cut of notebook [03/03](../03_complex_functions/03_the_complex_logarithm_and_branch_cuts.ipynb).

### 2.7 Limits at infinity and the Riemann sphere

"$z \to \infty$" means $\lvert z \rvert \to \infty$, in any direction. Define

$$
\lim_{z \to \infty} f(z) = L \iff \lim_{w \to 0} f(1/w) = L,
\qquad
\lim_{z \to z_0} f(z) = \infty \iff \lim_{z \to z_0} \frac{1}{f(z)} = 0 .
$$

So $\mathbb{C}$ has **one** point at infinity, not a $+\infty$ and a $-\infty$. The picture behind this is the **Riemann sphere**: put the unit sphere $X^2 + Y^2 + Z^2 = 1$ on top of the plane, and join the north pole $N = (0,0,1)$ to $z = x + iy$ by a straight line. It meets the sphere at

$$
(X, Y, Z) = \left( \frac{2x}{\lvert z \rvert^2 + 1},\; \frac{2y}{\lvert z \rvert^2 + 1},\; \frac{\lvert z \rvert^2 - 1}{\lvert z \rvert^2 + 1} \right).
$$

The unit circle maps to the equator, $0$ to the south pole, and as $\lvert z \rvert \to \infty$ in any direction the point climbs to $N$. Adding $N$ as the point $\infty$ gives the **extended plane** $\hat{\mathbb{C}} = \mathbb{C} \cup \{\infty\}$ — the natural home of the Möbius transformations of notebook [03/06](../03_complex_functions/06_mobius_transformations.ipynb).

## 3. Worked Examples

### Example 1 — an $\varepsilon$–$\delta$ proof: $\lim_{z \to 1+i} z^2 = 2i$

Let $z_0 = 1 + i$, so $z_0^2 = 2i$. Factor:

$$
\lvert z^2 - z_0^2 \rvert = \lvert z - z_0 \rvert \, \lvert z + z_0 \rvert .
$$

If $\lvert z - z_0 \rvert < 1$ then $\lvert z + z_0 \rvert = \lvert (z - z_0) + 2z_0 \rvert \le 1 + 2\sqrt{2} < 4$. So $\lvert z^2 - 2i \rvert < 4 \lvert z - z_0 \rvert$, and

$$
\delta = \min\!\left(1, \frac{\varepsilon}{4}\right)
$$

works. For $\varepsilon = 0.1$: $\delta = 0.025$.

### Example 2 — $\bar z / z$ has no limit at $0$

Let $f(z) = \bar z / z$ for $z \neq 0$. Approach along the ray $z = t e^{i\theta}$, $t \to 0^+$:

$$
f(t e^{i\theta}) = \frac{t e^{-i\theta}}{t e^{i\theta}} = e^{-2i\theta} .
$$

The value does not depend on $t$ at all — it is constant on each ray — but it depends on the direction:

| path | $\theta$ | $f$ on the path |
|---|---|---|
| positive real axis | $0$ | $1$ |
| positive imaginary axis | $\pi/2$ | $e^{-i\pi} = -1$ |
| diagonal $y = x$ | $\pi/4$ | $e^{-i\pi/2} = -i$ |

Two paths give different values, so the limit does not exist. Every punctured disc around $0$ is mapped onto the **whole unit circle**, which no small $\varepsilon$-disc can contain.

### Example 3 — a removable hole: $\lim_{z \to i} \dfrac{z^2 + 1}{z - i}$

Both numerator and denominator vanish at $i$. Factor $z^2 + 1 = (z - i)(z + i)$; for $z \neq i$ the quotient equals $z + i$, which is continuous, so the limit is $2i$.

### Example 4 — reducing to real parts: $\lim_{z \to 0} \dfrac{x^2 y}{x^2 + y^2} + i\,(x + y)$

The imaginary part $x + y \to 0$. For the real part, $\lvert x^2 y \rvert / (x^2 + y^2) \le \lvert y \rvert \le \lvert z \rvert \to 0$. Both parts go to $0$, so the limit is $0$.

### Example 5 — $\operatorname{Arg}$ jumps at $-1$

Take $z_n = -1 + i/n$ and $w_n = -1 - i/n$. Both tend to $-1$. But $\operatorname{Arg} z_n \to \pi$ while $\operatorname{Arg} w_n \to -\pi$. Two paths, two limits: $\operatorname{Arg}$ has no limit at $-1$ and is discontinuous there (even though $\operatorname{Arg}(-1) = \pi$ is defined). At $z_0 = 1$, by contrast, $\operatorname{Arg}$ is continuous.

### Example 6 — limits at infinity

$\displaystyle \lim_{z \to \infty} \frac{2z + 1}{z - 3}$: substitute $z = 1/w$,

$$
\frac{2/w + 1}{1/w - 3} = \frac{2 + w}{1 - 3w} \xrightarrow{w \to 0} 2 .
$$

And $\displaystyle \lim_{z \to 3} \frac{2z+1}{z-3} = \infty$, because the reciprocal $\frac{z-3}{2z+1} \to 0$. Note that $e^z$ has **no** limit at $\infty$: along the positive real axis it blows up, along the negative real axis it goes to $0$, and along the imaginary axis it circles forever.

## 4. Python Examples

### 4.1 Checking an $\varepsilon$–$\delta$ proof numerically

Sample many points in the punctured disc $D^*(1+i, \delta)$ with $\delta = \min(1, \varepsilon/4)$ and check that $\lvert z^2 - 2i \rvert < \varepsilon$ for all of them.

```python
import numpy as np

rng = np.random.default_rng(0)
z0 = 1 + 1j
limit_value = 2j

for epsilon in [1.0, 0.1, 0.001]:
    delta = min(1.0, epsilon / 4)
    radii = delta * np.sqrt(rng.uniform(0, 1, 100_000))      # uniform in the disc
    angles = rng.uniform(-np.pi, np.pi, 100_000)
    z = z0 + radii * np.exp(1j * angles)
    worst = np.max(np.abs(z**2 - limit_value))
    print(f"epsilon = {epsilon:<6}  delta = {delta:<8}  max |z^2 - 2i| = {worst:.6f}  < epsilon: {worst < epsilon}")
```

### 4.2 $\bar z / z$ along different paths

Evaluate $f(z) = \bar z / z$ at points marching toward $0$ along several rays and along a spiral. The values on each ray are constant; they never settle on a common value.

```python
import numpy as np


def conj_over_z(z):
    return np.conj(z) / z


t = np.array([1e-1, 1e-3, 1e-6, 1e-9])
paths = {
    "real axis   (theta = 0)   ": t,
    "imag axis   (theta = pi/2)": 1j * t,
    "diagonal    (theta = pi/4)": t * np.exp(1j * np.pi / 4),
}
for name, z in paths.items():
    values = conj_over_z(z)
    print(name, np.round(values, 6))

spiral = t * np.exp(1j / t)            # spirals into 0, angle keeps changing
print("spiral z = t e^{i/t}        ", np.round(conj_over_z(spiral), 3))
```

Along the spiral the values keep wandering around the unit circle: there is not even a limit along that single path.

### 4.3 Symbolic path limits with sympy

Substitute $z = x + i m x$ (the line through $0$ with slope $m$) and let $x \to 0$. The answer depends on $m$, which proves the limit does not exist.

```python
import sympy as sp

x, m = sp.symbols("x m", real=True)
z = x + sp.I * m * x
f_on_line = sp.conjugate(z) / z

limit_on_line = sp.simplify(sp.limit(sp.simplify(f_on_line), x, 0))
print("limit along y = m x :", limit_on_line)
print("equals (1 - i m)^2 / (1 + m^2):", sp.simplify(limit_on_line - (1 - sp.I * m) ** 2 / (1 + m**2)) == 0)
for slope in [0, 1, -1, sp.oo]:
    if slope == sp.oo:
        print("  m = oo (imag axis):", sp.limit(limit_on_line, m, sp.oo))
    else:
        print(f"  m = {slope}:", sp.simplify(limit_on_line.subs(m, slope)))
```

The formula $\frac{(1 - im)^2}{1 + m^2}$ equals $e^{-2i\theta}$ with $\tan\theta = m$, matching Example 2.

### 4.4 Limits that do exist: sympy and the removable hole

```python
import sympy as sp

z = sp.symbols("z")
print("lim_{z->i} (z^2+1)/(z-i)      =", sp.limit((z**2 + 1) / (z - sp.I), z, sp.I))
print("lim_{z->1+i} z^2              =", sp.expand(sp.limit(z**2, z, 1 + sp.I)))

w = sp.symbols("w")
f_of_inverse = ((2 / w + 1) / (1 / w - 3)).simplify()
print("(2z+1)/(z-3) with z = 1/w     =", f_of_inverse)
print("lim_{z->oo} (2z+1)/(z-3)      =", sp.limit(f_of_inverse, w, 0))
```

### 4.5 The jump of $\operatorname{Arg}$ across the negative real axis

```python
import numpy as np

for n in [10, 1_000, 1_000_000]:
    above = -1 + 1j / n
    below = -1 - 1j / n
    print(f"n = {n:>9}:  Arg(-1 + i/n) = {np.angle(above):+.6f}   Arg(-1 - i/n) = {np.angle(below):+.6f}")
print("Arg(-1) itself =", np.angle(-1 + 0j), " (principal value pi)")
```

## 5. Visualizations

### 5.1 The $\varepsilon$–$\delta$ picture: two discs

Left: the $\delta$-disc around $z_0 = 1+i$. Right: its image under $z \mapsto z^2$ (a slightly distorted disc around $2i$), which sits inside the $\varepsilon$-disc. Here $\varepsilon = 0.4$ and $\delta = \min(1, \varepsilon/4) = 0.1$.

```python
import matplotlib.pyplot as plt
import numpy as np

z0 = 1 + 1j
epsilon = 0.4
delta = min(1.0, epsilon / 4)

radii = np.linspace(0, delta, 25)
angles = np.linspace(-np.pi, np.pi, 200)
disc = z0 + radii[:, None] * np.exp(1j * angles[None, :])
image = disc**2
boundary = np.exp(1j * angles)

fig, (left, right) = plt.subplots(1, 2, figsize=(12, 6))

left.fill(disc[-1].real, disc[-1].imag, color="tab:blue", alpha=0.3, label=rf"$\delta$-disc, $\delta = {delta}$")
left.plot(z0.real, z0.imag, "k.", markersize=10)
left.annotate(r"$z_0 = 1 + i$", (z0.real, z0.imag), xytext=(10, -15), textcoords="offset points")
left.set_xlim(0.5, 1.5)
left.set_ylim(0.5, 1.5)
left.set_title("Domain: points within $\\delta$ of $z_0$")

right.fill((2j + epsilon * boundary).real, (2j + epsilon * boundary).imag, color="tab:orange", alpha=0.25,
           label=rf"$\varepsilon$-disc, $\varepsilon = {epsilon}$")
right.fill(image[-1].real, image[-1].imag, color="tab:blue", alpha=0.5, label="image of the $\\delta$-disc")
right.plot(0, 2, "k.", markersize=10)
right.annotate(r"$L = 2i$", (0, 2), xytext=(10, -15), textcoords="offset points")
right.set_xlim(-0.5, 0.5)
right.set_ylim(1.5, 2.5)
right.set_title("Codomain: $f(z) = z^2$ stays within $\\varepsilon$ of $L$")

for ax in (left, right):
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="upper left")
plt.tight_layout()
plt.show()
```

### 5.2 Domain colouring of $\bar z / z$

In a domain-colouring plot each point $z$ is coloured by the **argument** of $f(z)$ (hue) and the brightness encodes $\lvert f(z) \rvert$. For $\bar z / z = e^{-2i\theta}$ the modulus is always $1$, so only the hue changes, and it changes **only with direction**. Every colour of the wheel appears in every disc around $0$, however small — a picture of "no limit". Compare $z^2 / \lvert z \rvert$ (right), whose colours also swirl but whose brightness fades to black at $0$: that function does tend to $0$.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import hsv_to_rgb


def domain_colouring(values, modulus_scale=None):
    hue = (np.angle(values) / (2 * np.pi)) % 1.0
    if modulus_scale is None:
        brightness = np.ones_like(hue)
    else:
        brightness = np.clip(np.abs(values) / modulus_scale, 0, 1) ** 0.5
    saturation = np.full_like(hue, 0.9)
    return hsv_to_rgb(np.dstack([hue, saturation, brightness]))


extent = 1.0
grid = np.linspace(-extent, extent, 601)
X, Y = np.meshgrid(grid, grid)
Z = X + 1j * Y
Z[np.abs(Z) == 0] = 1e-12        # avoid dividing by zero at the centre pixel

fig, axes = plt.subplots(1, 2, figsize=(12, 6))
panels = [
    (np.conj(Z) / Z, None, r"$\bar{z}/z$: every colour near 0 (no limit)"),
    (Z**2 / np.abs(Z), 1.0, r"$z^2/|z|$: fades to black (limit 0)"),
]
for ax, (values, scale, title) in zip(axes, panels):
    ax.imshow(domain_colouring(values, scale), extent=[-extent, extent, -extent, extent], origin="lower")
    for radius in [0.5, 0.25, 0.1]:
        circle = radius * np.exp(1j * np.linspace(0, 2 * np.pi, 200))
        ax.plot(circle.real, circle.imag, color="white", linewidth=0.8, linestyle="--")
    ax.axhline(0, color="black", linewidth=0.6)
    ax.axvline(0, color="black", linewidth=0.6)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
plt.tight_layout()
plt.show()
```

Hue legend: red = argument $0$ (positive real), yellow ≈ $\pi/3$, green ≈ $2\pi/3$, cyan = $\pi$ (negative real), blue ≈ $-2\pi/3$, magenta ≈ $-\pi/3$. On the left the colour wheel is traversed **twice, clockwise** as you walk once counter-clockwise around $0$ — the signature of $e^{-2i\theta}$.

### 5.3 Values along paths

Plot the values $f(z(t))$ in the $w$-plane as $t \to 0$ for several paths into $0$. Each ray lands on its own fixed point of the unit circle (big dots: the value is the same at every point of the ray); the spiral's values (thin line) sweep around the whole circle.

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.geomspace(1, 1e-3, 400)
paths = {
    r"ray $\theta = 0$": t,
    r"ray $\theta = \pi/4$": t * np.exp(1j * np.pi / 4),
    r"ray $\theta = \pi/2$": 1j * t,
    r"ray $\theta = 3\pi/4$": t * np.exp(3j * np.pi / 4),
    r"spiral $t\,e^{i \ln t}$": t * np.exp(1j * np.log(t)),
}

fig, (zplane, wplane) = plt.subplots(1, 2, figsize=(12, 6))
circle = np.exp(1j * np.linspace(0, 2 * np.pi, 300))
wplane.plot(circle.real, circle.imag, color="lightgray", linewidth=4, zorder=0)

for label, z in paths.items():
    line, = zplane.plot(z.real, z.imag, label=label)
    w = np.conj(z) / z
    if "spiral" in label:
        wplane.plot(w.real, w.imag, color=line.get_color(), linewidth=1, label=label, zorder=1)
    else:
        # constant along the ray: one big marker
        wplane.plot(w.real[-1], w.imag[-1], "o", markersize=14, color=line.get_color(), label=label, zorder=3)

zplane.set_title("Paths into $z_0 = 0$")
wplane.set_title(r"Values of $\bar{z}/z$ along each path")
for ax, lim in ((zplane, 1.1), (wplane, 1.4)):
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_xlim(-lim, lim)
    ax.set_ylim(-lim, lim)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
zplane.legend(loc="lower left", fontsize=9)
wplane.legend(loc="lower left", fontsize=9)
plt.tight_layout()
plt.show()
```

A heat map of $\operatorname{Arg} z$: smooth everywhere except across the negative real axis, where the colour jumps from $+\pi$ (top) to $-\pi$ (bottom).

### 5.5 The Riemann sphere

Stereographic projection lifts the plane onto the sphere. Circles $\lvert z \rvert = r$ become circles of latitude: $r < 1$ in the southern hemisphere, $r = 1$ on the equator, and large $r$ crowding toward the north pole $N = \infty$. Straight lines through $0$ become meridians that all pass through $N$.

```python
import matplotlib.pyplot as plt
import numpy as np


def to_sphere(z):
    modulus_squared = np.abs(z) ** 2
    return (2 * z.real / (modulus_squared + 1),
            2 * z.imag / (modulus_squared + 1),
            (modulus_squared - 1) / (modulus_squared + 1))


fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(projection="3d")

u = np.linspace(0, 2 * np.pi, 60)
v = np.linspace(0, np.pi, 30)
ax.plot_wireframe(np.outer(np.cos(u), np.sin(v)), np.outer(np.sin(u), np.sin(v)),
                  np.outer(np.ones_like(u), np.cos(v)), color="lightgray", linewidth=0.4)

theta = np.linspace(0, 2 * np.pi, 300)
for radius, colour in [(0.3, "tab:blue"), (1.0, "tab:green"), (3.0, "tab:orange"), (10.0, "tab:red")]:
    X, Y, Zs = to_sphere(radius * np.exp(1j * theta))
    ax.plot(X, Y, Zs, color=colour, linewidth=2, label=f"|z| = {radius}")

t = np.geomspace(1e-3, 1e3, 300)
for angle in [0, np.pi / 2, np.pi, 3 * np.pi / 2]:
    X, Y, Zs = to_sphere(t * np.exp(1j * angle))
    ax.plot(X, Y, Zs, color="black", linewidth=1)

ax.scatter([0], [0], [1], color="black", s=60)
ax.text(0, 0, 1.15, "N = infinity", ha="center")
ax.scatter([0], [0], [-1], color="black", s=60)
ax.text(0, 0, -1.25, "S = 0", ha="center")
ax.set_box_aspect((1, 1, 1))
ax.set_xlabel("X (Re direction)")
ax.set_ylabel("Y (Im direction)")
ax.set_zlabel("Z")
ax.set_title("Stereographic projection: circles |z| = r on the Riemann sphere")
ax.legend(loc="upper left")
plt.show()
```

## 7. Solutions

1. Polynomials are continuous, so substitute: $(2 - i)^2 = 3 - 4i$ and $3(2 - i) = 6 - 3i$, giving $9 - 7i$.
2. $z^2 + 4 = (z - 2i)(z + 2i)$, so for $z \neq -2i$ the quotient is $z - 2i \to -4i$.
3. With $z = 1/w$: $\dfrac{i/w^2 - 1}{3/w^2 + 1/w} = \dfrac{i - w^2}{3 + w} \to \dfrac{i}{3}$.
4. On the real axis $z = x$: $\operatorname{Re} z / z = x/x = 1$. On the imaginary axis $z = iy$: $0 / (iy) = 0$. Different, so no limit. (On the ray at angle $\theta$ the value is $\cos\theta \, e^{-i\theta}$.)
5. $\lvert \bar z - \bar z_0 \rvert = \lvert \overline{z - z_0} \rvert = \lvert z - z_0 \rvert$, so $\delta = \varepsilon$ works, and no larger $\delta$ works in general: conjugation is an isometry (a reflection), so a $\delta$-disc maps onto a $\delta$-disc.
6. $z / \lvert z \rvert = e^{i\theta}$ has modulus $1$ and argument $\theta$, so the sketch is a full colour wheel traversed **once counter-clockwise**, with constant brightness — every colour appears in every disc around $0$, so no limit. Along $\theta = 0$ the value is $1$; along $\theta = \pi$ it is $-1$. Confirmed.
7. $\lvert z^2 / \lvert z \rvert \rvert = \lvert z \rvert^2 / \lvert z \rvert = \lvert z \rvert$. So $\delta = \varepsilon$ works.
8. **Continuous** at every $z_0$ not on the closed negative real axis: there $z_0 = r_0 e^{i\theta_0}$ with $-\pi < \theta_0 < \pi$; a small enough disc around $z_0$ stays inside the open sector $\lvert \theta - \theta_0 \rvert < \eta$ (with $\eta < \pi - \lvert \theta_0 \rvert$), on which $\operatorname{Arg}$ takes values within $\eta$ of $\theta_0$ — and $\eta$ can be made smaller than any $\varepsilon$. **Discontinuous** at every $z_0 = -r$, $r > 0$: the sequences $-r + i/n$ and $-r - i/n$ give limits $\pi$ and $-\pi$, as in Example 5.
9. On $y = mx$: $f = \dfrac{m x^3}{x^4 + m^2 x^2} = \dfrac{m x}{x^2 + m^2} \to 0$ (for $m \neq 0$; for $m = 0$ and on the $y$-axis $f \equiv 0$). On $y = x^2$: $f = \dfrac{x^4}{2x^4} = \tfrac{1}{2}$. Since $\tfrac12 \neq 0 = f(0)$, $f$ is not continuous at $0$; agreement along all straight lines is not enough.
10. Let $\rho = \lvert z \rvert$. For $1/\bar z = z / \rho^2$ the modulus is $1/\rho$, so its sphere point is $\left( \frac{2x/\rho^2}{1/\rho^2 + 1}, \frac{2y/\rho^2}{1/\rho^2 + 1}, \frac{1/\rho^2 - 1}{1/\rho^2 + 1} \right) = \left( \frac{2x}{1 + \rho^2}, \frac{2y}{1 + \rho^2}, \frac{1 - \rho^2}{1 + \rho^2} \right)$ — same $X$ and $Y$, opposite $Z$. So inversion in the unit circle, $z \mapsto 1/\bar z$, is just the reflection of the sphere in its equator; it swaps $0$ and $\infty$.

## 8. Conclusion

- $\lim_{z \to z_0} f(z) = L$ means: every $\varepsilon$-disc around $L$ contains the image of some punctured $\delta$-disc around $z_0$.
- A limit must be the same along **every** path into $z_0$. Two paths with different limits (like the axes for $\bar z / z$) prove the limit does not exist; agreement along straight lines alone proves nothing.
- A complex limit is the same as two real limits, one for $\operatorname{Re} f$ and one for $\operatorname{Im} f$; the limit laws carry over unchanged.
- Polynomials, rational functions (off their poles), $\bar z$, $\lvert z \rvert$, $e^z$ are continuous; $\operatorname{Arg} z$ jumps by $2\pi$ across the negative real axis.
- There is one point at infinity. Limits at $\infty$ are computed with $w = 1/z$, and the Riemann sphere makes $\infty$ an ordinary point, the north pole.
