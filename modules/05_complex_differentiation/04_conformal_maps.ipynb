# Conformal Maps: Angles Are Preserved

A holomorphic function's derivative $f'(z_0)$ is a single complex number, and multiplying by a complex number rotates by its argument and scales by its modulus. So near any point where $f'(z_0) \neq 0$, a holomorphic map acts on every tiny shape in the same way: **rotate by $\arg f'(z_0)$, scale by $\lvert f'(z_0) \rvert$**. Shapes get bent and resized over large distances, but at small scales nothing gets sheared — and in particular the **angle between any two curves is preserved**. Maps with this property are called **conformal**.

This notebook proves the angle-preservation property, watches it happen on the images of square grids under $z^2$, $e^z$ and the Joukowski map (which turns a circle into an aircraft wing profile), and looks at what goes wrong at the **critical points** where $f' = 0$: there $z^2$ doubles angles instead of preserving them.

This is notebook 04 of module 05 (**Complex differentiation**) in the **Learning Complex Numbers** series.

## 1. Introduction

Mercator's map of the Earth distorts areas enormously (Greenland looks as big as Africa) but it gets **angles** right: a ship holding a constant compass bearing follows a straight line on the chart. That is exactly the trade-off of a conformal map — sizes change from place to place, angles never do. Conformal maps appear wherever a problem's geometry is awkward but its physics only cares about angles and local shape:

- **Aerodynamics**: the Joukowski map carries the easy flow around a cylinder to the flow around a wing profile (this notebook and notebook 05).
- **Electrostatics and heat flow**: Laplace's equation is preserved by conformal maps (notebook 05), so a hard boundary can be mapped to a disc or a half-plane.
- **Cartography and computer graphics**: texture mapping and mesh parametrisation without local shear.

## 2. Theory

### 2.1 What a holomorphic map does to a curve

Let $\gamma(t)$ be a smooth curve with $\gamma(0) = z_0$ and tangent vector $\gamma'(0) \neq 0$ (a complex number pointing along the curve). The image curve $f(\gamma(t))$ has, by the chain rule,

$$
\frac{d}{dt} f(\gamma(t)) \Big|_{t=0} = f'(z_0)\, \gamma'(0).
$$

Write $f'(z_0) = \rho\, e^{i\varphi}$ with $\rho > 0$. Then the image tangent is the original tangent **rotated by $\varphi = \arg f'(z_0)$** and **scaled by $\rho = \lvert f'(z_0) \rvert$** — and crucially, $\varphi$ and $\rho$ depend only on the point $z_0$, not on which curve we picked.

### 2.2 Angles are preserved

The **angle from curve $\gamma_1$ to curve $\gamma_2$** at $z_0$ is the angle from $\gamma_1'(0)$ to $\gamma_2'(0)$:

$$
\alpha = \arg \frac{\gamma_2'(0)}{\gamma_1'(0)}.
$$

After mapping:

$$
\arg \frac{f'(z_0) \gamma_2'(0)}{f'(z_0) \gamma_1'(0)} = \arg \frac{\gamma_2'(0)}{\gamma_1'(0)} = \alpha .
$$

The factor $f'(z_0)$ cancels. **Theorem.** A holomorphic map is **conformal** at every point where $f'(z_0) \neq 0$: it preserves both the size and the orientation (sense) of angles between curves.

Consequences:

- The grid lines $x = \text{const}$ and $y = \text{const}$ meet at right angles, so their images under a conformal map are two families of curves that **also meet at right angles**.
- Lengths of tiny segments are multiplied by $\lvert f'(z_0) \rvert$ and areas by $\lvert f'(z_0) \rvert^2 = \det J$ — tiny squares become tiny squares, of different sizes in different places.

### 2.3 Conformal $\iff$ holomorphic with $f' \neq 0$

Conversely, if $f$ is real-differentiable with invertible Jacobian $J$ and preserves angles and their orientation, then $J$ must map every pair of perpendicular vectors to a pair of perpendicular vectors of equal length ratio, with positive determinant — which forces $J = \begin{pmatrix} a & -b \\ b & a \end{pmatrix}$, i.e. the CR equations. So for smooth maps of the plane, "conformal" and "holomorphic with nonzero derivative" are the same thing. The reflection $z \mapsto \bar z$ preserves angle **sizes** but reverses their orientation; it is *anti*-conformal.

### 2.4 Critical points multiply angles

A **critical point** of a holomorphic $f$ is a point with $f'(z_0) = 0$. Near such a point, the Taylor expansion (module 06) starts with a higher power:

$$
f(z) = f(z_0) + c\,(z - z_0)^m + \dots, \qquad c \neq 0,\ m \ge 2 .
$$

Then a ray leaving $z_0$ at angle $\theta$, $z = z_0 + t e^{i\theta}$, is mapped to approximately $f(z_0) + c\, t^m e^{im\theta}$, which leaves $f(z_0)$ at angle $\arg c + m\theta$. **Angles at $z_0$ are multiplied by $m$.** The simplest case is $f(z) = z^2$ at $0$ ($m = 2$): two rays at angle $\alpha$ become two rays at angle $2\alpha$, and the right angle between the positive real and positive imaginary axes opens into a straight angle (both axes land on the real line, the positive one on $[0, \infty)$ and the imaginary one on $(-\infty, 0]$).

### 2.5 Three model maps

| map | $f'(z)$ | critical points | grid lines become |
|---|---|---|---|
| $z^2$ | $2z$ | $0$ | $x = a$ and $y = b$ become two orthogonal families of parabolas |
| $e^z$ | $e^z$ (never $0$) | none | $x = a$ becomes the circle $\lvert w \rvert = e^a$; $y = b$ becomes the ray $\arg w = b$ |
| $J(z) = z + \dfrac{1}{z}$ | $1 - \dfrac{1}{z^2}$ | $\pm 1$ | circles $\lvert z \rvert = R$ become ellipses; rays become hyperbola branches |

The **Joukowski map** $J$ sends the unit circle $z = e^{i\theta}$ to $2\cos\theta$, i.e. the segment $[-2, 2]$ traversed twice. A circle that passes through the critical point $z = 1$ but is slightly offset from the origin is mapped to a closed curve with a rounded front and a **sharp cusp** at $J(1) = 2$: an airfoil. The cusp is §2.4 in action — at $z = 1$, $m = 2$, so the straight angle $\pi$ of the circle's tangent there becomes an angle of $2\pi$: the two sides of the image meet tangentially.

## 3. Worked Examples

### Example 1 — local rotation and scaling of $z^2$ at $1 + i$

$f'(1 + i) = 2(1 + i) = 2\sqrt{2}\, e^{i\pi/4}$. Near $1 + i$, the map rotates by $45°$ and stretches by $2\sqrt2 \approx 2.83$; areas grow by $8$. A small horizontal step $h = 0.01$ goes to $f(1 + i + 0.01) - f(1+i) \approx 0.02 + 0.02i$, pointing at $45°$.

### Example 2 — $e^z$ at $z_0 = i\pi/2$

$f'(i\pi/2) = e^{i\pi/2} = i$: rotate by $90°$, scale by $1$. Indeed the horizontal line $y = \pi/2$ maps to the ray $\arg w = \pi/2$ (the positive imaginary axis), which is the horizontal direction rotated by $90°$.

### Example 3 — the image of the grid under $e^z$

$e^{a + iy} = e^a e^{iy}$: the vertical line $x = a$ maps to the circle of radius $e^a$. $e^{x + ib} = e^x e^{ib}$: the horizontal line $y = b$ maps to the ray at angle $b$. Circles centred at $0$ and rays from $0$ meet at right angles everywhere ✓. The scaling factor $\lvert e^z \rvert = e^x$ grows to the right: the image "squares" (polar rectangles) get bigger further from the origin.

### Example 4 — grid lines under $z^2$

$w = z^2 = (x^2 - y^2) + 2ixy$. On $x = a \neq 0$: $u = a^2 - y^2$, $v = 2ay$, so $y = v/(2a)$ and

$$
u = a^2 - \frac{v^2}{4a^2},
$$

a parabola opening to the left. On $y = b \neq 0$: $u = \frac{v^2}{4b^2} - b^2$, a parabola opening to the right. At the image of $a + ib$ the two parabolas have slopes $\frac{du}{dv} = -\frac{v}{2a^2} = -\frac{b}{a}$ and $\frac{v}{2b^2} = \frac{a}{b}$ — product $-1$, perpendicular ✓.

### Example 5 — angle doubling at the critical point

Under $z^2$ the rays $\arg z = 0$ and $\arg z = \pi/6$ meet at $0$ at $30°$. Their images are $\arg w = 0$ and $\arg w = \pi/3$, meeting at $60°$. Away from $0$ — say at $z_0 = 1$, where $f'(1) = 2 \neq 0$ — angles are preserved.

### Example 6 — Joukowski critical points and the circle

$J'(z) = 1 - z^{-2} = 0 \iff z = \pm 1$, with $J(\pm 1) = \pm 2$. On the unit circle $J(e^{i\theta}) = e^{i\theta} + e^{-i\theta} = 2\cos\theta$, a flat segment. On $\lvert z \rvert = R > 1$: $J(Re^{i\theta}) = (R + R^{-1})\cos\theta + i(R - R^{-1})\sin\theta$, an ellipse with semi-axes $R + 1/R$ and $R - 1/R$ and foci $\pm 2$.

## 4. Python Examples

### 4.1 Mapping tangent vectors

Push two tangent vectors at $z_0$ through $f$ numerically (by mapping $z_0 + \varepsilon v$) and compare with $f'(z_0)\, v$. Then compare the angles between them before and after.

```python
import numpy as np


def pushed_tangent(f, z0, direction, eps=1e-7):
    return (f(z0 + eps * direction) - f(z0 - eps * direction)) / (2 * eps)


def angle_between(v1, v2):
    return np.angle(v2 / v1)


z0 = 1 + 1j
f = lambda z: z**2
derivative = 2 * z0

v1 = np.exp(1j * 0.2)
v2 = np.exp(1j * 1.4)
w1 = pushed_tangent(f, z0, v1)
w2 = pushed_tangent(f, z0, v2)

print("f'(z0)                    =", derivative, "  |f'| =", abs(derivative), "  arg f' =", np.angle(derivative))
print("pushed v1 vs f'(z0) v1    :", np.round(w1, 6), np.round(derivative * v1, 6))
print("length ratio |w1|/|v1|    =", round(abs(w1) / abs(v1), 6))
print("rotation arg(w1/v1)       =", round(np.angle(w1 / v1), 6))
print("angle v1 -> v2 before     =", round(angle_between(v1, v2), 6))
print("angle w1 -> w2 after      =", round(angle_between(w1, w2), 6))
```

### 4.2 Angle preservation, tested at random

Pick random points and random pairs of directions for several maps, and record the largest change of angle. Conformal maps give rounding noise; the non-holomorphic map $x^2 + iy^2$ does not.

```python
import numpy as np

rng = np.random.default_rng(1)
maps = {
    "z^2": lambda z: z**2,
    "exp(z)": np.exp,
    "z + 1/z": lambda z: z + 1 / z,
    "conj(z)": np.conj,
    "x^2 + i y^2": lambda z: z.real**2 + 1j * z.imag**2,
}
points = rng.uniform(0.3, 2, 200) * np.exp(1j * rng.uniform(-np.pi, np.pi, 200))
directions_1 = np.exp(1j * rng.uniform(-np.pi, np.pi, 200))
directions_2 = np.exp(1j * rng.uniform(-np.pi, np.pi, 200))
eps = 1e-6

for name, f in maps.items():
    w1 = (f(points + eps * directions_1) - f(points - eps * directions_1)) / (2 * eps)
    w2 = (f(points + eps * directions_2) - f(points - eps * directions_2)) / (2 * eps)
    angle_before = np.angle(directions_2 / directions_1)
    angle_after = np.angle(w2 / w1)
    size_change = np.max(np.abs(np.abs(angle_after) - np.abs(angle_before)))
    signed_change = np.max(np.abs(angle_after - angle_before))
    print(f"{name:12}  max change of |angle| = {size_change:.2e}   max change of signed angle = {signed_change:.2e}")
```

$\bar z$ keeps every angle's **size** but flips its sign (the signed change is up to $2\pi$), and $x^2 + iy^2$ changes even the sizes. The three holomorphic maps change angles only at the level of finite-difference rounding (the random points almost surely miss their critical points $0$ and $\pm 1$).

### 4.3 Symbolic check: the parabola families of $z^2$ are orthogonal

The images of $x = a$ and $y = b$ are the level curves $\{\operatorname{Re} z = a\}$ and $\{\operatorname{Im} z = b\}$ seen in the $w$-plane. Their tangent directions at $w = (a + ib)^2$ are $f' \cdot i$ (from the vertical line) and $f' \cdot 1$ (from the horizontal line).

```python
import sympy as sp

a, b, s = sp.symbols("a b s", real=True)

vertical_image = sp.expand((a + sp.I * s) ** 2)         # x = a, parameter s = y
horizontal_image = sp.expand((s + sp.I * b) ** 2)       # y = b, parameter s = x

tangent_1 = sp.diff(vertical_image, s).subs(s, b)
tangent_2 = sp.diff(horizontal_image, s).subs(s, a)
dot_product = sp.simplify(sp.re(tangent_1) * sp.re(tangent_2) + sp.im(tangent_1) * sp.im(tangent_2))

print("tangent of image of x = a :", sp.simplify(tangent_1), "  (= f'(a+ib) * i)")
print("tangent of image of y = b :", sp.simplify(tangent_2), "  (= f'(a+ib) * 1)")
print("dot product               :", dot_product)
```

### 4.4 Joukowski: critical points and images of circles

```python
import numpy as np
import sympy as sp

z = sp.symbols("z")
joukowski = z + 1 / z
print("J'(z)          =", sp.diff(joukowski, z))
print("critical points:", sp.solve(sp.diff(joukowski, z), z), "  images:", [joukowski.subs(z, c) for c in [-1, 1]])

theta = np.linspace(0, 2 * np.pi, 9)
unit_circle_image = np.exp(1j * theta) + np.exp(-1j * theta)
print("J on |z| = 1   :", np.round(unit_circle_image.real, 4), " imaginary parts all ~0:", np.allclose(unit_circle_image.imag, 0))

radius = 1.5
ellipse = radius * np.exp(1j * theta) + 1 / (radius * np.exp(1j * theta))
print(f"J on |z| = {radius}: semi-axes {radius + 1 / radius:.4f}, {radius - 1 / radius:.4f};",
      "max Re =", round(ellipse.real.max(), 4), " max Im =", round(ellipse.imag.max(), 4))
```

### 4.5 Angle multiplication at a critical point

Rays from $0$ at angles $\theta_1, \theta_2$ under $z^2$ and $z^3$: the angle between the images is $2(\theta_2 - \theta_1)$ and $3(\theta_2 - \theta_1)$.

```python
import numpy as np

theta_1, theta_2 = 0.0, np.pi / 6
t = 1e-3
for power in [2, 3]:
    image_1 = (t * np.exp(1j * theta_1)) ** power
    image_2 = (t * np.exp(1j * theta_2)) ** power
    angle_after = np.angle(image_2 / image_1)
    print(f"z^{power}: angle before = {np.degrees(theta_2 - theta_1):.1f} deg, after = {np.degrees(angle_after):.1f} deg")

z0 = 1.0
for power in [2, 3]:
    tangent_1 = power * z0 ** (power - 1) * np.exp(1j * theta_1)
    tangent_2 = power * z0 ** (power - 1) * np.exp(1j * theta_2)
    print(f"z^{power} at z0 = 1 (not critical): angle after = {np.degrees(np.angle(tangent_2 / tangent_1)):.1f} deg")
```

## 5. Visualizations

A helper draws a grid in the $z$-plane (vertical lines in blue, horizontal lines in orange) and its image in the $w$-plane. Wherever $f' \neq 0$, blue and orange curves cross at right angles.

```python
import matplotlib.pyplot as plt
import numpy as np


def plot_grid_and_image(f, x_values, y_values, title, w_limits, z_limits=None, samples=400, axes=None):
    if axes is None:
        fig, axes = plt.subplots(1, 2, figsize=(13, 6))
    zplane, wplane = axes
    x_range = (min(x_values), max(x_values))
    y_range = (min(y_values), max(y_values))
    for a in x_values:
        line = a + 1j * np.linspace(*y_range, samples)
        zplane.plot(line.real, line.imag, color="tab:blue", linewidth=1)
        image = f(line)
        wplane.plot(image.real, image.imag, color="tab:blue", linewidth=1)
    for b in y_values:
        line = np.linspace(*x_range, samples) + 1j * b
        zplane.plot(line.real, line.imag, color="tab:orange", linewidth=1)
        image = f(line)
        wplane.plot(image.real, image.imag, color="tab:orange", linewidth=1)
    zplane.set_title("z-plane grid")
    wplane.set_title(title)
    if z_limits is not None:
        zplane.set_xlim(*z_limits[0])
        zplane.set_ylim(*z_limits[1])
    wplane.set_xlim(*w_limits[0])
    wplane.set_ylim(*w_limits[1])
    for ax in axes:
        ax.axhline(0, color="black", linewidth=0.6)
        ax.axvline(0, color="black", linewidth=0.6)
        ax.set_aspect("equal")
        ax.set_xlabel("Re")
        ax.set_ylabel("Im")
        ax.grid(True, alpha=0.2)
    return axes


fig, axes = plt.subplots(1, 2, figsize=(13, 6))
plot_grid_and_image(lambda z: z**2, np.linspace(0.2, 2, 10), np.linspace(0.2, 2, 10),
                    r"image under $w = z^2$: orthogonal parabolas", ((-4, 4), (0, 8.5)), axes=axes)
plt.tight_layout()
plt.show()
```

### 5.1 $e^z$: lines become circles and rays

The strip $0 \le y \le \pi$ maps onto the upper half-plane minus the origin. Vertical lines (blue) become semicircles of radius $e^a$; horizontal lines (orange) become rays. Each little grid square becomes a little "polar square", larger further out.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(1, 2, figsize=(13, 6))
plot_grid_and_image(np.exp, np.linspace(-1.5, 1.2, 10), np.linspace(0, np.pi, 10),
                    r"image under $w = e^z$: circles and rays", ((-3.6, 3.6), (-0.3, 3.6)), axes=axes)
plt.tight_layout()
plt.show()
```

### 5.2 Zooming in: small squares stay squares

Take a tiny grid around $z_0 = 1 + i$ and map it by $z^2$, then undo the predicted local action by dividing by $f'(z_0)$. If the map is conformal, the result is (nearly) the original grid: all the map did locally was multiply by $f'(z_0) = 2 + 2i$. The non-holomorphic map $x^2 + iy^2$, treated the same way with its Jacobian, gives a sheared grid.

```python
import matplotlib.pyplot as plt
import numpy as np

z0 = 1 + 0.4j
half_width = 0.05
offsets = np.linspace(-half_width, half_width, 6)
param = np.linspace(-half_width, half_width, 100)

holomorphic = lambda z: z**2
non_holomorphic = lambda z: z.real**2 + 1j * z.imag**2

fig, axes = plt.subplots(1, 3, figsize=(17, 5.5))
titles = ["tiny grid around $z_0 = 1 + 0.4i$",
          r"$(z^2 - z_0^2) / f'(z_0)$: same grid back",
          r"$x^2 + i y^2$, divided by its value $2x_0$ on the real direction"]
for k, (ax, title) in enumerate(zip(axes, titles)):
    for c in offsets:
        for line, colour in [(z0 + c + 1j * param, "tab:blue"), (z0 + param + 1j * c, "tab:orange")]:
            if k == 0:
                image = line - z0
            elif k == 1:
                image = (holomorphic(line) - holomorphic(z0)) / (2 * z0)
            else:
                image = (non_holomorphic(line) - non_holomorphic(z0)) / (2 * z0.real)
            ax.plot(image.real, image.imag, color=colour)
    ax.set_xlim(-0.07, 0.07)
    ax.set_ylim(-0.07, 0.07)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Re (offset)")
    ax.set_ylabel("Im (offset)")
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

For $x^2 + iy^2$ at $z_0 = 1 + 0.4i$ the Jacobian is $\operatorname{diag}(2, 0.8)$: horizontal steps are stretched by $2$ and vertical ones by only $0.8$, so the squares become flat rectangles. No single complex number can undo that.

### 5.3 The critical point of $z^2$: angles double

Left: rays from $0$ every $30°$ in the first quadrant, plus a small arc showing the $90°$ corner. Right: their images under $z^2$ fan out every $60°$, and the corner opens up to $180°$. A second family of rays from $z_0 = 1$ (not critical) is mapped with angles intact.

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 1.2, 100)
ray_angles = np.radians([0, 30, 60, 90])
colours = ["tab:blue", "tab:orange", "tab:green", "tab:red"]

fig, (zplane, wplane) = plt.subplots(1, 2, figsize=(13, 6))
for angle, colour in zip(ray_angles, colours):
    ray = t * np.exp(1j * angle)
    zplane.plot(ray.real, ray.imag, color=colour, linewidth=2, label=f"{np.degrees(angle):.0f} deg")
    wplane.plot((ray**2).real, (ray**2).imag, color=colour, linewidth=2, label=f"image: {2 * np.degrees(angle):.0f} deg")
    short_ray = 1 + 0.35 * t * np.exp(1j * angle)
    zplane.plot(short_ray.real, short_ray.imag, color=colour, linewidth=1, linestyle="--")
    wplane.plot((short_ray**2).real, (short_ray**2).imag, color=colour, linewidth=1, linestyle="--")

arc = 0.3 * np.exp(1j * np.linspace(0, np.pi / 2, 50))
zplane.plot(arc.real, arc.imag, color="black")
arc_image = 0.3 * np.exp(1j * np.linspace(0, np.pi, 50))
wplane.plot(arc_image.real, arc_image.imag, color="black")
zplane.text(0.25, 0.25, "90 deg", fontsize=10)
wplane.text(-0.15, 0.35, "180 deg", fontsize=10)

zplane.set_title("rays at 30 deg spacing from 0 (solid) and from 1 (dashed)")
wplane.set_title(r"under $z^2$: doubled at 0, preserved at $1 \mapsto 1$")
for ax, (xlim, ylim) in [(zplane, ((-0.5, 1.6), (-0.3, 1.4))), (wplane, ((-1.6, 2.6), (-0.3, 1.8)))]:
    ax.axhline(0, color="black", linewidth=0.6)
    ax.axvline(0, color="black", linewidth=0.6)
    ax.set_xlim(*xlim)
    ax.set_ylim(*ylim)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="upper right", fontsize=9)
plt.tight_layout()
plt.show()
```

### 5.4 The Joukowski airfoil

Take the circle through $z = 1$ with centre $c = -0.1 + 0.12i$ (radius $\lvert 1 - c \rvert$). Its image under $J(z) = z + 1/z$ is a cambered airfoil with a cusp at $J(1) = 2$. We also map a polar grid around the circle (concentric circles about $c$ in blue and radial lines in orange): outside the critical point the images still cross at right angles.

```python
import matplotlib.pyplot as plt
import numpy as np


def joukowski(z):
    return z + 1 / z


centre = -0.1 + 0.12j
radius = abs(1 - centre)
theta = np.linspace(0, 2 * np.pi, 600)

fig, (zplane, wplane) = plt.subplots(1, 2, figsize=(15, 6))
for scale in [1.0, 1.15, 1.35, 1.6, 2.0, 2.6]:
    circle = centre + scale * radius * np.exp(1j * theta)
    style = dict(color="black", linewidth=2.5) if scale == 1.0 else dict(color="tab:blue", linewidth=0.9)
    zplane.plot(circle.real, circle.imag, **style)
    image = joukowski(circle)
    wplane.plot(image.real, image.imag, **style)

radial_parameter = np.linspace(1.0, 2.6, 100) * radius
for angle in np.linspace(0, 2 * np.pi, 24, endpoint=False):
    spoke = centre + radial_parameter * np.exp(1j * angle)
    zplane.plot(spoke.real, spoke.imag, color="tab:orange", linewidth=0.9)
    image = joukowski(spoke)
    wplane.plot(image.real, image.imag, color="tab:orange", linewidth=0.9)

zplane.plot([1, -1], [0, 0], "rx", markersize=10, markeredgewidth=2, label=r"critical points $\pm 1$")
wplane.plot([2], [0], "rx", markersize=10, markeredgewidth=2, label=r"cusp at $J(1) = 2$")
zplane.set_title(r"circle through $z = 1$, centre $-0.1 + 0.12i$")
wplane.set_title(r"image under $J(z) = z + 1/z$: an airfoil")
for ax, limit in [(zplane, 3.2), (wplane, 4.0)]:
    ax.axhline(0, color="black", linewidth=0.6)
    ax.axvline(0, color="black", linewidth=0.6)
    ax.set_xlim(-limit, limit)
    ax.set_ylim(-limit * 0.8, limit * 0.8)
    ax.set_aspect("equal")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.legend(loc="lower left")
plt.tight_layout()
plt.show()
```

Zooming in on the trailing edge shows the cusp. At moderate zoom the edge looks like a thin wedge, but the wedge closes up as we zoom further: the two sides of the airfoil meet at angle $0$ (they are tangent), because the circle's straight angle $\pi$ at the critical point $z = 1$ has been doubled to $2\pi$.

```python
import matplotlib.pyplot as plt
import numpy as np

centre = -0.1 + 0.12j
radius = abs(1 - centre)
theta = np.linspace(0, 2 * np.pi, 200001)
circle = centre + radius * np.exp(1j * theta)
airfoil = circle + 1 / circle

fig, axes = plt.subplots(1, 2, figsize=(15, 4.5))
windows = [((1.6, 2.03), (-0.04, 0.12)), ((1.986, 2.001), (-0.0005, 0.0035))]
titles = ["near the trailing edge", "zoomed 30x further: the sides become tangent (a cusp)"]
for ax, (xlim, ylim), title in zip(axes, windows, titles):
    ax.plot(airfoil.real, airfoil.imag, color="black", linewidth=1.5)
    ax.plot([2], [0], "rx", markersize=12, markeredgewidth=2)
    ax.set_xlim(*xlim)
    ax.set_ylim(*ylim)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 7. Solutions

1. $f'(i) = 3i^2 = -3 = 3e^{i\pi}$: rotation by $\pi$ ($180°$), scaling by $3$ (areas by $9$).
2. $f'(z) = 3z^2 - 3 = 0 \iff z = \pm 1$. Conformal on $\mathbb{C} \setminus \{1, -1\}$; at $\pm 1$ the derivative vanishes ($m = 2$) and angles are doubled.
3. $e^{iy}$ for $0 \le y \le \pi$: the upper half of the unit circle, from $1$ to $-1$. $e^{x + i\pi/4} = e^x e^{i\pi/4}$: the open ray from $0$ at angle $45°$ (not including $0$).
4. $1/z$ sends lines not through $0$ to circles through $0$. The vertical line $x = a$ goes to the circle with centre $\frac{1}{2a}$ and radius $\frac{1}{2a}$ (tangent to the imaginary axis at $0$); the horizontal line $y = b$ goes to the circle with centre $-\frac{i}{2b}$ and radius $\frac{1}{2b}$ (tangent to the real axis at $0$, below it). Each blue circle meets each orange circle at right angles. See the code below.
5. $\arg(\bar v_2 / \bar v_1) = \arg \overline{(v_2 / v_1)} = -\arg(v_2/v_1)$. Same size, opposite sign: orientation reversed.
6. $\lvert f'(1 + i) \rvert^2 = \lvert 2 + 2i \rvert^2 = 8$, so the image area is about $8 \times 10^{-4}$. The exact value (computed below) agrees to about $1\%$ — the square is small but not infinitesimal.
7. At $z = \pm 1$ (and $J$ is not defined at $0$). $J(2e^{i\theta}) = 2.5\cos\theta + 1.5 i \sin\theta$: an ellipse with semi-axes $2.5$ and $1.5$.
8. $y = 1$: $z = s + i$, $w = s^2 - 1 + 2is$, tangent $\frac{dw}{ds} = 2s + 2i$, which at $s = 1$ is $2 + 2i$. $x = 1$: $z = 1 + is$, $w = 1 - s^2 + 2is$, tangent $-2s + 2i$, at $s = 1$: $-2 + 2i$. Dot product: $(2)(-2) + (2)(2) = 0$ ✓. Both pass through $w = 2i$.
9. Solving $c(z - z_0)^m \approx w - f(z_0)$ gives $z - z_0 \approx \left( \frac{w - f(z_0)}{c} \right)^{1/m}$, and there are $m$ different $m$-th roots (notebook 01/05). So each nearby $w$ has $m$ nearby preimages, equally spaced in angle — the map wraps a small disc $m$ times around $f(z_0)$, which is why it multiplies angles by $m$. For $z^3 = 10^{-3}$: preimages $0.1, 0.1e^{2\pi i/3}, 0.1e^{4\pi i/3}$.
10. If the circle misses $z = 1$, every point of it has $J' \neq 0$, so $J$ is conformal along the whole curve and the smooth circle maps to a smooth closed curve without corners: the trailing edge is rounded. (In aerodynamics the sharp edge matters: it fixes the circulation via the Kutta condition.) See the plot below.

## 8. Conclusion

- At a point with $f'(z_0) \neq 0$, a holomorphic map acts on tangent vectors by multiplication with $f'(z_0)$: rotate by $\arg f'(z_0)$, scale by $\lvert f'(z_0) \rvert$, the same for every direction.
- Hence it preserves angles between curves, in size and orientation (**conformal**); orthogonal grids map to orthogonal curve families, and tiny squares to tiny squares with area factor $\lvert f' \rvert^2$.
- $z^2$ turns the grid into orthogonal parabolas, $e^z$ into circles and rays, the Joukowski map $z + 1/z$ turns offset circles into airfoils.
- At a critical point where $f(z) - f(z_0) \sim c(z - z_0)^m$, angles are multiplied by $m$: $z^2$ doubles angles at $0$, and the Joukowski map's critical point at $1$ produces the airfoil's cusp.
- $\bar z$ preserves angle sizes but flips orientation; non-holomorphic maps such as $x^2 + iy^2$ shear.
