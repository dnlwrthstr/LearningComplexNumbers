# Holomorphic Functions

Notebook 02 found functions that are complex-differentiable at a single point ($\lvert z \rvert^2$ at $0$) or along a single line ($x^2 + iy^2$ on $y = x$). Those are curiosities. The functions complex analysis is built on are differentiable at **every point of an open set** — they are called **holomorphic**. This notebook defines them, shows that sums, products, quotients and compositions of holomorphic functions stay holomorphic, and introduces a one-line test that replaces the two Cauchy–Riemann equations: the **Wirtinger derivative** $\partial f / \partial \bar z$ must vanish. In words: *a holomorphic function is a function of $z$ alone, with no $\bar z$ in it.* That is why $\bar z$, $\lvert z \rvert^2 = z \bar z$ and $\operatorname{Re} z = \frac{z + \bar z}{2}$ are not holomorphic.

This is notebook 03 of module 05 (**Complex differentiation**) in the **Learning Complex Numbers** series.

## 1. Introduction

In real analysis there is a ladder of smoothness: continuous, differentiable, differentiable with continuous derivative ($C^1$), twice differentiable, …, infinitely differentiable ($C^\infty$), and finally analytic (equal to its Taylor series). Each rung is strictly stronger than the one below: $x \lvert x \rvert$ is $C^1$ but not twice differentiable; $e^{-1/x^2}$ is $C^\infty$ but not analytic at $0$.

In complex analysis the ladder **collapses**. Being complex-differentiable on an open set — holomorphic — already implies infinitely many derivatives and a convergent power series. That is a theorem of module 06 (Cauchy's integral formula), but it is the reason this notebook's definition matters so much: one derivative on an open set buys everything.

## 2. Theory

### 2.1 Definitions

A set $U \subseteq \mathbb{C}$ is **open** if every point of $U$ is the centre of some disc contained in $U$. A **domain** is an open set that is connected (any two points can be joined by a path inside it).

- $f$ is **holomorphic on an open set $U$** if $f'(z)$ exists at every $z \in U$.
- $f$ is **holomorphic at $z_0$** if it is holomorphic on some disc $D(z_0, r)$.
- $f$ is **entire** if it is holomorphic on all of $\mathbb{C}$.

"Analytic" is often used as a synonym; module 06 shows the two notions coincide.

The open-set requirement rules out the examples of notebook 02: $\lvert z \rvert^2$ is differentiable only at $0$ and $x^2 + iy^2$ only on the line $y = x$, and neither a point nor a line contains a disc. So **neither is holomorphic anywhere**.

### 2.2 Closure rules

If $f$ and $g$ are holomorphic on $U$, then so are $f + g$, $f - g$, $fg$, $cf$ for constants $c$, and $f/g$ on the open set where $g \neq 0$. If $g$ is holomorphic on $U$, $f$ is holomorphic on $V$, and $g(U) \subseteq V$, then $f \circ g$ is holomorphic on $U$ with $(f \circ g)' = (f' \circ g)\, g'$. All of these follow pointwise from the differentiation rules of notebook 02.

Starting from the building blocks $z$, constants and $e^z$:

| function | holomorphic on |
|---|---|
| polynomials | $\mathbb{C}$ (entire) |
| rational $p/q$ | $\mathbb{C}$ minus the zeros of $q$ |
| $e^z$, $\sin z = \frac{e^{iz} - e^{-iz}}{2i}$, $\cos z$, $\sinh z$, $\cosh z$ | $\mathbb{C}$ (entire) |
| $\tan z = \sin z / \cos z$ | $\mathbb{C}$ minus $\{\frac{\pi}{2} + k\pi\}$ |
| $e^{z^2}$, $\sin(e^z)$, $\frac{1}{1 + e^z}$ | by composition / quotient |
| $\operatorname{Log} z$ (principal branch) | $\mathbb{C} \setminus (-\infty, 0]$, with $(\operatorname{Log} z)' = 1/z$ |

### 2.3 The Wirtinger derivatives

Since $x = \frac{z + \bar z}{2}$ and $y = \frac{z - \bar z}{2i}$, any function of $(x, y)$ can be rewritten as a function of $z$ and $\bar z$. Treating $z$ and $\bar z$ **formally as independent variables** and applying the chain rule suggests the definitions

$$
\frac{\partial}{\partial z} = \frac{1}{2}\left( \frac{\partial}{\partial x} - i \frac{\partial}{\partial y} \right),
\qquad
\frac{\partial}{\partial \bar z} = \frac{1}{2}\left( \frac{\partial}{\partial x} + i \frac{\partial}{\partial y} \right).
$$

These are honest operators on real-differentiable functions (no "independence" assumption is needed to *use* them). They satisfy $\frac{\partial z}{\partial z} = 1$, $\frac{\partial \bar z}{\partial z} = 0$, $\frac{\partial z}{\partial \bar z} = 0$, $\frac{\partial \bar z}{\partial \bar z} = 1$, and the usual sum, product and chain rules.

Expanding with $f = u + iv$:

$$
\frac{\partial f}{\partial \bar z} = \frac{1}{2}\big[ (u_x - v_y) + i (v_x + u_y) \big].
$$

The real part is the first CR residual and the imaginary part is the second. Therefore

$$
\boxed{\; \text{CR equations at } z_0 \iff \frac{\partial f}{\partial \bar z}(z_0) = 0 \;}
\qquad\text{and then}\qquad
f'(z_0) = \frac{\partial f}{\partial z}(z_0).
$$

These are exactly the numbers $B$ and $A$ of notebook 02, §2.6: the directional difference quotients trace $\frac{\partial f}{\partial z} + \frac{\partial f}{\partial \bar z} e^{-2i\theta}$.

**Practical test.** Write $f$ using $z$ and $\bar z$ and differentiate with respect to $\bar z$ as if $z$ were a constant.

| $f$ | in $z, \bar z$ | $\partial f / \partial \bar z$ | holomorphic? |
|---|---|---|---|
| $z^3 + 2z$ | $z^3 + 2z$ | $0$ | yes, entire |
| $\bar z$ | $\bar z$ | $1$ | nowhere |
| $\lvert z \rvert^2$ | $z \bar z$ | $z$ | nowhere (CR only at $0$) |
| $\operatorname{Re} z$ | $\frac{z + \bar z}{2}$ | $\frac{1}{2}$ | nowhere |
| $\lvert z \rvert$ | $(z \bar z)^{1/2}$ | $\frac{z}{2\lvert z \rvert}$ | nowhere |
| $x^2 + iy^2$ | $\frac{(z + \bar z)^2}{4} - i\frac{(z - \bar z)^2}{4}$ | $x - y = \frac{(1+i)z + (1-i)\bar z}{2}$ | nowhere (zero on $y = x$) |

### 2.4 Rigidity: holomorphic functions are hard to bend

On a domain $U$:

1. If $f' \equiv 0$, then $f$ is constant. ($u_x = u_y = v_x = v_y = 0$, so $u$ and $v$ are constant along every path.)
2. If $f$ is real-valued (or $\operatorname{Re} f$ is constant, or $\operatorname{Im} f$ is constant), then $f$ is constant. (If $v \equiv 0$, CR gives $u_x = v_y = 0$ and $u_y = -v_x = 0$.)
3. If $\lvert f \rvert$ is constant, then $f$ is constant. (Exercise 9.)

Real functions have no such constraints — $\sin x$ is real-valued and far from constant. Holomorphy couples $u$ and $v$ so tightly that you cannot change one without the other.

### 2.5 Preview: one derivative gives all of them

**Theorem (proved in module 06).** If $f$ is holomorphic on an open set $U$, then $f'$ is also holomorphic on $U$. Hence $f$ has derivatives of every order, and around each $z_0 \in U$ it equals its Taylor series $\sum_{n} \frac{f^{(n)}(z_0)}{n!} (z - z_0)^n$ on the largest disc inside $U$.

The proof needs integration: Cauchy's integral formula ([06/03](../06_complex_integration/03_cauchys_integral_formula.ipynb)) expresses $f^{(n)}(z_0)$ as an integral of $f$ around a circle, and power series ([06/04](../06_complex_integration/04_power_series_and_radius_of_convergence.ipynb)) follow. Nothing like this holds for real functions: $x \lvert x \rvert$ has a derivative $2\lvert x \rvert$ everywhere, but no second derivative at $0$. Its complex analogue $z \lvert z \rvert$ is simply not holomorphic.

## 3. Worked Examples

### Example 1 — where is $f(z) = \dfrac{e^z}{z^2 + 1}$ holomorphic?

$e^z$ and $z^2 + 1$ are entire, so the quotient is holomorphic wherever $z^2 + 1 \neq 0$: on $\mathbb{C} \setminus \{i, -i\}$. Its derivative by the quotient rule is $\dfrac{e^z(z^2 + 1) - e^z \cdot 2z}{(z^2+1)^2} = \dfrac{e^z (z - 1)^2}{(z^2 + 1)^2}$.

### Example 2 — the Wirtinger test on $\lvert z \rvert^2$

$\lvert z \rvert^2 = z \bar z$, so $\frac{\partial}{\partial \bar z}(z \bar z) = z$. This vanishes only at $z = 0$: CR holds only at the origin (as in notebook 02), and since $\{0\}$ is not open, $\lvert z \rvert^2$ is holomorphic nowhere. Also $\frac{\partial}{\partial z}(z\bar z) = \bar z$, which equals $f'(0) = 0$ at the origin.

### Example 3 — the Wirtinger test on $\operatorname{Re} z$ and $\operatorname{Im} z$

$\operatorname{Re} z = \frac{z + \bar z}{2}$ gives $\partial / \partial \bar z = \frac12 \neq 0$. $\operatorname{Im} z = \frac{z - \bar z}{2i}$ gives $-\frac{1}{2i} = \frac{i}{2} \neq 0$. Neither is holomorphic anywhere — as rigidity rule 2 predicts for any non-constant real-valued function.

### Example 4 — a function in disguise

Is $f(x + iy) = x^2 - y^2 + 2ixy + 3x + 3iy$ holomorphic? Substitute $x = \frac{z + \bar z}{2}$, $y = \frac{z - \bar z}{2i}$:

- $x^2 - y^2 + 2ixy = (x + iy)^2 = z^2$,
- $3x + 3iy = 3z$.

So $f = z^2 + 3z$, with no $\bar z$: entire, $f' = 2z + 3$.

### Example 5 — composition with a non-holomorphic piece

$g(z) = e^{\bar z}$. By the chain rule for Wirtinger derivatives, $\frac{\partial}{\partial \bar z} e^{\bar z} = e^{\bar z} \neq 0$ everywhere, so $g$ is nowhere holomorphic, even though $e^w$ is entire. A composition $f \circ g$ is guaranteed holomorphic only when **both** pieces are.

### Example 6 — rigidity in action

Suppose $f$ is entire and $\operatorname{Re} f(z) = x^2 - y^2$. Then $u = x^2 - y^2$, and CR forces $v_y = u_x = 2x$, $v_x = -u_y = 2y$, so $v = 2xy + C$ and $f = z^2 + iC$. The real part alone determines $f$ up to an imaginary constant (notebook 05 makes this systematic: the *harmonic conjugate*).

## 4. Python Examples

### 4.1 Wirtinger derivatives in sympy

Introduce independent symbols `z` and `zb` (for $\bar z$), rewrite $f(x, y)$ using $x = \frac{z + \bar z}{2}$, $y = \frac{z - \bar z}{2i}$, and differentiate.

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
z, zb = sp.symbols("z zb")          # zb plays the role of conj(z), treated as independent


def in_z_and_zbar(f_xy):
    return sp.simplify(sp.expand(f_xy.subs({x: (z + zb) / 2, y: (z - zb) / (2 * sp.I)}, simultaneous=True)))


def wirtinger(f_xy):
    g = in_z_and_zbar(f_xy)
    return g, sp.simplify(sp.diff(g, z)), sp.simplify(sp.diff(g, zb))


examples = {
    "z^3 + 2z": sp.expand((x + sp.I * y) ** 3 + 2 * (x + sp.I * y)),
    "conj(z)": x - sp.I * y,
    "|z|^2": x**2 + y**2,
    "Re z": x,
    "Im z": y,
    "x^2 + i y^2": x**2 + sp.I * y**2,
    "x^2-y^2+2ixy+3x+3iy": x**2 - y**2 + 2 * sp.I * x * y + 3 * x + 3 * sp.I * y,
}
for name, f in examples.items():
    g, d_dz, d_dzb = wirtinger(f)
    print(f"{name:22} = {str(g):34} d/dz = {str(d_dz):22} d/dzbar = {d_dzb}")
```

### 4.2 The Wirtinger operator from partial derivatives

The same numbers come straight from the definition $\frac{\partial}{\partial \bar z} = \frac12(\partial_x + i\partial_y)$ applied to $f(x, y)$, without any substitution. Both routes agree.

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)


def d_dzbar(f_xy):
    return sp.simplify((sp.diff(f_xy, x) + sp.I * sp.diff(f_xy, y)) / 2)


def d_dz(f_xy):
    return sp.simplify((sp.diff(f_xy, x) - sp.I * sp.diff(f_xy, y)) / 2)


for name, f in {"|z|^2": x**2 + y**2, "exp(z)": sp.exp(x) * (sp.cos(y) + sp.I * sp.sin(y)),
                "exp(conj z)": sp.exp(x) * (sp.cos(y) - sp.I * sp.sin(y)), "|z|": sp.sqrt(x**2 + y**2)}.items():
    print(f"{name:12}  d/dz = {str(d_dz(f)):42}  d/dzbar = {d_dzbar(f)}")
```

For $e^z$: $\partial / \partial \bar z = 0$ and $\partial / \partial z = e^x(\cos y + i\sin y) = e^z$, the derivative. For $\lvert z \rvert$: $\partial / \partial \bar z = \frac{x + iy}{2\lvert z \rvert} = \frac{z}{2\lvert z \rvert}$, never zero.

### 4.3 Closure: holomorphic in, holomorphic out

Build compositions and quotients of holomorphic functions and confirm $\partial f / \partial \bar z = 0$; mix in one $\bar z$ and watch it fail.

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
w = sp.symbols("w")
z_xy = x + sp.I * y


def d_dzbar_of(f_of_w, argument):
    expression = sp.expand_complex(f_of_w.subs(w, argument))
    return sp.simplify(sp.expand_complex((sp.diff(expression, x) + sp.I * sp.diff(expression, y)) / 2))


tests = {
    "exp(z^2)": (sp.exp(w), z_xy**2),
    "sin(exp(z))": (sp.sin(sp.exp(w)), z_xy),
    "1/(1 + z^2)": (1 / (1 + w**2), z_xy),
    "exp(conj z)": (sp.exp(w), sp.conjugate(z_xy)),
    "z * conj z": (w * sp.conjugate(w), z_xy),
}
for name, (outer, inner) in tests.items():
    print(f"{name:12}  d/dzbar = {d_dzbar_of(outer, inner)}")
```

### 4.4 A numerical Wirtinger derivative

For a function given only as numbers (no formula), approximate $\partial f/\partial \bar z$ with central differences. This is how one would test holomorphy of measured or simulated data.

```python
import numpy as np


def numerical_d_dzbar(f, z, step=1e-6):
    d_dx = (f(z + step) - f(z - step)) / (2 * step)
    d_dy = (f(z + 1j * step) - f(z - 1j * step)) / (2 * step)
    return 0.5 * (d_dx + 1j * d_dy)


points = np.array([0.3 + 0.2j, -1 + 2j, 2 - 0.5j])
for name, f in {"z^3": lambda z: z**3, "exp(z)": np.exp, "conj z": np.conj,
                "|z|^2": lambda z: np.abs(z) ** 2, "Re z": lambda z: z.real + 0j}.items():
    values = np.round(numerical_d_dzbar(f, points), 6) + 0.0     # "+ 0.0" turns -0.0 into 0.0
    print(f"{name:7}  d/dzbar at sample points: {values}")
print("expected for |z|^2: z itself =", points)
```

### 4.5 The principal logarithm is holomorphic off its branch cut

$\operatorname{Log} z = \ln\lvert z \rvert + i \operatorname{Arg} z$, so $u = \tfrac12 \ln(x^2 + y^2)$ and $v = \operatorname{atan2}(y, x)$. Away from the negative real axis (where $\operatorname{Arg}$ jumps, notebook 01) $v$ is smooth, and the CR equations hold with $f' = 1/z$.

```python
import numpy as np
import sympy as sp

x, y = sp.symbols("x y", real=True)
u = sp.log(x**2 + y**2) / 2
v = sp.atan2(y, x)

residual_1 = sp.simplify(sp.diff(u, x) - sp.diff(v, y))
residual_2 = sp.simplify(sp.diff(u, y) + sp.diff(v, x))
derivative = sp.simplify(sp.diff(u, x) + sp.I * sp.diff(v, x))
print("u_x - v_y =", residual_1, "   u_y + v_x =", residual_2)
print("u_x + i v_x =", derivative, "  equals 1/z:", sp.simplify(derivative - 1 / (x + sp.I * y)) == 0)

# Numerically: a vertical difference quotient straddling the cut is huge, one away from the cut is not
step = 1e-6
for point in [1 + 0j, 1j, -1 + 0j]:
    vertical = (np.log(point + 1j * step) - np.log(point - 1j * step)) / (2j * step)
    print(f"z0 = {point}:  vertical central difference of Log = {vertical:.4g}   1/z0 = {1 / point:.4g}")
```

At $z_0 = -1$ the quotient is about $\pi / 10^{-6}$: the jump of $2\pi$ in $\operatorname{Arg}$ divided by the step. $\operatorname{Log}$ is holomorphic on $\mathbb{C} \setminus (-\infty, 0]$ but not on the cut.

### 4.6 Rigidity: the real part determines the function

Reproduce Example 6 with sympy. From $u$ alone, the CR equations give $v_y = u_x$ and $v_x = -u_y$; integrate the first in $y$ and fix the leftover function of $x$ with the second.

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
z = sp.symbols("z")


def complete_from_real_part(u):
    v_partial = sp.integrate(sp.diff(u, x), y)                    # v = integral of u_x dy + phi(x)
    phi_prime = sp.simplify(-sp.diff(u, y) - sp.diff(v_partial, x))
    phi = sp.integrate(phi_prime, x)
    v = sp.simplify(v_partial + phi)
    return v


for u in [x**2 - y**2, sp.exp(x) * sp.cos(y), x**3 - 3 * x * y**2 + x]:
    v = complete_from_real_part(u)
    f_xy = sp.simplify(u + sp.I * v)
    in_z = sp.simplify(f_xy.subs({x: z, y: 0}))        # a holomorphic f is determined by its values on the real axis
    print(f"u = {str(u):24} ->  v = {str(v):18} ->  f = {in_z}  (+ i C)")
```

(The substitution $y = 0$, $x = z$ recovers $f$ from its values on the real axis — legitimate for holomorphic $f$ by the identity theorem of module 06, and checked here because each result has no $\bar z$.)

## 5. Visualizations

### 5.1 Maps of $\lvert \partial f / \partial \bar z \rvert$

Colour each point by $\lvert \partial f/\partial \bar z \rvert$. Where it is zero (dark), the CR equations hold. A holomorphic function is dark on a whole open region; the non-examples are dark only on a point, a line, or nowhere.

```python
import matplotlib.pyplot as plt
import numpy as np

grid = np.linspace(-2, 2, 301)
X, Y = np.meshgrid(grid, grid)

panels = [
    (r"$z^2 + e^z$", np.zeros_like(X)),
    (r"$|z|^2$: zero only at 0", np.abs(X + 1j * Y)),
    (r"$x^2 + i y^2$: zero on $y = x$", np.abs(X - Y)),
    (r"$\mathrm{Re}\, z$: never zero", np.full_like(X, 0.5)),
]

fig, axes = plt.subplots(1, 4, figsize=(20, 5), layout="constrained")
for ax, (title, magnitude) in zip(axes, panels):
    image = ax.imshow(magnitude, extent=[-2, 2, -2, 2], origin="lower", cmap="magma", vmin=0, vmax=3)
    ax.axhline(0, color="white", linewidth=0.6)
    ax.axvline(0, color="white", linewidth=0.6)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
fig.colorbar(image, ax=axes, label=r"$|\partial f / \partial \bar{z}|$", shrink=0.8)
plt.show()
```

### 5.2 Domain colouring: holomorphic functions keep the colour wheel turning the same way

In a domain-colouring plot (hue = $\arg f$, brightness rising with $\lvert f \rvert$), walk once counter-clockwise around a zero of a holomorphic function: the colours run red → yellow → green → cyan → blue → magenta, the **same** counter-clockwise order as the colour wheel, once per order of the zero. Replace $z$ by $\bar z$ and every zero has the colours running **backwards**: the map reverses orientation, which no holomorphic map does (its Jacobian determinant is $\lvert f' \rvert^2 \ge 0$).

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import hsv_to_rgb


def domain_colouring(values):
    hue = (np.angle(values) / (2 * np.pi)) % 1.0
    brightness = 1 - 1 / (1 + np.abs(values) ** 0.6)
    return hsv_to_rgb(np.dstack([hue, np.full_like(hue, 0.9), 0.15 + 0.85 * brightness]))


grid = np.linspace(-2, 2, 601)
X, Y = np.meshgrid(grid, grid)
Z = X + 1j * Y

panels = [
    (Z**2 - 1, r"$z^2 - 1$ (holomorphic)"),
    (np.conj(Z) ** 2 - 1, r"$\bar{z}^2 - 1$ (not holomorphic)"),
    (Z * np.abs(Z) - 1, r"$z|z| - 1$ (not holomorphic)"),
]
fig, axes = plt.subplots(1, 3, figsize=(18, 6), layout="constrained")
for ax, (values, title) in zip(axes, panels):
    ax.imshow(domain_colouring(values), extent=[-2, 2, -2, 2], origin="lower")
    loop = 1 + 0.5 * np.exp(1j * np.linspace(0, 2 * np.pi, 100))
    ax.plot(loop.real, loop.imag, color="white", linewidth=1)
    ax.annotate("", xy=(1.0, 0.5), xytext=(1.05, 0.497), arrowprops=dict(arrowstyle="->", color="white", lw=2))
    ax.axhline(0, color="black", linewidth=0.5)
    ax.axvline(0, color="black", linewidth=0.5)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
plt.show()
```

Follow the white counter-clockwise loop around the zero at $z = 1$. On the left the hue goes red → yellow → green → … (counter-clockwise on the colour wheel); in the middle it goes red → magenta → blue → … (clockwise). The right panel, $z\lvert z \rvert - 1$, preserves orientation but is still not holomorphic — the colour bands are stretched unevenly in different directions, which notebook 04 identifies as a failure to preserve angles.

### 5.3 The ladder collapses: real $x\lvert x \rvert$ versus complex $z^2$

$x \lvert x \rvert$ agrees with $x^2$ for $x > 0$ and is differentiable on $\mathbb{R}$, but its derivative $2\lvert x \rvert$ has a corner and its second derivative jumps. The complex function $z^2$ that it imitates has derivatives $2z$, $2$, $0, \dots$ — all smooth. And the "obvious" complex version $z \lvert z \rvert$ is not holomorphic at all (previous plot).

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-2, 2, 401)
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))
axes[0].plot(x, x * np.abs(x), label=r"$x|x|$")
axes[0].plot(x, x**2, "--", label=r"$x^2$")
axes[1].plot(x, 2 * np.abs(x), label=r"$(x|x|)' = 2|x|$")
axes[1].plot(x, 2 * x, "--", label=r"$(x^2)' = 2x$")
axes[2].plot(x[x < 0], -2 * np.ones(np.sum(x < 0)), color="tab:blue", label=r"$(x|x|)'' = 2\,\mathrm{sign}(x)$")
axes[2].plot(x[x > 0], 2 * np.ones(np.sum(x > 0)), color="tab:blue")
axes[2].plot(x, 2 * np.ones_like(x), "--", color="tab:orange", label=r"$(x^2)'' = 2$")
for ax, title in zip(axes, ["function", "first derivative", "second derivative (jumps at 0)"]):
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_title(title)
    ax.set_xlabel("x")
    ax.grid(True, alpha=0.3)
    ax.legend()
plt.tight_layout()
plt.show()
```

## 7. Solutions

1. $z^2 - 3z + 2 = (z - 1)(z - 2)$, so $f$ is holomorphic on $\mathbb{C} \setminus \{1, 2\}$.
2. $\frac{\partial}{\partial \bar z}(z^2 \bar z) = z^2$, which vanishes only at $z = 0$. CR holds only at the origin; $f$ is holomorphic nowhere.
3. $\lvert z \rvert$ is real-valued and not constant on any disc, while a real-valued holomorphic function on a disc must be constant.
4. $x^3 - 3xy^2 = \operatorname{Re}(x + iy)^3$ and $3x^2 y - y^3 = \operatorname{Im}(x + iy)^3$, so $f = z^3$ — no $\bar z$, entire, $f' = 3z^2$.
5. With $g = u + iv$, $\bar g = u - iv$: $\frac{\partial \bar g}{\partial \bar z} = \frac12[(u_x + v_y) + i(u_y - v_x)]$. Using CR ($v_y = u_x$, $u_y = -v_x$) this is $u_x - i v_x = \overline{u_x + i v_x} = \overline{g'}$. So $\bar g$ satisfies CR at a point exactly when $g' = 0$ there; if $\bar g$ is holomorphic on an open connected set, $g' \equiv 0$ and $g$ is constant.
6. Since $\bar z^3 = r^3 e^{-3i\theta}$, the argument decreases by $3 \cdot 2\pi$ in one counter-clockwise loop: the colour wheel goes around **three times, clockwise**, and the brightness fades to black at $0$.
7. $u = e^x \cos y$; CR: $v_y = u_x = e^x \cos y \Rightarrow v = e^x \sin y + \phi(x)$; $v_x = -u_y = e^x \sin y \Rightarrow \phi'(x) = 0$. So $f = e^x \cos y + i(e^x \sin y + C) = e^z + iC$, $C \in \mathbb{R}$.
8. $4 \cdot \frac12(\partial_x - i\partial_y) \cdot \frac12(\partial_x + i\partial_y) = \partial_x^2 + i\partial_x\partial_y - i\partial_y\partial_x + \partial_y^2 = \partial_x^2 + \partial_y^2$, using equality of mixed partials.
9. $u u_x + v v_x = 0$ and $u u_y + v v_y = 0$. CR turns the second into $-u v_x + v u_x = 0$. So $\begin{pmatrix} u & v \\ v & -u \end{pmatrix} \begin{pmatrix} u_x \\ v_x \end{pmatrix} = 0$, with determinant $-(u^2 + v^2) = -c$. If $c = 0$ then $f \equiv 0$. Otherwise the only solution is $u_x = v_x = 0$, so $f' = u_x + i v_x \equiv 0$ and $f$ is constant.
10. On $z = iy$: $z^2 = -y^2$, so $F(iy) = e^{1/y^2} \to \infty$ as $y \to 0$, while along the real axis $F \to 0$. Different limits along two paths (notebook 01): $F$ has no limit at $0$.

## 8. Conclusion

- **Holomorphic** = complex-differentiable at every point of an open set; **entire** = holomorphic on $\mathbb{C}$. Differentiability on a point or a line does not count.
- Sums, products, quotients (off zeros of the denominator) and compositions of holomorphic functions are holomorphic, so polynomials, rational functions, $e^z$, $\sin z$, $\cos z$ and everything built from them are.
- The Wirtinger derivative $\frac{\partial}{\partial \bar z} = \frac12(\partial_x + i\partial_y)$ packages both CR equations: $f$ is holomorphic exactly where $\partial f / \partial \bar z = 0$ on an open set, and then $f' = \partial f / \partial z$. Anything with a $\bar z$ in it — $\bar z$, $\lvert z \rvert^2 = z\bar z$, $\operatorname{Re} z$ — fails.
- Holomorphic functions are rigid: constant real part, constant modulus, or zero derivative on a domain forces the function to be constant; they never reverse orientation.
