# The Complex Derivative and the Cauchy–Riemann Equations

The complex derivative is defined by the same formula as the real one, $f'(z) = \lim_{h \to 0} \frac{f(z+h) - f(z)}{h}$. The only difference is that the step $h$ is now a complex number, so it can shrink to $0$ from **every direction in the plane**, and the limit has to come out the same for all of them. That one requirement is surprisingly strong. Comparing just two directions — a horizontal step and a vertical step — forces the real and imaginary parts of $f = u + iv$ to satisfy a pair of partial differential equations, the **Cauchy–Riemann equations**.

This notebook derives the equations, checks them symbolically for $z^2$ and $e^z$, watches them fail for $\bar z$, and plots difference quotients from many directions to see what "differentiable" and "not differentiable" look like.

This is notebook 02 of module 05 (**Complex differentiation**) in the **Learning Complex Numbers** series.

## 1. Introduction

Think of $f: \mathbb{C} \to \mathbb{C}$ as a map of the plane, $(x, y) \mapsto (u(x,y), v(x,y))$. Multivariable calculus already gives it a derivative: the $2 \times 2$ **Jacobian matrix** $\begin{pmatrix} u_x & u_y \\ v_x & v_y \end{pmatrix}$, which can stretch, shear and reflect a small square in any linear way. A complex derivative $f'(z)$ is a single complex number, and multiplying by a complex number can only **rotate and scale** (notebook [01/02](../01_polar_form_and_euler/02_multiplication_is_rotation_and_scaling.ipynb)). So complex differentiability asks the Jacobian to be a rotation-and-scaling matrix — and the Cauchy–Riemann equations are precisely that condition.

The payoff for this strictness comes later: a function with a complex derivative on an open set automatically has derivatives of every order, a convergent power series, and integrals that obey Cauchy's theorem (module 06). None of that is true of real-differentiable functions.

## 2. Theory

### 2.1 Definition

Let $f$ be defined on a disc around $z_0$. Then $f$ is **(complex) differentiable at $z_0$** if

$$
f'(z_0) = \lim_{h \to 0} \frac{f(z_0 + h) - f(z_0)}{h}
$$

exists, where $h \to 0$ through **complex** values — in the sense of notebook 01, from every direction and along every path.

**Example: $f(z) = z^2$.** $\dfrac{(z+h)^2 - z^2}{h} = \dfrac{2zh + h^2}{h} = 2z + h \to 2z$, whatever the direction of $h$. So $(z^2)' = 2z$.

**Non-example: $f(z) = \bar z$.** $\dfrac{\overline{z + h} - \bar z}{h} = \dfrac{\bar h}{h}$, which is $1$ for real $h$ and $-1$ for imaginary $h$ — the function of notebook 01 that has no limit at $0$. So $\bar z$ is differentiable **nowhere**, even though it is about the simplest continuous function there is.

### 2.2 The usual rules hold

Because the definition is algebraically identical to the real one, the proofs carry over word for word:

$$
(f \pm g)' = f' \pm g', \quad (fg)' = f'g + fg', \quad \left( \frac{f}{g} \right)' = \frac{f'g - fg'}{g^2}, \quad (f \circ g)'(z) = f'(g(z))\, g'(z),
$$

and $(z^n)' = n z^{n-1}$ for every integer $n$ ($z \neq 0$ if $n < 0$). Differentiable at $z_0$ implies continuous at $z_0$: $f(z_0 + h) - f(z_0) = h \cdot \frac{f(z_0 + h) - f(z_0)}{h} \to 0 \cdot f'(z_0) = 0$.

### 2.3 Two directions give the Cauchy–Riemann equations

Write $z = x + iy$ and $f = u(x, y) + i v(x, y)$, and suppose $f'(z_0)$ exists. Then the limit can be computed along any path we like.

**Horizontal step** $h = t$ (real, $t \to 0$):

$$
f'(z_0) = \lim_{t \to 0} \frac{u(x_0 + t, y_0) - u(x_0, y_0)}{t} + i\, \frac{v(x_0 + t, y_0) - v(x_0, y_0)}{t} = u_x + i v_x .
$$

**Vertical step** $h = it$:

$$
f'(z_0) = \lim_{t \to 0} \frac{[u(x_0, y_0 + t) - u] + i[v(x_0, y_0 + t) - v]}{it} = \frac{u_y + i v_y}{i} = v_y - i u_y .
$$

Both must equal $f'(z_0)$. Comparing real and imaginary parts:

$$
\boxed{\; u_x = v_y, \qquad u_y = -v_x \;}
\qquad\text{and}\qquad
f'(z_0) = u_x + i v_x = v_y - i u_y .
$$

These are the **Cauchy–Riemann (CR) equations**. They are *necessary* for differentiability.

### 2.4 When are they sufficient?

**Theorem.** If $u$ and $v$ have **continuous** first partial derivatives on a disc around $z_0$ and satisfy the CR equations at $z_0$, then $f$ is complex-differentiable at $z_0$, with $f'(z_0) = u_x + i v_x$.

*Sketch.* Continuous partials make $(u, v)$ real-differentiable: for a step $h = \Delta x + i\Delta y$,

$$
\Delta u = u_x \Delta x + u_y \Delta y + o(\lvert h \rvert), \qquad \Delta v = v_x \Delta x + v_y \Delta y + o(\lvert h \rvert).
$$

Put $a = u_x = v_y$ and $b = v_x = -u_y$. Then $\Delta u + i \Delta v = (a \Delta x - b \Delta y) + i(b \Delta x + a \Delta y) + o(\lvert h \rvert) = (a + ib)(\Delta x + i \Delta y) + o(\lvert h \rvert)$, so $\frac{\Delta f}{h} \to a + ib$. $\square$

(Without some extra hypothesis the CR equations alone are not enough — exercise 10 has a function that satisfies them at $0$ and is not differentiable there.)

### 2.5 The matrix picture

The Jacobian of $(x,y) \mapsto (u,v)$ under the CR equations is

$$
J = \begin{pmatrix} u_x & u_y \\ v_x & v_y \end{pmatrix} = \begin{pmatrix} a & -b \\ b & a \end{pmatrix},
$$

which is exactly the matrix of "multiply by $a + ib$" from notebook [04/01](../04_complex_linear_algebra/01_complex_numbers_as_matrices.ipynb): a rotation by $\arg f'$ composed with a scaling by $\lvert f' \rvert$. A general real-differentiable map has a general $J$; a complex-differentiable map has a $J$ with only two free numbers instead of four. Note $\det J = a^2 + b^2 = \lvert f' \rvert^2 \ge 0$, so complex-differentiable maps never reverse orientation.

### 2.6 Difference quotients from every direction

For a real-differentiable $f$ and a step $h = t e^{i\theta}$, the linear approximation gives

$$
\lim_{t \to 0^+} \frac{f(z_0 + t e^{i\theta}) - f(z_0)}{t e^{i\theta}} = A + B\, e^{-2i\theta},
\qquad A = \tfrac{1}{2}(f_x - i f_y),\quad B = \tfrac{1}{2}(f_x + i f_y),
$$

where $f_x = u_x + i v_x$ and $f_y = u_y + i v_y$. (Check: $\theta = 0$ gives $A + B = f_x$, and $\theta = \pi/2$ gives $A - B = f_y / i$.) As the direction $\theta$ turns, the directional limit **traces a circle** of radius $\lvert B \rvert$ around $A$ — twice. The function is complex-differentiable exactly when the circle collapses to a point, $B = 0$, and expanding $B = 0$ gives back the CR equations. The numbers $A$ and $B$ are the Wirtinger derivatives $\partial f / \partial z$ and $\partial f / \partial \bar z$ of notebook 03.

## 3. Worked Examples

### Example 1 — $f(z) = z^2$

$z^2 = (x + iy)^2 = (x^2 - y^2) + i(2xy)$, so $u = x^2 - y^2$, $v = 2xy$.

| | $\partial_x$ | $\partial_y$ |
|---|---|---|
| $u$ | $2x$ | $-2y$ |
| $v$ | $2y$ | $2x$ |

$u_x = 2x = v_y$ ✓ and $u_y = -2y = -v_x$ ✓ everywhere. The partials are continuous, so $f$ is differentiable everywhere with $f' = u_x + i v_x = 2x + 2iy = 2z$, matching §2.1.

### Example 2 — $f(z) = e^z$

$e^{x + iy} = e^x \cos y + i\, e^x \sin y$, so $u = e^x \cos y$, $v = e^x \sin y$.

- $u_x = e^x \cos y$, $v_y = e^x \cos y$ ✓
- $u_y = -e^x \sin y$, $v_x = e^x \sin y$, so $u_y = -v_x$ ✓

Hence $(e^z)' = u_x + i v_x = e^x \cos y + i e^x \sin y = e^z$. The exponential is its own derivative in $\mathbb{C}$ too.

### Example 3 — $f(z) = \bar z$ fails everywhere

$u = x$, $v = -y$: $u_x = 1$ but $v_y = -1$. The first CR equation fails at **every** point, so $\bar z$ is nowhere differentiable — in agreement with the $\bar h / h$ computation.

### Example 4 — $f(z) = \lvert z \rvert^2$ is differentiable at exactly one point

$u = x^2 + y^2$, $v = 0$. CR: $u_x = 2x = 0 = v_y$ and $u_y = 2y = -0 = -v_x$, so we need $x = y = 0$. The partials are continuous, so $f$ is differentiable at $z = 0$ (with $f'(0) = 0$) and nowhere else. Directly: $\frac{\lvert h \rvert^2}{h} = \bar h \to 0$.

### Example 5 — $f(z) = x^2 + i y^2$ is differentiable on a line

$u = x^2$, $v = y^2$: $u_x = 2x$, $v_y = 2y$, $u_y = 0$, $v_x = 0$. The second equation always holds; the first needs $x = y$. So $f$ is differentiable exactly on the line $y = x$, with $f'(x + ix) = u_x + i v_x = 2x$. (It is still not "holomorphic" anywhere, because no open disc fits inside a line — notebook 03.)

### Example 6 — $f(z) = 1/z$ via CR

$\frac{1}{z} = \frac{\bar z}{\lvert z \rvert^2} = \frac{x}{x^2 + y^2} - i\,\frac{y}{x^2 + y^2}$. Then

$$
u_x = \frac{y^2 - x^2}{(x^2 + y^2)^2} = v_y, \qquad u_y = \frac{-2xy}{(x^2 + y^2)^2} = -v_x,
$$

so $1/z$ is differentiable for $z \neq 0$, and $f' = u_x + i v_x = \frac{(y^2 - x^2) + 2ixy}{(x^2+y^2)^2} = \frac{-(x - iy)^2}{\lvert z \rvert^4} = -\frac{\bar z^2}{\lvert z \rvert^4} = -\frac{1}{z^2}$ ✓.

## 4. Python Examples

### 4.1 A Cauchy–Riemann checker in sympy

Given $f$ as an expression in $z$ (or directly in $x, y$), split into $u$ and $v$, and return the two CR residuals $u_x - v_y$ and $u_y + v_x$ together with $u_x + i v_x$.

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
z = sp.symbols("z")


def cauchy_riemann(f_of_z):
    expression = sp.expand_complex(f_of_z.subs(z, x + sp.I * y))
    u = sp.simplify(sp.re(expression))
    v = sp.simplify(sp.im(expression))
    residual_1 = sp.simplify(sp.diff(u, x) - sp.diff(v, y))
    residual_2 = sp.simplify(sp.diff(u, y) + sp.diff(v, x))
    derivative = sp.simplify(sp.diff(u, x) + sp.I * sp.diff(v, x))
    return u, v, residual_1, residual_2, derivative


examples = {
    "z^2": z**2,
    "z^3": z**3,
    "exp(z)": sp.exp(z),
    "1/z": 1 / z,
    "conj(z)": sp.conjugate(z),
    "|z|^2": z * sp.conjugate(z),
}
for name, f in examples.items():
    u, v, r1, r2, derivative = cauchy_riemann(f)
    print(f"{name:8}  u = {str(u):28} v = {str(v):24} u_x - v_y = {str(r1):10} u_y + v_x = {str(r2):10}")
```

Where both residuals are $0$, compare $u_x + i v_x$ with the derivative sympy computes in $z$:

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
z = sp.symbols("z")

for f in [z**2, z**3, sp.exp(z), 1 / z, sp.sin(z)]:
    expression = sp.expand_complex(f.subs(z, x + sp.I * y))
    u, v = sp.re(expression), sp.im(expression)
    from_cr = sp.diff(u, x) + sp.I * sp.diff(v, x)
    expected = sp.diff(f, z).subs(z, x + sp.I * y)
    agrees = sp.simplify(sp.expand_complex(from_cr - expected)) == 0
    print(f"f = {str(f):8}  f' = {str(sp.diff(f, z)):10}  u_x + i v_x agrees: {agrees}")
```

The rules of §2.2 can be checked the same way. $e^{z^2}$ is a composition, so the chain rule predicts $(e^{z^2})' = 2z\,e^{z^2}$; $z^2 e^z$ is a product, so $(z^2 e^z)' = (2z + z^2) e^z$. Both are verified below through $u$ and $v$ alone, without ever differentiating with respect to $z$.

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
z = sp.symbols("z")

for f, predicted in [(sp.exp(z**2), 2 * z * sp.exp(z**2)), (z**2 * sp.exp(z), (2 * z + z**2) * sp.exp(z))]:
    expression = sp.expand_complex(f.subs(z, x + sp.I * y))
    u, v = sp.re(expression), sp.im(expression)
    residuals = [sp.simplify(sp.diff(u, x) - sp.diff(v, y)), sp.simplify(sp.diff(u, y) + sp.diff(v, x))]
    from_cr = sp.diff(u, x) + sp.I * sp.diff(v, x)
    matches = sp.simplify(sp.expand_complex(from_cr - predicted.subs(z, x + sp.I * y))) == 0
    print(f"f = {str(f):14} CR residuals = {residuals}   u_x + i v_x equals {predicted}: {matches}")
```

### 4.2 Where do the CR equations hold?

Solve the two residual equations for $x, y$ to find the points of differentiability.

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)

candidates = {
    "|z|^2        (u = x^2 + y^2, v = 0)": (x**2 + y**2, sp.Integer(0)),
    "x^2 + i y^2  ": (x**2, y**2),
    "x^3 + i y^3   ": (x**3, y**3),
}
for name, (u, v) in candidates.items():
    equations = [sp.Eq(sp.diff(u, x), sp.diff(v, y)), sp.Eq(sp.diff(u, y), -sp.diff(v, x))]
    reduced = [sp.simplify(eq) for eq in equations]
    print(f"{name}:  {reduced[0]},  {reduced[1]}   ->  solutions: {sp.solve(equations, [x, y], dict=True)}")
```

For $x^2 + iy^2$ sympy returns $x = y$ (the whole line); for $x^3 + iy^3$ the first equation is $3x^2 = 3y^2$, so the function is differentiable on the two lines $y = \pm x$.

### 4.3 Directional difference quotients, numerically

Evaluate $Q(\theta, t) = \dfrac{f(z_0 + t e^{i\theta}) - f(z_0)}{t e^{i\theta}}$ for a few directions and shrinking $t$. For $z^2$ and $e^z$ all directions converge to the same number; for $\bar z$ they don't.

```python
import numpy as np


def difference_quotient(f, z0, theta, t):
    h = t * np.exp(1j * theta)
    return (f(z0 + h) - f(z0)) / h


z0 = 1 + 1j
functions = {"z^2": lambda z: z**2, "exp(z)": np.exp, "conj(z)": np.conj}
exact = {"z^2": 2 * z0, "exp(z)": np.exp(z0), "conj(z)": None}
directions = [0, np.pi / 4, np.pi / 2, np.pi]

for name, f in functions.items():
    print(f"f = {name}, z0 = {z0}, exact f'(z0) = {exact[name]}")
    for t in [1e-2, 1e-6]:
        row = [difference_quotient(f, z0, theta, t) for theta in directions]
        print(f"   t = {t:.0e}:", "  ".join(f"{q.real:+.6f}{q.imag:+.6f}i" for q in row))
```

### 4.4 The circle formula $A + B e^{-2i\theta}$

Compute $A = \tfrac12 (f_x - i f_y)$ and $B = \tfrac12 (f_x + i f_y)$ for $f = \lvert z \rvert^2$ at $z_0 = 1 + i$ and compare with numerical quotients.

```python
import numpy as np
import sympy as sp

x, y = sp.symbols("x y", real=True)
f_xy = x**2 + y**2                       # |z|^2 written in x, y
z0 = 1 + 1j
point = {x: z0.real, y: z0.imag}

f_x = complex(sp.diff(f_xy, x).subs(point))
f_y = complex(sp.diff(f_xy, y).subs(point))
A = (f_x - 1j * f_y) / 2
B = (f_x + 1j * f_y) / 2
print("A =", A, "  B =", B, "  (A = conj(z0), B = z0)")

t = 1e-7
for theta in np.linspace(0, np.pi, 5):
    h = t * np.exp(1j * theta)
    numeric = (abs(z0 + h) ** 2 - abs(z0) ** 2) / h
    formula = A + B * np.exp(-2j * theta)
    print(f"theta = {theta:.3f}:  numeric = {numeric:.6f}   A + B e^(-2i theta) = {formula:.6f}   close: {np.isclose(numeric, formula, atol=1e-5)}")
```

## 5. Visualizations

### 5.1 Directional difference quotients: point or circle?

For each of 72 directions $\theta$, plot the difference quotient $Q(\theta, t)$ at a small step $t$ in the complex plane. A differentiable function gives a single dot (all directions agree); a non-differentiable one gives a circle of radius $\lvert B \rvert$, traced twice.

```python
import matplotlib.pyplot as plt
import numpy as np

z0 = 1 + 1j
t = 1e-4
theta = np.linspace(0, 2 * np.pi, 72, endpoint=False)
h = t * np.exp(1j * theta)

cases = [
    (r"$f = z^2$", lambda z: z**2),
    (r"$f = e^z$", np.exp),
    (r"$f = \bar{z}$", np.conj),
    (r"$f = |z|^2$", lambda z: np.abs(z) ** 2),
    (r"$f = x^2 + i y^2$", lambda z: z.real**2 + 1j * z.imag**2),
]

fig, axes = plt.subplots(1, len(cases), figsize=(22, 5), layout="constrained")
for ax, (title, f) in zip(axes, cases):
    quotients = (f(z0 + h) - f(z0)) / h
    scatter = ax.scatter(quotients.real, quotients.imag, c=theta, cmap="hsv", s=25, zorder=3)
    centre = quotients.mean()
    half_width = max(1.5, 1.2 * np.max(np.abs(quotients - centre)))
    ax.set_xlim(centre.real - half_width, centre.real + half_width)
    ax.set_ylim(centre.imag - half_width, centre.imag + half_width)
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_aspect("equal")
    ax.set_title(title + r" at $z_0 = 1 + i$")
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
fig.colorbar(scatter, ax=axes, label=r"direction $\theta$ of the step $h$", shrink=0.8)
plt.show()
```

$z^2$ and $e^z$ collapse to the points $2z_0 = 2 + 2i$ and $e^{1+i}$. $\bar z$ gives the unit circle ($A = 0$, $B = 1$), $\lvert z \rvert^2$ gives the circle of radius $\lvert z_0 \rvert = \sqrt2$ around $\bar z_0 = 1 - i$, and $x^2 + iy^2$ — which is differentiable only on the line $y = x$, where $z_0 = 1 + i$ happens to lie — collapses to the point $2$. The colours run around each circle **twice** as $\theta$ goes once around: the $e^{-2i\theta}$.

### 5.2 How fast do the quotients converge?

For a differentiable function the error $\lvert Q(\theta, t) - f'(z_0) \rvert$ shrinks like $t$ in every direction (slope 1 on a log–log plot). For $\bar z$ there is no candidate $f'(z_0)$; the spread between directions stays constant.

```python
import matplotlib.pyplot as plt
import numpy as np

z0 = 1 + 1j
steps = np.geomspace(1e-1, 1e-7, 40)
directions = {r"$\theta = 0$": 0.0, r"$\theta = \pi/3$": np.pi / 3, r"$\theta = \pi/2$": np.pi / 2, r"$\theta = 5\pi/4$": 5 * np.pi / 4}

fig, (left, right) = plt.subplots(1, 2, figsize=(13, 5))
for label, theta in directions.items():
    h = steps * np.exp(1j * theta)
    left.loglog(steps, np.abs((np.exp(z0 + h) - np.exp(z0)) / h - np.exp(z0)), label=label)
left.loglog(steps, steps, "k--", linewidth=1, label="slope 1 reference")
left.set_title(r"$e^z$: error $|Q - f'(z_0)|$ shrinks like $t$ in every direction")
left.set_xlabel("step size t = |h|")
left.set_ylabel("error")
left.invert_xaxis()
left.grid(True, which="both", alpha=0.3)
left.legend()

spread = []
for t in steps:
    quotients = [np.conj(t * np.exp(1j * th)) / (t * np.exp(1j * th)) for th in np.linspace(0, np.pi, 50)]
    spread.append(np.max(np.abs(np.subtract.outer(quotients, quotients))))
right.semilogx(steps, spread, color="crimson")
right.set_ylim(0, 2.5)
right.set_title(r"$\bar{z}$: spread of $Q$ over directions never shrinks")
right.set_xlabel("step size t = |h|")
right.set_ylabel(r"max $|Q(\theta_1) - Q(\theta_2)|$")
right.invert_xaxis()
right.grid(True, which="both", alpha=0.3)
plt.tight_layout()
plt.show()
```

### 5.3 The Jacobian picture: what happens to a tiny square

Take a small square centred at $z_0 = 1 + 0.5i$ and map it. A complex-differentiable map sends it (to first order) to another **square**, rotated by $\arg f'(z_0)$ and scaled by $\lvert f'(z_0) \rvert$. $\bar z$ flips it (orientation reversed — note the order of the coloured corners), and $x^2 + iy^2$ at a point off the line $y = x$ stretches it into a rectangle.

```python
import matplotlib.pyplot as plt
import numpy as np

z0 = 1 + 0.5j
side = 0.12
corners = z0 + side * np.array([-0.5 - 0.5j, 0.5 - 0.5j, 0.5 + 0.5j, -0.5 + 0.5j])
edge_parameter = np.linspace(0, 1, 50)
outline = np.concatenate([corners[k] + edge_parameter * (corners[(k + 1) % 4] - corners[k]) for k in range(4)])

cases = [
    (r"original square", lambda z: z),
    (r"$z^2$", lambda z: z**2),
    (r"$e^z$", np.exp),
    (r"$\bar{z}$", np.conj),
    (r"$x^2 + i y^2$", lambda z: z.real**2 + 1j * z.imag**2),
]
corner_colours = ["tab:red", "tab:green", "tab:blue", "tab:orange"]

fig, axes = plt.subplots(1, len(cases), figsize=(20, 4.5))
for ax, (title, f) in zip(axes, cases):
    image = f(outline)
    ax.fill(image.real, image.imag, color="tab:gray", alpha=0.3)
    ax.plot(image.real, image.imag, color="black", linewidth=1)
    for corner, colour in zip(corners, corner_colours):
        mapped = f(np.array([corner]))[0]
        ax.plot(mapped.real, mapped.imag, "o", color=colour, markersize=8)
    centre = f(np.array([z0]))[0]
    ax.set_xlim(centre.real - 0.25, centre.real + 0.25)
    ax.set_ylim(centre.imag - 0.25, centre.imag + 0.25)
    ax.set_aspect("equal")
    ax.set_title(title)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
    ax.grid(True, alpha=0.3)
plt.suptitle("Image of a small square around $z_0 = 1 + 0.5i$ (corners: red, green, blue, orange counter-clockwise)")
plt.tight_layout()
plt.show()
```

## 7. Solutions

1. $\dfrac{(z+h)^3 - z^3}{h} = \dfrac{3z^2 h + 3zh^2 + h^3}{h} = 3z^2 + 3zh + h^2 \to 3z^2$ as $h \to 0$ from any direction.
2. $u = x^2 - y^2 + 2x$, $v = 2xy + 2y$. $u_x = 2x + 2 = v_y$ ✓, $u_y = -2y = -v_x$ ✓.
3. $u = x$, $v = 0$: $u_x = 1 \neq 0 = v_y$ at every point.
4. $u = x^2 - y^2$, $v = x^2 + y^2$. CR: $u_x = 2x = 2y = v_y \Rightarrow x = y$; $u_y = -2y = -v_x = -2x \Rightarrow x = y$. Differentiable exactly on the line $y = x$, where $f'(x + ix) = u_x + i v_x = 2x + 2ix = 2x(1 + i)$.
5. Writing $Q = A + B e^{-2i\theta}$ with $f = y$: $f_x = 0$, $f_y = 1$, so $A = -\frac{i}{2}$, $B = \frac{i}{2}$. Prediction: a circle of radius $\tfrac12$ centred at $-\tfrac{i}{2}$, passing through $0$ (horizontal steps, $\theta = 0$) and $-i$ (vertical steps, $\theta = \pi/2$). Closed form: $Q(\theta) = \dfrac{\sin\theta}{e^{i\theta}} = \sin\theta\,e^{-i\theta} = \tfrac{i}{2}(e^{-2i\theta} - 1)$ ✓.
6. $u = \sin x \cosh y$, $v = \cos x \sinh y$. $u_x = \cos x \cosh y = v_y$ ✓; $u_y = \sin x \sinh y$ and $v_x = -\sin x \sinh y$, so $u_y = -v_x$ ✓. Then $f' = u_x + i v_x = \cos x \cosh y - i \sin x \sinh y = \cos(x + iy)$.
7. Real-valued means $v \equiv 0$, so $v_x = v_y = 0$. Then $f' = u_x + i v_x = u_x$ and also $f' = v_y - i u_y = -i u_y$. The first is real, the second purely imaginary, so both are $0$: $u_x = u_y = 0$ and $f'(z_0) = 0$.
8. Chain rule with $x = r\cos\theta$, $y = r\sin\theta$: $u_r = u_x \cos\theta + u_y \sin\theta$, $u_\theta = r(-u_x \sin\theta + u_y \cos\theta)$, and the same for $v$. Substituting $v_y = u_x$, $v_x = -u_y$: $\frac{1}{r} v_\theta = -v_x \sin\theta + v_y \cos\theta = u_y \sin\theta + u_x \cos\theta = u_r$ ✓, and similarly $v_r = -\frac1r u_\theta$. For $z^n$: $u_r = n r^{n-1} \cos n\theta$ and $\frac1r v_\theta = \frac1r r^n n \cos n\theta$ ✓; $v_r = n r^{n-1} \sin n\theta$ and $-\frac1r u_\theta = -\frac1r r^n(-n \sin n\theta)$ ✓.
9. $\det J = u_x v_y - u_y v_x = u_x^2 + v_x^2 = \lvert u_x + i v_x \rvert^2 = \lvert f' \rvert^2$, using $v_y = u_x$ and $u_y = -v_x$. It is never negative, and positive wherever $f' \neq 0$, so the local linear map never reflects: orientation is preserved.
10. On the real axis $f(x) = x^2/x = x$, so $u(x, 0) = x$, $v(x, 0) = 0$: $u_x(0) = 1$, $v_x(0) = 0$. On the imaginary axis $f(iy) = (-iy)^2/(iy) = -y^2/(iy) = iy$, so $u(0, y) = 0$, $v(0, y) = y$: $u_y(0) = 0$, $v_y(0) = 1$. CR holds at $0$: $u_x = 1 = v_y$, $u_y = 0 = -v_x$. But $\frac{f(h) - f(0)}{h} = \frac{\bar h^2}{h^2} = e^{-4i\theta}$ depends on the direction (it is $-1$ on the diagonal $\theta = \pi/4$), so $f'(0)$ does not exist. The theorem's hypothesis that fails is **continuity of the partial derivatives** near $0$ (indeed $u, v$ are not even real-differentiable at $0$).

## 8. Conclusion

- $f'(z_0) = \lim_{h \to 0} \frac{f(z_0 + h) - f(z_0)}{h}$ with complex $h$; the usual differentiation rules all hold.
- Comparing horizontal and vertical steps gives the **Cauchy–Riemann equations** $u_x = v_y$, $u_y = -v_x$, and $f' = u_x + i v_x$. With continuous partials they are also sufficient.
- Geometrically: the Jacobian must be a rotation-scaling matrix $\begin{pmatrix} a & -b \\ b & a \end{pmatrix}$, i.e. multiplication by $f'(z_0)$.
- Directional difference quotients of a real-differentiable map trace a circle $A + Be^{-2i\theta}$; differentiable means the circle shrinks to a point.
- $z^n$, $e^z$, $\sin z$, $1/z$ pass; $\bar z$, $\operatorname{Re} z$ fail everywhere; $\lvert z \rvert^2$ and $x^2 + iy^2$ are differentiable only on a point or a line.
