# Harmonic Functions and the Link to 2-D Physics

A real function $u(x, y)$ is **harmonic** if it satisfies Laplace's equation $u_{xx} + u_{yy} = 0$. This single equation governs steady-state temperature, electrostatic potential in a charge-free region, the velocity potential of an ideal fluid, and the shape of a soap film stretched over a nearly flat wire. Complex analysis gives a factory for its solutions: **the real and imaginary parts of every holomorphic function are harmonic**, and their level curves cut each other at right angles — equipotentials and field lines, or potential lines and streamlines.

This notebook proves that fact from the Cauchy–Riemann equations, constructs **harmonic conjugates** with sympy, draws the orthogonal level-curve families, models flows and fields with a complex potential, and checks the **mean-value property** numerically: a harmonic function's value at a point is the average of its values on any circle around it.

This is notebook 05 of module 05 (**Complex differentiation**) in the **Learning Complex Numbers** series.

## 1. Introduction

Laplace's equation says, informally, that $u$ has **no local bumps**: at each point the surface curves up in one direction exactly as much as it curves down in the perpendicular one. Its solutions are the "most relaxed" functions with given boundary values — the temperature of a metal plate once it has stopped changing, or the electric potential between conductors held at fixed voltages.

In two dimensions there is a shortcut to these solutions. If $f = u + iv$ is holomorphic, the Cauchy–Riemann equations tie $u$ and $v$ together so tightly that both are forced to satisfy Laplace's equation. Conversely, every harmonic $u$ on a disc is the real part of some holomorphic $f$. So the whole toolbox of the previous notebooks — $z^n$, $e^z$, $\operatorname{Log} z$, $z + 1/z$, conformal maps — becomes a toolbox for 2-D physics.

## 2. Theory

### 2.1 Harmonic functions

The **Laplacian** of a twice-differentiable $u(x, y)$ is

$$
\Delta u = \frac{\partial^2 u}{\partial x^2} + \frac{\partial^2 u}{\partial y^2},
$$

the trace of the Hessian. $u$ is **harmonic** on an open set $U$ if it has continuous second partials and $\Delta u = 0$ on $U$. Examples: $x$, $y$, $xy$, $x^2 - y^2$, $e^x \cos y$, $\ln(x^2 + y^2)$ (away from $0$). Non-examples: $x^2 + y^2$ ($\Delta = 4$), $x^2$ ($\Delta = 2$).

### 2.2 Real and imaginary parts of holomorphic functions are harmonic

Let $f = u + iv$ be holomorphic. By the theorem previewed in notebook 03 (proved in module 06), $f'$ is again holomorphic, so $u$ and $v$ have continuous partial derivatives of every order. Differentiate the CR equations $u_x = v_y$ and $u_y = -v_x$:

$$
u_{xx} = v_{yx}, \qquad u_{yy} = -v_{xy}.
$$

Adding and using equality of mixed partials ($v_{xy} = v_{yx}$):

$$
\boxed{\;\Delta u = u_{xx} + u_{yy} = 0\;}
\qquad\text{and similarly}\qquad
\Delta v = v_{xx} + v_{yy} = -u_{yx} + u_{xy} = 0 .
$$

In Wirtinger form (notebook 03, exercise 8), $\Delta = 4\, \partial_z \partial_{\bar z}$, and $\partial_{\bar z} f = 0$ immediately gives $\Delta f = 0$.

### 2.3 Harmonic conjugates

If $u$ is harmonic and $u + iv$ is holomorphic, $v$ is called a **harmonic conjugate** of $u$. To find it, solve the CR equations for $v$:

$$
v_y = u_x, \qquad v_x = -u_y .
$$

**Recipe.** Integrate the first equation in $y$: $v = \int u_x \, dy + \phi(x)$. Substitute into the second to get $\phi'(x)$, and integrate. The answer is unique up to an additive constant. The compatibility condition that makes this work — $(u_x)_x = (-u_y)_y$ — is exactly $\Delta u = 0$.

**Theorem.** On a disc (more generally, on a simply connected domain) every harmonic $u$ has a harmonic conjugate. On a domain with a hole it may not: $u = \ln\lvert z \rvert$ is harmonic on $\mathbb{C} \setminus \{0\}$, its local conjugates are the branches of $\arg z$, and no single continuous choice exists around the whole puncture. (Going once around the origin increases $\arg z$ by $2\pi$ — the theme of contour integration in module 06.)

### 2.4 The level curves of $u$ and $v$ are orthogonal

The gradient $\nabla u = (u_x, u_y)$ is perpendicular to the level curves of $u$. With CR,

$$
\nabla u \cdot \nabla v = u_x v_x + u_y v_y = u_x v_x + (-v_x)(u_x) = 0,
\qquad
\lvert \nabla u \rvert = \lvert \nabla v \rvert = \lvert f' \rvert .
$$

So wherever $f' \neq 0$ the level curves $u = c_1$ and $v = c_2$ cross at right angles. This is notebook 04 again: they are the images, under the conformal map $f^{-1}$, of the orthogonal grid lines $\operatorname{Re} w = c_1$, $\operatorname{Im} w = c_2$.

### 2.5 Physics: the complex potential

**Ideal 2-D fluid flow** (incompressible, irrotational) has a velocity field $\mathbf{V} = \nabla \varphi$ for a **velocity potential** $\varphi$ with $\Delta \varphi = 0$, and a **stream function** $\psi$ (the harmonic conjugate) whose level curves are the **streamlines**. Their combination is the **complex potential**

$$
\Omega(z) = \varphi + i\psi, \qquad \text{velocity } V_x + i V_y = \overline{\Omega'(z)} .
$$

(Check: $\Omega' = \varphi_x + i\psi_x = \varphi_x - i\varphi_y$, so $\overline{\Omega'} = \varphi_x + i\varphi_y$.)

| $\Omega(z)$ | flow |
|---|---|
| $Uz$ | uniform flow with speed $U$ to the right |
| $\frac{m}{2\pi} \operatorname{Log} z$ | source of strength $m$ at $0$ (streamlines: rays) |
| $-\frac{i\Gamma}{2\pi} \operatorname{Log} z$ | vortex with circulation $\Gamma$ (streamlines: circles) |
| $U\left(z + \frac{a^2}{z}\right)$ | uniform flow around a cylinder of radius $a$ |

For the cylinder, $\psi = U\left(y - \frac{a^2 y}{x^2 + y^2}\right)$ vanishes on $\lvert z \rvert = a$, so the circle is a streamline: fluid flows around it, not through it. With $a = 1$ this is $U$ times the Joukowski map — composing with the inverse Joukowski map turns it into flow past an airfoil.

**Electrostatics** uses the same mathematics with $\varphi$ as the electric potential (equipotentials $\varphi = c$) and $\psi$ labelling the **field lines**. A line charge at $z_0$ has potential proportional to $-\ln\lvert z - z_0 \rvert$.

### 2.6 The mean-value property and the maximum principle

**Mean-value property.** If $u$ is harmonic on a region containing the closed disc $\lvert z - z_0 \rvert \le r$, then

$$
u(z_0) = \frac{1}{2\pi} \int_0^{2\pi} u\!\left(z_0 + r e^{i\theta}\right) d\theta .
$$

The value at the centre is the average over the circle. (Proof sketch: $u = \operatorname{Re} f$, and Cauchy's integral formula of module 06 gives $f(z_0) = \frac{1}{2\pi}\int_0^{2\pi} f(z_0 + re^{i\theta})\,d\theta$; take real parts.)

**Maximum principle.** A non-constant harmonic function on a bounded domain attains its maximum and minimum only on the boundary — an interior maximum would have to equal the average of its neighbours, all of which are smaller. Physically: in a steady-state temperature distribution, the hottest point is on the edge.

## 3. Worked Examples

### Example 1 — $u = x^2 - y^2$ and its conjugate

$u_{xx} = 2$, $u_{yy} = -2$, so $\Delta u = 0$. Conjugate: $v_y = u_x = 2x \Rightarrow v = 2xy + \phi(x)$; $v_x = 2y + \phi'(x)$ must equal $-u_y = 2y$, so $\phi' = 0$. Hence $v = 2xy + C$ and $u + iv = z^2 + iC$. Level curves: $x^2 - y^2 = c_1$ and $2xy = c_2$ — two families of hyperbolas, crossing at right angles except at $0$, the critical point of $z^2$.

### Example 2 — $u = e^x \cos y$

$u_{xx} = e^x\cos y$, $u_{yy} = -e^x \cos y$, sum $0$. $v_y = e^x \cos y \Rightarrow v = e^x \sin y + \phi(x)$, and $v_x = e^x \sin y = -u_y$ ✓ gives $\phi' = 0$. So $u + iv = e^z$.

### Example 3 — $u = x^3 - 3xy^2 + 2y$

$\Delta u = 6x - 6x + 0 = 0$. $v_y = u_x = 3x^2 - 3y^2 \Rightarrow v = 3x^2 y - y^3 + \phi(x)$. Then $v_x = 6xy + \phi'(x)$ must equal $-u_y = 6xy - 2$, so $\phi(x) = -2x$. Thus $v = 3x^2y - y^3 - 2x$ and $u + iv = z^3 - 2iz$ (check: $-2i(x + iy) = 2y - 2ix$ ✓).

### Example 4 — $u = x^2 + y^2$ is not harmonic, and the recipe breaks

$\Delta u = 4 \neq 0$. Trying anyway: $v = \int 2x\, dy = 2xy + \phi(x)$, then $v_x = 2y + \phi'(x)$ must equal $-u_y = -2y$, giving $\phi'(x) = -4y$ — impossible, since $\phi$ depends only on $x$. The obstruction is exactly $\Delta u$.

### Example 5 — $\ln\lvert z \rvert$ and the missing global conjugate

$u = \tfrac12 \ln(x^2 + y^2)$: $u_x = \frac{x}{x^2+y^2}$, $u_{xx} = \frac{y^2 - x^2}{(x^2+y^2)^2}$, and by symmetry $u_{yy} = \frac{x^2 - y^2}{(x^2+y^2)^2}$, so $\Delta u = 0$ on $\mathbb{C}\setminus\{0\}$. The recipe gives $v = \arctan(y/x) + C$ locally — a branch of $\arg z$ — and $u + iv$ is a branch of $\log z$. No branch is continuous all the way around $0$.

### Example 6 — flow around a cylinder

$\Omega = U(z + 1/z)$. Velocity: $\overline{\Omega'} = U\,\overline{(1 - z^{-2})}$. Far away ($\lvert z \rvert \to \infty$) it tends to $U$, uniform flow. On the cylinder $z = e^{i\theta}$: $\overline{\Omega'} = U(1 - e^{2i\theta})$, with speed $2U\lvert\sin\theta\rvert$: zero at the **stagnation points** $\theta = 0, \pi$ (front and back) and maximal, $2U$, at the top and bottom. These are the critical points $z = \pm 1$ of the Joukowski map.

### Example 7 — the mean value of $x^2 - y^2$ on a circle

On $z_0 + re^{i\theta}$ with $z_0 = x_0 + iy_0$: $u = (x_0 + r\cos\theta)^2 - (y_0 + r\sin\theta)^2$. Averaging over $\theta$, the linear terms $\cos\theta$, $\sin\theta$ average to $0$, and $\cos^2\theta$, $\sin^2\theta$ both average to $\tfrac12$: mean $= x_0^2 - y_0^2 + \tfrac{r^2}{2} - \tfrac{r^2}{2} = u(z_0)$ ✓. For the non-harmonic $x^2 + y^2$ the same computation gives $x_0^2 + y_0^2 + r^2 \neq u(z_0)$.

## 4. Python Examples

### 4.1 Testing harmonicity and the real/imaginary parts of holomorphic functions

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
z = sp.symbols("z")


def laplacian(u):
    return sp.simplify(sp.diff(u, x, 2) + sp.diff(u, y, 2))


for f in [z**2, z**3, sp.exp(z), sp.sin(z), 1 / z, z + 1 / z]:
    expression = sp.expand_complex(f.subs(z, x + sp.I * y))
    u, v = sp.simplify(sp.re(expression)), sp.simplify(sp.im(expression))
    print(f"f = {str(f):8}  Laplacian(Re f) = {laplacian(u)},  Laplacian(Im f) = {laplacian(v)}")

print()
for u in [x**2 + y**2, x**2, x * y, sp.log(x**2 + y**2), sp.exp(x) * sp.sin(y), sp.exp(x) * sp.sin(x)]:
    print(f"u = {str(u):20} Laplacian = {laplacian(u)}")
```

### 4.2 Constructing a harmonic conjugate

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
z = sp.symbols("z")


def harmonic_conjugate(u):
    if sp.simplify(sp.diff(u, x, 2) + sp.diff(u, y, 2)) != 0:
        raise ValueError("u is not harmonic, so it has no harmonic conjugate")
    v_partial = sp.integrate(sp.diff(u, x), y)                    # v = int u_x dy + phi(x)
    phi_prime = sp.simplify(-sp.diff(u, y) - sp.diff(v_partial, x))
    assert not phi_prime.has(y), "phi' must not depend on y"
    return sp.simplify(v_partial + sp.integrate(phi_prime, x))


for u in [x**2 - y**2, sp.exp(x) * sp.cos(y), x**3 - 3 * x * y**2 + 2 * y, x / (x**2 + y**2), sp.sin(x) * sp.cosh(y)]:
    v = harmonic_conjugate(u)
    f_xy = u + sp.I * v
    as_function_of_z = sp.simplify(f_xy.subs({x: z, y: 0}))
    check = sp.simplify(sp.expand_complex(as_function_of_z.subs(z, x + sp.I * y) - f_xy))
    print(f"u = {str(u):24} v = {str(v):26} u + iv = {str(as_function_of_z):12} (check: {check})")

try:
    harmonic_conjugate(x**2 + y**2)
except ValueError as error:
    print("u = x^2 + y^2:", error)
```

### 4.3 Orthogonality of the gradients

```python
import sympy as sp

x, y = sp.symbols("x y", real=True)
z = sp.symbols("z")

for f in [z**2, sp.exp(z), z + 1 / z]:
    expression = sp.expand_complex(f.subs(z, x + sp.I * y))
    u, v = sp.re(expression), sp.im(expression)
    grad_u = sp.Matrix([sp.diff(u, x), sp.diff(u, y)])
    grad_v = sp.Matrix([sp.diff(v, x), sp.diff(v, y)])
    derivative = sp.diff(f, z).subs(z, x + sp.I * y)
    print(f"f = {str(f):8}  grad u . grad v = {sp.simplify(grad_u.dot(grad_v))}   "
          f"|grad u|^2 - |f'|^2 = {sp.simplify(grad_u.dot(grad_u) - sp.expand_complex(derivative * sp.conjugate(derivative)))}")
```

### 4.4 The mean-value property, numerically

Average $u$ over circles of several radii around $z_0 = 0.3 + 0.2i$ with the trapezoidal rule and compare with $u(z_0)$. For harmonic $u$ the average does not depend on $r$; for $x^2 + y^2$ it grows like $r^2$.

```python
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

z0 = 0.3 + 0.2j
theta = np.linspace(0, 2 * np.pi, 2001)

functions = {
    "Re z^2 = x^2 - y^2": lambda z: (z**2).real,
    "Re e^z": lambda z: np.exp(z).real,
    "Im(z^3 - 2iz)": lambda z: (z**3 - 2j * z).imag,
    "ln|z - 2|": lambda z: np.log(np.abs(z - 2)),
    "x^2 + y^2 (not harmonic)": lambda z: np.abs(z) ** 2,
}
for name, u in functions.items():
    averages = []
    for r in [0.1, 0.5, 1.0]:
        circle = z0 + r * np.exp(1j * theta)
        averages.append(np.trapezoid(u(circle), theta) / (2 * np.pi))
    print(f"{name:26} u(z0) = {u(z0):+.8f}   averages r = 0.1, 0.5, 1.0: " + ", ".join(f"{a:+.8f}" for a in averages))
```

$\ln\lvert z - 2 \rvert$ is harmonic only away from $z = 2$; the circle of radius $1$ around $z_0$ stays clear of $2$ (distance $\lvert 2 - z_0 \rvert \approx 1.71$), so the property still holds. Try $r = 2$ and it fails.

### 4.5 A harmonic function is the average of its neighbours: relaxation

The discrete version of the mean-value property is "each grid value equals the average of its four neighbours". Repeatedly replacing every interior value by that average (**Jacobi relaxation**) converges to the harmonic function with the given boundary values. With boundary values taken from $x^2 - y^2$ on the square $[-1, 1]^2$, the relaxed grid should reproduce $x^2 - y^2$ inside.

```python
import numpy as np

n = 41
grid = np.linspace(-1, 1, n)
X, Y = np.meshgrid(grid, grid)
exact = X**2 - Y**2

u = np.zeros_like(exact)
u[0, :], u[-1, :], u[:, 0], u[:, -1] = exact[0, :], exact[-1, :], exact[:, 0], exact[:, -1]

for iteration in range(1, 4001):
    u[1:-1, 1:-1] = 0.25 * (u[:-2, 1:-1] + u[2:, 1:-1] + u[1:-1, :-2] + u[1:-1, 2:])
    if iteration in (10, 100, 1000, 4000):
        print(f"after {iteration:>4} sweeps: max |u - (x^2 - y^2)| = {np.max(np.abs(u - exact)):.2e}")
```

(Five-point averaging is exact for quadratics, so the only error left is from stopping the iteration early.)

They cross at right angles everywhere except at critical points ($z = 0$ for $z^2$); for $\operatorname{Log}$ the $v$-curves are rays and the $u$-curves are circles, and the orange contours bunch up along the branch cut where $\operatorname{Arg}$ jumps.

### 5.2 A harmonic surface has no bumps

$u = x^2 - y^2$ is a saddle: wherever it curves up in one direction it curves down by the same amount in the perpendicular one. Its maximum and minimum over the disc $\lvert z \rvert \le 1$ are on the boundary circle (the maximum principle). Compare $x^2 + y^2$, a bowl with an interior minimum.

```python
import matplotlib.pyplot as plt
import numpy as np

radius = np.linspace(0, 1, 40)
angle = np.linspace(0, 2 * np.pi, 80)
R, T = np.meshgrid(radius, angle)
X, Y = R * np.cos(T), R * np.sin(T)

fig = plt.figure(figsize=(14, 6))
for k, (values, title) in enumerate([(X**2 - Y**2, r"harmonic $x^2 - y^2$: extremes on the boundary"),
                                     (X**2 + Y**2, r"not harmonic $x^2 + y^2$: interior minimum")]):
    ax = fig.add_subplot(1, 2, k + 1, projection="3d")
    ax.plot_surface(X, Y, values, cmap="coolwarm", linewidth=0, alpha=0.9)
    ax.plot(np.cos(angle), np.sin(angle), values[:, -1], color="black", linewidth=2)
    ax.set_title(title)
    ax.set_xlabel("x (Re)")
    ax.set_ylabel("y (Im)")
    ax.set_zlabel("u")
plt.tight_layout()
plt.show()
```

### 5.3 Flow around a cylinder

The complex potential $\Omega = U(z + 1/z)$ with $U = 1$: streamlines are contours of $\psi = \operatorname{Im} \Omega$ (black), equipotentials are contours of $\varphi = \operatorname{Re}\Omega$ (dashed blue), and the colour shows the speed $\lvert \Omega'(z) \rvert$. The fluid stagnates at $z = \pm 1$ and is fastest ($2U$) at $\pm i$.

```python
import matplotlib.pyplot as plt
import numpy as np

grid = np.linspace(-3, 3, 501)
X, Y = np.meshgrid(grid, grid)
Z = X + 1j * Y
inside = np.abs(Z) < 1
Z_safe = np.where(inside, np.nan, Z)

with np.errstate(invalid="ignore"):   # the NaN-masked disc divides NaN by NaN
    potential = Z_safe + 1 / Z_safe
    velocity = np.conj(1 - 1 / Z_safe**2)
speed = np.abs(velocity)

fig, ax = plt.subplots(figsize=(10, 8))
image = ax.imshow(speed, extent=[-3, 3, -3, 3], origin="lower", cmap="viridis", vmin=0, vmax=2)
ax.contour(X, Y, potential.imag, levels=np.linspace(-3, 3, 25), colors="black", linewidths=1, linestyles="solid")
ax.contour(X, Y, potential.real, levels=np.linspace(-6, 6, 25), colors="lightskyblue", linewidths=0.8, linestyles="dashed")
cylinder = np.exp(1j * np.linspace(0, 2 * np.pi, 200))
ax.fill(cylinder.real, cylinder.imag, color="lightgray", zorder=3)
ax.plot([-1, 1], [0, 0], "ro", zorder=4, label="stagnation points $z = \\pm 1$")
fig.colorbar(image, ax=ax, label=r"speed $|\Omega'(z)|$")
ax.set_aspect("equal")
ax.set_title(r"Flow past a cylinder: $\Omega = z + 1/z$ (streamlines black, equipotentials dashed)")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.legend(loc="lower left")
plt.show()
```

### 5.4 Sources, vortices and electrostatics

Left: a source at $-1$ and a sink at $+1$, $\Omega = \operatorname{Log}(z + 1) - \operatorname{Log}(z - 1)$; read electrostatically, the equipotentials (blue) are circles around two opposite line charges and the field lines (black) are circular arcs from one to the other. Middle: a vortex $\Omega = -i \operatorname{Log} z$, whose streamlines are circles. Right: a source plus a vortex, $\Omega = (1 - i)\operatorname{Log} z$, with spiral streamlines. Arrows show the velocity $\overline{\Omega'}$.

```python
import matplotlib.pyplot as plt
import numpy as np

grid = np.linspace(-2.5, 2.5, 400)      # an even number of points keeps 0 and +-1 off the grid
X, Y = np.meshgrid(grid, grid)
Z = X + 1j * Y

cases = [
    (lambda z: np.log(z + 1) - np.log(z - 1), lambda z: 1 / (z + 1) - 1 / (z - 1), "source at -1, sink at +1"),
    (lambda z: -1j * np.log(z), lambda z: -1j / z, "vortex at 0"),
    (lambda z: (1 - 1j) * np.log(z), lambda z: (1 - 1j) / z, "source + vortex: spirals"),
]
fig, axes = plt.subplots(1, 3, figsize=(19, 6.5))
for ax, (omega, omega_prime, title) in zip(axes, cases):
    values = omega(Z)
    velocity = np.conj(omega_prime(Z))
    # Arg jumps by 2 pi across the branch cut of Log. Contours of sin(value - shift) draw the curves
    # value = shift (mod pi), which do not care about that jump.
    for shift in np.linspace(0, np.pi, 8, endpoint=False):
        ax.contour(X, Y, np.sin(values.real - shift), levels=[0], colors="tab:blue", linewidths=0.9, linestyles="solid")
        ax.contour(X, Y, np.sin(values.imag - shift), levels=[0], colors="black", linewidths=1, linestyles="solid")
    step = 25
    ax.quiver(X[::step, ::step], Y[::step, ::step],
              (velocity / (np.abs(velocity) + 1e-12)).real[::step, ::step],
              (velocity / (np.abs(velocity) + 1e-12)).imag[::step, ::step], color="tab:red", scale=25, width=0.004)
    ax.set_aspect("equal")
    ax.set_xlim(-2.5, 2.5)
    ax.set_ylim(-2.5, 2.5)
    ax.set_title(title)
    ax.set_xlabel("Re")
    ax.set_ylabel("Im")
plt.tight_layout()
plt.show()
```

Blue curves are equipotentials ($\varphi = $ const), black curves are streamlines ($\psi = $ const), and they cross at right angles. Using contours of $\sin(\psi - s)$ instead of $\psi$ (and likewise for $\varphi$) draws the curves $\psi \equiv s \pmod \pi$, which hides the artificial $2\pi$ jump of $\operatorname{Arg}$ along the branch cut — the multivalued potentials of §2.3 in practice (for the vortex it is $\varphi = \operatorname{Arg} z$ that is multivalued).

### 5.5 The mean-value property on a circle

Plot $u = \operatorname{Re} e^z$ along the circle $\lvert z - z_0 \rvert = 1$ against $\theta$: the curve goes above and below $u(z_0)$ and the signed areas cancel exactly.

```python
import matplotlib.pyplot as plt
import numpy as np

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

z0 = 0.3 + 0.2j
theta = np.linspace(0, 2 * np.pi, 600)
circle = z0 + np.exp(1j * theta)

fig, (left, right) = plt.subplots(1, 2, figsize=(14, 5.5))
grid = np.linspace(-1.2, 1.8, 301)
X, Y = np.meshgrid(grid, grid - 0.3)
image = left.contourf(X, Y, np.exp(X + 1j * Y).real, levels=30, cmap="coolwarm")
left.plot(circle.real, circle.imag, color="black", linewidth=2)
left.plot(z0.real, z0.imag, "ko")
left.annotate("$z_0$", (z0.real, z0.imag), xytext=(6, 6), textcoords="offset points")
fig.colorbar(image, ax=left, label=r"$u = \mathrm{Re}\, e^z$")
left.set_aspect("equal")
left.set_title(r"$u = \mathrm{Re}\, e^z$ and the circle $|z - z_0| = 1$")
left.set_xlabel("Re")
left.set_ylabel("Im")

values = np.exp(circle).real
centre_value = np.exp(z0).real
right.plot(theta, values, label=r"$u(z_0 + e^{i\theta})$")
right.axhline(centre_value, color="black", linestyle="--", label=rf"$u(z_0) = {centre_value:.4f}$")
right.fill_between(theta, values, centre_value, where=values >= centre_value, color="tab:red", alpha=0.3)
right.fill_between(theta, values, centre_value, where=values < centre_value, color="tab:blue", alpha=0.3)
average = np.trapezoid(values, theta) / (2 * np.pi)
right.set_title(f"average over the circle = {average:.4f}")
right.set_xlabel(r"$\theta$")
right.set_ylabel("u")
right.grid(True, alpha=0.3)
right.legend()
plt.tight_layout()
plt.show()
```

## 7. Solutions

1. (a) yes, linear. (b) $\Delta = 2 - 2 + 0 = 0$, yes. (c) $\Delta = 6x + 6y$, no. (d) $u_{xx} = -e^{-y}\sin x$, $u_{yy} = e^{-y}\sin x$, yes (it is $\operatorname{Im} e^{iz}$).
2. $u_x = 2 - 2y$, $u_y = -2x$. $v_y = 2 - 2y \Rightarrow v = 2y - y^2 + \phi(x)$; $v_x = \phi'(x) = -u_y = 2x \Rightarrow \phi = x^2$. So $v = x^2 - y^2 + 2y$, and $f = 2x - 2xy + i(x^2 - y^2 + 2y) = iz^2 + 2z$. (Check: $iz^2 = -2xy + i(x^2 - y^2)$, $2z = 2x + 2iy$ ✓.)
3. $-if = -i(u + iv) = v - iu$ is holomorphic, with real part $v$ and imaginary part $-u$.
4. $u = \frac{y}{x^2+y^2} = -\operatorname{Im}\frac{1}{z}$, and $\frac{i}{z} = \frac{i(x - iy)}{x^2+y^2} = \frac{y + ix}{x^2 + y^2}$. So $f = \frac{i}{z}$, $v = \frac{x}{x^2+y^2}$ (plus a constant).
5. $u = \frac{x}{x^2 + y^2} = c$ gives $\left(x - \frac{1}{2c}\right)^2 + y^2 = \frac{1}{4c^2}$: circles through $0$ centred on the real axis. $v = \frac{-y}{x^2 + y^2} = c$ gives circles through $0$ centred on the imaginary axis. Two families of circles all tangent at $0$ (the two "bundles"), crossing at right angles — a picture of the electric dipole. Plot below.
6. $e^{\cos\theta}\cos(\sin\theta) = \operatorname{Re} e^{e^{i\theta}}$, the harmonic $\operatorname{Re} e^z$ averaged over the unit circle around $0$. By the mean-value property the answer is $\operatorname{Re} e^0 = 1$.
7. $\Omega' = U(1 - a^2/z^2)$; on $z = ae^{i\theta}$, $\Omega' = U(1 - e^{-2i\theta})$, so $\lvert \Omega' \rvert = U\lvert e^{i\theta} - e^{-i\theta}\rvert = 2U\lvert\sin\theta\rvert$. Stagnation where $\Omega' = 0$: $z = \pm a$.
8. Near a point, $u = \operatorname{Re} f$ for some holomorphic $f$ (harmonic conjugate on a small disc). Then $u \circ g = \operatorname{Re}(f \circ g)$, and $f \circ g$ is holomorphic, so $u \circ g$ is harmonic. To solve Laplace's equation on a complicated domain $D$, find a conformal $g$ from $D$ onto a disc or half-plane where the solution $u$ is known; then $u \circ g$ solves it on $D$ (boundary values are carried along).
9. $T = \frac{1}{\pi}\operatorname{Arg} z = \frac{1}{\pi}\operatorname{Im}\operatorname{Log} z$ is harmonic in the upper half-plane (imaginary part of a holomorphic function, away from the branch cut on the negative real axis, which is on the boundary). On $x > 0$: $\operatorname{Arg} = 0$, $T = 0$. On $x < 0$ (approached from above): $\operatorname{Arg} \to \pi$, $T \to 1$ ✓. Isotherms are rays from $0$.
10. In polar coordinates around $z_0$: $\frac{1}{\pi R^2}\int_0^R \int_0^{2\pi} u\, r\, d\theta\, dr = \frac{1}{\pi R^2}\int_0^R 2\pi r\, u(z_0)\, dr = u(z_0)$, using the circle mean value for each $r$. Numerically below.

Numerical checks:

```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

x, y = sp.symbols("x y", real=True)
for name, u in {"a": 3 * x - 5 * y + 2, "b": x**2 - y**2 + x * y, "c": x**3 + y**3, "d": sp.exp(-y) * sp.sin(x)}.items():
    print(f"Ex 1{name}: Laplacian =", sp.simplify(sp.diff(u, x, 2) + sp.diff(u, y, 2)))

z_xy = x + sp.I * y
print("Ex 2: i z^2 + 2z =", sp.expand(sp.expand_complex(sp.I * z_xy**2 + 2 * z_xy)))
i_over_z = sp.expand_complex(sp.I / z_xy)
print("Ex 4: Re(i/z) =", sp.simplify(sp.re(i_over_z)), "  Im(i/z) =", sp.simplify(sp.im(i_over_z)))

theta = np.linspace(0, 2 * np.pi, 4001)
print("Ex 6:", np.trapezoid(np.exp(np.cos(theta)) * np.cos(np.sin(theta)), theta) / (2 * np.pi))

on_cylinder = 2 * np.exp(1j * theta[::500])
U, a = 1.5, 2.0
speed = np.abs(U * (1 - a**2 / on_cylinder**2))
print("Ex 7: speed equals 2U|sin(theta)| on the cylinder:", np.allclose(speed, 2 * U * np.abs(np.sin(theta[::500]))))

T = sp.atan2(y, x) / sp.pi
print("Ex 9: Laplacian of Arg/pi =", sp.simplify(sp.diff(T, x, 2) + sp.diff(T, y, 2)),
      "  T(1, 0+) =", T.subs({x: 1, y: 0}), "  T(-1, 0+) =", sp.limit(T.subs(x, -1), y, 0, "+"))

z0, R = 0.5 - 0.2j, 1.0
radii = np.linspace(0, R, 801)
angles = np.linspace(0, 2 * np.pi, 801)
RR, AA = np.meshgrid(radii, angles)
points = z0 + RR * np.exp(1j * AA)
u_values = points.real**3 - 3 * points.real * points.imag**2
disc_average = np.trapezoid(np.trapezoid(u_values * RR, angles, axis=0), radii) / (np.pi * R**2)
print(f"Ex 10: disc average = {disc_average:.6f}   u(z0) = {z0.real**3 - 3 * z0.real * z0.imag**2:.6f}")

grid = np.linspace(-2, 2, 401)
X, Y = np.meshgrid(grid, grid)
with np.errstate(divide="ignore", invalid="ignore"):
    W = 1 / (X + 1j * Y)
fig, ax = plt.subplots(figsize=(7, 7))
levels = [-3, -2, -1, -0.5, -0.3, 0.3, 0.5, 1, 2, 3]
ax.contour(X, Y, W.real, levels=levels, colors="tab:blue", linewidths=1, linestyles="solid")
ax.contour(X, Y, W.imag, levels=levels, colors="tab:orange", linewidths=1, linestyles="solid")
ax.axhline(0, color="black", linewidth=0.5)
ax.axvline(0, color="black", linewidth=0.5)
ax.set_aspect("equal")
ax.set_title(r"Ex 5: level curves of Re$(1/z)$ (blue) and Im$(1/z)$ (orange)")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
plt.show()
```

## 8. Conclusion

- $u$ is harmonic when $\Delta u = u_{xx} + u_{yy} = 0$. The real and imaginary parts of every holomorphic function are harmonic — a two-line consequence of the Cauchy–Riemann equations.
- Every harmonic $u$ on a disc has a **harmonic conjugate** $v$ (unique up to a constant) with $u + iv$ holomorphic, found by integrating $v_y = u_x$, $v_x = -u_y$. On domains with holes it can be multivalued: $\ln\lvert z \rvert$ has conjugate $\arg z$.
- The level curves of $u$ and $v$ are orthogonal wherever $f' \neq 0$, with $\lvert \nabla u \rvert = \lvert \nabla v \rvert = \lvert f' \rvert$.
- A **complex potential** $\Omega = \varphi + i\psi$ describes ideal 2-D flow (velocity $\overline{\Omega'}$, streamlines $\psi = c$) and electrostatics (equipotentials $\varphi = c$, field lines $\psi = c$): uniform flow, sources, vortices and the cylinder $z + 1/z$.
- Harmonic functions satisfy the **mean-value property** and the **maximum principle**: no interior bumps, extremes on the boundary, and each value is the average around it — which is also why relaxation solves Laplace's equation numerically.
